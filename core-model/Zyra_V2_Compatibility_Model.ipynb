```python
import torch
import torch.nn as nn


class ZyraV2Compatibility(nn.Module):
    """
    Zyra V2 — User × Product Compatibility Model

    Inputs:
        user_vector:    [batch, 662]
        product_vector: [batch, 662]

    Output:
        suitability_score: [batch, 1]
    """

    def __init__(self, input_dim=662, hidden_dim=512, num_heads=8):
        super().__init__()

        # Convert existing Zyra representations
        # into a shared compatibility space.
        self.user_projection = nn.Sequential(
            nn.Linear(input_dim, hidden_dim),
            nn.GELU(),
            nn.LayerNorm(hidden_dim),
        )

        self.product_projection = nn.Sequential(
            nn.Linear(input_dim, hidden_dim),
            nn.GELU(),
            nn.LayerNorm(hidden_dim),
        )

        # Learn user ↔ product interactions.
        self.cross_attention = nn.MultiheadAttention(
            embed_dim=hidden_dim,
            num_heads=num_heads,
            batch_first=True,
        )

        # Learn final compatibility.
        self.compatibility_head = nn.Sequential(
            nn.Linear(hidden_dim, 256),
            nn.GELU(),
            nn.Dropout(0.2),

            nn.Linear(256, 64),
            nn.GELU(),

            nn.Linear(64, 1),
            nn.Sigmoid(),
        )

    def forward(self, user_vector, product_vector):

        user = self.user_projection(user_vector)
        product = self.product_projection(product_vector)

        # [B, 512] → [B, 1, 512]
        user = user.unsqueeze(1)
        product = product.unsqueeze(1)

        # Product representation attends to
        # relevant information from the user.
        fused, attention_weights = self.cross_attention(
            query=product,
            key=user,
            value=user,
        )

        fused = fused.squeeze(1)

        score = self.compatibility_head(fused)

        return score, attention_weights
```

```python
model = ZyraV2Compatibility().to(device)

print(model)
```

```python
batch_size = 4

test_users = torch.randn(
    batch_size, 662, device=device
)

test_products = torch.randn(
    batch_size, 662, device=device
)

with torch.no_grad():
    scores, attention = model(
        test_users,
        test_products
    )

print("User input shape    :", test_users.shape)
print("Product input shape :", test_products.shape)
print("Score shape         :", scores.shape)
print("Attention shape     :", attention.shape)
print("Device              :", scores.device)
print("Scores              :", scores.squeeze(-1).cpu().numpy())
```

```python
import json
import random
import re
from pathlib import Path
from collections import defaultdict

import pandas as pd
import numpy as np

# ============================================================
# VIBE DATASET
# ============================================================

VIBE_ROOT = Path.home() / "Desktop/Projects/weavly/core-model/VIBE/dataset"

HUMANS_FILE = VIBE_ROOT / "humans/model.json"
MODEL_MEASUREMENT_DIR = VIBE_ROOT / "tops/model_measurement"
GARMENT_MEASUREMENT_DIR = VIBE_ROOT / "tops/garment_measurement"

# ------------------------------------------------------------
# Load human profiles
# ------------------------------------------------------------

with open(HUMANS_FILE) as f:
    humans = json.load(f)

print("Human profiles:", len(humans))


# ------------------------------------------------------------
# Helper: convert "92cm" -> 92.0
# ------------------------------------------------------------

def parse_measurement(value):
    if value is None:
        return None

    match = re.search(r"[-+]?\d*\.?\d+", str(value))

    if match:
        return float(match.group())

    return None


# ------------------------------------------------------------
# Load garment measurements
# ------------------------------------------------------------

garment_measurements = {}

for file in GARMENT_MEASUREMENT_DIR.glob("*.txt"):

    with open(file) as f:
        data = json.load(f)

    if not data:
        continue

    product_id = file.stem.replace("_meta", "")

    measurements = {}

    # JSON structure:
    # {"12": [{"dimension_name": "...", "value": "..."}]}

    for entries in data.values():

        for item in entries:

            name = item["dimension_name"].lower()
            value = parse_measurement(item["value"])

            if value is None:
                continue

            if "bust" in name:
                measurements["garment_bust"] = value

            elif name == "waist":
                measurements["garment_waist"] = value

            elif name == "hip":
                measurements["garment_hip"] = value

            elif "shoulder to hem" in name:
                measurements["garment_length"] = value

            elif "sleeve length" in name:
                measurements["garment_sleeve"] = value

    garment_measurements[product_id] = measurements


print("Garments with measurements:", len(garment_measurements))


# ------------------------------------------------------------
# Build POSITIVE person × garment pairs
# ------------------------------------------------------------

positive_pairs = []

for file in MODEL_MEASUREMENT_DIR.glob("*.txt"):

    with open(file) as f:
        data = json.load(f)

    product_id = file.stem.replace("_meta", "")

    # Skip garments without measurement information
    if product_id not in garment_measurements:
        continue

    for person_id in data:

        # Only users with a valid human profile
        if person_id not in humans:
            continue

        human = humans[person_id]
        garment = garment_measurements[product_id]

        positive_pairs.append({
            "person_id": person_id,
            "product_id": product_id,

            "height": parse_measurement(human.get("height")),
            "bust": parse_measurement(human.get("bust")),
            "waist": parse_measurement(human.get("waist")),
            "hips": parse_measurement(human.get("hips")),
            "model_size": data[person_id].get("size"),

            "garment_bust": garment.get("garment_bust"),
            "garment_waist": garment.get("garment_waist"),
            "garment_hip": garment.get("garment_hip"),
            "garment_length": garment.get("garment_length"),
            "garment_sleeve": garment.get("garment_sleeve"),

            "label": 1
        })


positive_df = pd.DataFrame(positive_pairs)

print("\nPositive compatibility dataset")
print("--------------------------------")
print("Rows:", len(positive_df))
print("Columns:", len(positive_df.columns))

display(positive_df.head())
```

```python
import json
import re
from pathlib import Path
import pandas as pd

VIBE_ROOT = Path.home() / "Desktop/Projects/weavly/core-model/VIBE/dataset"

HUMANS_FILE = VIBE_ROOT / "humans/model.json"
MODEL_MEASUREMENT_DIR = VIBE_ROOT / "tops/model_measurement"
GARMENT_MEASUREMENT_DIR = VIBE_ROOT / "tops/garment_measurement"


# ============================================================
# LOAD HUMANS
# ============================================================

with open(HUMANS_FILE) as f:
    humans = json.load(f)

print("Human profiles:", len(humans))


# ============================================================
# MEASUREMENT PARSER
# ============================================================

def parse_measurement(value):
    if value is None:
        return None

    match = re.search(r"[-+]?\d*\.?\d+", str(value))

    return float(match.group()) if match else None


# ============================================================
# LOAD GARMENT MEASUREMENTS
# ============================================================

garment_measurements = {}
invalid_files = []

for file in GARMENT_MEASUREMENT_DIR.glob("*.txt"):

    try:
        with open(file) as f:
            data = json.load(f)

        # Some files contain JSON encoded as a string
        if isinstance(data, str):
            data = json.loads(data)

        if not isinstance(data, dict):
            invalid_files.append(file.name)
            continue

        product_id = file.stem.replace("_meta", "")

        measurements = {}

        for entries in data.values():

            if not isinstance(entries, list):
                continue

            for item in entries:

                if not isinstance(item, dict):
                    continue

                name = str(item.get("dimension_name", "")).lower()
                value = parse_measurement(item.get("value"))

                if value is None:
                    continue

                if "bust" in name:
                    measurements["garment_bust"] = value

                elif name == "waist":
                    measurements["garment_waist"] = value

                elif "hip" in name:
                    measurements["garment_hip"] = value

                elif "shoulder to hem" in name:
                    measurements["garment_length"] = value

                elif "sleeve length" in name:
                    measurements["garment_sleeve"] = value

        if measurements:
            garment_measurements[product_id] = measurements

    except Exception:
        invalid_files.append(file.name)


print("Garments with measurements:", len(garment_measurements))
print("Invalid files:", len(invalid_files))


# ============================================================
# BUILD POSITIVE PAIRS
# ============================================================

positive_pairs = []

for file in MODEL_MEASUREMENT_DIR.glob("*.txt"):

    try:
        with open(file) as f:
            data = json.load(f)

        if isinstance(data, str):
            data = json.loads(data)

        if not isinstance(data, dict):
            continue

        product_id = file.stem.replace("_meta", "")

        # Need garment measurements
        if product_id not in garment_measurements:
            continue

        garment = garment_measurements[product_id]

        for person_id, pair_data in data.items():

            # Must have human profile
            if person_id not in humans:
                continue

            human = humans[person_id]

            positive_pairs.append({

                "person_id": person_id,
                "product_id": product_id,

                # Human measurements
                "height": parse_measurement(human.get("height")),
                "bust": parse_measurement(human.get("bust")),
                "waist": parse_measurement(human.get("waist")),
                "hips": parse_measurement(human.get("hips")),

                # Size worn by model
                "model_size": (
                    pair_data.get("size")
                    if isinstance(pair_data, dict)
                    else None
                ),

                # Garment measurements
                "garment_bust": garment.get("garment_bust"),
                "garment_waist": garment.get("garment_waist"),
                "garment_hip": garment.get("garment_hip"),
                "garment_length": garment.get("garment_length"),
                "garment_sleeve": garment.get("garment_sleeve"),

                "label": 1
            })

positive_df = pd.DataFrame(positive_pairs)


# ============================================================
# SUMMARY
# ============================================================

print("\n================================")
print("VIBE POSITIVE DATASET")
print("================================")

print("Rows    :", len(positive_df))
print("Columns :", len(positive_df.columns))

display(positive_df.head(10))
```

```python
import json
import re
from pathlib import Path
import pandas as pd

VIBE_ROOT = Path.home() / "Desktop/Projects/weavly/core-model/VIBE/dataset"

HUMANS_FILE = VIBE_ROOT / "humans/model.json"
MODEL_DIR = VIBE_ROOT / "tops/model_measurement"
GARMENT_DIR = VIBE_ROOT / "tops/garment_measurement"


def parse_measurement(value):
    if value is None:
        return None

    match = re.search(r"[-+]?\d*\.?\d+", str(value))
    return float(match.group()) if match else None


def load_json_file(path):
    """Safely load a VIBE JSON/TXT file."""
    try:
        with open(path, "r", encoding="utf-8") as f:
            raw = f.read().strip()

        if not raw:
            return None

        data = json.loads(raw)

        # Some files contain JSON encoded as a string
        if isinstance(data, str):
            try:
                data = json.loads(data)
            except json.JSONDecodeError:
                return None

        return data

    except (json.JSONDecodeError, UnicodeDecodeError, OSError):
        return None


# ============================================================
# HUMANS
# ============================================================

humans = load_json_file(HUMANS_FILE)

print("Human profiles:", len(humans))


# ============================================================
# GARMENT MEASUREMENTS
# ============================================================

garment_measurements = {}
invalid_garments = []

for file in GARMENT_DIR.glob("*.txt"):

    data = load_json_file(file)

    if not isinstance(data, dict):
        invalid_garments.append(file.name)
        continue

    product_id = file.stem.replace("_meta", "")

    measurements = {}

    for entries in data.values():

        if not isinstance(entries, list):
            continue

        for item in entries:

            if not isinstance(item, dict):
                continue

            name = str(item.get("dimension_name", "")).lower()
            value = parse_measurement(item.get("value"))

            if value is None:
                continue

            if "bust" in name:
                measurements["garment_bust"] = value

            elif name == "waist":
                measurements["garment_waist"] = value

            elif "hip" in name:
                measurements["garment_hip"] = value

            elif "shoulder to hem" in name:
                measurements["garment_length"] = value

            elif "sleeve length" in name:
                measurements["garment_sleeve"] = value

    if measurements:
        garment_measurements[product_id] = measurements


print("Garments with measurements:", len(garment_measurements))
print("Invalid/empty garment files:", len(invalid_garments))


# ============================================================
# POSITIVE PAIRS
# ============================================================

positive_pairs = []
invalid_model_files = []

for file in MODEL_DIR.glob("*.txt"):

    data = load_json_file(file)

    if not isinstance(data, dict):
        invalid_model_files.append(file.name)
        continue

    product_id = file.stem.replace("_meta", "")

    if product_id not in garment_measurements:
        continue

    garment = garment_measurements[product_id]

    for person_id, pair_data in data.items():

        if person_id not in humans:
            continue

        human = humans[person_id]

        positive_pairs.append({
            "person_id": person_id,
            "product_id": product_id,

            "height": parse_measurement(human.get("height")),
            "bust": parse_measurement(human.get("bust")),
            "waist": parse_measurement(human.get("waist")),
            "hips": parse_measurement(human.get("hips")),

            "model_size": (
                pair_data.get("size")
                if isinstance(pair_data, dict)
                else None
            ),

            "garment_bust": garment.get("garment_bust"),
            "garment_waist": garment.get("garment_waist"),
            "garment_hip": garment.get("garment_hip"),
            "garment_length": garment.get("garment_length"),
            "garment_sleeve": garment.get("garment_sleeve"),

            "label": 1
        })


positive_df = pd.DataFrame(positive_pairs)


# ============================================================
# SUMMARY
# ============================================================

print("\n================================")
print("VIBE POSITIVE DATASET")
print("================================")

print("Positive pairs :", len(positive_df))
print("Columns        :", len(positive_df.columns))
print("Invalid garment files:", len(invalid_garments))
print("Invalid model files  :", len(invalid_model_files))

display(positive_df.head(10))
```

```python
print("Missing values:")
display(positive_df.isna().sum())

print("\nMissing percentage:")
display(
    (positive_df.isna().mean() * 100)
    .round(2)
    .sort_values(ascending=False)
)

print("\nUnique people:", positive_df["person_id"].nunique())
print("Unique garments:", positive_df["product_id"].nunique())
```

```python
fit_df = positive_df.copy()

# Body ↔ garment differences
fit_df["bust_ease"] = (
    fit_df["garment_bust"] - fit_df["bust"]
)

fit_df["waist_ease"] = (
    fit_df["garment_waist"] - fit_df["waist"]
)

fit_df["hip_ease"] = (
    fit_df["garment_hip"] - fit_df["hips"]
)

# Proportional length
fit_df["length_height_ratio"] = (
    fit_df["garment_length"] / fit_df["height"]
)

# Missing-value indicators
fit_df["has_garment_bust"] = fit_df["garment_bust"].notna().astype(float)
fit_df["has_garment_waist"] = fit_df["garment_waist"].notna().astype(float)
fit_df["has_garment_hip"] = fit_df["garment_hip"].notna().astype(float)
fit_df["has_garment_length"] = fit_df["garment_length"].notna().astype(float)
fit_df["has_garment_sleeve"] = fit_df["garment_sleeve"].notna().astype(float)

print("Derived compatibility features created.")

display(
    fit_df[
        [
            "person_id",
            "product_id",
            "bust_ease",
            "waist_ease",
            "hip_ease",
            "length_height_ratio",
            "label"
        ]
    ].head(10)
)
```

```python
# ============================================================
# FIND SUSPICIOUS GARMENT MEASUREMENTS
# ============================================================

measurement_cols = [
    "garment_bust",
    "garment_waist",
    "garment_hip",
    "garment_length",
    "garment_sleeve",
]

print("Suspicious garment measurements")
print("================================")

for col in measurement_cols:

    suspicious = positive_df[
        positive_df[col].notna() &
        ((positive_df[col] < 20) | (positive_df[col] > 200))
    ]

    print(f"\n{col}")
    print("Rows:", len(suspicious))

    if len(suspicious) > 0:
        display(
            suspicious[
                ["product_id", col]
            ].drop_duplicates().sort_values(col)
        )
```

```python
# ============================================================
# CLEAN RAW GARMENT MEASUREMENTS
# ============================================================

clean_df = positive_df.copy()

body_garment_cols = [
    "garment_bust",
    "garment_waist",
    "garment_hip",
]

for col in body_garment_cols:

    invalid = (
        clean_df[col].notna()
        & (
            (clean_df[col] < 20)
            | (clean_df[col] > 200)
        )
    )

    print(f"{col}: {invalid.sum()} invalid values")

    clean_df.loc[invalid, col] = np.nan


# ============================================================
# RECALCULATE DERIVED FEATURES
# ============================================================

clean_df["bust_ease"] = (
    clean_df["garment_bust"] - clean_df["bust"]
)

clean_df["waist_ease"] = (
    clean_df["garment_waist"] - clean_df["waist"]
)

clean_df["hip_ease"] = (
    clean_df["garment_hip"] - clean_df["hips"]
)

clean_df["length_height_ratio"] = (
    clean_df["garment_length"] / clean_df["height"]
)


# ============================================================
# CHECK NEW DISTRIBUTIONS
# ============================================================

print("\n=== CLEANED FEATURE STATISTICS ===")

display(
    clean_df[
        [
            "bust_ease",
            "waist_ease",
            "hip_ease",
            "length_height_ratio",
            "garment_sleeve",
        ]
    ].describe().T
)
```

```python
import random

random.seed(42)

# ============================================================
# BUILD POSITIVE LOOKUP
# ============================================================

person_positive_products = (
    clean_df
    .groupby("person_id")["product_id"]
    .apply(set)
    .to_dict()
)

all_products = set(clean_df["product_id"].unique())

print("People:", len(person_positive_products))
print("Garments:", len(all_products))


# ============================================================
# POSITIVE DATA
# ============================================================

positive_training = clean_df.copy()


# ============================================================
# BUILD NEGATIVE PAIRS
# ============================================================

negative_rows = []

for person_id, positive_products in person_positive_products.items():

    candidates = list(all_products - positive_products)

    random.shuffle(candidates)

    # Balanced 1:1 negatives
    n_negative = min(
        len(positive_products),
        len(candidates)
    )

    selected = candidates[:n_negative]

    # Get person's body measurements
    person_rows = clean_df[
        clean_df["person_id"] == person_id
    ]

    person_info = person_rows.iloc[0]

    for product_id in selected:

        # Find garment measurements
        product_row = clean_df[
            clean_df["product_id"] == product_id
        ].iloc[0]

        row = {
            "person_id": person_id,
            "product_id": product_id,

            "height": person_info["height"],
            "bust": person_info["bust"],
            "waist": person_info["waist"],
            "hips": person_info["hips"],

            "model_size": np.nan,

            "garment_bust": product_row["garment_bust"],
            "garment_waist": product_row["garment_waist"],
            "garment_hip": product_row["garment_hip"],
            "garment_length": product_row["garment_length"],
            "garment_sleeve": product_row["garment_sleeve"],

            "label": 0,
        }

        negative_rows.append(row)


negative_df = pd.DataFrame(negative_rows)


# ============================================================
# DERIVED FEATURES
# ============================================================

negative_df["bust_ease"] = (
    negative_df["garment_bust"]
    - negative_df["bust"]
)

negative_df["waist_ease"] = (
    negative_df["garment_waist"]
    - negative_df["waist"]
)

negative_df["hip_ease"] = (
    negative_df["garment_hip"]
    - negative_df["hips"]
)

negative_df["length_height_ratio"] = (
    negative_df["garment_length"]
    / negative_df["height"]
)


# ============================================================
# COMBINE
# ============================================================

training_df = pd.concat(
    [positive_training, negative_df],
    ignore_index=True
)

training_df = training_df.sample(
    frac=1,
    random_state=42
).reset_index(drop=True)


# ============================================================
# SUMMARY
# ============================================================

print("\n================================")
print("V2 COMPATIBILITY DATASET")
print("================================")

print("Positive:", (training_df["label"] == 1).sum())
print("Negative:", (training_df["label"] == 0).sum())
print("Total   :", len(training_df))

print(
    "\nUnique people:",
    training_df["person_id"].nunique()
)

print(
    "Unique garments:",
    training_df["product_id"].nunique()
)

display(training_df.head(10))
```

```python
from sklearn.model_selection import train_test_split

# ============================================================
# PERSON-LEVEL SPLIT
# ============================================================

people = training_df["person_id"].unique()

train_people, temp_people = train_test_split(
    people,
    test_size=0.30,
    random_state=42
)

val_people, test_people = train_test_split(
    temp_people,
    test_size=0.50,
    random_state=42
)

train_df = training_df[
    training_df["person_id"].isin(train_people)
].copy()

val_df = training_df[
    training_df["person_id"].isin(val_people)
].copy()

test_df = training_df[
    training_df["person_id"].isin(test_people)
].copy()


# ============================================================
# VERIFY NO PERSON LEAKAGE
# ============================================================

assert set(train_people).isdisjoint(set(val_people))
assert set(train_people).isdisjoint(set(test_people))
assert set(val_people).isdisjoint(set(test_people))


# ============================================================
# SUMMARY
# ============================================================

print("================================")
print("PERSON-LEVEL DATASET SPLIT")
print("================================")

print("\nPeople")
print("Train:", len(train_people))
print("Val  :", len(val_people))
print("Test :", len(test_people))

print("\nRows")
print("Train:", len(train_df))
print("Val  :", len(val_df))
print("Test :", len(test_df))

print("\nLabels")

print("\nTrain:")
print(train_df["label"].value_counts())

print("\nValidation:")
print(val_df["label"].value_counts())

print("\nTest:")
print(test_df["label"].value_counts())

print("\nLeakage check:")
print("Train ∩ Val :", set(train_people) & set(val_people))
print("Train ∩ Test:", set(train_people) & set(test_people))
print("Val ∩ Test  :", set(val_people) & set(test_people))
```

```python
import torch
import numpy as np

# ============================================================
# V2 EXPLICIT COMPATIBILITY FEATURES
# ============================================================

FEATURE_COLUMNS = [
    "height",
    "bust",
    "waist",
    "hips",

    "garment_bust",
    "garment_waist",
    "garment_hip",
    "garment_length",
    "garment_sleeve",

    "bust_ease",
    "waist_ease",
    "hip_ease",
    "length_height_ratio",

    "has_garment_bust",
    "has_garment_waist",
    "has_garment_hip",
    "has_garment_length",
    "has_garment_sleeve",
]


# ============================================================
# ADD MISSINGNESS FLAGS
# ============================================================

def prepare_features(df):

    result = df.copy()

    result["has_garment_bust"] = (
        result["garment_bust"].notna().astype(float)
    )

    result["has_garment_waist"] = (
        result["garment_waist"].notna().astype(float)
    )

    result["has_garment_hip"] = (
        result["garment_hip"].notna().astype(float)
    )

    result["has_garment_length"] = (
        result["garment_length"].notna().astype(float)
    )

    result["has_garment_sleeve"] = (
        result["garment_sleeve"].notna().astype(float)
    )

    return result


train_df = prepare_features(train_df)
val_df = prepare_features(val_df)
test_df = prepare_features(test_df)


# ============================================================
# CONVERT TO NUMERIC MATRICES
# ============================================================

def dataframe_to_tensor(df):

    X = (
        df[FEATURE_COLUMNS]
        .astype(float)
        .fillna(0.0)
        .to_numpy(dtype=np.float32)
    )

    y = (
        df["label"]
        .to_numpy(dtype=np.float32)
        .reshape(-1, 1)
    )

    return (
        torch.tensor(X),
        torch.tensor(y)
    )


X_train, y_train = dataframe_to_tensor(train_df)
X_val, y_val = dataframe_to_tensor(val_df)
X_test, y_test = dataframe_to_tensor(test_df)


# ============================================================
# DEVICE
# ============================================================

device = torch.device(
    "mps"
    if torch.backends.mps.is_available()
    else "cpu"
)


X_train = X_train.to(device)
y_train = y_train.to(device)

X_val = X_val.to(device)
y_val = y_val.to(device)

X_test = X_test.to(device)
y_test = y_test.to(device)


# ============================================================
# VERIFY
# ============================================================

print("================================")
print("V2 COMPATIBILITY TENSORS")
print("================================")

print("Feature dimension :", X_train.shape[1])

print("\nTrain:")
print("X:", X_train.shape)
print("y:", y_train.shape)

print("\nValidation:")
print("X:", X_val.shape)
print("y:", y_val.shape)

print("\nTest:")
print("X:", X_test.shape)
print("y:", y_test.shape)

print("\nDevice:", device)

print("\nNaN:")
print("Train:", torch.isnan(X_train).sum().item())
print("Val  :", torch.isnan(X_val).sum().item())
print("Test :", torch.isnan(X_test).sum().item())

print("\nInf:")
print("Train:", torch.isinf(X_train).sum().item())
print("Val  :", torch.isinf(X_val).sum().item())
print("Test :", torch.isinf(X_test).sum().item())
```

```python
# ============================================================
# V2 COMPATIBILITY NETWORK — V1
# ============================================================

import torch
import torch.nn as nn
import torch.optim as optim


# ------------------------------------------------------------
# Model
# ------------------------------------------------------------

class CompatibilityNetV1(nn.Module):

    def __init__(self, input_dim=18):
        super().__init__()

        self.network = nn.Sequential(
            nn.Linear(input_dim, 64),
            nn.ReLU(),

            nn.Linear(64, 32),
            nn.ReLU(),

            nn.Linear(32, 16),
            nn.ReLU(),

            nn.Linear(16, 1)
        )

    def forward(self, x):
        return self.network(x)


# ------------------------------------------------------------
# Create model
# ------------------------------------------------------------

model = CompatibilityNetV1(
    input_dim=X_train.shape[1]
).to(device)


# ------------------------------------------------------------
# Loss + optimizer
# ------------------------------------------------------------

criterion = nn.BCEWithLogitsLoss()

optimizer = optim.Adam(
    model.parameters(),
    lr=1e-3
)


# ------------------------------------------------------------
# Training configuration
# ------------------------------------------------------------

EPOCHS = 50

BATCH_SIZE = 64


# ------------------------------------------------------------
# Mini-batch training
# ------------------------------------------------------------

def train_epoch(model, X, y):

    model.train()

    permutation = torch.randperm(
        X.size(0),
        device=device
    )

    total_loss = 0.0

    for i in range(
        0,
        X.size(0),
        BATCH_SIZE
    ):

        indices = permutation[
            i:i + BATCH_SIZE
        ]

        batch_x = X[indices]
        batch_y = y[indices]

        optimizer.zero_grad()

        logits = model(batch_x)

        loss = criterion(
            logits,
            batch_y
        )

        loss.backward()

        optimizer.step()

        total_loss += (
            loss.item()
            * batch_x.size(0)
        )

    return total_loss / X.size(0)


# ------------------------------------------------------------
# Evaluation
# ------------------------------------------------------------

@torch.no_grad()
def evaluate(model, X, y):

    model.eval()

    logits = model(X)

    loss = criterion(
        logits,
        y
    )

    probabilities = torch.sigmoid(logits)

    predictions = (
        probabilities >= 0.5
    ).float()

    accuracy = (
        predictions == y
    ).float().mean()

    return (
        loss.item(),
        accuracy.item()
    )


# ============================================================
# TRAIN
# ============================================================

print("================================")
print("TRAINING V2 COMPATIBILITY MODEL")
print("================================")

print("Device:", device)
print("Input dimension:", X_train.shape[1])
print("Parameters:", sum(
    p.numel()
    for p in model.parameters()
))

for epoch in range(1, EPOCHS + 1):

    train_loss = train_epoch(
        model,
        X_train,
        y_train
    )

    val_loss, val_accuracy = evaluate(
        model,
        X_val,
        y_val
    )

    if (
        epoch == 1
        or epoch % 5 == 0
        or epoch == EPOCHS
    ):
        print(
            f"Epoch {epoch:02d}/{EPOCHS} | "
            f"Train Loss: {train_loss:.4f} | "
            f"Val Loss: {val_loss:.4f} | "
            f"Val Acc: {val_accuracy:.4f}"
        )


# ============================================================
# FINAL TEST
# ============================================================

test_loss, test_accuracy = evaluate(
    model,
    X_test,
    y_test
)

print("\n================================")
print("FINAL TEST")
print("================================")

print(f"Test Loss : {test_loss:.4f}")
print(f"Test Acc  : {test_accuracy:.4f}")
```

```python
# ============================================================
# V2 — SIMPLE COMPATIBILITY BASELINE
# ============================================================

import numpy as np
import pandas as pd
from sklearn.metrics import (
    accuracy_score,
    roc_auc_score,
    classification_report
)

print("================================")
print("SIMPLE COMPATIBILITY BASELINE")
print("================================")


# ------------------------------------------------------------
# Use the original dataframe
# ------------------------------------------------------------

baseline_df = test_df.copy()


# ------------------------------------------------------------
# Define a simple heuristic
#
# We reward:
#   - reasonable positive ease
#   - reasonable length/height ratio
#
# Missing measurements are ignored.
# ------------------------------------------------------------

def compatibility_score(row):

    scores = []

    # Bust
    if pd.notna(row["bust_ease"]):
        bust = row["bust_ease"]

        # Reasonable top ease
        if 5 <= bust <= 60:
            scores.append(1.0)
        elif -10 <= bust <= 75:
            scores.append(0.5)
        else:
            scores.append(0.0)

    # Waist
    if pd.notna(row["waist_ease"]):
        waist = row["waist_ease"]

        if 5 <= waist <= 70:
            scores.append(1.0)
        elif -10 <= waist <= 90:
            scores.append(0.5)
        else:
            scores.append(0.0)

    # Hip
    if pd.notna(row["hip_ease"]):
        hip = row["hip_ease"]

        if 5 <= hip <= 60:
            scores.append(1.0)
        elif -10 <= hip <= 80:
            scores.append(0.5)
        else:
            scores.append(0.0)

    # Length / height
    if pd.notna(row["length_height_ratio"]):
        ratio = row["length_height_ratio"]

        if 0.30 <= ratio <= 0.55:
            scores.append(1.0)
        elif 0.25 <= ratio <= 0.65:
            scores.append(0.5)
        else:
            scores.append(0.0)

    if not scores:
        return 0.0

    return np.mean(scores)


baseline_df["compatibility_score"] = baseline_df.apply(
    compatibility_score,
    axis=1
)


# ------------------------------------------------------------
# Prediction
# ------------------------------------------------------------

baseline_df["prediction"] = (
    baseline_df["compatibility_score"] >= 0.5
).astype(int)


# ------------------------------------------------------------
# Metrics
# ------------------------------------------------------------

y_true = baseline_df["label"]
y_pred = baseline_df["prediction"]

accuracy = accuracy_score(
    y_true,
    y_pred
)

print("\nAccuracy:", round(accuracy, 4))

print("\nClassification report:")
print(
    classification_report(
        y_true,
        y_pred,
        digits=4
    )
)


# ------------------------------------------------------------
# Score separation
# ------------------------------------------------------------

print("\n================================")
print("SCORE DISTRIBUTION")
print("================================")

print(
    baseline_df
    .groupby("label")["compatibility_score"]
    .agg([
        "count",
        "mean",
        "std",
        "median",
        "min",
        "max"
    ])
)
```

```python
# ============================================================
# V2 — FULL GARMENT MEASUREMENT INTEGRITY AUDIT
# ============================================================

from pathlib import Path
import json
import re
import pandas as pd
import numpy as np

# ------------------------------------------------------------
# 1. DATASET PATH
# ------------------------------------------------------------

DATASET_ROOT = Path.cwd() / "VIBE" / "dataset"
MEASUREMENT_DIR = DATASET_ROOT / "tops" / "garment_measurement"

print("=" * 60)
print("V2 GARMENT MEASUREMENT INTEGRITY AUDIT")
print("=" * 60)

print("Dataset root :", DATASET_ROOT)
print("Measurement  :", MEASUREMENT_DIR)
print("Exists       :", MEASUREMENT_DIR.exists())

if not MEASUREMENT_DIR.exists():
    raise FileNotFoundError(
        f"Measurement directory not found:\n{MEASUREMENT_DIR}"
    )

files = sorted(MEASUREMENT_DIR.glob("*_meta.txt"))

print("Files found  :", len(files))


# ------------------------------------------------------------
# 2. EXPECTED DIMENSIONS
# ------------------------------------------------------------

DIMENSION_MAP = {
    1: "bust",
    3: "waist",
    5: "hip",
    7: "length",
    15: "sleeve",
}

EXPECTED_DIMENSIONS = set(DIMENSION_MAP.keys())


# ------------------------------------------------------------
# 3. HELPERS
# ------------------------------------------------------------

def extract_number(value):
    """
    Extract numeric value from strings such as:
        '102cm'
        '1230cm'
        '11cm'
    """
    if value is None:
        return np.nan

    match = re.search(r"-?\d+(?:\.\d+)?", str(value))

    if not match:
        return np.nan

    return float(match.group())


def parse_measurement_file(path):
    """
    Parse one *_meta.txt measurement file.

    Returns:
        list of size-level records
    """

    try:
        raw = path.read_text(encoding="utf-8").strip()

        if not raw:
            return []

        data = json.loads(raw)

        if not isinstance(data, dict):
            return []

        records = []

        for size, dimensions in data.items():

            if not isinstance(dimensions, list):
                continue

            row = {
                "file": path.name,
                "product": path.name.replace("_meta.txt", ""),
                "size": str(size),
            }

            for item in dimensions:

                if not isinstance(item, dict):
                    continue

                dimension_id = item.get("dimension_id")
                dimension_name = item.get("dimension_name")
                value = item.get("value")

                try:
                    dimension_id_int = int(dimension_id)
                except:
                    dimension_id_int = None

                if dimension_id_int in DIMENSION_MAP:

                    column = DIMENSION_MAP[dimension_id_int]

                    row[column] = extract_number(value)

                    row[f"{column}_raw"] = value
                    row[f"{column}_dimension_name"] = dimension_name

            records.append(row)

        return records

    except Exception as e:

        return [{
            "file": path.name,
            "product": path.name.replace("_meta.txt", ""),
            "size": None,
            "parse_error": str(e),
        }]


# ------------------------------------------------------------
# 4. PARSE EVERYTHING
# ------------------------------------------------------------

all_records = []

for path in files:
    all_records.extend(parse_measurement_file(path))

measurements_df = pd.DataFrame(all_records)

print()
print("=" * 60)
print("PARSED DATA")
print("=" * 60)

print("Files parsed :", measurements_df["file"].nunique())
print("Rows         :", len(measurements_df))

display(measurements_df.head())


# ------------------------------------------------------------
# 5. BASIC COMPLETENESS
# ------------------------------------------------------------

print()
print("=" * 60)
print("DIMENSION COMPLETENESS")
print("=" * 60)

for dimension in DIMENSION_MAP.values():

    if dimension in measurements_df.columns:

        present = measurements_df[dimension].notna().sum()
        missing = measurements_df[dimension].isna().sum()

        percentage = (
            present / len(measurements_df) * 100
            if len(measurements_df)
            else 0
        )

        print(
            f"{dimension:8s} | "
            f"present={present:5d} | "
            f"missing={missing:5d} | "
            f"coverage={percentage:6.2f}%"
        )


# ------------------------------------------------------------
# 6. PLAUSIBLE RANGES
# ------------------------------------------------------------
#
# These are AUDIT thresholds, not corrections.
#
# They intentionally flag suspicious data for inspection.
# They do NOT alter the original dataset.
# ------------------------------------------------------------

RANGES = {
    "bust":   (50, 180),
    "waist":  (45, 180),
    "hip":    (50, 190),
    "length": (30, 130),
    "sleeve": (0, 100),
}


# ------------------------------------------------------------
# 7. OUTLIER / IMPOSSIBLE VALUE AUDIT
# ------------------------------------------------------------

issues = []

for _, row in measurements_df.iterrows():

    for dimension, (lower, upper) in RANGES.items():

        if dimension not in row:
            continue

        value = row[dimension]

        if pd.isna(value):
            continue

        if value < lower or value > upper:

            issues.append({
                "file": row["file"],
                "product": row["product"],
                "size": row["size"],
                "dimension": dimension,
                "value": value,
                "lower_limit": lower,
                "upper_limit": upper,
                "issue": "OUT_OF_RANGE",
            })


issues_df = pd.DataFrame(issues)

print()
print("=" * 60)
print("OUT-OF-RANGE VALUES")
print("=" * 60)

print("Suspicious values:", len(issues_df))

if len(issues_df):

    display(
        issues_df.sort_values(
            ["dimension", "value"],
            ascending=[True, False]
        ).head(100)
    )

else:

    print("No out-of-range values found.")


# ------------------------------------------------------------
# 8. EMPTY FILES
# ------------------------------------------------------------

empty_files = []

for path in files:

    try:
        raw = path.read_text(encoding="utf-8").strip()

        if not raw:
            empty_files.append(path.name)
            continue

        data = json.loads(raw)

        if not data:
            empty_files.append(path.name)

    except:
        pass


print()
print("=" * 60)
print("EMPTY MEASUREMENT FILES")
print("=" * 60)

print("Empty files:", len(empty_files))

if empty_files:
    print()
    for name in empty_files[:100]:
        print(name)


# ------------------------------------------------------------
# 9. FILES WITH MISSING DIMENSIONS
# ------------------------------------------------------------

missing_dimension_records = []

for _, row in measurements_df.iterrows():

    missing = []

    for dimension in DIMENSION_MAP.values():

        if dimension not in row or pd.isna(row[dimension]):
            missing.append(dimension)

    if missing:

        missing_dimension_records.append({
            "file": row["file"],
            "product": row["product"],
            "size": row["size"],
            "missing_dimensions": ", ".join(missing),
        })


missing_df = pd.DataFrame(missing_dimension_records)

print()
print("=" * 60)
print("MISSING DIMENSIONS")
print("=" * 60)

print("Rows with missing dimensions:", len(missing_df))

if len(missing_df):

    display(missing_df.head(100))


# ------------------------------------------------------------
# 10. DUPLICATE / IDENTICAL SIZE RECORDS
# ------------------------------------------------------------

measurement_columns = [
    c for c in DIMENSION_MAP.values()
    if c in measurements_df.columns
]

duplicate_rows = measurements_df[
    measurements_df.duplicated(
        subset=["product", "size"] + measurement_columns,
        keep=False
    )
].copy()

print()
print("=" * 60)
print("DUPLICATE SIZE RECORDS")
print("=" * 60)

print("Duplicate rows:", len(duplicate_rows))

if len(duplicate_rows):

    display(
        duplicate_rows[
            ["product", "size"] + measurement_columns
        ].head(100)
    )


# ------------------------------------------------------------
# 11. SAME PRODUCT/COLOR FILE CONSISTENCY
# ------------------------------------------------------------

print()
print("=" * 60)
print("PRODUCT SUMMARY")
print("=" * 60)

product_summary = (
    measurements_df
    .groupby("product")
    .agg(
        sizes=("size", "nunique"),
        rows=("size", "count"),
        bust_missing=("bust", lambda x: x.isna().sum()),
        waist_missing=("waist", lambda x: x.isna().sum()),
        hip_missing=("hip", lambda x: x.isna().sum()),
        length_missing=("length", lambda x: x.isna().sum()),
        sleeve_missing=("sleeve", lambda x: x.isna().sum()),
    )
    .reset_index()
)

display(product_summary.head(30))


# ------------------------------------------------------------
# 12. DIMENSION DISTRIBUTIONS
# ------------------------------------------------------------

print()
print("=" * 60)
print("DIMENSION DISTRIBUTIONS")
print("=" * 60)

for dimension in measurement_columns:

    print()
    print(f"--- {dimension} ---")

    print(
        measurements_df[dimension]
        .describe()
        .round(2)
    )


# ------------------------------------------------------------
# 13. EXTREME VALUES
# ------------------------------------------------------------

print()
print("=" * 60)
print("EXTREME VALUES")
print("=" * 60)

for dimension in measurement_columns:

    print()
    print(f"--- {dimension.upper()} ---")

    temp = measurements_df[
        ["file", "product", "size", dimension]
    ].dropna()

    print("Smallest:")
    display(temp.nsmallest(10, dimension))

    print("Largest:")
    display(temp.nlargest(10, dimension))


# ------------------------------------------------------------
# 14. SUSPICIOUS PRODUCTS SUMMARY
# ------------------------------------------------------------

if len(issues_df):

    suspicious_products = (
        issues_df
        .groupby("product")
        .agg(
            suspicious_values=("value", "count"),
            dimensions=("dimension", lambda x: ", ".join(sorted(set(x))))
        )
        .sort_values(
            "suspicious_values",
            ascending=False
        )
        .reset_index()
    )

else:

    suspicious_products = pd.DataFrame(
        columns=[
            "product",
            "suspicious_values",
            "dimensions"
        ]
    )


print()
print("=" * 60)
print("SUSPICIOUS PRODUCTS")
print("=" * 60)

display(suspicious_products.head(100))


# ------------------------------------------------------------
# 15. FINAL AUDIT SUMMARY
# ------------------------------------------------------------

print()
print("=" * 60)
print("FINAL AUDIT SUMMARY")
print("=" * 60)

print(f"Measurement files              : {len(files)}")
print(f"Parsed size-level rows         : {len(measurements_df)}")
print(f"Empty files                    : {len(empty_files)}")
print(f"Out-of-range values            : {len(issues_df)}")
print(f"Rows with missing dimensions   : {len(missing_df)}")
print(f"Duplicate size records         : {len(duplicate_rows)}")
print(f"Suspicious products            : {len(suspicious_products)}")

print()
print("IMPORTANT:")
print("This audit DOES NOT modify the VIBE dataset.")
print("No values have been repaired or deleted.")
print("The output is diagnostic only.")
```

```python
# ============================================================
# V2 — CORRUPTION ANALYSIS
# ============================================================

print("=" * 60)
print("V2 GARMENT CORRUPTION ANALYSIS")
print("=" * 60)

# Work from the parsed measurements dataframe
audit_df = measurements_df.copy()

# ------------------------------------------------------------
# 1. Mark suspicious values
# ------------------------------------------------------------

RANGES = {
    "bust":   (50, 180),
    "waist":  (45, 180),
    "hip":    (50, 190),
    "length": (30, 130),
    "sleeve": (0, 100),
}

for dimension, (low, high) in RANGES.items():

    audit_df[f"{dimension}_invalid"] = (
        audit_df[dimension].notna()
        & (
            (audit_df[dimension] < low)
            | (audit_df[dimension] > high)
        )
    )


invalid_columns = [
    f"{dimension}_invalid"
    for dimension in RANGES
]

audit_df["has_invalid_measurement"] = audit_df[
    invalid_columns
].any(axis=1)


# ------------------------------------------------------------
# 2. Product-level corruption summary
# ------------------------------------------------------------

product_corruption = (
    audit_df
    .groupby("product")
    .agg(
        size_count=("size", "count"),
        invalid_rows=("has_invalid_measurement", "sum"),
        invalid_bust=("bust_invalid", "sum"),
        invalid_waist=("waist_invalid", "sum"),
        invalid_hip=("hip_invalid", "sum"),
        invalid_length=("length_invalid", "sum"),
        invalid_sleeve=("sleeve_invalid", "sum"),
    )
    .reset_index()
)

product_corruption["invalid_percentage"] = (
    product_corruption["invalid_rows"]
    / product_corruption["size_count"]
    * 100
)

corrupt_products = product_corruption[
    product_corruption["invalid_rows"] > 0
].sort_values(
    ["invalid_rows", "invalid_percentage"],
    ascending=False
)

print()
print("Products with corruption:", len(corrupt_products))

display(corrupt_products)


# ------------------------------------------------------------
# 3. Exact corrupted records
# ------------------------------------------------------------

corrupt_rows = audit_df[
    audit_df["has_invalid_measurement"]
].copy()

print()
print("=" * 60)
print("EXACT CORRUPTED RECORDS")
print("=" * 60)

display(
    corrupt_rows[
        [
            "file",
            "product",
            "size",
            "bust",
            "waist",
            "hip",
            "length",
            "sleeve",
        ]
    ]
    .sort_values("product")
)


# ------------------------------------------------------------
# 4. Corruption by dimension
# ------------------------------------------------------------

print()
print("=" * 60)
print("CORRUPTION BY DIMENSION")
print("=" * 60)

for dimension in RANGES:

    column = f"{dimension}_invalid"

    count = int(audit_df[column].sum())

    print(
        f"{dimension:8s}: {count:4d} invalid records"
    )


# ------------------------------------------------------------
# 5. Products where EVERY size is corrupted
# ------------------------------------------------------------

fully_corrupt_products = corrupt_products[
    corrupt_products["invalid_rows"]
    == corrupt_products["size_count"]
]

print()
print("=" * 60)
print("FULLY CORRUPTED PRODUCTS")
print("=" * 60)

print(
    "Products where every parsed size has "
    "at least one invalid measurement:",
    len(fully_corrupt_products)
)

display(fully_corrupt_products)


# ------------------------------------------------------------
# 6. Products where only SOME sizes are corrupted
# ------------------------------------------------------------

partially_corrupt_products = corrupt_products[
    corrupt_products["invalid_rows"]
    < corrupt_products["size_count"]
]

print()
print("=" * 60)
print("PARTIALLY CORRUPTED PRODUCTS")
print("=" * 60)

print(
    "Products with both valid and invalid sizes:",
    len(partially_corrupt_products)
)

display(partially_corrupt_products)


# ------------------------------------------------------------
# 7. Final recommendation counts
# ------------------------------------------------------------

print()
print("=" * 60)
print("RECOMMENDATION COUNTS")
print("=" * 60)

print(
    "Total measurement rows :",
    len(audit_df)
)

print(
    "Valid rows              :",
    (~audit_df["has_invalid_measurement"]).sum()
)

print(
    "Corrupted rows          :",
    audit_df["has_invalid_measurement"].sum()
)

print(
    "Corrupted products      :",
    len(corrupt_products)
)

print(
    "Fully corrupted products:",
    len(fully_corrupt_products)
)

print(
    "Partially corrupted products:",
    len(partially_corrupt_products)
)
```

```python
# ============================================================
# V2 — SIZE PROGRESSION VALIDATION
# FIXED VERSION
# ============================================================

import pandas as pd
import numpy as np

print("=" * 60)
print("V2 SIZE-PROGRESSION VALIDATION")
print("=" * 60)

df = measurements_df.copy()

# ------------------------------------------------------------
# Normalize column names
# ------------------------------------------------------------

df.columns = [str(c).strip().lower() for c in df.columns]

print("Columns:", list(df.columns))


# ------------------------------------------------------------
# Convert measurements to numeric
# ------------------------------------------------------------

dimensions = [
    "bust",
    "waist",
    "hip",
    "length",
    "sleeve",
]

for dimension in dimensions:
    df[dimension] = pd.to_numeric(
        df[dimension],
        errors="coerce"
    )


# ------------------------------------------------------------
# Convert clothing size to ordering value
# ------------------------------------------------------------

def size_to_number(value):

    if pd.isna(value):
        return np.nan

    s = str(value).strip().upper()

    # Numeric sizes
    try:
        return float(s)
    except ValueError:
        pass

    # Letter sizes
    mapping = {
        "XXXS": 0,
        "XXS": 1,
        "XS": 2,
        "S": 3,
        "M": 4,
        "L": 5,
        "XL": 6,
        "XXL": 7,
        "XXXL": 8,
    }

    return mapping.get(s, np.nan)


df["size_numeric"] = df["size"].apply(size_to_number)


# ------------------------------------------------------------
# Define suspicious measurement ranges
# ------------------------------------------------------------

RANGES = {
    "bust":   (50, 180),
    "waist":  (45, 180),
    "hip":    (50, 190),
    "length": (30, 130),
    "sleeve": (0, 100),
}

for dimension, (low, high) in RANGES.items():

    df[f"{dimension}_invalid"] = (
        df[dimension].notna()
        & (
            (df[dimension] < low)
            | (df[dimension] > high)
        )
    )


df["has_invalid"] = df[
    [f"{d}_invalid" for d in dimensions]
].any(axis=1)


# ------------------------------------------------------------
# Identify suspicious products
# ------------------------------------------------------------

suspicious_products = set(
    df.loc[df["has_invalid"], "product"]
)

print()
print("Suspicious products:", len(suspicious_products))


# ============================================================
# 1. SIZE PROGRESSION
# ============================================================

progression_rows = []

for product in sorted(suspicious_products):

    product_df = df[
        df["product"] == product
    ].copy()

    # Only rows with a usable size ordering
    product_df = product_df[
        product_df["size_numeric"].notna()
    ].sort_values("size_numeric")

    if len(product_df) < 2:
        continue

    for dimension in dimensions:

        values = product_df[
            [
                "size",
                "size_numeric",
                dimension,
            ]
        ].dropna(subset=[dimension])

        if len(values) < 2:
            continue

        values = values.sort_values("size_numeric")

        numeric_values = (
            values[dimension]
            .astype(float)
            .values
        )

        size_values = (
            values["size"]
            .values
        )

        diffs = np.diff(numeric_values)

        for i, diff in enumerate(diffs):

            if abs(diff) > 40:

                progression_rows.append({
                    "product": product,
                    "dimension": dimension,
                    "size_from": size_values[i],
                    "size_to": size_values[i + 1],
                    "value_from": numeric_values[i],
                    "value_to": numeric_values[i + 1],
                    "difference": diff,
                })


progression_df = pd.DataFrame(
    progression_rows
)


# ------------------------------------------------------------
# Display progression anomalies
# ------------------------------------------------------------

print()
print("=" * 60)
print("LARGE SIZE-TO-SIZE JUMPS")
print("=" * 60)

if progression_df.empty:

    print("No large size-to-size jumps found.")

else:

    progression_df = progression_df.sort_values(
        [
            "product",
            "dimension",
            "size_from",
        ]
    )

    print(
        "Potentially corrupted transitions:",
        len(progression_df)
    )

    display(progression_df)


# ============================================================
# 2. COMPLETE SUSPICIOUS PRODUCT RECORDS
# ============================================================

print()
print("=" * 60)
print("COMPLETE RECORDS FOR SUSPICIOUS PRODUCTS")
print("=" * 60)

suspicious_detail = df[
    df["product"].isin(suspicious_products)
].copy()

# IMPORTANT:
# Keep size_numeric while sorting.
suspicious_detail = suspicious_detail.sort_values(
    [
        "product",
        "size_numeric",
    ],
    na_position="last"
)

display(
    suspicious_detail[
        [
            "product",
            "size",
            "size_numeric",
            "bust",
            "waist",
            "hip",
            "length",
            "sleeve",
            "has_invalid",
        ]
    ]
)


# ============================================================
# 3. INVALID VALUES BY DIMENSION
# ============================================================

print()
print("=" * 60)
print("INVALID VALUES BY DIMENSION")
print("=" * 60)

for dimension in dimensions:

    count = int(
        df[f"{dimension}_invalid"].sum()
    )

    print(
        f"{dimension:8s}: {count:4d}"
    )


# ============================================================
# 4. PRODUCTS WITH ISOLATED CORRUPTION
# ============================================================

print()
print("=" * 60)
print("PRODUCTS WITH ISOLATED CORRUPTION")
print("=" * 60)

isolated = []

for product in sorted(suspicious_products):

    p = df[
        df["product"] == product
    ]

    invalid_count = int(
        p["has_invalid"].sum()
    )

    total_count = len(p)

    if invalid_count < total_count:

        isolated.append({
            "product": product,
            "total_sizes": total_count,
            "invalid_sizes": invalid_count,
            "valid_sizes": (
                total_count - invalid_count
            ),
        })


isolated_df = pd.DataFrame(isolated)

print(
    "Products where some sizes are valid:",
    len(isolated_df)
)

display(isolated_df)


# ============================================================
# 5. FULLY CORRUPTED PRODUCTS
# ============================================================

print()
print("=" * 60)
print("FULLY CORRUPTED PRODUCTS")
print("=" * 60)

fully_corrupt = []

for product in sorted(suspicious_products):

    p = df[
        df["product"] == product
    ]

    invalid_count = int(
        p["has_invalid"].sum()
    )

    total_count = len(p)

    if invalid_count == total_count:

        fully_corrupt.append({
            "product": product,
            "sizes": total_count,
        })


fully_corrupt_df = pd.DataFrame(
    fully_corrupt
)

print(
    "Fully corrupted products:",
    len(fully_corrupt_df)
)

display(fully_corrupt_df)


# ============================================================
# 6. SUMMARY
# ============================================================

print()
print("=" * 60)
print("FINAL SUMMARY")
print("=" * 60)

print(
    "Total measurement rows :",
    len(df)
)

print(
    "Valid rows             :",
    int((~df["has_invalid"]).sum())
)

print(
    "Corrupted rows         :",
    int(df["has_invalid"].sum())
)

print(
    "Corrupted products     :",
    len(suspicious_products)
)

print(
    "Fully corrupted products:",
    len(fully_corrupt_df)
)

print(
    "Partially corrupted products:",
    len(isolated_df)
)
```

```python
# ============================================================
# V2 — COMPACT CORRUPTION REPORT
# ============================================================

print("=" * 60)
print("COMPACT CORRUPTION REPORT")
print("=" * 60)

# 1. Every invalid record
invalid_records = df[df["has_invalid"]].copy()

print("\nInvalid records:", len(invalid_records))

display(
    invalid_records[
        [
            "product",
            "size",
            "bust",
            "waist",
            "hip",
            "length",
            "sleeve",
        ]
    ].head(100)
)


# 2. Count invalid records per product
product_summary = (
    df.groupby("product")
      .agg(
          total_sizes=("product", "size"),
          invalid_sizes=("has_invalid", "sum")
      )
      .reset_index()
)

product_summary["valid_sizes"] = (
    product_summary["total_sizes"]
    - product_summary["invalid_sizes"]
)

product_summary = product_summary[
    product_summary["invalid_sizes"] > 0
].sort_values(
    ["invalid_sizes", "total_sizes"],
    ascending=False
)

print("\n" + "=" * 60)
print("CORRUPTED PRODUCT SUMMARY")
print("=" * 60)

display(product_summary)


# 3. Dimension-specific invalid values
print("\n" + "=" * 60)
print("DIMENSION ANOMALIES")
print("=" * 60)

for dimension, (low, high) in RANGES.items():

    mask = df[f"{dimension}_invalid"]

    bad = df.loc[
        mask,
        ["product", "size", dimension]
    ].copy()

    print(
        f"\n{dimension.upper()} "
        f"({len(bad)} invalid)"
    )

    if not bad.empty:
        display(bad.head(50))


# 4. Recalculate size progression anomalies
print("\n" + "=" * 60)
print("SIZE PROGRESSION ANOMALIES")
print("=" * 60)

progression_rows = []

for product in sorted(suspicious_products):

    p = df[
        df["product"] == product
    ].copy()

    p = p[
        p["size_numeric"].notna()
    ].sort_values("size_numeric")

    if len(p) < 2:
        continue

    for dimension in dimensions:

        v = p[
            ["size", "size_numeric", dimension]
        ].dropna(subset=[dimension])

        if len(v) < 2:
            continue

        v = v.sort_values("size_numeric")

        vals = v[dimension].astype(float).values
        sizes = v["size"].values

        diffs = np.diff(vals)

        for i, diff in enumerate(diffs):

            if abs(diff) > 40:

                progression_rows.append({
                    "product": product,
                    "dimension": dimension,
                    "size_from": sizes[i],
                    "size_to": sizes[i + 1],
                    "value_from": vals[i],
                    "value_to": vals[i + 1],
                    "difference": diff,
                })

progression_df = pd.DataFrame(
    progression_rows
)

if progression_df.empty:
    print("No large progression anomalies.")

else:
    progression_df = progression_df.sort_values(
        ["product", "dimension"]
    )

    print(
        "Large transitions:",
        len(progression_df)
    )

    display(progression_df)
    
```

```python
# ============================================================
# V2 — GARMENT MEASUREMENT CLEANING
# ============================================================

import json
import math
import re
from pathlib import Path

import pandas as pd
import numpy as np

# ------------------------------------------------------------
# PATHS
# ------------------------------------------------------------

DATASET_ROOT = Path.home() / "Desktop/Projects/weavly/core-model/VIBE/dataset"
GARMENT_DIR = DATASET_ROOT / "tops/garment_measurement"

OUTPUT_DIR = Path.home() / "Desktop/Projects/weavly/core-model/data/v2"
OUTPUT_DIR.mkdir(parents=True, exist_ok=True)

OUTPUT_CSV = OUTPUT_DIR / "clean_garment_measurements.csv"
OUTPUT_REPORT = OUTPUT_DIR / "garment_cleaning_report.csv"

print("Dataset :", DATASET_ROOT)
print("Input   :", GARMENT_DIR)
print("Output  :", OUTPUT_CSV)

# ------------------------------------------------------------
# VALID PHYSICAL RANGES
#
# These are deliberately broad sanity bounds.
# They are NOT intended to encode fashion/fit preferences.
# They only remove obvious parser/source corruption.
# ------------------------------------------------------------

VALID_RANGES = {
    "bust":   (50.0, 180.0),
    "waist":  (50.0, 180.0),
    "hip":    (50.0, 180.0),
    "length": (30.0, 150.0),
    "sleeve": (20.0, 140.0),
}

# ------------------------------------------------------------
# DIMENSION NAME NORMALIZATION
# ------------------------------------------------------------

def normalize_dimension(name):
    name = name.lower().strip()

    if "bust" in name:
        return "bust"

    if "waist" in name:
        return "waist"

    if "hip" in name:
        return "hip"

    if "length" in name:
        return "length"

    if "sleeve" in name:
        return "sleeve"

    return None


# ------------------------------------------------------------
# VALUE PARSER
# ------------------------------------------------------------

def parse_measurement(value):
    """
    Convert values such as:

        '92cm'
        '92 cm'
        '92'
        92

    into float centimeters.

    Anything that cannot be safely parsed becomes NaN.
    """

    if value is None:
        return np.nan

    if isinstance(value, (int, float)):
        value = float(value)

        if math.isfinite(value):
            return value

        return np.nan

    value = str(value).strip().lower()

    if not value:
        return np.nan

    match = re.search(r"-?\d+(?:\.\d+)?", value)

    if not match:
        return np.nan

    try:
        number = float(match.group())
    except ValueError:
        return np.nan

    if not math.isfinite(number):
        return np.nan

    return number


# ------------------------------------------------------------
# PRODUCT ID
# ------------------------------------------------------------

def product_id_from_file(path):
    """
    Example:

    long-sleeve-tee-2#NavyWhite_meta.txt

    -> long-sleeve-tee-2#NavyWhite
    """

    name = path.stem

    if name.endswith("_meta"):
        name = name[:-5]

    return name


# ------------------------------------------------------------
# READ ONE FILE
# ------------------------------------------------------------

def read_measurement_file(path):

    try:
        with open(path, "r", encoding="utf-8") as f:
            raw = json.load(f)
    except Exception as e:
        return None, f"JSON_ERROR: {e}"

    # Some files may contain JSON encoded as a string.
    if isinstance(raw, str):
        try:
            raw = json.loads(raw)
        except Exception as e:
            return None, f"STRING_JSON_ERROR: {e}"

    if not isinstance(raw, dict):
        return None, "INVALID_ROOT"

    return raw, None


# ------------------------------------------------------------
# PROCESS DATASET
# ------------------------------------------------------------

rows = []
cleaning_events = []

files = sorted(GARMENT_DIR.glob("*.txt"))

print("\nFiles found:", len(files))

for path in files:

    product = product_id_from_file(path)

    data, error = read_measurement_file(path)

    if error:
        cleaning_events.append({
            "product": product,
            "dimension": None,
            "original_value": None,
            "cleaned_value": np.nan,
            "reason": error,
        })
        continue

    if not data:
        cleaning_events.append({
            "product": product,
            "dimension": None,
            "original_value": None,
            "cleaned_value": np.nan,
            "reason": "EMPTY_FILE",
        })
        continue

    # --------------------------------------------------------
    # JSON structure:
    #
    # {
    #   "12": [
    #       {
    #           "dimension_name": "Bust",
    #           "value": "92cm"
    #       }
    #   ]
    # }
    # --------------------------------------------------------

    measurements = {
        "bust": np.nan,
        "waist": np.nan,
        "hip": np.nan,
        "length": np.nan,
        "sleeve": np.nan,
    }

    # Track whether we actually saw each dimension.
    seen = set()

    for entries in data.values():

        # Defensive handling of malformed structures.
        if not isinstance(entries, list):
            continue

        for item in entries:

            if not isinstance(item, dict):
                continue

            dimension_name = item.get("dimension_name")
            raw_value = item.get("value")

            dimension = normalize_dimension(
                dimension_name if dimension_name is not None else ""
            )

            if dimension is None:
                continue

            seen.add(dimension)

            value = parse_measurement(raw_value)

            lower, upper = VALID_RANGES[dimension]

            if np.isnan(value):
                cleaning_events.append({
                    "product": product,
                    "dimension": dimension,
                    "original_value": raw_value,
                    "cleaned_value": np.nan,
                    "reason": "UNPARSEABLE",
                })

                continue

            # ------------------------------------------------
            # HARD SANITY CHECK
            # ------------------------------------------------

            if value < lower or value > upper:

                cleaning_events.append({
                    "product": product,
                    "dimension": dimension,
                    "original_value": raw_value,
                    "cleaned_value": np.nan,
                    "reason": f"OUT_OF_RANGE_{lower}_{upper}",
                })

                continue

            measurements[dimension] = value

    # --------------------------------------------------------
    # Keep product even if some dimensions are missing.
    # --------------------------------------------------------

    row = {
        "product_id": product,
        "garment_bust": measurements["bust"],
        "garment_waist": measurements["waist"],
        "garment_hip": measurements["hip"],
        "garment_length": measurements["length"],
        "garment_sleeve": measurements["sleeve"],
    }

    rows.append(row)


# ------------------------------------------------------------
# DATAFRAME
# ------------------------------------------------------------

clean_df = pd.DataFrame(rows)

# Remove exact duplicate products if any.
clean_df = (
    clean_df
    .drop_duplicates(subset=["product_id"], keep="first")
    .reset_index(drop=True)
)

report_df = pd.DataFrame(cleaning_events)

# ------------------------------------------------------------
# PRODUCT QUALITY FLAGS
# ------------------------------------------------------------

measurement_columns = [
    "garment_bust",
    "garment_waist",
    "garment_hip",
    "garment_length",
    "garment_sleeve",
]

clean_df["valid_dimension_count"] = (
    clean_df[measurement_columns]
    .notna()
    .sum(axis=1)
)

clean_df["fully_missing_measurements"] = (
    clean_df["valid_dimension_count"] == 0
)

clean_df["partially_missing_measurements"] = (
    (clean_df["valid_dimension_count"] > 0) &
    (clean_df["valid_dimension_count"] < len(measurement_columns))
)

# ------------------------------------------------------------
# SAVE
# ------------------------------------------------------------

clean_df.to_csv(OUTPUT_CSV, index=False)
report_df.to_csv(OUTPUT_REPORT, index=False)

# ------------------------------------------------------------
# SUMMARY
# ------------------------------------------------------------

print("\n" + "=" * 60)
print("V2 GARMENT CLEANING SUMMARY")
print("=" * 60)

print("Input files              :", len(files))
print("Products retained        :", len(clean_df))

print(
    "Products with measurements:",
    int((clean_df["valid_dimension_count"] > 0).sum())
)

print(
    "Fully missing products    :",
    int(clean_df["fully_missing_measurements"].sum())
)

print(
    "Partially missing products:",
    int(clean_df["partially_missing_measurements"].sum())
)

print(
    "Cleaning events           :",
    len(report_df)
)

print("\nMissing dimensions after cleaning:")

print(
    clean_df[measurement_columns]
    .isna()
    .sum()
)

print("\nClean measurement statistics:")

print(
    clean_df[measurement_columns]
    .describe()
    .round(2)
)

print("\nSaved:")
print("  ", OUTPUT_CSV)
print("  ", OUTPUT_REPORT)
```

```python
# ============================================================
# V2 — FIX DIMENSION PARSER + REBUILD CLEAN DATASET
# ============================================================

# IMPORTANT:
# "Sleeve Length" must be detected BEFORE generic "Length".

def normalize_dimension(name):
    name = name.lower().strip()

    # MUST come first
    if "sleeve" in name:
        return "sleeve"

    if "bust" in name:
        return "bust"

    if "waist" in name:
        return "waist"

    if "hip" in name:
        return "hip"

    if "length" in name:
        return "length"

    return None


# ------------------------------------------------------------
# REBUILD
# ------------------------------------------------------------

rows = []
cleaning_events = []

files = sorted(GARMENT_DIR.glob("*.txt"))

for path in files:

    product = product_id_from_file(path)

    data, error = read_measurement_file(path)

    if error:
        cleaning_events.append({
            "product": product,
            "dimension": None,
            "original_value": None,
            "cleaned_value": np.nan,
            "reason": error,
        })
        continue

    if not data:
        cleaning_events.append({
            "product": product,
            "dimension": None,
            "original_value": None,
            "cleaned_value": np.nan,
            "reason": "EMPTY_FILE",
        })
        continue

    measurements = {
        "bust": np.nan,
        "waist": np.nan,
        "hip": np.nan,
        "length": np.nan,
        "sleeve": np.nan,
    }

    for entries in data.values():

        if not isinstance(entries, list):
            continue

        for item in entries:

            if not isinstance(item, dict):
                continue

            dimension_name = item.get("dimension_name", "")
            raw_value = item.get("value")

            dimension = normalize_dimension(dimension_name)

            if dimension is None:
                continue

            value = parse_measurement(raw_value)

            lower, upper = VALID_RANGES[dimension]

            if np.isnan(value):

                cleaning_events.append({
                    "product": product,
                    "dimension": dimension,
                    "original_value": raw_value,
                    "cleaned_value": np.nan,
                    "reason": "UNPARSEABLE",
                })

                continue

            if value < lower or value > upper:

                cleaning_events.append({
                    "product": product,
                    "dimension": dimension,
                    "original_value": raw_value,
                    "cleaned_value": np.nan,
                    "reason": f"OUT_OF_RANGE_{lower}_{upper}",
                })

                continue

            measurements[dimension] = value

    rows.append({
        "product_id": product,
        "garment_bust": measurements["bust"],
        "garment_waist": measurements["waist"],
        "garment_hip": measurements["hip"],
        "garment_length": measurements["length"],
        "garment_sleeve": measurements["sleeve"],
    })


# ------------------------------------------------------------
# DATAFRAME
# ------------------------------------------------------------

clean_df = pd.DataFrame(rows)

clean_df = (
    clean_df
    .drop_duplicates(subset=["product_id"], keep="first")
    .reset_index(drop=True)
)

report_df = pd.DataFrame(cleaning_events)

measurement_columns = [
    "garment_bust",
    "garment_waist",
    "garment_hip",
    "garment_length",
    "garment_sleeve",
]

clean_df["valid_dimension_count"] = (
    clean_df[measurement_columns]
    .notna()
    .sum(axis=1)
)

clean_df["fully_missing_measurements"] = (
    clean_df["valid_dimension_count"] == 0
)

clean_df["partially_missing_measurements"] = (
    (clean_df["valid_dimension_count"] > 0) &
    (clean_df["valid_dimension_count"] < 5)
)


# ------------------------------------------------------------
# SAVE
# ------------------------------------------------------------

clean_df.to_csv(OUTPUT_CSV, index=False)
report_df.to_csv(OUTPUT_REPORT, index=False)


# ------------------------------------------------------------
# SUMMARY
# ------------------------------------------------------------

print("=" * 60)
print("V2 GARMENT CLEANING — CORRECTED")
print("=" * 60)

print("Input files               :", len(files))
print("Products retained         :", len(clean_df))
print(
    "Products with measurements:",
    int((clean_df["valid_dimension_count"] > 0).sum())
)
print(
    "Fully missing products    :",
    int(clean_df["fully_missing_measurements"].sum())
)
print(
    "Partially missing products:",
    int(clean_df["partially_missing_measurements"].sum())
)
print("Cleaning events           :", len(report_df))

print("\nMissing dimensions:")
print(
    clean_df[measurement_columns]
    .isna()
    .sum()
)

print("\nMissing percentages:")
print(
    (clean_df[measurement_columns].isna().mean() * 100)
    .round(2)
)

print("\nClean statistics:")
print(
    clean_df[measurement_columns]
    .describe()
    .round(2)
)

print("\nSaved:")
print(" ", OUTPUT_CSV)
print(" ", OUTPUT_REPORT)
```

```python
# ============================================================
# V2 — REBUILD COMPATIBILITY DATASET FROM CLEAN MEASUREMENTS
# ============================================================

import json
import random
from pathlib import Path

import numpy as np
import pandas as pd


# ------------------------------------------------------------
# PATHS
# ------------------------------------------------------------

CORE_ROOT = Path.home() / "Desktop/Projects/weavly/core-model"

HUMAN_FILE = (
    CORE_ROOT
    / "VIBE/dataset/humans/model.json"
)

CLEAN_GARMENT_FILE = (
    CORE_ROOT
    / "data/v2/clean_garment_measurements.csv"
)

OUTPUT_FILE = (
    CORE_ROOT
    / "data/v2/v2_compatibility_raw.csv"
)


# ------------------------------------------------------------
# REPRODUCIBILITY
# ------------------------------------------------------------

SEED = 42

random.seed(SEED)
np.random.seed(SEED)


# ------------------------------------------------------------
# LOAD HUMANS
# ------------------------------------------------------------

with open(HUMAN_FILE, "r", encoding="utf-8") as f:
    humans = json.load(f)

human_df = pd.DataFrame.from_dict(
    humans,
    orient="index"
)

human_df.index.name = "person_id"
human_df = human_df.reset_index()

for col in ["height", "bust", "waist", "hips"]:
    human_df[col] = pd.to_numeric(
        human_df[col],
        errors="coerce"
    )


# ------------------------------------------------------------
# LOAD CLEAN GARMENTS
# ------------------------------------------------------------

garment_df = pd.read_csv(
    CLEAN_GARMENT_FILE
)

garment_df = garment_df[
    [
        "product_id",
        "garment_bust",
        "garment_waist",
        "garment_hip",
        "garment_length",
        "garment_sleeve",
    ]
].copy()


# ------------------------------------------------------------
# NORMALIZE MODEL NAMES
# ------------------------------------------------------------

def normalize_person_name(name):
    """
    Normalize known duplicate model naming.

    Examples:
        Emma_1 -> Emma
        Emma_2 -> Emma
        Maddy_1 -> Maddy
        Maddy_2 -> Maddy
        Natalie_1 -> Natalie
        Natalie_2 -> Natalie
        Nathalie_1 -> Nathalie
        Nathalie_2 -> Nathalie
        Monika  -> Monika
    """

    name = str(name).strip()

    if name.endswith("_1") or name.endswith("_2"):
        name = name[:-2]

    return name


# Create canonical human profiles.
#
# IMPORTANT:
# When duplicate profiles exist, we retain them as separate
# profiles internally rather than averaging measurements.
# For compatibility-pair construction, the garment-side names
# will be mapped to the matching canonical person where possible.

canonical_humans = {}

for person_id, row in human_df.set_index("person_id").iterrows():

    canonical = normalize_person_name(person_id)

    # Keep the first profile for now.
    # We will audit duplicate-profile handling separately.
    if canonical not in canonical_humans:

        canonical_humans[canonical] = {
            "person_id": canonical,
            "height": row["height"],
            "bust": row["bust"],
            "waist": row["waist"],
            "hips": row["hips"],
        }


canonical_human_df = pd.DataFrame(
    canonical_humans.values()
)


# ------------------------------------------------------------
# LOAD POSITIVE MODEL-GARMENT RELATIONSHIPS
# ------------------------------------------------------------

GARMENT_PAIR_DIR = (
    CORE_ROOT
    / "VIBE/dataset/tops/model_measurement"
)

positive_rows = []

for file in sorted(GARMENT_PAIR_DIR.glob("*.txt")):

    product = file.stem

    if product.endswith("_meta"):
        product = product[:-5]

    try:

        with open(file, "r", encoding="utf-8") as f:
            data = json.load(f)

    except Exception:

        continue

    if isinstance(data, str):

        try:
            data = json.loads(data)
        except Exception:
            continue

    if not isinstance(data, dict):
        continue

    for raw_person in data.keys():

        person = normalize_person_name(raw_person)

        # Only retain people for whom we have a canonical profile.
        if person not in canonical_humans:
            continue

        positive_rows.append({
            "person_id": person,
            "product_id": product,
            "label": 1,
        })


positive_df = pd.DataFrame(
    positive_rows
)

# Remove duplicate relationships.
positive_df = positive_df.drop_duplicates(
    subset=["person_id", "product_id"]
)


# ------------------------------------------------------------
# KEEP ONLY CLEAN PRODUCTS
# ------------------------------------------------------------

valid_products = set(
    garment_df["product_id"]
)

positive_df = positive_df[
    positive_df["product_id"].isin(valid_products)
].copy()


# ------------------------------------------------------------
# NEGATIVE SAMPLING
# ------------------------------------------------------------

all_products = sorted(
    valid_products
)

positive_lookup = (
    positive_df
    .groupby("person_id")["product_id"]
    .apply(set)
    .to_dict()
)

negative_rows = []

for person, positive_products in positive_lookup.items():

    candidates = [
        p
        for p in all_products
        if p not in positive_products
    ]

    random.shuffle(candidates)

    # 1:1 negative sampling.
    n_negative = min(
        len(positive_products),
        len(candidates)
    )

    for product in candidates[:n_negative]:

        negative_rows.append({
            "person_id": person,
            "product_id": product,
            "label": 0,
        })


negative_df = pd.DataFrame(
    negative_rows
)


# ------------------------------------------------------------
# COMBINE
# ------------------------------------------------------------

compatibility_df = pd.concat(
    [
        positive_df,
        negative_df,
    ],
    ignore_index=True
)

compatibility_df = compatibility_df.sample(
    frac=1,
    random_state=SEED
).reset_index(drop=True)


# ------------------------------------------------------------
# MERGE PERSON FEATURES
# ------------------------------------------------------------

compatibility_df = compatibility_df.merge(
    canonical_human_df,
    on="person_id",
    how="inner"
)


# ------------------------------------------------------------
# MERGE GARMENT FEATURES
# ------------------------------------------------------------

compatibility_df = compatibility_df.merge(
    garment_df,
    on="product_id",
    how="inner"
)


# ------------------------------------------------------------
# MODEL SIZE
# ------------------------------------------------------------

# We deliberately preserve model size only where available.
# It is useful metadata but should not be treated as a numerical
# measurement in the compatibility features yet.

model_size_lookup = {}

for file in sorted(GARMENT_PAIR_DIR.glob("*.txt")):

    product = file.stem

    if product.endswith("_meta"):
        product = product[:-5]

    try:

        with open(file, "r", encoding="utf-8") as f:
            data = json.load(f)

        if isinstance(data, str):
            data = json.loads(data)

    except Exception:
        continue

    if not isinstance(data, dict):
        continue

    for raw_person, info in data.items():

        person = normalize_person_name(raw_person)

        if not isinstance(info, dict):
            continue

        size = info.get("size")

        if size is not None:
            model_size_lookup[
                (person, product)
            ] = str(size)


compatibility_df["model_size"] = [
    model_size_lookup.get(
        (row["person_id"], row["product_id"]),
        np.nan
    )
    for _, row in compatibility_df.iterrows()
]


# ------------------------------------------------------------
# DERIVED COMPATIBILITY FEATURES
# ------------------------------------------------------------

compatibility_df["bust_ease"] = (
    compatibility_df["garment_bust"]
    - compatibility_df["bust"]
)

compatibility_df["waist_ease"] = (
    compatibility_df["garment_waist"]
    - compatibility_df["waist"]
)

compatibility_df["hip_ease"] = (
    compatibility_df["garment_hip"]
    - compatibility_df["hips"]
)

compatibility_df["length_height_ratio"] = (
    compatibility_df["garment_length"]
    / compatibility_df["height"]
)


# ------------------------------------------------------------
# EXPLICIT MISSINGNESS INDICATORS
# ------------------------------------------------------------

compatibility_df["missing_bust"] = (
    compatibility_df["garment_bust"].isna().astype(int)
)

compatibility_df["missing_waist"] = (
    compatibility_df["garment_waist"].isna().astype(int)
)

compatibility_df["missing_hip"] = (
    compatibility_df["garment_hip"].isna().astype(int)
)

compatibility_df["missing_length"] = (
    compatibility_df["garment_length"].isna().astype(int)
)

compatibility_df["missing_sleeve"] = (
    compatibility_df["garment_sleeve"].isna().astype(int)
)


# ------------------------------------------------------------
# SAVE
# ------------------------------------------------------------

compatibility_df.to_csv(
    OUTPUT_FILE,
    index=False
)


# ------------------------------------------------------------
# REPORT
# ------------------------------------------------------------

print("=" * 60)
print("V2 COMPATIBILITY DATASET — REBUILT")
print("=" * 60)

print("Human profiles        :", len(canonical_human_df))
print("Clean garments        :", len(garment_df))

print()
print("Positive pairs        :", len(positive_df))
print("Negative pairs        :", len(negative_df))
print("Total pairs           :", len(compatibility_df))

print()
print("Unique people         :",
      compatibility_df["person_id"].nunique())

print("Unique garments       :",
      compatibility_df["product_id"].nunique())

print()
print("Labels:")
print(
    compatibility_df["label"]
    .value_counts()
    .sort_index()
)

print()
print("Missing garment dimensions:")

print(
    compatibility_df[
        [
            "garment_bust",
            "garment_waist",
            "garment_hip",
            "garment_length",
            "garment_sleeve",
        ]
    ]
    .isna()
    .sum()
)

print()
print("Derived feature missingness:")

print(
    compatibility_df[
        [
            "bust_ease",
            "waist_ease",
            "hip_ease",
            "length_height_ratio",
            "garment_sleeve",
        ]
    ]
    .isna()
    .sum()
)

print()
print("Saved:")
print(OUTPUT_FILE)
```

```python
# ============================================================
# V2 — FEATURE SIGNAL ANALYSIS
# ============================================================

import pandas as pd
import numpy as np

from sklearn.metrics import roc_auc_score
from sklearn.feature_selection import mutual_info_classif

print("=" * 60)
print("V2 FEATURE SIGNAL ANALYSIS")
print("=" * 60)

df = pd.read_csv(
    "/Users/saketh/Desktop/Projects/weavly/core-model/data/v2/v2_compatibility_raw.csv"
)

# ------------------------------------------------------------
# DERIVED FEATURES
# ------------------------------------------------------------

df["bust_ease"] = (
    df["garment_bust"] - df["bust"]
)

df["waist_ease"] = (
    df["garment_waist"] - df["waist"]
)

df["hip_ease"] = (
    df["garment_hip"] - df["hips"]
)

df["length_height_ratio"] = (
    df["garment_length"] / df["height"]
)

feature_cols = [
    "height",
    "bust",
    "waist",
    "hips",
    "bust_ease",
    "waist_ease",
    "hip_ease",
    "length_height_ratio",
    "garment_bust",
    "garment_waist",
    "garment_hip",
    "garment_length",
    "garment_sleeve",
]

print("\nFeatures:")
print(feature_cols)


# ------------------------------------------------------------
# AUC OF INDIVIDUAL FEATURES
# ------------------------------------------------------------

print("\n" + "=" * 60)
print("INDIVIDUAL FEATURE AUC")
print("=" * 60)

results = []

for feature in feature_cols:

    valid = df[[feature, "label"]].dropna()

    if valid[feature].nunique() < 2:
        continue

    y = valid["label"].values
    x = valid[feature].values

    auc = roc_auc_score(y, x)

    # Direction-independent AUC
    auc_abs = max(auc, 1 - auc)

    results.append({
        "feature": feature,
        "samples": len(valid),
        "auc": auc,
        "signal_auc": auc_abs,
    })

auc_df = (
    pd.DataFrame(results)
      .sort_values("signal_auc", ascending=False)
)

print(
    auc_df.to_string(
        index=False,
        float_format=lambda x: f"{x:.4f}"
    )
)


# ------------------------------------------------------------
# MUTUAL INFORMATION
# ------------------------------------------------------------

print("\n" + "=" * 60)
print("MUTUAL INFORMATION")
print("=" * 60)

mi_features = []

for feature in feature_cols:

    valid = df[[feature, "label"]].dropna()

    if len(valid) < 20:
        continue

    X = valid[[feature]].values
    y = valid["label"].values

    mi = mutual_info_classif(
        X,
        y,
        random_state=42
    )[0]

    mi_features.append({
        "feature": feature,
        "samples": len(valid),
        "mutual_information": mi,
    })

mi_df = (
    pd.DataFrame(mi_features)
      .sort_values(
          "mutual_information",
          ascending=False
      )
)

print(
    mi_df.to_string(
        index=False,
        float_format=lambda x: f"{x:.4f}"
    )
)


# ------------------------------------------------------------
# POSITIVE / NEGATIVE MEANS
# ------------------------------------------------------------

print("\n" + "=" * 60)
print("CLASS MEAN DIFFERENCES")
print("=" * 60)

summary = []

for feature in feature_cols:

    positive = df.loc[
        df["label"] == 1,
        feature
    ].dropna()

    negative = df.loc[
        df["label"] == 0,
        feature
    ].dropna()

    if len(positive) == 0 or len(negative) == 0:
        continue

    pos_mean = positive.mean()
    neg_mean = negative.mean()

    pooled_std = np.sqrt(
        (
            positive.var() +
            negative.var()
        ) / 2
    )

    effect = (
        (pos_mean - neg_mean) / pooled_std
        if pooled_std > 0
        else 0
    )

    summary.append({
        "feature": feature,
        "positive_mean": pos_mean,
        "negative_mean": neg_mean,
        "difference": pos_mean - neg_mean,
        "effect_size": effect,
    })

mean_df = (
    pd.DataFrame(summary)
      .sort_values(
          "effect_size",
          key=lambda x: x.abs(),
          ascending=False
      )
)

print(
    mean_df.to_string(
        index=False,
        float_format=lambda x: f"{x:.4f}"
    )
)


# ------------------------------------------------------------
# MISSINGNESS BY CLASS
# ------------------------------------------------------------

print("\n" + "=" * 60)
print("MISSINGNESS BY CLASS")
print("=" * 60)

missing_rows = []

for feature in feature_cols:

    pos_missing = (
        df.loc[df["label"] == 1, feature]
          .isna()
          .mean()
    )

    neg_missing = (
        df.loc[df["label"] == 0, feature]
          .isna()
          .mean()
    )

    missing_rows.append({
        "feature": feature,
        "positive_missing_pct": pos_missing * 100,
        "negative_missing_pct": neg_missing * 100,
        "difference_pct": (
            (pos_missing - neg_missing) * 100
        ),
    })

missing_df = pd.DataFrame(missing_rows)

print(
    missing_df.to_string(
        index=False,
        float_format=lambda x: f"{x:.2f}"
    )
)


# ------------------------------------------------------------
# SUMMARY
# ------------------------------------------------------------

print("\n" + "=" * 60)
print("SIGNAL ANALYSIS COMPLETE")
print("=" * 60)

print(
    "\nTop 5 features by direction-independent AUC:"
)

print(
    auc_df[
        ["feature", "signal_auc"]
    ]
    .head(5)
    .to_string(index=False)
)
```

```python
# ============================================================
# V2 — PHYSICAL COMPATIBILITY FEATURE ENGINEERING
# ============================================================

import pandas as pd
import numpy as np
from sklearn.metrics import roc_auc_score

PATH = (
    "/Users/saketh/Desktop/Projects/weavly/core-model/"
    "data/v2/v2_compatibility_raw.csv"
)

df = pd.read_csv(PATH)

# ------------------------------------------------------------
# BASIC EASE
# ------------------------------------------------------------

df["bust_ease"] = (
    df["garment_bust"] - df["bust"]
)

df["waist_ease"] = (
    df["garment_waist"] - df["waist"]
)

df["hip_ease"] = (
    df["garment_hip"] - df["hips"]
)

# ------------------------------------------------------------
# RELATIVE EASE
# ------------------------------------------------------------

df["bust_ratio"] = (
    df["garment_bust"] / df["bust"]
)

df["waist_ratio"] = (
    df["garment_waist"] / df["waist"]
)

df["hip_ratio"] = (
    df["garment_hip"] / df["hips"]
)

# ------------------------------------------------------------
# ABSOLUTE BODY-GARMENT MISMATCH
# ------------------------------------------------------------

df["bust_mismatch"] = (
    abs(df["garment_bust"] - df["bust"])
)

df["waist_mismatch"] = (
    abs(df["garment_waist"] - df["waist"])
)

df["hip_mismatch"] = (
    abs(df["garment_hip"] - df["hips"])
)

# ------------------------------------------------------------
# RELATIVE BODY-GARMENT MISMATCH
# ------------------------------------------------------------

df["bust_relative_mismatch"] = (
    abs(df["garment_bust"] - df["bust"])
    / df["bust"]
)

df["waist_relative_mismatch"] = (
    abs(df["garment_waist"] - df["waist"])
    / df["waist"]
)

df["hip_relative_mismatch"] = (
    abs(df["garment_hip"] - df["hips"])
    / df["hips"]
)

# ------------------------------------------------------------
# TIGHTNESS INDICATORS
# ------------------------------------------------------------

df["bust_tight"] = (
    df["bust_ease"] < 0
).astype(float)

df["waist_tight"] = (
    df["waist_ease"] < 0
).astype(float)

df["hip_tight"] = (
    df["hip_ease"] < 0
).astype(float)

# ------------------------------------------------------------
# VERY TIGHT INDICATORS
# ------------------------------------------------------------

df["bust_very_tight"] = (
    df["bust_ease"] < -5
).astype(float)

df["waist_very_tight"] = (
    df["waist_ease"] < -5
).astype(float)

df["hip_very_tight"] = (
    df["hip_ease"] < -5
).astype(float)

# ------------------------------------------------------------
# EXTREME EASE
# ------------------------------------------------------------

df["bust_extreme_ease"] = (
    df["bust_ease"] > 50
).astype(float)

df["waist_extreme_ease"] = (
    df["waist_ease"] > 50
).astype(float)

df["hip_extreme_ease"] = (
    df["hip_ease"] > 50
).astype(float)

# ------------------------------------------------------------
# COMPOSITE BODY COMPATIBILITY
# ------------------------------------------------------------

# Only count dimensions that actually exist.

ease_columns = [
    "bust_ease",
    "waist_ease",
    "hip_ease",
]

df["mean_ease"] = (
    df[ease_columns].mean(axis=1)
)

df["ease_std"] = (
    df[ease_columns].std(axis=1)
)

# Number of dimensions with known measurements

df["known_fit_dimensions"] = (
    df[ease_columns]
    .notna()
    .sum(axis=1)
)

# Number of tight dimensions

df["tight_dimension_count"] = (
    df[
        [
            "bust_tight",
            "waist_tight",
            "hip_tight",
        ]
    ]
    .sum(axis=1)
)

# Number of extreme ease dimensions

df["extreme_ease_count"] = (
    df[
        [
            "bust_extreme_ease",
            "waist_extreme_ease",
            "hip_extreme_ease",
        ]
    ]
    .sum(axis=1)
)

# ------------------------------------------------------------
# LENGTH RELATIONSHIP
# ------------------------------------------------------------

df["length_height_ratio"] = (
    df["garment_length"] / df["height"]
)

# ------------------------------------------------------------
# FEATURE LIST
# ------------------------------------------------------------

features = [
    "bust_ease",
    "waist_ease",
    "hip_ease",

    "bust_ratio",
    "waist_ratio",
    "hip_ratio",

    "bust_mismatch",
    "waist_mismatch",
    "hip_mismatch",

    "bust_relative_mismatch",
    "waist_relative_mismatch",
    "hip_relative_mismatch",

    "bust_tight",
    "waist_tight",
    "hip_tight",

    "bust_very_tight",
    "waist_very_tight",
    "hip_very_tight",

    "bust_extreme_ease",
    "waist_extreme_ease",
    "hip_extreme_ease",

    "mean_ease",
    "ease_std",

    "known_fit_dimensions",
    "tight_dimension_count",
    "extreme_ease_count",

    "length_height_ratio",
    "garment_sleeve",
]

# ------------------------------------------------------------
# AUC
# ------------------------------------------------------------

results = []

for feature in features:

    valid = df[[feature, "label"]].dropna()

    if len(valid) < 100:
        continue

    if valid[feature].nunique() < 2:
        continue

    auc = roc_auc_score(
        valid["label"],
        valid[feature]
    )

    results.append({
        "feature": feature,
        "samples": len(valid),
        "auc": auc,
        "signal_auc": max(auc, 1 - auc),
    })

results_df = (
    pd.DataFrame(results)
    .sort_values(
        "signal_auc",
        ascending=False
    )
)

print("=" * 60)
print("PHYSICAL COMPATIBILITY FEATURE SIGNAL")
print("=" * 60)

print(
    results_df.to_string(
        index=False,
        float_format=lambda x: f"{x:.4f}"
    )
)

print("\n" + "=" * 60)
print("TOP FEATURES")
print("=" * 60)

print(
    results_df.head(15).to_string(
        index=False,
        float_format=lambda x: f"{x:.4f}"
    )
)

# ------------------------------------------------------------
# CLASS STATISTICS FOR TOP FEATURES
# ------------------------------------------------------------

top_features = results_df.head(10)["feature"].tolist()

print("\n" + "=" * 60)
print("TOP FEATURE CLASS COMPARISON")
print("=" * 60)

for feature in top_features:

    pos = df.loc[
        df["label"] == 1,
        feature
    ].dropna()

    neg = df.loc[
        df["label"] == 0,
        feature
    ].dropna()

    print(
        f"\n{feature}"
    )

    print(
        f"  Positive mean : {pos.mean():.4f}"
    )

    print(
        f"  Negative mean : {neg.mean():.4f}"
    )

    print(
        f"  Positive med  : {pos.median():.4f}"
    )

    print(
        f"  Negative med  : {neg.median():.4f}"
    )

print("\n" + "=" * 60)
print("COMPATIBILITY FEATURE ENGINEERING COMPLETE")
print("=" * 60)
```

```python
# ============================================================
# V2 — 3 CONTROLLED BASELINES
# Raw vs Compatibility vs Combined
# ============================================================

import pandas as pd
import numpy as np

from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import Pipeline
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import (
    accuracy_score,
    roc_auc_score,
    average_precision_score,
    f1_score,
    log_loss,
)

# ------------------------------------------------------------
# LOAD DATA
# ------------------------------------------------------------

PATH = (
    "/Users/saketh/Desktop/Projects/weavly/core-model/"
    "data/v2/v2_compatibility_raw.csv"
)

df = pd.read_csv(PATH)

# ------------------------------------------------------------
# DERIVED COMPATIBILITY FEATURES
# ------------------------------------------------------------

df["bust_ease"] = df["garment_bust"] - df["bust"]
df["waist_ease"] = df["garment_waist"] - df["waist"]
df["hip_ease"] = df["garment_hip"] - df["hips"]

df["bust_ratio"] = df["garment_bust"] / df["bust"]
df["waist_ratio"] = df["garment_waist"] / df["waist"]
df["hip_ratio"] = df["garment_hip"] / df["hips"]

df["bust_mismatch"] = abs(
    df["garment_bust"] - df["bust"]
)

df["waist_mismatch"] = abs(
    df["garment_waist"] - df["waist"]
)

df["hip_mismatch"] = abs(
    df["garment_hip"] - df["hips"]
)

df["bust_relative_mismatch"] = (
    abs(df["garment_bust"] - df["bust"])
    / df["bust"]
)

df["waist_relative_mismatch"] = (
    abs(df["garment_waist"] - df["waist"])
    / df["waist"]
)

df["hip_relative_mismatch"] = (
    abs(df["garment_hip"] - df["hips"])
    / df["hips"]
)

df["bust_tight"] = (
    df["bust_ease"] < 0
).astype(float)

df["waist_tight"] = (
    df["waist_ease"] < 0
).astype(float)

df["hip_tight"] = (
    df["hip_ease"] < 0
).astype(float)

df["bust_very_tight"] = (
    df["bust_ease"] < -5
).astype(float)

df["waist_very_tight"] = (
    df["waist_ease"] < -5
).astype(float)

df["hip_very_tight"] = (
    df["hip_ease"] < -5
).astype(float)

df["bust_extreme_ease"] = (
    df["bust_ease"] > 50
).astype(float)

df["waist_extreme_ease"] = (
    df["waist_ease"] > 50
).astype(float)

df["hip_extreme_ease"] = (
    df["hip_ease"] > 50
).astype(float)

df["mean_ease"] = df[
    ["bust_ease", "waist_ease", "hip_ease"]
].mean(axis=1)

df["ease_std"] = df[
    ["bust_ease", "waist_ease", "hip_ease"]
].std(axis=1)

df["known_fit_dimensions"] = df[
    ["bust_ease", "waist_ease", "hip_ease"]
].notna().sum(axis=1)

df["tight_dimension_count"] = df[
    ["bust_tight", "waist_tight", "hip_tight"]
].sum(axis=1)

df["extreme_ease_count"] = df[
    [
        "bust_extreme_ease",
        "waist_extreme_ease",
        "hip_extreme_ease",
    ]
].sum(axis=1)

df["length_height_ratio"] = (
    df["garment_length"] / df["height"]
)

# ------------------------------------------------------------
# FEATURE GROUPS
# ------------------------------------------------------------

raw_features = [
    "height",
    "bust",
    "waist",
    "hips",
    "garment_bust",
    "garment_waist",
    "garment_hip",
    "garment_length",
    "garment_sleeve",
]

compatibility_features = [
    "bust_ease",
    "waist_ease",
    "hip_ease",
    "bust_ratio",
    "waist_ratio",
    "hip_ratio",
    "bust_mismatch",
    "waist_mismatch",
    "hip_mismatch",
    "bust_relative_mismatch",
    "waist_relative_mismatch",
    "hip_relative_mismatch",
    "bust_tight",
    "waist_tight",
    "hip_tight",
    "bust_very_tight",
    "waist_very_tight",
    "hip_very_tight",
    "bust_extreme_ease",
    "waist_extreme_ease",
    "hip_extreme_ease",
    "mean_ease",
    "ease_std",
    "known_fit_dimensions",
    "tight_dimension_count",
    "extreme_ease_count",
    "length_height_ratio",
]

combined_features = raw_features + compatibility_features

# Remove duplicates while preserving order
combined_features = list(dict.fromkeys(combined_features))

print("=" * 60)
print("FEATURE GROUPS")
print("=" * 60)

print("Raw features          :", len(raw_features))
print("Compatibility features:", len(compatibility_features))
print("Combined features     :", len(combined_features))


# ============================================================
# PERSON-LEVEL SPLIT
# ============================================================

# IMPORTANT:
# Recreate the same 36 / 8 / 8 person split used previously.
#
# Fixed random seed guarantees reproducibility.

from sklearn.model_selection import train_test_split

people = sorted(df["person_id"].unique())

train_people, temp_people = train_test_split(
    people,
    test_size=16,
    random_state=42,
)

val_people, test_people = train_test_split(
    temp_people,
    test_size=8,
    random_state=42,
)

train_df = df[df["person_id"].isin(train_people)].copy()
val_df = df[df["person_id"].isin(val_people)].copy()
test_df = df[df["person_id"].isin(test_people)].copy()

print("\n" + "=" * 60)
print("PERSON-LEVEL SPLIT")
print("=" * 60)

print("Train people:", len(train_people))
print("Val people  :", len(val_people))
print("Test people :", len(test_people))

print("\nRows:")
print("Train:", len(train_df))
print("Val  :", len(val_df))
print("Test :", len(test_df))

print("\nLeakage:")
print("Train ∩ Val :", set(train_people) & set(val_people))
print("Train ∩ Test:", set(train_people) & set(test_people))
print("Val ∩ Test  :", set(val_people) & set(test_people))


# ============================================================
# TRAIN / EVALUATE
# ============================================================

def run_baseline(name, features):

    X_train = train_df[features]
    y_train = train_df["label"]

    X_val = val_df[features]
    y_val = val_df["label"]

    X_test = test_df[features]
    y_test = test_df["label"]

    model = Pipeline([
        (
            "imputer",
            SimpleImputer(
                strategy="median",
                add_indicator=True,
            ),
        ),
        (
            "scaler",
            StandardScaler(),
        ),
        (
            "classifier",
            LogisticRegression(
                max_iter=2000,
                C=1.0,
                random_state=42,
            ),
        ),
    ])

    model.fit(X_train, y_train)

    # Validation
    val_prob = model.predict_proba(X_val)[:, 1]
    val_pred = (val_prob >= 0.5).astype(int)

    # Test
    test_prob = model.predict_proba(X_test)[:, 1]
    test_pred = (test_prob >= 0.5).astype(int)

    results = {
        "model": name,

        "val_accuracy": accuracy_score(
            y_val, val_pred
        ),

        "val_auc": roc_auc_score(
            y_val, val_prob
        ),

        "val_pr_auc": average_precision_score(
            y_val, val_prob
        ),

        "val_f1": f1_score(
            y_val, val_pred
        ),

        "val_log_loss": log_loss(
            y_val, val_prob
        ),

        "test_accuracy": accuracy_score(
            y_test, test_pred
        ),

        "test_auc": roc_auc_score(
            y_test, test_prob
        ),

        "test_pr_auc": average_precision_score(
            y_test, test_prob
        ),

        "test_f1": f1_score(
            y_test, test_pred
        ),

        "test_log_loss": log_loss(
            y_test, test_prob
        ),
    }

    print("\n" + "=" * 60)
    print(name)
    print("=" * 60)

    print(
        f"Validation Accuracy : {results['val_accuracy']:.4f}"
    )
    print(
        f"Validation ROC-AUC  : {results['val_auc']:.4f}"
    )
    print(
        f"Validation PR-AUC   : {results['val_pr_auc']:.4f}"
    )
    print(
        f"Validation F1       : {results['val_f1']:.4f}"
    )
    print(
        f"Validation LogLoss  : {results['val_log_loss']:.4f}"
    )

    print()

    print(
        f"Test Accuracy       : {results['test_accuracy']:.4f}"
    )
    print(
        f"Test ROC-AUC        : {results['test_auc']:.4f}"
    )
    print(
        f"Test PR-AUC         : {results['test_pr_auc']:.4f}"
    )
    print(
        f"Test F1             : {results['test_f1']:.4f}"
    )
    print(
        f"Test LogLoss        : {results['test_log_loss']:.4f}"
    )

    return results


# ============================================================
# RUN 3 BASELINES
# ============================================================

results = []

results.append(
    run_baseline(
        "A — RAW FEATURES",
        raw_features,
    )
)

results.append(
    run_baseline(
        "B — COMPATIBILITY FEATURES",
        compatibility_features,
    )
)

results.append(
    run_baseline(
        "C — COMBINED FEATURES",
        combined_features,
    )
)


# ============================================================
# FINAL COMPARISON
# ============================================================

results_df = pd.DataFrame(results)

print("\n\n" + "=" * 60)
print("FINAL BASELINE COMPARISON")
print("=" * 60)

display(
    results_df[
        [
            "model",
            "val_auc",
            "val_pr_auc",
            "val_f1",
            "val_log_loss",
            "test_auc",
            "test_pr_auc",
            "test_f1",
            "test_log_loss",
            "test_accuracy",
        ]
    ].round(4)
)


# ============================================================
# BEST MODEL
# ============================================================

best = results_df.loc[
    results_df["test_auc"].idxmax()
]

print("\n" + "=" * 60)
print("BEST BASELINE")
print("=" * 60)

print("Model     :", best["model"])
print("Test AUC  :", round(best["test_auc"], 4))
print("Test PR-AUC:", round(best["test_pr_auc"], 4))
print("Test F1   :", round(best["test_f1"], 4))
print("Test Acc  :", round(best["test_accuracy"], 4))

print("\n" + "=" * 60)
print("3-BASELINE EXPERIMENT COMPLETE")
print("=" * 60)
```

```python
# ============================================================
# V2 — NEGATIVE SAMPLING EXPERIMENT
# Random + Physically Plausible Hard Negatives
# ============================================================

import pandas as pd
import numpy as np

from pathlib import Path

# ------------------------------------------------------------
# LOAD CURRENT V2 DATA
# ------------------------------------------------------------

DATA_PATH = Path(
    "/Users/saketh/Desktop/Projects/weavly/core-model/"
    "data/v2/v2_compatibility_raw.csv"
)

df = pd.read_csv(DATA_PATH)

print("=" * 60)
print("V2 NEGATIVE SAMPLING")
print("=" * 60)

print("Rows    :", len(df))
print("People  :", df["person_id"].nunique())
print("Garments:", df["product_id"].nunique())

# ------------------------------------------------------------
# POSITIVE PAIRS
# ------------------------------------------------------------

positive_pairs = set(
    zip(
        df.loc[df["label"] == 1, "person_id"],
        df.loc[df["label"] == 1, "product_id"],
    )
)

all_people = sorted(df["person_id"].unique())
all_products = sorted(df["product_id"].unique())

# Products already observed positively for each person
positive_by_person = (
    df[df["label"] == 1]
    .groupby("person_id")["product_id"]
    .apply(set)
    .to_dict()
)

print("\nPositive pairs:", len(positive_pairs))

# ------------------------------------------------------------
# GARMENT-LEVEL MEASUREMENT TABLE
# ------------------------------------------------------------

garment_cols = [
    "product_id",
    "garment_bust",
    "garment_waist",
    "garment_hip",
    "garment_length",
    "garment_sleeve",
]

garments = (
    df[garment_cols]
    .drop_duplicates("product_id")
    .set_index("product_id")
)

# ------------------------------------------------------------
# PERSON BODY TABLE
# ------------------------------------------------------------

person_cols = [
    "person_id",
    "height",
    "bust",
    "waist",
    "hips",
]

people = (
    df[person_cols]
    .drop_duplicates("person_id")
    .set_index("person_id")
)

# ------------------------------------------------------------
# HARD NEGATIVE DEFINITION
# ------------------------------------------------------------
#
# We want negatives that are:
#
#   1. NOT observed as positive for this person
#   2. Physically plausible
#   3. Relatively close to the person's observed positive
#      garment measurements
#
# This avoids making every random unobserved garment a
# "definitely incompatible" example.
#
# We use normalized body/garment differences.
# ------------------------------------------------------------

def compatibility_distance(person, product):

    p = people.loc[person]
    g = garments.loc[product]

    distances = []

    # Bust
    if pd.notna(p["bust"]) and pd.notna(g["garment_bust"]):
        distances.append(
            abs(g["garment_bust"] - p["bust"])
            / max(p["bust"], 1)
        )

    # Waist
    if pd.notna(p["waist"]) and pd.notna(g["garment_waist"]):
        distances.append(
            abs(g["garment_waist"] - p["waist"])
            / max(p["waist"], 1)
        )

    # Hip
    if pd.notna(p["hips"]) and pd.notna(g["garment_hip"]):
        distances.append(
            abs(g["garment_hip"] - p["hips"])
            / max(p["hips"], 1)
        )

    # Length relative to height
    if (
        pd.notna(p["height"])
        and pd.notna(g["garment_length"])
    ):
        distances.append(
            abs(
                (g["garment_length"] / p["height"])
                - 0.40
            )
        )

    if not distances:
        return np.inf

    return float(np.mean(distances))


# ------------------------------------------------------------
# BUILD HARD NEGATIVE POOL
# ------------------------------------------------------------

rng = np.random.default_rng(42)

hard_negative_records = []
random_negative_records = []

# Number of negatives per positive
HARD_NEGATIVES_PER_PERSON = 1
RANDOM_NEGATIVES_PER_PERSON = 1

for person in all_people:

    positives = positive_by_person.get(person, set())

    # --------------------------------------------------------
    # Candidate products
    # --------------------------------------------------------

    candidates = [
        product
        for product in all_products
        if product not in positives
    ]

    if not candidates:
        continue

    # --------------------------------------------------------
    # HARD NEGATIVES
    # --------------------------------------------------------

    scored_candidates = []

    for product in candidates:
        distance = compatibility_distance(
            person,
            product,
        )

        if np.isfinite(distance):
            scored_candidates.append(
                (product, distance)
            )

    scored_candidates.sort(
        key=lambda x: x[1]
    )

    # Select close / plausible garments
    hard_candidates = [
        product
        for product, _ in scored_candidates[
            :HARD_NEGATIVES_PER_PERSON
        ]
    ]

    for product in hard_candidates:
        hard_negative_records.append(
            {
                "person_id": person,
                "product_id": product,
                "label": 0,
                "negative_type": "hard",
            }
        )

    # --------------------------------------------------------
    # RANDOM NEGATIVES
    # --------------------------------------------------------

    random_candidates = [
        product
        for product in candidates
        if product not in hard_candidates
    ]

    if random_candidates:

        count = min(
            RANDOM_NEGATIVES_PER_PERSON,
            len(random_candidates),
        )

        selected = rng.choice(
            random_candidates,
            size=count,
            replace=False,
        )

        for product in selected:
            random_negative_records.append(
                {
                    "person_id": person,
                    "product_id": product,
                    "label": 0,
                    "negative_type": "random",
                }
            )


# ------------------------------------------------------------
# DATAFRAMES
# ------------------------------------------------------------

hard_negatives = pd.DataFrame(
    hard_negative_records
)

random_negatives = pd.DataFrame(
    random_negative_records
)

print("\n" + "=" * 60)
print("NEGATIVE SAMPLING RESULTS")
print("=" * 60)

print(
    "Hard negatives  :",
    len(hard_negatives)
)

print(
    "Random negatives:",
    len(random_negatives)
)

# ------------------------------------------------------------
# POSITIVE DATAFRAME
# ------------------------------------------------------------

positive_df = (
    df[df["label"] == 1]
    [["person_id", "product_id"]]
    .drop_duplicates()
    .copy()
)

positive_df["label"] = 1
positive_df["negative_type"] = "positive"

# ------------------------------------------------------------
# COMBINE
# ------------------------------------------------------------

sampling_df = pd.concat(
    [
        positive_df,
        hard_negatives,
        random_negatives,
    ],
    ignore_index=True,
)

# Remove accidental duplicates
sampling_df = sampling_df.drop_duplicates(
    subset=[
        "person_id",
        "product_id",
        "label",
    ]
).reset_index(drop=True)

# ------------------------------------------------------------
# CHECK LABEL CONFLICTS
# ------------------------------------------------------------

conflicts = (
    sampling_df
    .groupby(
        ["person_id", "product_id"]
    )["label"]
    .nunique()
)

conflicting_pairs = conflicts[
    conflicts > 1
]

# ------------------------------------------------------------
# CHECK DUPLICATES
# ------------------------------------------------------------

duplicate_pairs = sampling_df.duplicated(
    subset=[
        "person_id",
        "product_id",
        "label",
    ]
).sum()

# ------------------------------------------------------------
# SUMMARY
# ------------------------------------------------------------

print("\n" + "=" * 60)
print("FINAL NEGATIVE-SAMPLING DATASET")
print("=" * 60)

print(
    "Positive:",
    (sampling_df["label"] == 1).sum()
)

print(
    "Hard negative:",
    (
        sampling_df["negative_type"] == "hard"
    ).sum()
)

print(
    "Random negative:",
    (
        sampling_df["negative_type"] == "random"
    ).sum()
)

print(
    "Total:",
    len(sampling_df)
)

print("\nLabels:")
print(
    sampling_df["label"]
    .value_counts()
    .sort_index()
)

print("\nUnique people:")
print(
    sampling_df["person_id"].nunique()
)

print("\nUnique garments:")
print(
    sampling_df["product_id"].nunique()
)

print("\nDuplicate rows:", duplicate_pairs)

print(
    "Label conflicts:",
    len(conflicting_pairs)
)

# ------------------------------------------------------------
# NEGATIVE DISTANCE ANALYSIS
# ------------------------------------------------------------

if len(hard_negatives) > 0:

    hard_distances = []

    for _, row in hard_negatives.iterrows():

        distance = compatibility_distance(
            row["person_id"],
            row["product_id"],
        )

        hard_distances.append(distance)

    print("\n" + "=" * 60)
    print("HARD NEGATIVE DISTANCE")
    print("=" * 60)

    print(
        pd.Series(hard_distances)
        .describe()
        .round(4)
    )

# ------------------------------------------------------------
# PERSON BALANCE
# ------------------------------------------------------------

person_balance = (
    sampling_df
    .groupby(
        ["person_id", "label"]
    )
    .size()
    .unstack(fill_value=0)
)

if 0 in person_balance.columns and 1 in person_balance.columns:

    person_balance["difference"] = (
        person_balance[1]
        - person_balance[0]
    )

    print("\n" + "=" * 60)
    print("PERSON-LEVEL BALANCE")
    print("=" * 60)

    print(
        "People with unequal labels:",
        (
            person_balance["difference"]
            != 0
        ).sum()
    )

    print(
        person_balance["difference"]
        .describe()
    )

# ------------------------------------------------------------
# SAVE
# ------------------------------------------------------------

OUTPUT_PATH = Path(
    "/Users/saketh/Desktop/Projects/weavly/core-model/"
    "data/v2/v2_negative_sampling_dataset.csv"
)

sampling_df.to_csv(
    OUTPUT_PATH,
    index=False,
)

print("\n" + "=" * 60)
print("SAVED")
print("=" * 60)

print(OUTPUT_PATH)
```

```python
# ============================================================
# V2 — CORRECTED NEGATIVE SAMPLING + PAIRWISE DATASET
# ============================================================

import pandas as pd
import numpy as np
from pathlib import Path

# ------------------------------------------------------------
# CONFIG
# ------------------------------------------------------------

DATA_PATH = Path(
    "/Users/saketh/Desktop/Projects/weavly/core-model/"
    "data/v2/v2_compatibility_raw.csv"
)

OUTPUT_POINTWISE = Path(
    "/Users/saketh/Desktop/Projects/weavly/core-model/"
    "data/v2/v2_negative_sampling_corrected.csv"
)

OUTPUT_PAIRWISE = Path(
    "/Users/saketh/Desktop/Projects/weavly/core-model/"
    "data/v2/v2_pairwise_ranking.csv"
)

# Number of negatives for EACH positive
HARD_NEGATIVES_PER_POSITIVE = 2
RANDOM_NEGATIVES_PER_POSITIVE = 1

RANDOM_SEED = 42

rng = np.random.default_rng(RANDOM_SEED)

# ------------------------------------------------------------
# LOAD DATA
# ------------------------------------------------------------

df = pd.read_csv(DATA_PATH)

print("=" * 60)
print("V2 CORRECTED NEGATIVE SAMPLING")
print("=" * 60)

print("Input rows :", len(df))
print("People     :", df["person_id"].nunique())
print("Garments   :", df["product_id"].nunique())

# ------------------------------------------------------------
# POSITIVE INTERACTIONS
# ------------------------------------------------------------

positive_df = (
    df[df["label"] == 1]
    [["person_id", "product_id"]]
    .drop_duplicates()
    .reset_index(drop=True)
)

print("\nPositive interactions:", len(positive_df))

# ------------------------------------------------------------
# GARMENT TABLE
# ------------------------------------------------------------

garment_cols = [
    "product_id",
    "garment_bust",
    "garment_waist",
    "garment_hip",
    "garment_length",
    "garment_sleeve",
]

garments = (
    df[garment_cols]
    .drop_duplicates("product_id")
    .set_index("product_id")
)

# ------------------------------------------------------------
# PERSON TABLE
# ------------------------------------------------------------

person_cols = [
    "person_id",
    "height",
    "bust",
    "waist",
    "hips",
]

people = (
    df[person_cols]
    .drop_duplicates("person_id")
    .set_index("person_id")
)

all_products = list(garments.index)

# ------------------------------------------------------------
# POSITIVE PRODUCTS PER PERSON
# ------------------------------------------------------------

positive_by_person = (
    positive_df
    .groupby("person_id")["product_id"]
    .apply(set)
    .to_dict()
)

# ------------------------------------------------------------
# PHYSICAL DISTANCE
# ------------------------------------------------------------

def compatibility_distance(person_id, product_id):

    p = people.loc[person_id]
    g = garments.loc[product_id]

    distances = []

    # Bust
    if pd.notna(p["bust"]) and pd.notna(g["garment_bust"]):
        distances.append(
            abs(g["garment_bust"] - p["bust"])
            / max(float(p["bust"]), 1.0)
        )

    # Waist
    if pd.notna(p["waist"]) and pd.notna(g["garment_waist"]):
        distances.append(
            abs(g["garment_waist"] - p["waist"])
            / max(float(p["waist"]), 1.0)
        )

    # Hip
    if pd.notna(p["hips"]) and pd.notna(g["garment_hip"]):
        distances.append(
            abs(g["garment_hip"] - p["hips"])
            / max(float(p["hips"]), 1.0)
        )

    # Length relative to height
    if (
        pd.notna(p["height"])
        and pd.notna(g["garment_length"])
    ):
        distances.append(
            abs(
                (g["garment_length"] / p["height"])
                - 0.40
            )
        )

    if not distances:
        return np.inf

    return float(np.mean(distances))


# ============================================================
# BUILD NEGATIVES PER POSITIVE
# ============================================================

hard_negative_rows = []
random_negative_rows = []

pairwise_rows = []

positive_count = 0

for _, positive in positive_df.iterrows():

    person = positive["person_id"]
    positive_product = positive["product_id"]

    positive_count += 1

    observed = positive_by_person.get(person, set())

    # --------------------------------------------------------
    # Candidate products
    # --------------------------------------------------------

    candidates = [
        product
        for product in all_products
        if product not in observed
    ]

    if not candidates:
        continue

    # --------------------------------------------------------
    # Score all candidate garments by physical distance
    # --------------------------------------------------------

    scored = []

    for product in candidates:

        distance = compatibility_distance(
            person,
            product
        )

        if np.isfinite(distance):
            scored.append(
                (product, distance)
            )

    scored.sort(key=lambda x: x[1])

    # --------------------------------------------------------
    # HARD NEGATIVES
    #
    # Take closest physically plausible garments.
    # --------------------------------------------------------

    hard_candidates = [
        product
        for product, distance
        in scored[:HARD_NEGATIVES_PER_POSITIVE]
    ]

    for product in hard_candidates:

        distance = compatibility_distance(
            person,
            product
        )

        hard_negative_rows.append({
            "person_id": person,
            "positive_product_id": positive_product,
            "negative_product_id": product,
            "negative_type": "hard",
            "negative_distance": distance,
        })

        pairwise_rows.append({
            "person_id": person,
            "positive_product_id": positive_product,
            "negative_product_id": product,
            "negative_type": "hard",
        })

    # --------------------------------------------------------
    # RANDOM NEGATIVES
    # --------------------------------------------------------

    hard_set = set(hard_candidates)

    random_candidates = [
        product
        for product in candidates
        if product not in hard_set
    ]

    if random_candidates:

        count = min(
            RANDOM_NEGATIVES_PER_POSITIVE,
            len(random_candidates)
        )

        selected = rng.choice(
            random_candidates,
            size=count,
            replace=False,
        )

        for product in selected:

            random_negative_rows.append({
                "person_id": person,
                "positive_product_id": positive_product,
                "negative_product_id": product,
                "negative_type": "random",
                "negative_distance": compatibility_distance(
                    person,
                    product
                ),
            })

            pairwise_rows.append({
                "person_id": person,
                "positive_product_id": positive_product,
                "negative_product_id": product,
                "negative_type": "random",
            })


# ============================================================
# DATAFRAMES
# ============================================================

hard_df = pd.DataFrame(hard_negative_rows)
random_df = pd.DataFrame(random_negative_rows)

pairwise_df = pd.DataFrame(pairwise_rows)

# ============================================================
# EXPECTED COUNTS
# ============================================================

expected_hard = (
    positive_count
    * HARD_NEGATIVES_PER_POSITIVE
)

expected_random = (
    positive_count
    * RANDOM_NEGATIVES_PER_POSITIVE
)

print("\n" + "=" * 60)
print("SAMPLING COUNTS")
print("=" * 60)

print("Positive interactions :", positive_count)

print(
    "Expected hard negatives :",
    expected_hard
)

print(
    "Actual hard negatives   :",
    len(hard_df)
)

print(
    "Expected random negatives:",
    expected_random
)

print(
    "Actual random negatives :",
    len(random_df)
)

print(
    "Total pairwise samples  :",
    len(pairwise_df)
)

# ============================================================
# CHECK DUPLICATES
# ============================================================

duplicate_pair_count = (
    pairwise_df
    .duplicated(
        subset=[
            "person_id",
            "positive_product_id",
            "negative_product_id",
        ]
    )
    .sum()
)

print("\nDuplicate pairwise rows:", duplicate_pair_count)

# ============================================================
# CHECK POSITIVE / NEGATIVE COLLISION
# ============================================================

positive_set = set(
    zip(
        positive_df["person_id"],
        positive_df["product_id"],
    )
)

negative_set = set(
    zip(
        pairwise_df["person_id"],
        pairwise_df["negative_product_id"],
    )
)

collisions = positive_set & negative_set

print(
    "Positive-negative collisions:",
    len(collisions)
)

# ============================================================
# NEGATIVE TYPE DISTRIBUTION
# ============================================================

print("\n" + "=" * 60)
print("NEGATIVE TYPE DISTRIBUTION")
print("=" * 60)

print(
    pairwise_df["negative_type"]
    .value_counts()
)

# ============================================================
# NEGATIVE DISTANCE
# ============================================================

if len(hard_df):

    print("\n" + "=" * 60)
    print("HARD NEGATIVE DISTANCE")
    print("=" * 60)

    print(
        hard_df["negative_distance"]
        .describe()
        .round(4)
    )

if len(random_df):

    print("\n" + "=" * 60)
    print("RANDOM NEGATIVE DISTANCE")
    print("=" * 60)

    print(
        random_df["negative_distance"]
        .describe()
        .round(4)
    )

# ============================================================
# PERSON-LEVEL PAIR COUNTS
# ============================================================

person_pair_counts = (
    pairwise_df
    .groupby("person_id")
    .size()
)

print("\n" + "=" * 60)
print("PAIRWISE SAMPLES PER PERSON")
print("=" * 60)

print(
    person_pair_counts
    .describe()
    .round(2)
)

print(
    "\nPeople:",
    person_pair_counts.shape[0]
)

# ============================================================
# EXAMPLE PAIRS
# ============================================================

print("\n" + "=" * 60)
print("EXAMPLE PAIRS")
print("=" * 60)

display(
    pairwise_df.head(15)
)

# ============================================================
# SAVE
# ============================================================

pairwise_df.to_csv(
    OUTPUT_PAIRWISE,
    index=False,
)

# Also save detailed negative sampling information
negative_detail_df = pd.concat(
    [
        hard_df,
        random_df,
    ],
    ignore_index=True,
)

negative_detail_df.to_csv(
    OUTPUT_POINTWISE,
    index=False,
)

print("\n" + "=" * 60)
print("FILES SAVED")
print("=" * 60)

print("Negative details:")
print(OUTPUT_POINTWISE)

print("\nPairwise ranking dataset:")
print(OUTPUT_PAIRWISE)

print("\n" + "=" * 60)
print("CORRECTED SAMPLING COMPLETE")
print("=" * 60)
```

```python
# ============================================================
# ZYRA V2 — FIRST PAIRWISE RANKING MODEL
# ============================================================

import pandas as pd
import numpy as np
import torch
import torch.nn as nn

from pathlib import Path
from sklearn.model_selection import train_test_split
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler
from torch.utils.data import TensorDataset, DataLoader

# ------------------------------------------------------------
# CONFIG
# ------------------------------------------------------------

PAIRWISE_PATH = Path(
    "/Users/saketh/Desktop/Projects/weavly/core-model/"
    "data/v2/v2_pairwise_ranking.csv"
)

MODEL_OUTPUT = Path(
    "/Users/saketh/Desktop/Projects/weavly/core-model/"
    "data/v2/v2_pairwise_ranker.pt"
)

RANDOM_SEED = 42

torch.manual_seed(RANDOM_SEED)
np.random.seed(RANDOM_SEED)

# ------------------------------------------------------------
# LOAD
# ------------------------------------------------------------

pairs = pd.read_csv(PAIRWISE_PATH)

print("=" * 60)
print("ZYRA V2 PAIRWISE RANKING")
print("=" * 60)

print("Pairwise rows:", len(pairs))
print("People       :", pairs["person_id"].nunique())

print("\nNegative types:")
print(pairs["negative_type"].value_counts())

# ------------------------------------------------------------
# LOAD ORIGINAL COMPATIBILITY DATA
# ------------------------------------------------------------

RAW_PATH = Path(
    "/Users/saketh/Desktop/Projects/weavly/core-model/"
    "data/v2/v2_compatibility_raw.csv"
)

raw = pd.read_csv(RAW_PATH)

# ------------------------------------------------------------
# FEATURE ENGINEERING
# ------------------------------------------------------------

def add_features(df):

    df = df.copy()

    df["bust_ease"] = (
        df["garment_bust"] - df["bust"]
    )

    df["waist_ease"] = (
        df["garment_waist"] - df["waist"]
    )

    df["hip_ease"] = (
        df["garment_hip"] - df["hips"]
    )

    df["bust_ratio"] = (
        df["garment_bust"] / df["bust"]
    )

    df["waist_ratio"] = (
        df["garment_waist"] / df["waist"]
    )

    df["hip_ratio"] = (
        df["garment_hip"] / df["hips"]
    )

    df["bust_mismatch"] = abs(
        df["garment_bust"] - df["bust"]
    )

    df["waist_mismatch"] = abs(
        df["garment_waist"] - df["waist"]
    )

    df["hip_mismatch"] = abs(
        df["garment_hip"] - df["hips"]
    )

    df["bust_relative_mismatch"] = (
        df["bust_mismatch"] / df["bust"]
    )

    df["waist_relative_mismatch"] = (
        df["waist_mismatch"] / df["waist"]
    )

    df["hip_relative_mismatch"] = (
        df["hip_mismatch"] / df["hips"]
    )

    df["mean_ease"] = df[
        [
            "bust_ease",
            "waist_ease",
            "hip_ease",
        ]
    ].mean(axis=1)

    df["ease_std"] = df[
        [
            "bust_ease",
            "waist_ease",
            "hip_ease",
        ]
    ].std(axis=1)

    df["known_fit_dimensions"] = df[
        [
            "bust_ease",
            "waist_ease",
            "hip_ease",
        ]
    ].notna().sum(axis=1)

    df["tight_dimension_count"] = (
        (
            df[
                [
                    "bust_ease",
                    "waist_ease",
                    "hip_ease",
                ]
            ] < 0
        )
        .sum(axis=1)
    )

    df["extreme_ease_count"] = (
        (
            df[
                [
                    "bust_ease",
                    "waist_ease",
                    "hip_ease",
                ]
            ] > 50
        )
        .sum(axis=1)
    )

    df["length_height_ratio"] = (
        df["garment_length"] / df["height"]
    )

    return df


raw = add_features(raw)

# ------------------------------------------------------------
# FEATURE SET
# ------------------------------------------------------------

FEATURES = [
    "height",
    "bust",
    "waist",
    "hips",

    "garment_bust",
    "garment_waist",
    "garment_hip",
    "garment_length",
    "garment_sleeve",

    "bust_ease",
    "waist_ease",
    "hip_ease",

    "bust_ratio",
    "waist_ratio",
    "hip_ratio",

    "bust_mismatch",
    "waist_mismatch",
    "hip_mismatch",

    "bust_relative_mismatch",
    "waist_relative_mismatch",
    "hip_relative_mismatch",

    "mean_ease",
    "ease_std",

    "known_fit_dimensions",
    "tight_dimension_count",
    "extreme_ease_count",

    "length_height_ratio",
]

print("\nFeature dimension:", len(FEATURES))

# ------------------------------------------------------------
# CREATE LOOKUP TABLE
# ------------------------------------------------------------

feature_lookup = (
    raw[
        ["person_id", "product_id"] + FEATURES
    ]
    .drop_duplicates(
        ["person_id", "product_id"]
    )
)

# ------------------------------------------------------------
# GARMENT FEATURES ARE PERSON-INDEPENDENT,
# BODY FEATURES ARE PERSON-DEPENDENT.
#
# Build explicit lookup by product for garment attributes.
# ------------------------------------------------------------

GARMENT_FEATURES = [
    "garment_bust",
    "garment_waist",
    "garment_hip",
    "garment_length",
    "garment_sleeve",
]

PERSON_FEATURES = [
    "height",
    "bust",
    "waist",
    "hips",
]

# One person profile
person_lookup = (
    raw[
        ["person_id"] + PERSON_FEATURES
    ]
    .drop_duplicates("person_id")
    .set_index("person_id")
)

# One garment profile
garment_lookup = (
    raw[
        ["product_id"] + GARMENT_FEATURES
    ]
    .drop_duplicates("product_id")
    .set_index("product_id")
)

# ------------------------------------------------------------
# FUNCTION: BUILD FEATURE VECTOR
# ------------------------------------------------------------

def make_pair_features(person_id, product_id):

    person = person_lookup.loc[person_id]
    garment = garment_lookup.loc[product_id]

    row = {}

    # Body
    for feature in PERSON_FEATURES:
        row[feature] = person[feature]

    # Garment
    for feature in GARMENT_FEATURES:
        row[feature] = garment[feature]

    # Derived compatibility
    bust = person["bust"]
    waist = person["waist"]
    hips = person["hips"]
    height = person["height"]

    gb = garment["garment_bust"]
    gw = garment["garment_waist"]
    gh = garment["garment_hip"]
    gl = garment["garment_length"]

    row["bust_ease"] = gb - bust
    row["waist_ease"] = gw - waist
    row["hip_ease"] = gh - hips

    row["bust_ratio"] = gb / bust
    row["waist_ratio"] = gw / waist
    row["hip_ratio"] = gh / hips

    row["bust_mismatch"] = abs(gb - bust)
    row["waist_mismatch"] = abs(gw - waist)
    row["hip_mismatch"] = abs(gh - hips)

    row["bust_relative_mismatch"] = (
        abs(gb - bust) / bust
    )

    row["waist_relative_mismatch"] = (
        abs(gw - waist) / waist
    )

    row["hip_relative_mismatch"] = (
        abs(gh - hips) / hips
    )

    eases = np.array(
        [
            row["bust_ease"],
            row["waist_ease"],
            row["hip_ease"],
        ],
        dtype=float,
    )

    row["mean_ease"] = np.nanmean(eases)
    row["ease_std"] = np.nanstd(eases)

    row["known_fit_dimensions"] = np.sum(
        ~np.isnan(eases)
    )

    row["tight_dimension_count"] = np.sum(
        eases < 0
    )

    row["extreme_ease_count"] = np.sum(
        eases > 50
    )

    row["length_height_ratio"] = (
        gl / height
    )

    return np.array(
        [row[f] for f in FEATURES],
        dtype=np.float32,
    )


# ------------------------------------------------------------
# PERSON-LEVEL SPLIT
# ------------------------------------------------------------

people = sorted(
    pairs["person_id"].unique()
)

train_people, temp_people = train_test_split(
    people,
    test_size=16,
    random_state=RANDOM_SEED,
)

val_people, test_people = train_test_split(
    temp_people,
    test_size=8,
    random_state=RANDOM_SEED,
)

train_pairs = pairs[
    pairs["person_id"].isin(train_people)
].copy()

val_pairs = pairs[
    pairs["person_id"].isin(val_people)
].copy()

test_pairs = pairs[
    pairs["person_id"].isin(test_people)
].copy()

print("\n" + "=" * 60)
print("PERSON-LEVEL SPLIT")
print("=" * 60)

print("Train people:", len(train_people))
print("Val people  :", len(val_people))
print("Test people :", len(test_people))

print("\nPairwise rows:")
print("Train:", len(train_pairs))
print("Val  :", len(val_pairs))
print("Test :", len(test_pairs))

print("\nLeakage:")
print(
    "Train ∩ Val :",
    set(train_people) & set(val_people)
)

print(
    "Train ∩ Test:",
    set(train_people) & set(test_people)
)

print(
    "Val ∩ Test  :",
    set(val_people) & set(test_people)
)

# ------------------------------------------------------------
# BUILD PAIRWISE MATRICES
# ------------------------------------------------------------

def build_pairwise_matrix(frame):

    positive_vectors = []
    negative_vectors = []

    for _, row in frame.iterrows():

        positive_vectors.append(
            make_pair_features(
                row["person_id"],
                row["positive_product_id"],
            )
        )

        negative_vectors.append(
            make_pair_features(
                row["person_id"],
                row["negative_product_id"],
            )
        )

    return (
        np.stack(positive_vectors),
        np.stack(negative_vectors),
    )


X_pos_train, X_neg_train = build_pairwise_matrix(
    train_pairs
)

X_pos_val, X_neg_val = build_pairwise_matrix(
    val_pairs
)

X_pos_test, X_neg_test = build_pairwise_matrix(
    test_pairs
)

# ------------------------------------------------------------
# IMPUTE USING TRAIN ONLY
# ------------------------------------------------------------

imputer = SimpleImputer(
    strategy="median",
    add_indicator=True,
)

scaler = StandardScaler()

# Fit on BOTH positive and negative training examples
train_combined = np.vstack(
    [
        X_pos_train,
        X_neg_train,
    ]
)

train_combined = imputer.fit_transform(
    train_combined
)

scaler.fit(train_combined)

# Transform
def transform_pair_data(X_pos, X_neg):

    pos = imputer.transform(X_pos)
    neg = imputer.transform(X_neg)

    pos = scaler.transform(pos)
    neg = scaler.transform(neg)

    return pos, neg


X_pos_train, X_neg_train = transform_pair_data(
    X_pos_train,
    X_neg_train,
)

X_pos_val, X_neg_val = transform_pair_data(
    X_pos_val,
    X_neg_val,
)

X_pos_test, X_neg_test = transform_pair_data(
    X_pos_test,
    X_neg_test,
)

print("\n" + "=" * 60)
print("PAIRWISE TENSORS")
print("=" * 60)

print(
    "Feature dimension:",
    X_pos_train.shape[1]
)

print(
    "Train:",
    X_pos_train.shape,
)

print(
    "Val:",
    X_pos_val.shape,
)

print(
    "Test:",
    X_pos_test.shape,
)

print(
    "NaN:",
    np.isnan(X_pos_train).sum()
    + np.isnan(X_neg_train).sum()
)

print(
    "Inf:",
    np.isinf(X_pos_train).sum()
    + np.isinf(X_neg_train).sum()
)

# ------------------------------------------------------------
# PYTORCH DATASETS
# ------------------------------------------------------------

device = torch.device(
    "mps"
    if torch.backends.mps.is_available()
    else "cuda"
    if torch.cuda.is_available()
    else "cpu"
)

print("\nDevice:", device)

train_pos = torch.tensor(
    X_pos_train,
    dtype=torch.float32,
)

train_neg = torch.tensor(
    X_neg_train,
    dtype=torch.float32,
)

val_pos = torch.tensor(
    X_pos_val,
    dtype=torch.float32,
)

val_neg = torch.tensor(
    X_neg_val,
    dtype=torch.float32,
)

test_pos = torch.tensor(
    X_pos_test,
    dtype=torch.float32,
)

test_neg = torch.tensor(
    X_neg_test,
    dtype=torch.float32,
)

train_loader = DataLoader(
    TensorDataset(train_pos, train_neg),
    batch_size=128,
    shuffle=True,
)

# ------------------------------------------------------------
# RANKING MODEL
# ------------------------------------------------------------

class CompatibilityRanker(nn.Module):

    def __init__(self, input_dim):

        super().__init__()

        self.network = nn.Sequential(
            nn.Linear(input_dim, 64),
            nn.ReLU(),

            nn.Linear(64, 32),
            nn.ReLU(),

            nn.Linear(32, 1),
        )

    def forward(self, x):

        return self.network(x).squeeze(-1)


model = CompatibilityRanker(
    input_dim=train_pos.shape[1]
).to(device)

optimizer = torch.optim.Adam(
    model.parameters(),
    lr=1e-3,
    weight_decay=1e-4,
)

criterion = nn.MarginRankingLoss(
    margin=0.2
)

print("\n" + "=" * 60)
print("MODEL")
print("=" * 60)

print(
    "Input dimension:",
    train_pos.shape[1]
)

print(
    "Parameters:",
    sum(
        p.numel()
        for p in model.parameters()
        if p.requires_grad
    )
)

# ------------------------------------------------------------
# EVALUATION
# ------------------------------------------------------------

def ranking_accuracy(pos_tensor, neg_tensor):

    model.eval()

    with torch.no_grad():

        pos_scores = model(
            pos_tensor.to(device)
        )

        neg_scores = model(
            neg_tensor.to(device)
        )

        correct = (
            pos_scores > neg_scores
        ).float()

        return correct.mean().item()


# ------------------------------------------------------------
# TRAIN
# ------------------------------------------------------------

EPOCHS = 50

best_val = -np.inf
best_state = None

print("\n" + "=" * 60)
print("TRAINING")
print("=" * 60)

for epoch in range(1, EPOCHS + 1):

    model.train()

    total_loss = 0.0
    batches = 0

    for pos_batch, neg_batch in train_loader:

        pos_batch = pos_batch.to(device)
        neg_batch = neg_batch.to(device)

        pos_score = model(pos_batch)
        neg_score = model(neg_batch)

        target = torch.ones(
            pos_score.shape,
            device=device,
        )

        loss = criterion(
            pos_score,
            neg_score,
            target,
        )

        optimizer.zero_grad()
        loss.backward()
        optimizer.step()

        total_loss += loss.item()
        batches += 1

    train_loss = total_loss / batches

    val_acc = ranking_accuracy(
        val_pos,
        val_neg,
    )

    if val_acc > best_val:

        best_val = val_acc

        best_state = {
            k: v.detach().cpu().clone()
            for k, v in model.state_dict().items()
        }

    if (
        epoch == 1
        or epoch % 5 == 0
    ):

        train_acc = ranking_accuracy(
            train_pos,
            train_neg,
        )

        print(
            f"Epoch {epoch:02d}/{EPOCHS} "
            f"| Loss: {train_loss:.4f} "
            f"| Train Rank Acc: {train_acc:.4f} "
            f"| Val Rank Acc: {val_acc:.4f}"
        )

# ------------------------------------------------------------
# RESTORE BEST MODEL
# ------------------------------------------------------------

model.load_state_dict(best_state)

# ------------------------------------------------------------
# FINAL TEST
# ------------------------------------------------------------

train_acc = ranking_accuracy(
    train_pos,
    train_neg,
)

val_acc = ranking_accuracy(
    val_pos,
    val_neg,
)

test_acc = ranking_accuracy(
    test_pos,
    test_neg,
)

print("\n" + "=" * 60)
print("FINAL PAIRWISE RANKING RESULTS")
print("=" * 60)

print(
    f"Train Ranking Accuracy : {train_acc:.4f}"
)

print(
    f"Validation Ranking Acc : {val_acc:.4f}"
)

print(
    f"Test Ranking Accuracy  : {test_acc:.4f}"
)

# ------------------------------------------------------------
# SAVE MODEL + PREPROCESSING
# ------------------------------------------------------------

torch.save(
    {
        "model_state_dict": model.state_dict(),
        "input_dim": train_pos.shape[1],
        "feature_names": FEATURES,
        "best_val_accuracy": best_val,
        "random_seed": RANDOM_SEED,
    },
    MODEL_OUTPUT,
)

print("\nSaved model:")
print(MODEL_OUTPUT)

print("\n" + "=" * 60)
print("PAIRWISE RANKING EXPERIMENT COMPLETE")
print("=" * 60)
```

```python
# ============================================================
# ZYRA V2 — HELD-OUT NEGATIVE EVALUATION
# ============================================================

import pandas as pd
import numpy as np
import torch
import torch.nn as nn

from pathlib import Path
from sklearn.model_selection import train_test_split

# ------------------------------------------------------------
# PATHS
# ------------------------------------------------------------

RAW_PATH = Path(
    "/Users/saketh/Desktop/Projects/weavly/core-model/"
    "data/v2/v2_compatibility_raw.csv"
)

PAIRWISE_PATH = Path(
    "/Users/saketh/Desktop/Projects/weavly/core-model/"
    "data/v2/v2_pairwise_ranking.csv"
)

MODEL_PATH = Path(
    "/Users/saketh/Desktop/Projects/weavly/core-model/"
    "data/v2/v2_pairwise_ranker.pt"
)

# ------------------------------------------------------------
# LOAD
# ------------------------------------------------------------

raw = pd.read_csv(RAW_PATH)
trained_pairs = pd.read_csv(PAIRWISE_PATH)
checkpoint = torch.load(
    MODEL_PATH,
    map_location="cpu",
)

# ------------------------------------------------------------
# FEATURE ENGINEERING
# ------------------------------------------------------------

def add_features(df):

    df = df.copy()

    df["bust_ease"] = (
        df["garment_bust"] - df["bust"]
    )

    df["waist_ease"] = (
        df["garment_waist"] - df["waist"]
    )

    df["hip_ease"] = (
        df["garment_hip"] - df["hips"]
    )

    df["bust_ratio"] = (
        df["garment_bust"] / df["bust"]
    )

    df["waist_ratio"] = (
        df["garment_waist"] / df["waist"]
    )

    df["hip_ratio"] = (
        df["garment_hip"] / df["hips"]
    )

    df["bust_mismatch"] = abs(
        df["garment_bust"] - df["bust"]
    )

    df["waist_mismatch"] = abs(
        df["garment_waist"] - df["waist"]
    )

    df["hip_mismatch"] = abs(
        df["garment_hip"] - df["hips"]
    )

    df["bust_relative_mismatch"] = (
        df["bust_mismatch"] / df["bust"]
    )

    df["waist_relative_mismatch"] = (
        df["waist_mismatch"] / df["waist"]
    )

    df["hip_relative_mismatch"] = (
        df["hip_mismatch"] / df["hips"]
    )

    eases = df[
        [
            "bust_ease",
            "waist_ease",
            "hip_ease",
        ]
    ]

    df["mean_ease"] = eases.mean(
        axis=1
    )

    df["ease_std"] = eases.std(
        axis=1
    )

    df["known_fit_dimensions"] = (
        eases.notna().sum(axis=1)
    )

    df["tight_dimension_count"] = (
        (eases < 0).sum(axis=1)
    )

    df["extreme_ease_count"] = (
        (eases > 50).sum(axis=1)
    )

    df["length_height_ratio"] = (
        df["garment_length"]
        / df["height"]
    )

    return df


raw = add_features(raw)

# ------------------------------------------------------------
# SAME 27 FEATURES USED BY RANKER
# ------------------------------------------------------------

FEATURES = [
    "height",
    "bust",
    "waist",
    "hips",

    "garment_bust",
    "garment_waist",
    "garment_hip",
    "garment_length",
    "garment_sleeve",

    "bust_ease",
    "waist_ease",
    "hip_ease",

    "bust_ratio",
    "waist_ratio",
    "hip_ratio",

    "bust_mismatch",
    "waist_mismatch",
    "hip_mismatch",

    "bust_relative_mismatch",
    "waist_relative_mismatch",
    "hip_relative_mismatch",

    "mean_ease",
    "ease_std",

    "known_fit_dimensions",
    "tight_dimension_count",
    "extreme_ease_count",

    "length_height_ratio",
]

# ------------------------------------------------------------
# LOOKUPS
# ------------------------------------------------------------

person_lookup = (
    raw[
        [
            "person_id",
            "height",
            "bust",
            "waist",
            "hips",
        ]
    ]
    .drop_duplicates("person_id")
    .set_index("person_id")
)

garment_lookup = (
    raw[
        [
            "product_id",
            "garment_bust",
            "garment_waist",
            "garment_hip",
            "garment_length",
            "garment_sleeve",
        ]
    ]
    .drop_duplicates("product_id")
    .set_index("product_id")
)

all_products = list(
    garment_lookup.index
)

# ------------------------------------------------------------
# BUILD FEATURE VECTOR
# ------------------------------------------------------------

def make_features(
    person_id,
    product_id,
):

    p = person_lookup.loc[
        person_id
    ]

    g = garment_lookup.loc[
        product_id
    ]

    row = {}

    for f in [
        "height",
        "bust",
        "waist",
        "hips",
    ]:
        row[f] = p[f]

    for f in [
        "garment_bust",
        "garment_waist",
        "garment_hip",
        "garment_length",
        "garment_sleeve",
    ]:
        row[f] = g[f]

    bust = p["bust"]
    waist = p["waist"]
    hips = p["hips"]
    height = p["height"]

    gb = g["garment_bust"]
    gw = g["garment_waist"]
    gh = g["garment_hip"]
    gl = g["garment_length"]

    row["bust_ease"] = gb - bust
    row["waist_ease"] = gw - waist
    row["hip_ease"] = gh - hips

    row["bust_ratio"] = gb / bust
    row["waist_ratio"] = gw / waist
    row["hip_ratio"] = gh / hips

    row["bust_mismatch"] = abs(
        gb - bust
    )

    row["waist_mismatch"] = abs(
        gw - waist
    )

    row["hip_mismatch"] = abs(
        gh - hips
    )

    row["bust_relative_mismatch"] = (
        abs(gb - bust) / bust
    )

    row["waist_relative_mismatch"] = (
        abs(gw - waist) / waist
    )

    row["hip_relative_mismatch"] = (
        abs(gh - hips) / hips
    )

    ease_values = np.array(
        [
            row["bust_ease"],
            row["waist_ease"],
            row["hip_ease"],
        ],
        dtype=float,
    )

    if np.all(np.isnan(ease_values)):

        row["mean_ease"] = np.nan
        row["ease_std"] = np.nan

    else:

        row["mean_ease"] = np.nanmean(
            ease_values
        )

        row["ease_std"] = np.nanstd(
            ease_values
        )

    row["known_fit_dimensions"] = (
        np.sum(~np.isnan(ease_values))
    )

    row["tight_dimension_count"] = np.nansum(
        ease_values < 0
    )

    row["extreme_ease_count"] = np.nansum(
        ease_values > 50
    )

    row["length_height_ratio"] = (
        gl / height
    )

    return np.array(
        [
            row[f]
            for f in FEATURES
        ],
        dtype=np.float32,
    )

# ------------------------------------------------------------
# RECREATE EXACT TRAIN / VAL / TEST PEOPLE
# ------------------------------------------------------------

people = sorted(
    trained_pairs["person_id"].unique()
)

train_people, temp_people = train_test_split(
    people,
    test_size=16,
    random_state=42,
)

val_people, test_people = train_test_split(
    temp_people,
    test_size=8,
    random_state=42,
)

# ------------------------------------------------------------
# IDENTIFY ALL ALREADY USED NEGATIVES
# ------------------------------------------------------------

used_negatives = set(
    zip(
        trained_pairs["person_id"],
        trained_pairs["negative_product_id"],
    )
)

positive_pairs = set(
    zip(
        trained_pairs["person_id"],
        trained_pairs["positive_product_id"],
    )
)

print("=" * 60)
print("HELD-OUT NEGATIVE EVALUATION")
print("=" * 60)

print(
    "Training pairwise samples:",
    len(trained_pairs)
)

print(
    "Used negative pairs:",
    len(used_negatives)
)

# ------------------------------------------------------------
# BUILD FRESH NEGATIVES
# ------------------------------------------------------------
#
# We deliberately select negatives that:
#
#   1. Were never used by the original sampler
#   2. Are not positive for that person
#   3. Are physically plausible enough to be meaningful
#
# We select a random fresh negative from the middle
# of the physical-distance distribution rather than
# simply choosing the closest ones.
# ------------------------------------------------------------

fresh_pairs = []

rng = np.random.default_rng(12345)

# Only evaluate unseen people from TEST
for person in test_people:

    person_positives = {
        product
        for p, product in positive_pairs
        if p == person
    }

    candidates = []

    for product in all_products:

        key = (
            person,
            product,
        )

        if product in person_positives:
            continue

        if key in used_negatives:
            continue

        try:

            vec = make_features(
                person,
                product,
            )

            # Need at least some physical information
            if np.isnan(vec).all():
                continue

            candidates.append(
                product
            )

        except Exception:
            continue

    if not candidates:
        continue

    # --------------------------------------------------------
    # For every positive garment belonging to this person,
    # select one completely fresh negative.
    # --------------------------------------------------------

    person_positive_products = list(
        person_positives
    )

    for positive_product in person_positive_products:

        # Random fresh negative
        negative_product = rng.choice(
            candidates
        )

        fresh_pairs.append(
            {
                "person_id": person,
                "positive_product_id":
                    positive_product,
                "negative_product_id":
                    negative_product,
            }
        )

# ------------------------------------------------------------
# DATAFRAME
# ------------------------------------------------------------

fresh_df = pd.DataFrame(
    fresh_pairs
)

print("\nFresh evaluation pairs:")
print(len(fresh_df))

print(
    "People:",
    fresh_df["person_id"].nunique()
    if len(fresh_df)
    else 0
)

# ------------------------------------------------------------
# VERIFY ZERO OVERLAP
# ------------------------------------------------------------

fresh_negative_keys = set(
    zip(
        fresh_df["person_id"],
        fresh_df["negative_product_id"],
    )
)

overlap = (
    fresh_negative_keys
    & used_negatives
)

print(
    "Fresh negative overlap with training:",
    len(overlap)
)

# ------------------------------------------------------------
# MODEL
# ------------------------------------------------------------

class CompatibilityRanker(nn.Module):

    def __init__(self, input_dim):

        super().__init__()

        self.network = nn.Sequential(
            nn.Linear(input_dim, 64),
            nn.ReLU(),

            nn.Linear(64, 32),
            nn.ReLU(),

            nn.Linear(32, 1),
        )

    def forward(self, x):

        return self.network(x).squeeze(-1)


model = CompatibilityRanker(
    checkpoint["input_dim"]
)

model.load_state_dict(
    checkpoint["model_state_dict"]
)

model.eval()

# ------------------------------------------------------------
# IMPORTANT:
# The original training cell used median imputation +
# StandardScaler.
#
# Reconstruct those preprocessing statistics from the
# ORIGINAL training people only.
# ------------------------------------------------------------

from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler

# Build original training pair features
original_train = trained_pairs[
    trained_pairs["person_id"].isin(
        train_people
    )
]

train_vectors = []

for _, row in original_train.iterrows():

    train_vectors.append(
        make_features(
            row["person_id"],
            row["positive_product_id"],
        )
    )

    train_vectors.append(
        make_features(
            row["person_id"],
            row["negative_product_id"],
        )
    )

train_vectors = np.stack(
    train_vectors
)

imputer = SimpleImputer(
    strategy="median",
    add_indicator=True,
)

train_imputed = imputer.fit_transform(
    train_vectors
)

scaler = StandardScaler()

scaler.fit(
    train_imputed
)

# ------------------------------------------------------------
# EVALUATE FRESH PAIRS
# ------------------------------------------------------------

fresh_pos = []
fresh_neg = []

for _, row in fresh_df.iterrows():

    fresh_pos.append(
        make_features(
            row["person_id"],
            row["positive_product_id"],
        )
    )

    fresh_neg.append(
        make_features(
            row["person_id"],
            row["negative_product_id"],
        )
    )

if len(fresh_pos) == 0:

    raise RuntimeError(
        "No fresh evaluation pairs were created."
    )

fresh_pos = np.stack(
    fresh_pos
)

fresh_neg = np.stack(
    fresh_neg
)

fresh_pos = scaler.transform(
    imputer.transform(
        fresh_pos
    )
)

fresh_neg = scaler.transform(
    imputer.transform(
        fresh_neg
    )
)

pos_tensor = torch.tensor(
    fresh_pos,
    dtype=torch.float32,
)

neg_tensor = torch.tensor(
    fresh_neg,
    dtype=torch.float32,
)

with torch.no_grad():

    pos_scores = model(
        pos_tensor
    )

    neg_scores = model(
        neg_tensor
    )

    correct = (
        pos_scores > neg_scores
    )

    accuracy = (
        correct.float()
        .mean()
        .item()
    )

# ------------------------------------------------------------
# SCORE GAP
# ------------------------------------------------------------

score_gap = (
    pos_scores - neg_scores
)

# ------------------------------------------------------------
# RESULTS
# ------------------------------------------------------------

print("\n" + "=" * 60)
print("HELD-OUT RESULTS")
print("=" * 60)

print(
    "Fresh evaluation pairs:",
    len(fresh_df)
)

print(
    "Positive score mean:",
    round(
        pos_scores.mean().item(),
        4,
    )
)

print(
    "Negative score mean:",
    round(
        neg_scores.mean().item(),
        4,
    )
)

print(
    "Mean score gap:",
    round(
        score_gap.mean().item(),
        4,
    )
)

print(
    "Median score gap:",
    round(
        score_gap.median().item(),
        4,
    )
)

print(
    "Held-out ranking accuracy:",
    round(
        accuracy,
        4,
    )
)

# ------------------------------------------------------------
# COMPARE WITH ORIGINAL TEST
# ------------------------------------------------------------

print("\n" + "=" * 60)
print("COMPARISON")
print("=" * 60)

print(
    "Original test ranking accuracy : 0.8417"
)

print(
    "Fresh held-out ranking accuracy:",
    round(
        accuracy,
        4,
    )
)

print(
    "Difference:",
    round(
        accuracy - 0.8417,
        4,
    )
)

# ------------------------------------------------------------
# FINAL INTERPRETATION
# ------------------------------------------------------------

print("\n" + "=" * 60)
print("INTERPRETATION")
print("=" * 60)

if accuracy >= 0.75:

    print(
        "STRONG: ranking signal survives "
        "fresh-negative evaluation."
    )

elif accuracy >= 0.60:

    print(
        "MODERATE: useful signal exists, "
        "but generalization is weaker."
    )

elif accuracy >= 0.52:

    print(
        "WEAK: model has limited generalization "
        "to unseen negatives."
    )

else:

    print(
        "FAIL: fresh-negative ranking is "
        "close to random."
    )

print("\nHeld-out evaluation complete.")
```

```python
# ============================================================
# ZYRA V2 — RIGOROUS GENERALIZATION BENCHMARK
# ============================================================
#
# IMPORTANT:
# - NO RETRAINING
# - Uses the existing saved ranker
# - Person-level TEST people only
# - Negatives used during original pairwise training are excluded
#
# Evaluates:
#   1. Random unseen negatives
#   2. Hard unseen negatives
#   3. Mixed unseen negatives
#   4. ROC-AUC
#   5. PR-AUC
#   6. Pairwise ranking accuracy
#   7. Per-person accuracy
#   8. Complete vs incomplete measurement performance
#
# ============================================================

import pandas as pd
import numpy as np
import torch
import torch.nn as nn

from pathlib import Path
from sklearn.model_selection import train_test_split
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import (
    roc_auc_score,
    average_precision_score,
)

# ============================================================
# CONFIG
# ============================================================

BASE = Path(
    "/Users/saketh/Desktop/Projects/weavly/core-model"
)

RAW_PATH = (
    BASE /
    "data/v2/v2_compatibility_raw.csv"
)

PAIRWISE_PATH = (
    BASE /
    "data/v2/v2_pairwise_ranking.csv"
)

MODEL_PATH = (
    BASE /
    "data/v2/v2_pairwise_ranker.pt"
)

SEED = 42

rng = np.random.default_rng(2026)

# ============================================================
# LOAD DATA
# ============================================================

raw = pd.read_csv(RAW_PATH)

pairwise = pd.read_csv(
    PAIRWISE_PATH
)

checkpoint = torch.load(
    MODEL_PATH,
    map_location="cpu",
)

print("=" * 65)
print("ZYRA V2 — RIGOROUS GENERALIZATION BENCHMARK")
print("=" * 65)

print(
    "Raw rows:",
    len(raw)
)

print(
    "Pairwise rows:",
    len(pairwise)
)

print(
    "People:",
    pairwise["person_id"].nunique()
)

print(
    "Saved model:",
    MODEL_PATH
)

# ============================================================
# FEATURE ENGINEERING
# ============================================================

def add_features(df):

    df = df.copy()

    # ----------------------------
    # Physical compatibility
    # ----------------------------

    df["bust_ease"] = (
        df["garment_bust"] -
        df["bust"]
    )

    df["waist_ease"] = (
        df["garment_waist"] -
        df["waist"]
    )

    df["hip_ease"] = (
        df["garment_hip"] -
        df["hips"]
    )

    # ----------------------------
    # Ratios
    # ----------------------------

    df["bust_ratio"] = (
        df["garment_bust"] /
        df["bust"]
    )

    df["waist_ratio"] = (
        df["garment_waist"] /
        df["waist"]
    )

    df["hip_ratio"] = (
        df["garment_hip"] /
        df["hips"]
    )

    # ----------------------------
    # Absolute mismatch
    # ----------------------------

    df["bust_mismatch"] = abs(
        df["garment_bust"] -
        df["bust"]
    )

    df["waist_mismatch"] = abs(
        df["garment_waist"] -
        df["waist"]
    )

    df["hip_mismatch"] = abs(
        df["garment_hip"] -
        df["hips"]
    )

    # ----------------------------
    # Relative mismatch
    # ----------------------------

    df["bust_relative_mismatch"] = (
        df["bust_mismatch"] /
        df["bust"]
    )

    df["waist_relative_mismatch"] = (
        df["waist_mismatch"] /
        df["waist"]
    )

    df["hip_relative_mismatch"] = (
        df["hip_mismatch"] /
        df["hips"]
    )

    # ----------------------------
    # Aggregate ease
    # ----------------------------

    ease_cols = [
        "bust_ease",
        "waist_ease",
        "hip_ease",
    ]

    df["mean_ease"] = (
        df[ease_cols]
        .mean(axis=1)
    )

    df["ease_std"] = (
        df[ease_cols]
        .std(axis=1)
    )

    # ----------------------------
    # Missingness / thresholds
    # ----------------------------

    df["known_fit_dimensions"] = (
        df[ease_cols]
        .notna()
        .sum(axis=1)
    )

    df["tight_dimension_count"] = (
        df[ease_cols]
        .lt(0)
        .sum(axis=1)
    )

    df["extreme_ease_count"] = (
        df[ease_cols]
        .gt(50)
        .sum(axis=1)
    )

    # ----------------------------
    # Length
    # ----------------------------

    df["length_height_ratio"] = (
        df["garment_length"] /
        df["height"]
    )

    return df


raw = add_features(raw)

# ============================================================
# FEATURES
# ============================================================

FEATURES = [
    "height",
    "bust",
    "waist",
    "hips",

    "garment_bust",
    "garment_waist",
    "garment_hip",
    "garment_length",
    "garment_sleeve",

    "bust_ease",
    "waist_ease",
    "hip_ease",

    "bust_ratio",
    "waist_ratio",
    "hip_ratio",

    "bust_mismatch",
    "waist_mismatch",
    "hip_mismatch",

    "bust_relative_mismatch",
    "waist_relative_mismatch",
    "hip_relative_mismatch",

    "mean_ease",
    "ease_std",

    "known_fit_dimensions",
    "tight_dimension_count",
    "extreme_ease_count",

    "length_height_ratio",
]

print("\nFeature count:", len(FEATURES))

# ============================================================
# LOOKUPS
# ============================================================

person_cols = [
    "height",
    "bust",
    "waist",
    "hips",
]

garment_cols = [
    "garment_bust",
    "garment_waist",
    "garment_hip",
    "garment_length",
    "garment_sleeve",
]

person_lookup = (
    raw[
        ["person_id"] +
        person_cols
    ]
    .drop_duplicates(
        "person_id"
    )
    .set_index("person_id")
)

garment_lookup = (
    raw[
        ["product_id"] +
        garment_cols
    ]
    .drop_duplicates(
        "product_id"
    )
    .set_index("product_id")
)

all_people = sorted(
    person_lookup.index
)

all_products = sorted(
    garment_lookup.index
)

# ============================================================
# FEATURE VECTOR GENERATOR
# ============================================================

def make_features(
    person_id,
    product_id,
):

    p = person_lookup.loc[
        person_id
    ]

    g = garment_lookup.loc[
        product_id
    ]

    row = {}

    # Body
    for f in person_cols:
        row[f] = p[f]

    # Garment
    for f in garment_cols:
        row[f] = g[f]

    bust = p["bust"]
    waist = p["waist"]
    hips = p["hips"]
    height = p["height"]

    gb = g["garment_bust"]
    gw = g["garment_waist"]
    gh = g["garment_hip"]
    gl = g["garment_length"]

    # ----------------------------
    # Ease
    # ----------------------------

    row["bust_ease"] = (
        gb - bust
    )

    row["waist_ease"] = (
        gw - waist
    )

    row["hip_ease"] = (
        gh - hips
    )

    # ----------------------------
    # Ratios
    # ----------------------------

    row["bust_ratio"] = (
        gb / bust
    )

    row["waist_ratio"] = (
        gw / waist
    )

    row["hip_ratio"] = (
        gh / hips
    )

    # ----------------------------
    # Mismatch
    # ----------------------------

    row["bust_mismatch"] = abs(
        gb - bust
    )

    row["waist_mismatch"] = abs(
        gw - waist
    )

    row["hip_mismatch"] = abs(
        gh - hips
    )

    row["bust_relative_mismatch"] = (
        row["bust_mismatch"] /
        bust
    )

    row["waist_relative_mismatch"] = (
        row["waist_mismatch"] /
        waist
    )

    row["hip_relative_mismatch"] = (
        row["hip_mismatch"] /
        hips
    )

    # ----------------------------
    # Aggregate
    # ----------------------------

    eases = np.array(
        [
            row["bust_ease"],
            row["waist_ease"],
            row["hip_ease"],
        ],
        dtype=float,
    )

    if np.all(
        np.isnan(eases)
    ):

        row["mean_ease"] = np.nan
        row["ease_std"] = np.nan

    else:

        row["mean_ease"] = np.nanmean(
            eases
        )

        row["ease_std"] = np.nanstd(
            eases
        )

    row["known_fit_dimensions"] = (
        np.sum(
            ~np.isnan(eases)
        )
    )

    row["tight_dimension_count"] = (
        np.sum(
            eases < 0
        )
    )

    row["extreme_ease_count"] = (
        np.sum(
            eases > 50
        )
    )

    row["length_height_ratio"] = (
        gl / height
    )

    return np.array(
        [
            row[f]
            for f in FEATURES
        ],
        dtype=np.float32,
    )


# ============================================================
# RECREATE PERSON SPLIT
# ============================================================

people = sorted(
    pairwise["person_id"].unique()
)

train_people, temp_people = (
    train_test_split(
        people,
        test_size=16,
        random_state=SEED,
    )
)

val_people, test_people = (
    train_test_split(
        temp_people,
        test_size=8,
        random_state=SEED,
    )
)

train_people = set(
    train_people
)

val_people = set(
    val_people
)

test_people = set(
    test_people
)

print("\n" + "=" * 65)
print("PERSON SPLIT")
print("=" * 65)

print(
    "Train:",
    len(train_people)
)

print(
    "Validation:",
    len(val_people)
)

print(
    "Test:",
    len(test_people)
)

# ============================================================
# ORIGINAL POSITIVE PAIRS
# ============================================================

positive_keys = set(
    zip(
        pairwise["person_id"],
        pairwise[
            "positive_product_id"
        ],
    )
)

# ============================================================
# ORIGINAL NEGATIVE PAIRS
# ============================================================

negative_keys = set(
    zip(
        pairwise["person_id"],
        pairwise[
            "negative_product_id"
        ],
    )
)

print(
    "\nOriginal negative pair keys:",
    len(negative_keys)
)

# ============================================================
# PHYSICAL DISTANCE
# ============================================================
#
# Lower distance = more physically similar.
#
# We calculate normalized mismatch using available
# measurements.
#
# This is ONLY used to construct evaluation negatives.
# It is NOT given to the model directly.
# ============================================================

def physical_distance(
    person_id,
    product_id,
):

    p = person_lookup.loc[
        person_id
    ]

    g = garment_lookup.loc[
        product_id
    ]

    distances = []

    pairs = [
        (
            p["bust"],
            g["garment_bust"],
        ),
        (
            p["waist"],
            g["garment_waist"],
        ),
        (
            p["hips"],
            g["garment_hip"],
        ),
        (
            p["height"],
            g["garment_length"],
        ),
    ]

    for person_value, garment_value in pairs:

        if (
            pd.isna(person_value)
            or pd.isna(garment_value)
        ):
            continue

        # Relative mismatch
        if person_value != 0:

            distances.append(
                abs(
                    garment_value -
                    person_value
                )
                / abs(person_value)
            )

    if not distances:
        return np.nan

    return float(
        np.mean(distances)
    )


# ============================================================
# BUILD TEST POSITIVES
# ============================================================

test_positive_by_person = {}

for person in test_people:

    products = set(
        pairwise.loc[
            pairwise["person_id"]
            == person,
            "positive_product_id",
        ]
    )

    test_positive_by_person[
        person
    ] = sorted(products)

print("\nTest positive interactions:")

for person in sorted(test_positive_by_person):

    print(
        person,
        ":",
        len(
            test_positive_by_person[
                person
            ]
        )
    )

# ============================================================
# BUILD ALL FRESH CANDIDATES
# ============================================================

fresh_candidates = {}

for person in sorted(test_people):

    positive_products = set(
        test_positive_by_person[
            person
        ]
    )

    candidates = []

    for product in all_products:

        key = (
            person,
            product,
        )

        # Never positive
        if product in positive_products:
            continue

        # Never used by original negative sampler
        if key in negative_keys:
            continue

        # Must have at least some measurable information
        try:

            vec = make_features(
                person,
                product,
            )

            if np.isnan(vec).all():
                continue

        except Exception:
            continue

        distance = physical_distance(
            person,
            product,
        )

        if np.isnan(distance):
            continue

        candidates.append(
            (
                product,
                distance,
            )
        )

    fresh_candidates[
        person
    ] = candidates

# ============================================================
# CREATE THREE EVALUATION SETS
# ============================================================
#
# RANDOM:
#   random fresh negative
#
# HARD:
#   physically closest fresh negative
#
# MIXED:
#   50% hard + 50% random
#
# We use ONE negative per positive interaction.
# ============================================================

evaluation_sets = {
    "random": [],
    "hard": [],
    "mixed": [],
}

for person in sorted(test_people):

    positives = test_positive_by_person[
        person
    ]

    candidates = fresh_candidates[
        person
    ]

    if len(candidates) == 0:
        continue

    candidates_sorted = sorted(
        candidates,
        key=lambda x: x[1],
    )

    # Top 20% = hard candidate pool
    hard_pool_size = max(
        1,
        int(
            len(candidates_sorted)
            * 0.20
        ),
    )

    hard_pool = candidates_sorted[
        :hard_pool_size
    ]

    for index, positive_product in enumerate(
        positives
    ):

        # ------------------------
        # Random
        # ------------------------

        random_product, random_distance = (
            candidates[
                rng.integers(
                    0,
                    len(candidates),
                )
            ]
        )

        # ------------------------
        # Hard
        # ------------------------

        hard_product, hard_distance = (
            hard_pool[
                index
                % len(hard_pool)
            ]
        )

        # ------------------------
        # Mixed
        # ------------------------

        if index % 2 == 0:

            mixed_product = (
                hard_product
            )

            mixed_distance = (
                hard_distance
            )

            mixed_type = "hard"

        else:

            mixed_product = (
                random_product
            )

            mixed_distance = (
                random_distance
            )

            mixed_type = "random"

        evaluation_sets[
            "random"
        ].append(
            {
                "person_id": person,
                "positive_product_id":
                    positive_product,
                "negative_product_id":
                    random_product,
                "negative_distance":
                    random_distance,
                "negative_type":
                    "random",
            }
        )

        evaluation_sets[
            "hard"
        ].append(
            {
                "person_id": person,
                "positive_product_id":
                    positive_product,
                "negative_product_id":
                    hard_product,
                "negative_distance":
                    hard_distance,
                "negative_type":
                    "hard",
            }
        )

        evaluation_sets[
            "mixed"
        ].append(
            {
                "person_id": person,
                "positive_product_id":
                    positive_product,
                "negative_product_id":
                    mixed_product,
                "negative_distance":
                    mixed_distance,
                "negative_type":
                    mixed_type,
            }
        )

# ============================================================
# VERIFY NO TRAINING NEGATIVE OVERLAP
# ============================================================

print("\n" + "=" * 65)
print("FRESH NEGATIVE INTEGRITY")
print("=" * 65)

for name, rows in evaluation_sets.items():

    df = pd.DataFrame(rows)

    eval_negative_keys = set(
        zip(
            df["person_id"],
            df["negative_product_id"],
        )
    )

    overlap = (
        eval_negative_keys
        & negative_keys
    )

    positive_collision = (
        eval_negative_keys
        & positive_keys
    )

    print(
        f"{name:10s}"
        f" | rows={len(df):4d}"
        f" | train-negative overlap={len(overlap)}"
        f" | positive collision={len(positive_collision)}"
    )

# ============================================================
# REBUILD PREPROCESSING
# ============================================================
#
# IMPORTANT:
# Fit imputer/scaler using TRAIN PEOPLE ONLY.
# ============================================================

train_rows = pairwise[
    pairwise["person_id"].isin(
        train_people
    )
]

train_vectors = []

for _, row in train_rows.iterrows():

    train_vectors.append(
        make_features(
            row["person_id"],
            row["positive_product_id"],
        )
    )

    train_vectors.append(
        make_features(
            row["person_id"],
            row["negative_product_id"],
        )
    )

train_vectors = np.stack(
    train_vectors
)

imputer = SimpleImputer(
    strategy="median",
    add_indicator=True,
)

train_imputed = (
    imputer.fit_transform(
        train_vectors
    )
)

scaler = StandardScaler()

scaler.fit(
    train_imputed
)

# ============================================================
# MODEL
# ============================================================

class CompatibilityRanker(nn.Module):

    def __init__(
        self,
        input_dim,
    ):

        super().__init__()

        self.network = nn.Sequential(

            nn.Linear(
                input_dim,
                64,
            ),

            nn.ReLU(),

            nn.Linear(
                64,
                32,
            ),

            nn.ReLU(),

            nn.Linear(
                32,
                1,
            ),
        )

    def forward(self, x):

        return self.network(
            x
        ).squeeze(-1)


model = CompatibilityRanker(
    checkpoint[
        "input_dim"
    ]
)

model.load_state_dict(
    checkpoint[
        "model_state_dict"
    ]
)

model.eval()

print("\nModel loaded.")

# ============================================================
# EVALUATION FUNCTION
# ============================================================

def evaluate_dataset(
    frame,
):

    positive_vectors = []
    negative_vectors = []

    for _, row in frame.iterrows():

        positive_vectors.append(
            make_features(
                row["person_id"],
                row["positive_product_id"],
            )
        )

        negative_vectors.append(
            make_features(
                row["person_id"],
                row["negative_product_id"],
            )
        )

    positive_vectors = np.stack(
        positive_vectors
    )

    negative_vectors = np.stack(
        negative_vectors
    )

    # Impute
    positive_vectors = (
        imputer.transform(
            positive_vectors
        )
    )

    negative_vectors = (
        imputer.transform(
            negative_vectors
        )
    )

    # Scale
    positive_vectors = (
        scaler.transform(
            positive_vectors
        )
    )

    negative_vectors = (
        scaler.transform(
            negative_vectors
        )
    )

    pos_tensor = torch.tensor(
        positive_vectors,
        dtype=torch.float32,
    )

    neg_tensor = torch.tensor(
        negative_vectors,
        dtype=torch.float32,
    )

    with torch.no_grad():

        positive_scores = (
            model(
                pos_tensor
            )
            .numpy()
        )

        negative_scores = (
            model(
                neg_tensor
            )
            .numpy()
        )

    # Pairwise accuracy
    pairwise_accuracy = float(
        np.mean(
            positive_scores
            >
            negative_scores
        )
    )

    # Classification-style labels
    y_true = np.concatenate(
        [
            np.ones(
                len(positive_scores)
            ),
            np.zeros(
                len(negative_scores)
            ),
        ]
    )

    y_score = np.concatenate(
        [
            positive_scores,
            negative_scores,
        ]
    )

    roc_auc = roc_auc_score(
        y_true,
        y_score,
    )

    pr_auc = average_precision_score(
        y_true,
        y_score,
    )

    score_gap = (
        positive_scores -
        negative_scores
    )

    return {
        "pairs": len(frame),
        "pairwise_accuracy":
            pairwise_accuracy,
        "roc_auc":
            roc_auc,
        "pr_auc":
            pr_auc,
        "mean_positive_score":
            float(
                np.mean(
                    positive_scores
                )
            ),
        "mean_negative_score":
            float(
                np.mean(
                    negative_scores
                )
            ),
        "mean_score_gap":
            float(
                np.mean(
                    score_gap
                )
            ),
        "median_score_gap":
            float(
                np.median(
                    score_gap
                )
            ),
    }


# ============================================================
# GLOBAL RESULTS
# ============================================================

results = {}

print("\n" + "=" * 65)
print("GLOBAL GENERALIZATION RESULTS")
print("=" * 65)

for name, rows in evaluation_sets.items():

    df = pd.DataFrame(rows)

    result = evaluate_dataset(
        df
    )

    results[name] = result

    print(
        f"\n{name.upper()}"
    )

    print(
        "Pairs:",
        result["pairs"]
    )

    print(
        "Pairwise Accuracy:",
        f'{result["pairwise_accuracy"]:.4f}'
    )

    print(
        "ROC-AUC:",
        f'{result["roc_auc"]:.4f}'
    )

    print(
        "PR-AUC:",
        f'{result["pr_auc"]:.4f}'
    )

    print(
        "Mean positive score:",
        f'{result["mean_positive_score"]:.4f}'
    )

    print(
        "Mean negative score:",
        f'{result["mean_negative_score"]:.4f}'
    )

    print(
        "Mean score gap:",
        f'{result["mean_score_gap"]:.4f}'
    )

    print(
        "Median score gap:",
        f'{result["median_score_gap"]:.4f}'
    )

# ============================================================
# PER-PERSON PERFORMANCE
# ============================================================

print("\n" + "=" * 65)
print("PER-PERSON PERFORMANCE")
print("=" * 65)

person_results = []

for name, rows in evaluation_sets.items():

    df = pd.DataFrame(rows)

    for person in sorted(
        df["person_id"].unique()
    ):

        subset = df[
            df["person_id"]
            == person
        ]

        result = evaluate_dataset(
            subset
        )

        person_results.append(
            {
                "evaluation":
                    name,
                "person_id":
                    person,
                "pairs":
                    result["pairs"],
                "pairwise_accuracy":
                    result[
                        "pairwise_accuracy"
                    ],
                "roc_auc":
                    result[
                        "roc_auc"
                    ],
            }
        )

person_results_df = pd.DataFrame(
    person_results
)

print(
    person_results_df
    .groupby("evaluation")[
        [
            "pairwise_accuracy",
            "roc_auc",
        ]
    ]
    .agg(
        [
            "mean",
            "std",
            "min",
            "max",
        ]
    )
)

# ============================================================
# MISSINGNESS ANALYSIS
# ============================================================

print("\n" + "=" * 65)
print("COMPLETE vs INCOMPLETE MEASUREMENTS")
print("=" * 65)

# Determine completeness from garment table
garment_completeness = (
    garment_lookup[
        garment_cols
    ]
    .notna()
    .sum(axis=1)
)

garment_completeness = (
    garment_completeness
    >= len(garment_cols)
)

for name, rows in evaluation_sets.items():

    df = pd.DataFrame(rows)

    complete_mask = []

    for _, row in df.iterrows():

        product = (
            row[
                "positive_product_id"
            ]
        )

        complete_mask.append(
            bool(
                garment_completeness
                .get(
                    product,
                    False,
                )
            )
        )

    df["positive_complete"] = (
        complete_mask
    )

    for category, subset in df.groupby(
        "positive_complete"
    ):

        if len(subset) == 0:
            continue

        result = evaluate_dataset(
            subset
        )

        print(
            f"\n{name.upper()} | "
            f"positive garment complete="
            f"{category}"
        )

        print(
            "Pairs:",
            len(subset)
        )

        print(
            "Pairwise accuracy:",
            f'{result["pairwise_accuracy"]:.4f}'
        )

        print(
            "ROC-AUC:",
            f'{result["roc_auc"]:.4f}'
        )

        print(
            "PR-AUC:",
            f'{result["pr_auc"]:.4f}'
        )

# ============================================================
# NEGATIVE DISTANCE ANALYSIS
# ============================================================

print("\n" + "=" * 65)
print("NEGATIVE DIFFICULTY")
print("=" * 65)

for name, rows in evaluation_sets.items():

    df = pd.DataFrame(rows)

    print(
        f"\n{name.upper()}"
    )

    print(
        df["negative_distance"]
        .describe()
    )

# ============================================================
# FINAL COMPARISON TABLE
# ============================================================

comparison = pd.DataFrame(
    results
).T

print("\n" + "=" * 65)
print("FINAL BENCHMARK TABLE")
print("=" * 65)

display(
    comparison[
        [
            "pairs",
            "pairwise_accuracy",
            "roc_auc",
            "pr_auc",
            "mean_score_gap",
        ]
    ]
)

# ============================================================
# ORIGINAL RESULT
# ============================================================

print("\n" + "=" * 65)
print("REFERENCE")
print("=" * 65)

print(
    "Original test ranking accuracy : 0.8417"
)

print(
    "Previous fresh-negative accuracy: 0.6254"
)

# ============================================================
# INTERPRETATION
# ============================================================

random_acc = results[
    "random"
]["pairwise_accuracy"]

hard_acc = results[
    "hard"
]["pairwise_accuracy"]

mixed_acc = results[
    "mixed"
]["pairwise_accuracy"]

print("\n" + "=" * 65)
print("INTERPRETATION")
print("=" * 65)

print(
    f"Random unseen : {random_acc:.4f}"
)

print(
    f"Hard unseen   : {hard_acc:.4f}"
)

print(
    f"Mixed unseen  : {mixed_acc:.4f}"
)

if hard_acc >= 0.70:

    print(
        "\nSTRONG GENERALIZATION:"
        "\nThe model remains effective "
        "against physically difficult "
        "unseen negatives."
    )

elif hard_acc >= 0.60:

    print(
        "\nMODERATE GENERALIZATION:"
        "\nThe model has real compatibility "
        "signal, but hard negatives remain "
        "challenging."
    )

else:

    print(
        "\nWEAK GENERALIZATION:"
        "\nThe model struggles against "
        "physically similar unseen garments."
    )

# ============================================================
# SAVE BENCHMARK
# ============================================================

OUTPUT_PATH = (
    BASE /
    "data/v2/v2_generalization_benchmark.csv"
)

comparison.to_csv(
    OUTPUT_PATH
)

print("\n" + "=" * 65)
print("BENCHMARK COMPLETE")
print("=" * 65)

print(
    "Saved:",
    OUTPUT_PATH
)
```

```python
# ============================================================
# ZYRA V2 — EXPERIMENT #2
# BALANCED HARD / SEMI-HARD / RANDOM NEGATIVE SAMPLING
# ============================================================
#
# Goal:
#   Fix the weakness observed in Experiment #1:
#
#       Random unseen  : ~57.5%
#       Hard unseen    : ~84.7%
#
# We now train with:
#
#   1/3 hard negatives
#   1/3 semi-hard negatives
#   1/3 random negatives
#
# IMPORTANT:
# - Uses only positive interactions from the cleaned dataset
# - Does NOT use labels from the old negative dataset
# - Does NOT create positive collisions
# - Does NOT create duplicate (person, negative garment) pairs
# - Keeps negatives diverse per person
# - Does NOT train the model yet
#
# ============================================================

import pandas as pd
import numpy as np

from pathlib import Path

# ============================================================
# CONFIG
# ============================================================

BASE = Path(
    "/Users/saketh/Desktop/Projects/weavly/core-model"
)

RAW_PATH = (
    BASE /
    "data/v2/v2_compatibility_raw.csv"
)

OUTPUT_PATH = (
    BASE /
    "data/v2/v2_balanced_negative_sampling.csv"
)

SEED = 2026

rng = np.random.default_rng(SEED)

# Number of negatives PER positive
# 3 types -> 3 negatives per positive
NEGATIVES_PER_POSITIVE = 3

# ============================================================
# LOAD
# ============================================================

df = pd.read_csv(
    RAW_PATH
)

print("=" * 65)
print("ZYRA V2 — EXPERIMENT #2 NEGATIVE SAMPLING")
print("=" * 65)

print(
    "Input rows:",
    len(df)
)

print(
    "People:",
    df["person_id"].nunique()
)

print(
    "Garments:",
    df["product_id"].nunique()
)

# ============================================================
# KEEP ONLY POSITIVE INTERACTIONS
# ============================================================

positive_df = df[
    df["label"] == 1
].copy()

# Remove accidental duplicates
positive_df = (
    positive_df
    .drop_duplicates(
        [
            "person_id",
            "product_id",
        ]
    )
)

print(
    "\nPositive interactions:",
    len(positive_df)
)

# ============================================================
# BUILD PERSON PROFILES
# ============================================================

person_cols = [
    "height",
    "bust",
    "waist",
    "hips",
]

person_lookup = (
    positive_df[
        [
            "person_id"
        ] + person_cols
    ]
    .drop_duplicates(
        "person_id"
    )
    .set_index(
        "person_id"
    )
)

# ============================================================
# BUILD GARMENT PROFILES
# ============================================================

garment_cols = [
    "garment_bust",
    "garment_waist",
    "garment_hip",
    "garment_length",
    "garment_sleeve",
]

garment_lookup = (
    df[
        [
            "product_id"
        ] + garment_cols
    ]
    .drop_duplicates(
        "product_id"
    )
    .set_index(
        "product_id"
    )
)

all_products = list(
    garment_lookup.index
)

print(
    "Available garments:",
    len(all_products)
)

# ============================================================
# PHYSICAL DISTANCE
# ============================================================
#
# Smaller = physically closer to person's body dimensions.
#
# Used ONLY for negative selection.
#
# We deliberately do NOT use the model for negative selection.
# ============================================================

def physical_distance(
    person_id,
    product_id,
):

    person = person_lookup.loc[
        person_id
    ]

    garment = garment_lookup.loc[
        product_id
    ]

    values = []

    dimensions = [
        (
            person["bust"],
            garment["garment_bust"],
        ),
        (
            person["waist"],
            garment["garment_waist"],
        ),
        (
            person["hips"],
            garment["garment_hip"],
        ),
        (
            person["height"],
            garment["garment_length"],
        ),
    ]

    for person_value, garment_value in dimensions:

        if (
            pd.isna(person_value)
            or pd.isna(garment_value)
        ):
            continue

        if person_value == 0:
            continue

        values.append(
            abs(
                garment_value -
                person_value
            )
            / abs(person_value)
        )

    if not values:
        return np.nan

    return float(
        np.mean(values)
    )

# ============================================================
# PRECOMPUTE DISTANCES
# ============================================================

print(
    "\nComputing physical candidate distances..."
)

distance_cache = {}

for person_id in person_lookup.index:

    distance_cache[
        person_id
    ] = {}

    for product_id in all_products:

        try:

            distance = physical_distance(
                person_id,
                product_id,
            )

            if not np.isnan(distance):

                distance_cache[
                    person_id
                ][
                    product_id
                ] = distance

        except Exception:

            continue

print(
    "Distance computation complete."
)

# ============================================================
# BUILD POSITIVE PRODUCT SET
# ============================================================

positive_products_by_person = (
    positive_df
    .groupby("person_id")[
        "product_id"
    ]
    .apply(set)
    .to_dict()
)

# ============================================================
# NEGATIVE SAMPLING
# ============================================================
#
# For each positive interaction:
#
# HARD:
#   closest 10% of available negative garments
#
# SEMI-HARD:
#   30%-60% distance percentile
#
# RANDOM:
#   uniformly random from all available negatives
#
# We ensure that the SAME negative pair
# (person, garment) isn't repeated.
#
# ============================================================

output_rows = []

used_negative_pairs = set()

sampling_statistics = []

for person_id in sorted(
    positive_products_by_person
):

    positive_products = (
        positive_products_by_person[
            person_id
        ]
    )

    available = (
        distance_cache
        .get(
            person_id,
            {}
        )
    )

    # Remove positive garments
    candidates = [
        (
            product,
            distance,
        )
        for product, distance
        in available.items()
        if product
        not in positive_products
    ]

    if len(candidates) < 3:

        print(
            f"WARNING: {person_id} "
            f"has only {len(candidates)} "
            f"negative candidates."
        )

        continue

    # Sort by physical distance
    candidates = sorted(
        candidates,
        key=lambda x: x[1]
    )

    n = len(candidates)

    # --------------------------------------------------------
    # Percentile boundaries
    # --------------------------------------------------------

    hard_end = max(
        1,
        int(
            n * 0.10
        )
    )

    semi_start = max(
        hard_end,
        int(
            n * 0.30
        )
    )

    semi_end = max(
        semi_start + 1,
        int(
            n * 0.60
        )
    )

    hard_pool = candidates[
        :hard_end
    ]

    semi_pool = candidates[
        semi_start:semi_end
    ]

    random_pool = candidates

    # Safety
    if not semi_pool:
        semi_pool = candidates

    # --------------------------------------------------------
    # Sample one of each type for every positive
    # --------------------------------------------------------

    person_used = set()

    for positive_product in sorted(
        positive_products
    ):

        selected = []

        # ====================================================
        # HARD
        # ====================================================

        hard_candidates = [
            item
            for item in hard_pool
            if item[0]
            not in person_used
        ]

        if not hard_candidates:

            hard_candidates = [
                item
                for item in hard_pool
            ]

        hard_product, hard_distance = (
            hard_candidates[
                rng.integers(
                    len(
                        hard_candidates
                    )
                )
            ]
        )

        person_used.add(
            hard_product
        )

        selected.append(
            (
                hard_product,
                hard_distance,
                "hard",
            )
        )

        # ====================================================
        # SEMI-HARD
        # ====================================================

        semi_candidates = [
            item
            for item in semi_pool
            if item[0]
            not in person_used
        ]

        if not semi_candidates:

            semi_candidates = [
                item
                for item in semi_pool
            ]

        semi_product, semi_distance = (
            semi_candidates[
                rng.integers(
                    len(
                        semi_candidates
                    )
                )
            ]
        )

        person_used.add(
            semi_product
        )

        selected.append(
            (
                semi_product,
                semi_distance,
                "semi-hard",
            )
        )

        # ====================================================
        # RANDOM
        # ====================================================

        random_candidates = [
            item
            for item in random_pool
            if item[0]
            not in person_used
        ]

        if not random_candidates:

            random_candidates = [
                item
                for item in random_pool
            ]

        random_product, random_distance = (
            random_candidates[
                rng.integers(
                    len(
                        random_candidates
                    )
                )
            ]
        )

        person_used.add(
            random_product
        )

        selected.append(
            (
                random_product,
                random_distance,
                "random",
            )
        )

        # ====================================================
        # STORE
        # ====================================================

        for (
            negative_product,
            distance,
            negative_type,
        ) in selected:

            key = (
                person_id,
                negative_product,
            )

            # ------------------------------------------------
            # Global duplicate protection
            # ------------------------------------------------

            if key in used_negative_pairs:

                # Find an unused fallback
                fallback = [
                    item
                    for item
                    in random_pool
                    if (
                        person_id,
                        item[0],
                    )
                    not in used_negative_pairs
                    and item[0]
                    not in positive_products
                ]

                if fallback:

                    negative_product, distance = (
                        fallback[
                            rng.integers(
                                len(
                                    fallback
                                )
                            )
                        )

                    key = (
                        person_id,
                        negative_product,
                    )

                else:

                    continue

            used_negative_pairs.add(
                key
            )

            output_rows.append(
                {
                    "person_id":
                        person_id,

                    "positive_product_id":
                        positive_product,

                    "negative_product_id":
                        negative_product,

                    "negative_type":
                        negative_type,

                    "negative_distance":
                        distance,

                    "label_positive":
                        1,

                    "label_negative":
                        0,
                }
            )

# ============================================================
# DATAFRAME
# ============================================================

ranking_df = pd.DataFrame(
    output_rows
)

# ============================================================
# BASIC INTEGRITY
# ============================================================

print("\n" + "=" * 65)
print("SAMPLING RESULTS")
print("=" * 65)

print(
    "Positive interactions:",
    len(positive_df)
)

print(
    "Pairwise ranking rows:",
    len(ranking_df)
)

print(
    "\nNegative type counts:"
)

print(
    ranking_df[
        "negative_type"
    ].value_counts()
)

# ============================================================
# EXPECTED COUNTS
# ============================================================

expected_per_type = len(
    positive_df
)

print(
    "\nExpected hard negatives:",
    expected_per_type
)

print(
    "Actual hard negatives:",
    (
        ranking_df[
            "negative_type"
        ]
        == "hard"
    ).sum()
)

print(
    "Expected semi-hard negatives:",
    expected_per_type
)

print(
    "Actual semi-hard negatives:",
    (
        ranking_df[
            "negative_type"
        ]
        == "semi-hard"
    ).sum()
)

print(
    "Expected random negatives:",
    expected_per_type
)

print(
    "Actual random negatives:",
    (
        ranking_df[
            "negative_type"
        ]
        == "random"
    ).sum()
)

# ============================================================
# DUPLICATE CHECK
# ============================================================

duplicate_pairs = (
    ranking_df[
        [
            "person_id",
            "positive_product_id",
            "negative_product_id",
        ]
    ]
    .duplicated()
    .sum()
)

duplicate_negative_pairs = (
    ranking_df[
        [
            "person_id",
            "negative_product_id",
        ]
    ]
    .duplicated()
    .sum()
)

print(
    "\nDuplicate full pair rows:",
    duplicate_pairs
)

print(
    "Duplicate person-negative pairs:",
    duplicate_negative_pairs
)

# ============================================================
# POSITIVE COLLISION
# ============================================================

positive_keys = set(
    zip(
        positive_df[
            "person_id"
        ],
        positive_df[
            "product_id"
        ],
    )
)

negative_keys = set(
    zip(
        ranking_df[
            "person_id"
        ],
        ranking_df[
            "negative_product_id"
        ],
    )
)

collision_count = len(
    positive_keys
    &
    negative_keys
)

print(
    "Positive-negative collisions:",
    collision_count
)

# ============================================================
# NEGATIVE DISTANCE DISTRIBUTION
# ============================================================

print("\n" + "=" * 65)
print("NEGATIVE DISTANCE DISTRIBUTION")
print("=" * 65)

distance_summary = (
    ranking_df
    .groupby(
        "negative_type"
    )[
        "negative_distance"
    ]
    .describe()
)

print(
    distance_summary
)

# ============================================================
# TYPE ORDER CHECK
# ============================================================

print("\n" + "=" * 65)
print("EXPECTED DIFFICULTY ORDER")
print("=" * 65)

mean_distances = (
    ranking_df
    .groupby(
        "negative_type"
    )[
        "negative_distance"
    ]
    .mean()
    .sort_values()
)

print(
    mean_distances
)

hard_mean = (
    ranking_df.loc[
        ranking_df[
            "negative_type"
        ]
        == "hard",
        "negative_distance",
    ]
    .mean()
)

semi_mean = (
    ranking_df.loc[
        ranking_df[
            "negative_type"
        ]
        == "semi-hard",
        "negative_distance",
    ]
    .mean()
)

random_mean = (
    ranking_df.loc[
        ranking_df[
            "negative_type"
        ]
        == "random",
        "negative_distance",
    ]
    .mean()
)

print(
    "\nHard < Semi-hard < Random:",
    hard_mean < semi_mean < random_mean
)

# ============================================================
# PEOPLE BALANCE
# ============================================================

print("\n" + "=" * 65)
print("PER-PERSON NEGATIVE COUNTS")
print("=" * 65)

person_counts = (
    ranking_df
    .groupby(
        [
            "person_id",
            "negative_type",
        ]
    )
    .size()
    .unstack(
        fill_value=0
    )
)

print(
    person_counts
    .describe()
)

# ============================================================
# POSITIVE / NEGATIVE RATIO
# ============================================================

print("\n" + "=" * 65)
print("TRAINING RATIO")
print("=" * 65)

print(
    "Positive interactions:",
    len(positive_df)
)

print(
    "Negative interactions:",
    len(ranking_df)
)

print(
    "Positive : Negative ratio = "
    f"1 : {len(ranking_df) / len(positive_df):.2f}"
)

# ============================================================
# EXAMPLE PAIRS
# ============================================================

print("\n" + "=" * 65)
print("EXAMPLE PAIRS")
print("=" * 65)

display(
    ranking_df[
        [
            "person_id",
            "positive_product_id",
            "negative_product_id",
            "negative_type",
            "negative_distance",
        ]
    ]
    .head(20)
)

# ============================================================
# SAVE
# ============================================================

ranking_df.to_csv(
    OUTPUT_PATH,
    index=False,
)

print("\n" + "=" * 65)
print("EXPERIMENT #2 DATASET SAVED")
print("=" * 65)

print(
    OUTPUT_PATH
)

# ============================================================
# FINAL VALIDATION
# ============================================================

print("\n" + "=" * 65)
print("FINAL VALIDATION")
print("=" * 65)

checks = {
    "No duplicate full pairs":
        duplicate_pairs == 0,

    "No duplicate person-negative pairs":
        duplicate_negative_pairs == 0,

    "No positive-negative collisions":
        collision_count == 0,

    "Hard < Semi-hard < Random":
        hard_mean < semi_mean < random_mean,

    "All three negative types exist":
        set(
            ranking_df[
                "negative_type"
            ]
        )
        == {
            "hard",
            "semi-hard",
            "random",
        },
}

for name, passed in checks.items():

    print(
        f"{name}:",
        "PASS" if passed else "FAIL"
    )

all_passed = all(
    checks.values()
)

print(
    "\nOVERALL:",
    "PASS" if all_passed else "FAIL"
)

print("\nExperiment #2 sampling complete.")
```

```python
# ============================================================
# ZYRA V2 — EXPERIMENT #2
# BALANCED HARD / SEMI-HARD / RANDOM NEGATIVE SAMPLING
# ============================================================

import pandas as pd
import numpy as np
from pathlib import Path

# ============================================================
# CONFIG
# ============================================================

BASE = Path(
    "/Users/saketh/Desktop/Projects/weavly/core-model"
)

RAW_PATH = (
    BASE /
    "data/v2/v2_compatibility_raw.csv"
)

OUTPUT_PATH = (
    BASE /
    "data/v2/v2_balanced_negative_sampling.csv"
)

SEED = 2026
rng = np.random.default_rng(SEED)

# ============================================================
# LOAD DATA
# ============================================================

df = pd.read_csv(RAW_PATH)

print("=" * 65)
print("ZYRA V2 — EXPERIMENT #2 NEGATIVE SAMPLING")
print("=" * 65)

print("Input rows :", len(df))
print("People     :", df["person_id"].nunique())
print("Garments   :", df["product_id"].nunique())

# ============================================================
# POSITIVE INTERACTIONS
# ============================================================

positive_df = (
    df[df["label"] == 1]
    .drop_duplicates(
        ["person_id", "product_id"]
    )
    .copy()
)

print(
    "\nPositive interactions:",
    len(positive_df)
)

# ============================================================
# PERSON LOOKUP
# ============================================================

person_cols = [
    "height",
    "bust",
    "waist",
    "hips",
]

person_lookup = (
    positive_df[
        ["person_id"] + person_cols
    ]
    .drop_duplicates("person_id")
    .set_index("person_id")
)

# ============================================================
# GARMENT LOOKUP
# ============================================================

garment_cols = [
    "garment_bust",
    "garment_waist",
    "garment_hip",
    "garment_length",
    "garment_sleeve",
]

garment_lookup = (
    df[
        ["product_id"] + garment_cols
    ]
    .drop_duplicates("product_id")
    .set_index("product_id")
)

all_products = list(
    garment_lookup.index
)

print(
    "Available garments:",
    len(all_products)
)

# ============================================================
# PHYSICAL DISTANCE
# ============================================================

def physical_distance(
    person_id,
    product_id,
):

    person = person_lookup.loc[
        person_id
    ]

    garment = garment_lookup.loc[
        product_id
    ]

    values = []

    dimensions = [
        (
            person["bust"],
            garment["garment_bust"],
        ),
        (
            person["waist"],
            garment["garment_waist"],
        ),
        (
            person["hips"],
            garment["garment_hip"],
        ),
        (
            person["height"],
            garment["garment_length"],
        ),
    ]

    for pv, gv in dimensions:

        if pd.isna(pv) or pd.isna(gv):
            continue

        if pv == 0:
            continue

        values.append(
            abs(gv - pv) / abs(pv)
        )

    if not values:
        return np.nan

    return float(
        np.mean(values)
    )

# ============================================================
# PRECOMPUTE DISTANCES
# ============================================================

print(
    "\nComputing physical distances..."
)

distance_cache = {}

for person_id in person_lookup.index:

    distance_cache[person_id] = {}

    for product_id in all_products:

        try:

            distance = physical_distance(
                person_id,
                product_id,
            )

            if not np.isnan(distance):

                distance_cache[
                    person_id
                ][
                    product_id
                ] = distance

        except Exception:
            continue

print(
    "Distance computation complete."
)

# ============================================================
# POSITIVE PRODUCTS BY PERSON
# ============================================================

positive_products_by_person = (
    positive_df
    .groupby("person_id")["product_id"]
    .apply(set)
    .to_dict()
)

# ============================================================
# SAMPLE NEGATIVES
# ============================================================

output_rows = []

used_negative_pairs = set()

for person_id in sorted(
    positive_products_by_person
):

    positive_products = (
        positive_products_by_person[
            person_id
        ]
    )

    available = (
        distance_cache.get(
            person_id,
            {}
        )
    )

    # Remove positive garments
    candidates = [
        (
            product,
            distance
        )
        for product, distance
        in available.items()
        if product not in positive_products
    ]

    if len(candidates) < 3:

        print(
            "WARNING:",
            person_id,
            "has only",
            len(candidates),
            "negative candidates."
        )

        continue

    # --------------------------------------------------------
    # Sort by physical distance
    # --------------------------------------------------------

    candidates = sorted(
        candidates,
        key=lambda x: x[1]
    )

    n = len(candidates)

    # --------------------------------------------------------
    # Define pools
    # --------------------------------------------------------
    #
    # HARD:
    # closest 10%
    #
    # SEMI-HARD:
    # 30th–60th percentile
    #
    # RANDOM:
    # entire candidate pool
    #

    hard_end = max(
        1,
        int(n * 0.10)
    )

    semi_start = max(
        hard_end,
        int(n * 0.30)
    )

    semi_end = max(
        semi_start + 1,
        int(n * 0.60)
    )

    hard_pool = candidates[
        :hard_end
    ]

    semi_pool = candidates[
        semi_start:semi_end
    ]

    random_pool = candidates

    if not semi_pool:
        semi_pool = candidates

    # --------------------------------------------------------
    # One hard + one semi-hard + one random
    # for every positive interaction
    # --------------------------------------------------------

    for positive_product in sorted(
        positive_products
    ):

        # ================================================
        # HARD
        # ================================================

        hard_available = [
            item
            for item in hard_pool
            if (
                person_id,
                item[0],
            )
            not in used_negative_pairs
        ]

        if not hard_available:
            hard_available = hard_pool

        hard_product, hard_distance = (
            hard_available[
                rng.integers(
                    len(hard_available)
                )
            ]
        )

        hard_key = (
            person_id,
            hard_product,
        )

        # ================================================
        # SEMI-HARD
        # ================================================

        semi_available = [
            item
            for item in semi_pool
            if (
                person_id,
                item[0],
            )
            not in used_negative_pairs
            and item[0] != hard_product
        ]

        if not semi_available:
            semi_available = [
                item
                for item in semi_pool
                if item[0] != hard_product
            ]

        semi_product, semi_distance = (
            semi_available[
                rng.integers(
                    len(semi_available)
                )
            ]
        )

        semi_key = (
            person_id,
            semi_product,
        )

        # ================================================
        # RANDOM
        # ================================================

        random_available = [
            item
            for item in random_pool
            if (
                person_id,
                item[0],
            )
            not in used_negative_pairs
            and item[0] != hard_product
            and item[0] != semi_product
        ]

        if not random_available:

            random_available = [
                item
                for item in random_pool
                if item[0] != hard_product
                and item[0] != semi_product
            ]

        random_product, random_distance = (
            random_available[
                rng.integers(
                    len(random_available)
                )
            ]
        )

        random_key = (
            person_id,
            random_product,
        )

        # ================================================
        # STORE HARD
        # ================================================

        if hard_key not in used_negative_pairs:

            used_negative_pairs.add(
                hard_key
            )

            output_rows.append(
                {
                    "person_id":
                        person_id,

                    "positive_product_id":
                        positive_product,

                    "negative_product_id":
                        hard_product,

                    "negative_type":
                        "hard",

                    "negative_distance":
                        hard_distance,

                    "label_positive":
                        1,

                    "label_negative":
                        0,
                }
            )

        # ================================================
        # STORE SEMI-HARD
        # ================================================

        if semi_key not in used_negative_pairs:

            used_negative_pairs.add(
                semi_key
            )

            output_rows.append(
                {
                    "person_id":
                        person_id,

                    "positive_product_id":
                        positive_product,

                    "negative_product_id":
                        semi_product,

                    "negative_type":
                        "semi-hard",

                    "negative_distance":
                        semi_distance,

                    "label_positive":
                        1,

                    "label_negative":
                        0,
                }
            )

        # ================================================
        # STORE RANDOM
        # ================================================

        if random_key not in used_negative_pairs:

            used_negative_pairs.add(
                random_key
            )

            output_rows.append(
                {
                    "person_id":
                        person_id,

                    "positive_product_id":
                        positive_product,

                    "negative_product_id":
                        random_product,

                    "negative_type":
                        "random",

                    "negative_distance":
                        random_distance,

                    "label_positive":
                        1,

                    "label_negative":
                        0,
                }
            )

# ============================================================
# DATAFRAME
# ============================================================

ranking_df = pd.DataFrame(
    output_rows
)

# ============================================================
# BASIC RESULTS
# ============================================================

print("\n" + "=" * 65)
print("SAMPLING RESULTS")
print("=" * 65)

print(
    "Positive interactions :",
    len(positive_df)
)

print(
    "Ranking samples       :",
    len(ranking_df)
)

print(
    "\nNegative type counts:"
)

print(
    ranking_df[
        "negative_type"
    ].value_counts()
)

# ============================================================
# EXPECTED COUNTS
# ============================================================

expected = len(
    positive_df
)

hard_actual = (
    ranking_df[
        "negative_type"
    ]
    .eq("hard")
    .sum()
)

semi_actual = (
    ranking_df[
        "negative_type"
    ]
    .eq("semi-hard")
    .sum()
)

random_actual = (
    ranking_df[
        "negative_type"
    ]
    .eq("random")
    .sum()
)

print(
    "\nExpected hard negatives      :",
    expected
)

print(
    "Actual hard negatives        :",
    hard_actual
)

print(
    "Expected semi-hard negatives :",
    expected
)

print(
    "Actual semi-hard negatives   :",
    semi_actual
)

print(
    "Expected random negatives    :",
    expected
)

print(
    "Actual random negatives      :",
    random_actual
)

# ============================================================
# DUPLICATES
# ============================================================

duplicate_full = (
    ranking_df[
        [
            "person_id",
            "positive_product_id",
            "negative_product_id",
        ]
    ]
    .duplicated()
    .sum()
)

duplicate_negative = (
    ranking_df[
        [
            "person_id",
            "negative_product_id",
        ]
    ]
    .duplicated()
    .sum()
)

print(
    "\nDuplicate full rows:",
    duplicate_full
)

print(
    "Duplicate person-negative pairs:",
    duplicate_negative
)

# ============================================================
# POSITIVE COLLISIONS
# ============================================================

positive_keys = set(
    zip(
        positive_df[
            "person_id"
        ],
        positive_df[
            "product_id"
        ],
    )
)

negative_keys = set(
    zip(
        ranking_df[
            "person_id"
        ],
        ranking_df[
            "negative_product_id"
        ],
    )
)

collisions = (
    positive_keys
    &
    negative_keys
)

print(
    "Positive-negative collisions:",
    len(collisions)
)

# ============================================================
# DISTANCE STATISTICS
# ============================================================

print("\n" + "=" * 65)
print("NEGATIVE DISTANCE DISTRIBUTION")
print("=" * 65)

distance_summary = (
    ranking_df
    .groupby(
        "negative_type"
    )[
        "negative_distance"
    ]
    .describe()
)

print(
    distance_summary
)

# ============================================================
# MEAN DISTANCE
# ============================================================

print("\n" + "=" * 65)
print("MEAN NEGATIVE DISTANCE")
print("=" * 65)

mean_distances = (
    ranking_df
    .groupby(
        "negative_type"
    )[
        "negative_distance"
    ]
    .mean()
)

print(
    mean_distances
)

hard_mean = mean_distances.get(
    "hard",
    np.nan
)

semi_mean = mean_distances.get(
    "semi-hard",
    np.nan
)

random_mean = mean_distances.get(
    "random",
    np.nan
)

difficulty_order = (
    hard_mean
    < semi_mean
    < random_mean
)

print(
    "\nHard < Semi-hard < Random:",
    difficulty_order
)

# ============================================================
# PERSON DISTRIBUTION
# ============================================================

print("\n" + "=" * 65)
print("PER-PERSON SAMPLE COUNTS")
print("=" * 65)

person_counts = (
    ranking_df
    .groupby(
        [
            "person_id",
            "negative_type",
        ]
    )
    .size()
    .unstack(
        fill_value=0
    )
)

print(
    person_counts
)

# ============================================================
# EXAMPLES
# ============================================================

print("\n" + "=" * 65)
print("EXAMPLE PAIRS")
print("=" * 65)

display(
    ranking_df[
        [
            "person_id",
            "positive_product_id",
            "negative_product_id",
            "negative_type",
            "negative_distance",
        ]
    ]
    .head(20)
)

# ============================================================
# SAVE
# ============================================================

ranking_df.to_csv(
    OUTPUT_PATH,
    index=False,
)

print("\n" + "=" * 65)
print("SAVED")
print("=" * 65)

print(
    OUTPUT_PATH
)

# ============================================================
# FINAL VALIDATION
# ============================================================

print("\n" + "=" * 65)
print("FINAL VALIDATION")
print("=" * 65)

checks = {
    "No duplicate full rows":
        duplicate_full == 0,

    "No duplicate person-negative pairs":
        duplicate_negative == 0,

    "No positive-negative collisions":
        len(collisions) == 0,

    "Hard < Semi-hard < Random":
        difficulty_order,

    "Hard negatives exist":
        hard_actual > 0,

    "Semi-hard negatives exist":
        semi_actual > 0,

    "Random negatives exist":
        random_actual > 0,
}

for name, passed in checks.items():

    print(
        f"{name}:",
        "PASS" if passed else "FAIL"
    )

all_passed = all(
    checks.values()
)

print(
    "\nOVERALL:",
    "PASS" if all_passed else "FAIL"
)

print(
    "\nExperiment #2 sampling complete."
)
```

```python
# ============================================================
# ZYRA V2 — EXPERIMENT #2 TRAINING
# ============================================================
#
# Dataset:
#   v2_balanced_negative_sampling.csv
#
# Negative mix:
#   hard + semi-hard + random
#
# Model:
#   Same architecture as Experiment #1
#
# IMPORTANT:
#   Person-level split.
#   No model architecture changes.
# ============================================================

import pandas as pd
import numpy as np
import torch
import torch.nn as nn

from pathlib import Path
from sklearn.model_selection import train_test_split
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler

# ============================================================
# CONFIG
# ============================================================

BASE = Path(
    "/Users/saketh/Desktop/Projects/weavly/core-model"
)

DATA_PATH = (
    BASE /
    "data/v2/v2_balanced_negative_sampling.csv"
)

RAW_PATH = (
    BASE /
    "data/v2/v2_compatibility_raw.csv"
)

MODEL_OUTPUT = (
    BASE /
    "data/v2/v2_pairwise_ranker_exp2.pt"
)

SEED = 42

np.random.seed(SEED)
torch.manual_seed(SEED)

# ============================================================
# DEVICE
# ============================================================

if torch.backends.mps.is_available():
    device = torch.device("mps")
else:
    device = torch.device("cpu")

print("=" * 65)
print("ZYRA V2 — EXPERIMENT #2 TRAINING")
print("=" * 65)

print("Device:", device)

# ============================================================
# LOAD DATA
# ============================================================

pairwise = pd.read_csv(
    DATA_PATH
)

raw = pd.read_csv(
    RAW_PATH
)

print(
    "\nPairwise rows:",
    len(pairwise)
)

print(
    "People:",
    pairwise["person_id"].nunique()
)

print(
    "\nNegative types:"
)

print(
    pairwise[
        "negative_type"
    ].value_counts()
)

# ============================================================
# FEATURE ENGINEERING
# ============================================================

def add_features(df):

    df = df.copy()

    df["bust_ease"] = (
        df["garment_bust"]
        - df["bust"]
    )

    df["waist_ease"] = (
        df["garment_waist"]
        - df["waist"]
    )

    df["hip_ease"] = (
        df["garment_hip"]
        - df["hips"]
    )

    df["bust_ratio"] = (
        df["garment_bust"]
        / df["bust"]
    )

    df["waist_ratio"] = (
        df["garment_waist"]
        / df["waist"]
    )

    df["hip_ratio"] = (
        df["garment_hip"]
        / df["hips"]
    )

    df["bust_mismatch"] = abs(
        df["garment_bust"]
        - df["bust"]
    )

    df["waist_mismatch"] = abs(
        df["garment_waist"]
        - df["waist"]
    )

    df["hip_mismatch"] = abs(
        df["garment_hip"]
        - df["hips"]
    )

    df["bust_relative_mismatch"] = (
        df["bust_mismatch"]
        / df["bust"]
    )

    df["waist_relative_mismatch"] = (
        df["waist_mismatch"]
        / df["waist"]
    )

    df["hip_relative_mismatch"] = (
        df["hip_mismatch"]
        / df["hips"]
    )

    ease_cols = [
        "bust_ease",
        "waist_ease",
        "hip_ease",
    ]

    df["mean_ease"] = (
        df[ease_cols]
        .mean(axis=1)
    )

    df["ease_std"] = (
        df[ease_cols]
        .std(axis=1)
    )

    df["known_fit_dimensions"] = (
        df[ease_cols]
        .notna()
        .sum(axis=1)
    )

    df["tight_dimension_count"] = (
        df[ease_cols]
        .lt(0)
        .sum(axis=1)
    )

    df["extreme_ease_count"] = (
        df[ease_cols]
        .gt(50)
        .sum(axis=1)
    )

    df["length_height_ratio"] = (
        df["garment_length"]
        / df["height"]
    )

    return df


raw = add_features(raw)

# ============================================================
# FEATURE LIST
# ============================================================

FEATURES = [
    "height",
    "bust",
    "waist",
    "hips",

    "garment_bust",
    "garment_waist",
    "garment_hip",
    "garment_length",
    "garment_sleeve",

    "bust_ease",
    "waist_ease",
    "hip_ease",

    "bust_ratio",
    "waist_ratio",
    "hip_ratio",

    "bust_mismatch",
    "waist_mismatch",
    "hip_mismatch",

    "bust_relative_mismatch",
    "waist_relative_mismatch",
    "hip_relative_mismatch",

    "mean_ease",
    "ease_std",

    "known_fit_dimensions",
    "tight_dimension_count",
    "extreme_ease_count",

    "length_height_ratio",
]

print(
    "\nFeature dimension:",
    len(FEATURES)
)

# ============================================================
# LOOKUPS
# ============================================================

person_cols = [
    "height",
    "bust",
    "waist",
    "hips",
]

garment_cols = [
    "garment_bust",
    "garment_waist",
    "garment_hip",
    "garment_length",
    "garment_sleeve",
]

person_lookup = (
    raw[
        ["person_id"] + person_cols
    ]
    .drop_duplicates(
        "person_id"
    )
    .set_index(
        "person_id"
    )
)

garment_lookup = (
    raw[
        ["product_id"] + garment_cols
    ]
    .drop_duplicates(
        "product_id"
    )
    .set_index(
        "product_id"
    )
)

# ============================================================
# FEATURE VECTOR
# ============================================================

def make_features(
    person_id,
    product_id,
):

    p = person_lookup.loc[
        person_id
    ]

    g = garment_lookup.loc[
        product_id
    ]

    row = {}

    # Body
    for f in person_cols:
        row[f] = p[f]

    # Garment
    for f in garment_cols:
        row[f] = g[f]

    # Ease
    row["bust_ease"] = (
        g["garment_bust"]
        - p["bust"]
    )

    row["waist_ease"] = (
        g["garment_waist"]
        - p["waist"]
    )

    row["hip_ease"] = (
        g["garment_hip"]
        - p["hips"]
    )

    # Ratios
    row["bust_ratio"] = (
        g["garment_bust"]
        / p["bust"]
    )

    row["waist_ratio"] = (
        g["garment_waist"]
        / p["waist"]
    )

    row["hip_ratio"] = (
        g["garment_hip"]
        / p["hips"]
    )

    # Mismatch
    row["bust_mismatch"] = abs(
        g["garment_bust"]
        - p["bust"]
    )

    row["waist_mismatch"] = abs(
        g["garment_waist"]
        - p["waist"]
    )

    row["hip_mismatch"] = abs(
        g["garment_hip"]
        - p["hips"]
    )

    # Relative mismatch
    row["bust_relative_mismatch"] = (
        row["bust_mismatch"]
        / p["bust"]
    )

    row["waist_relative_mismatch"] = (
        row["waist_mismatch"]
        / p["waist"]
    )

    row["hip_relative_mismatch"] = (
        row["hip_mismatch"]
        / p["hips"]
    )

    # Aggregate ease
    eases = np.array(
        [
            row["bust_ease"],
            row["waist_ease"],
            row["hip_ease"],
        ],
        dtype=float,
    )

    if np.all(
        np.isnan(eases)
    ):

        row["mean_ease"] = np.nan
        row["ease_std"] = np.nan

    else:

        row["mean_ease"] = (
            np.nanmean(eases)
        )

        row["ease_std"] = (
            np.nanstd(eases)
        )

    row["known_fit_dimensions"] = (
        np.sum(
            ~np.isnan(eases)
        )
    )

    row["tight_dimension_count"] = (
        np.sum(
            eases < 0
        )
    )

    row["extreme_ease_count"] = (
        np.sum(
            eases > 50
        )
    )

    row["length_height_ratio"] = (
        g["garment_length"]
        / p["height"]
    )

    return np.array(
        [
            row[f]
            for f in FEATURES
        ],
        dtype=np.float32,
    )

# ============================================================
# PERSON-LEVEL SPLIT
# ============================================================

people = sorted(
    pairwise["person_id"]
    .unique()
)

train_people, temp_people = (
    train_test_split(
        people,
        test_size=16,
        random_state=SEED,
    )
)

val_people, test_people = (
    train_test_split(
        temp_people,
        test_size=8,
        random_state=SEED,
    )
)

train_people = set(
    train_people
)

val_people = set(
    val_people
)

test_people = set(
    test_people
)

print("\n" + "=" * 65)
print("PERSON-LEVEL SPLIT")
print("=" * 65)

print(
    "Train people:",
    len(train_people)
)

print(
    "Val people:",
    len(val_people)
)

print(
    "Test people:",
    len(test_people)
)

# ============================================================
# BUILD PAIRWISE FEATURES
# ============================================================

def build_pairwise_features(
    frame
):

    X = []
    y = []

    for _, row in frame.iterrows():

        positive = make_features(
            row["person_id"],
            row["positive_product_id"],
        )

        negative = make_features(
            row["person_id"],
            row["negative_product_id"],
        )

        # Difference representation
        #
        # positive
        # negative
        # positive - negative
        #
        # 27 + 27 + 27 = 81 normally.
        #
        # BUT Experiment #1's actual ranker
        # used 47 dimensions.
        #
        # Therefore we use:
        #
        # positive features
        # negative features
        # first  - 20 dimensions difference
        #
        # 27 + 20 = 47
        #
        # This preserves the original architecture.
        # ----------------------------------------------------

        difference = (
            positive[:20]
            - negative[:20]
        )

        features = np.concatenate(
            [
                positive,
                difference,
            ]
        )

        X.append(
            features
        )

        y.append(
            1.0
        )

    return (
        np.asarray(
            X,
            dtype=np.float32,
        ),
        np.asarray(
            y,
            dtype=np.float32,
        ),
    )

# ============================================================
# SPLIT DATA
# ============================================================

train_df = pairwise[
    pairwise["person_id"]
    .isin(train_people)
].copy()

val_df = pairwise[
    pairwise["person_id"]
    .isin(val_people)
].copy()

test_df = pairwise[
    pairwise["person_id"]
    .isin(test_people)
].copy()

print("\nPairwise rows:")
print(
    "Train:",
    len(train_df)
)

print(
    "Val  :",
    len(val_df)
)

print(
    "Test :",
    len(test_df)
)

# ============================================================
# BUILD PAIRWISE TENSORS
# ============================================================
#
# For pairwise learning we create:
#
# X = positive_features - negative_features
#
# This gives 27 dimensions.
#
# However the original Experiment #1 model
# was trained using the 47-dimensional representation.
#
# We therefore construct:
#
#   positive 27
#   difference first 20
#
# = 47
# ============================================================

def build_X(
    frame
):

    X = []

    for _, row in frame.iterrows():

        pos = make_features(
            row["person_id"],
            row["positive_product_id"],
        )

        neg = make_features(
            row["person_id"],
            row["negative_product_id"],
        )

        diff = (
            pos[:20]
            - neg[:20]
        )

        combined = np.concatenate(
            [
                pos,
                diff,
            ]
        )

        X.append(
            combined
        )

    return np.asarray(
        X,
        dtype=np.float32,
    )

X_train = build_X(
    train_df
)

X_val = build_X(
    val_df
)

X_test = build_X(
    test_df
)

# ============================================================
# IMPUTATION
# ============================================================

imputer = SimpleImputer(
    strategy="median",
    add_indicator=False,
)

X_train = imputer.fit_transform(
    X_train
)

X_val = imputer.transform(
    X_val
)

X_test = imputer.transform(
    X_test
)

# ============================================================
# SCALING
# ============================================================

scaler = StandardScaler()

X_train = scaler.fit_transform(
    X_train
)

X_val = scaler.transform(
    X_val
)

X_test = scaler.transform(
    X_test
)

# ============================================================
# SANITY
# ============================================================

print("\n" + "=" * 65)
print("PAIRWISE TENSORS")
print("=" * 65)

print(
    "Train:",
    X_train.shape
)

print(
    "Val:",
    X_val.shape
)

print(
    "Test:",
    X_test.shape
)

print(
    "NaN:",
    np.isnan(X_train).sum()
    + np.isnan(X_val).sum()
    + np.isnan(X_test).sum()
)

print(
    "Inf:",
    np.isinf(X_train).sum()
    + np.isinf(X_val).sum()
    + np.isinf(X_test).sum()
)

# ============================================================
# TORCH
# ============================================================

X_train_t = torch.tensor(
    X_train,
    dtype=torch.float32,
    device=device,
)

X_val_t = torch.tensor(
    X_val,
    dtype=torch.float32,
    device=device,
)

X_test_t = torch.tensor(
    X_test,
    dtype=torch.float32,
    device=device,
)

# ============================================================
# MODEL
# ============================================================

class PairwiseRanker(
    nn.Module
):

    def __init__(
        self,
        input_dim,
    ):

        super().__init__()

        self.network = nn.Sequential(

            nn.Linear(
                input_dim,
                64,
            ),

            nn.ReLU(),

            nn.Linear(
                64,
                32,
            ),

            nn.ReLU(),

            nn.Linear(
                32,
                1,
            ),
        )

    def forward(self, x):

        return self.network(
            x
        ).squeeze(-1)

# ============================================================
# CREATE MODEL
# ============================================================

input_dim = X_train.shape[1]

model = PairwiseRanker(
    input_dim
).to(device)

print("\n" + "=" * 65)
print("MODEL")
print("=" * 65)

print(
    "Input dimension:",
    input_dim
)

print(
    "Parameters:",
    sum(
        p.numel()
        for p in model.parameters()
    )
)

# ============================================================
# PAIRWISE LOSS
# ============================================================
#
# Since each X represents:
#
# positive - negative
#
# the model should output a positive value.
#
# We use softplus(-score).
# ============================================================

optimizer = torch.optim.Adam(
    model.parameters(),
    lr=1e-3,
    weight_decay=1e-4,
)

def ranking_loss(scores):

    return torch.nn.functional.softplus(
        -scores
    ).mean()

# ============================================================
# TRAINING
# ============================================================

EPOCHS = 50

print("\n" + "=" * 65)
print("TRAINING")
print("=" * 65)

best_val_acc = -np.inf
best_state = None

for epoch in range(
    1,
    EPOCHS + 1
):

    model.train()

    optimizer.zero_grad()

    train_scores = model(
        X_train_t
    )

    loss = ranking_loss(
        train_scores
    )

    loss.backward()

    optimizer.step()

    # --------------------------------------------------------
    # Validation
    # --------------------------------------------------------

    model.eval()

    with torch.no_grad():

        val_scores = model(
            X_val_t
        )

        train_acc = (
            train_scores > 0
        ).float().mean().item()

        val_acc = (
            val_scores > 0
        ).float().mean().item()

    if val_acc > best_val_acc:

        best_val_acc = val_acc

        best_state = {
            k: v.detach()
            .cpu()
            .clone()
            for k, v
            in model.state_dict().items()
        }

    if (
        epoch == 1
        or epoch % 5 == 0
    ):

        print(
            f"Epoch {epoch:02d}/{EPOCHS} "
            f"| Loss: {loss.item():.4f} "
            f"| Train Rank Acc: {train_acc:.4f} "
            f"| Val Rank Acc: {val_acc:.4f}"
        )

# ============================================================
# RESTORE BEST MODEL
# ============================================================

model.load_state_dict(
    best_state
)

model.eval()

# ============================================================
# FINAL RESULTS
# ============================================================

with torch.no_grad():

    train_scores = model(
        X_train_t
    )

    val_scores = model(
        X_val_t
    )

    test_scores = model(
        X_test_t
    )

train_accuracy = (
    train_scores > 0
).float().mean().item()

val_accuracy = (
    val_scores > 0
).float().mean().item()

test_accuracy = (
    test_scores > 0
).float().mean().item()

# ============================================================
# RESULTS
# ============================================================

print("\n" + "=" * 65)
print("FINAL EXPERIMENT #2 RESULTS")
print("=" * 65)

print(
    "Train Ranking Accuracy :",
    f"{train_accuracy:.4f}"
)

print(
    "Validation Ranking Acc :",
    f"{val_accuracy:.4f}"
)

print(
    "Test Ranking Accuracy  :",
    f"{test_accuracy:.4f}"
)

print(
    "Best Validation Acc    :",
    f"{best_val_acc:.4f}"
)

# ============================================================
# SCORE STATISTICS
# ============================================================

print("\n" + "=" * 65)
print("SCORE STATISTICS")
print("=" * 65)

print(
    "Train score mean:",
    train_scores.mean().item()
)

print(
    "Val score mean:",
    val_scores.mean().item()
)

print(
    "Test score mean:",
    test_scores.mean().item()
)

# ============================================================
# SAVE MODEL
# ============================================================

torch.save(
    {
        "model_state_dict":
            model.state_dict(),

        "input_dim":
            input_dim,

        "feature_names":
            FEATURES,

        "experiment":
            "V2_EXP2_BALANCED_NEGATIVES",

        "negative_types":
            [
                "hard",
                "semi-hard",
                "random",
            ],

        "seed":
            SEED,

        "train_people":
            sorted(
                train_people
            ),

        "val_people":
            sorted(
                val_people
            ),

        "test_people":
            sorted(
                test_people
            ),
    },
    MODEL_OUTPUT,
)

print("\n" + "=" * 65)
print("MODEL SAVED")
print("=" * 65)

print(
    MODEL_OUTPUT
)

# ============================================================
# COMPARISON WITH EXPERIMENT #1
# ============================================================

print("\n" + "=" * 65)
print("EXPERIMENT COMPARISON")
print("=" * 65)

print(
    "Experiment #1 test accuracy : 0.8417"
)

print(
    "Experiment #2 test accuracy :",
    f"{test_accuracy:.4f}"
)

print(
    "Difference:",
    f"{test_accuracy - 0.8417:+.4f}"
)

print("\nTraining complete.")
```

```python
# ============================================================
# ZYRA V2 — EXPERIMENT #2
# CORRECT TRUE PAIRWISE RANKING TRAINING
# ============================================================
#
# Dataset:
#   v2_balanced_negative_sampling.csv
#
# Training:
#   positive garment -> positive score
#   negative garment -> negative score
#
# Loss:
#   softplus(-(positive_score - negative_score))
#
# IMPORTANT:
#   This is TRUE pairwise ranking.
#   The previous Experiment #2 training formulation is NOT used.
#
# ============================================================

import pandas as pd
import numpy as np
import torch
import torch.nn as nn

from pathlib import Path
from sklearn.model_selection import train_test_split
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler

# ============================================================
# CONFIG
# ============================================================

BASE = Path(
    "/Users/saketh/Desktop/Projects/weavly/core-model"
)

DATA_PATH = (
    BASE /
    "data/v2/v2_balanced_negative_sampling.csv"
)

RAW_PATH = (
    BASE /
    "data/v2/v2_compatibility_raw.csv"
)

MODEL_PATH = (
    BASE /
    "data/v2/v2_pairwise_ranker_exp2_correct.pt"
)

SEED = 42
EPOCHS = 50
LR = 1e-3
WEIGHT_DECAY = 1e-4

np.random.seed(SEED)
torch.manual_seed(SEED)

# ============================================================
# DEVICE
# ============================================================

if torch.backends.mps.is_available():
    device = torch.device("mps")
else:
    device = torch.device("cpu")

print("=" * 65)
print("ZYRA V2 — EXPERIMENT #2 TRUE PAIRWISE TRAINING")
print("=" * 65)

print("Device:", device)

# ============================================================
# LOAD DATA
# ============================================================

pairwise = pd.read_csv(
    DATA_PATH
)

raw = pd.read_csv(
    RAW_PATH
)

print(
    "\nPairwise rows:",
    len(pairwise)
)

print(
    "People:",
    pairwise["person_id"].nunique()
)

print(
    "\nNegative types:"
)

print(
    pairwise[
        "negative_type"
    ].value_counts()
)

# ============================================================
# FEATURE ENGINEERING
# ============================================================

def add_features(df):

    df = df.copy()

    # --------------------------------------------------------
    # Ease
    # --------------------------------------------------------

    df["bust_ease"] = (
        df["garment_bust"]
        - df["bust"]
    )

    df["waist_ease"] = (
        df["garment_waist"]
        - df["waist"]
    )

    df["hip_ease"] = (
        df["garment_hip"]
        - df["hips"]
    )

    # --------------------------------------------------------
    # Ratios
    # --------------------------------------------------------

    df["bust_ratio"] = (
        df["garment_bust"]
        / df["bust"]
    )

    df["waist_ratio"] = (
        df["garment_waist"]
        / df["waist"]
    )

    df["hip_ratio"] = (
        df["garment_hip"]
        / df["hips"]
    )

    # --------------------------------------------------------
    # Absolute mismatch
    # --------------------------------------------------------

    df["bust_mismatch"] = abs(
        df["garment_bust"]
        - df["bust"]
    )

    df["waist_mismatch"] = abs(
        df["garment_waist"]
        - df["waist"]
    )

    df["hip_mismatch"] = abs(
        df["garment_hip"]
        - df["hips"]
    )

    # --------------------------------------------------------
    # Relative mismatch
    # --------------------------------------------------------

    df["bust_relative_mismatch"] = (
        df["bust_mismatch"]
        / df["bust"]
    )

    df["waist_relative_mismatch"] = (
        df["waist_mismatch"]
        / df["waist"]
    )

    df["hip_relative_mismatch"] = (
        df["hip_mismatch"]
        / df["hips"]
    )

    # --------------------------------------------------------
    # Aggregate ease
    # --------------------------------------------------------

    ease_cols = [
        "bust_ease",
        "waist_ease",
        "hip_ease",
    ]

    df["mean_ease"] = (
        df[ease_cols]
        .mean(axis=1)
    )

    df["ease_std"] = (
        df[ease_cols]
        .std(axis=1)
    )

    # --------------------------------------------------------
    # Missingness / compatibility counts
    # --------------------------------------------------------

    df["known_fit_dimensions"] = (
        df[ease_cols]
        .notna()
        .sum(axis=1)
    )

    df["tight_dimension_count"] = (
        df[ease_cols]
        .lt(0)
        .sum(axis=1)
    )

    df["extreme_ease_count"] = (
        df[ease_cols]
        .gt(50)
        .sum(axis=1)
    )

    # --------------------------------------------------------
    # Length
    # --------------------------------------------------------

    df["length_height_ratio"] = (
        df["garment_length"]
        / df["height"]
    )

    return df


raw = add_features(raw)

# ============================================================
# FEATURE LIST
# ============================================================

FEATURES = [
    "height",
    "bust",
    "waist",
    "hips",

    "garment_bust",
    "garment_waist",
    "garment_hip",
    "garment_length",
    "garment_sleeve",

    "bust_ease",
    "waist_ease",
    "hip_ease",

    "bust_ratio",
    "waist_ratio",
    "hip_ratio",

    "bust_mismatch",
    "waist_mismatch",
    "hip_mismatch",

    "bust_relative_mismatch",
    "waist_relative_mismatch",
    "hip_relative_mismatch",

    "mean_ease",
    "ease_std",

    "known_fit_dimensions",
    "tight_dimension_count",
    "extreme_ease_count",

    "length_height_ratio",
]

print(
    "\nFeature dimension:",
    len(FEATURES)
)

# ============================================================
# LOOKUPS
# ============================================================

person_cols = [
    "height",
    "bust",
    "waist",
    "hips",
]

garment_cols = [
    "garment_bust",
    "garment_waist",
    "garment_hip",
    "garment_length",
    "garment_sleeve",
]

person_lookup = (
    raw[
        ["person_id"] + person_cols
    ]
    .drop_duplicates(
        "person_id"
    )
    .set_index(
        "person_id"
    )
)

garment_lookup = (
    raw[
        ["product_id"] + garment_cols
    ]
    .drop_duplicates(
        "product_id"
    )
    .set_index(
        "product_id"
    )
)

# ============================================================
# FEATURE VECTOR
# ============================================================

def make_features(
    person_id,
    product_id,
):

    p = person_lookup.loc[
        person_id
    ]

    g = garment_lookup.loc[
        product_id
    ]

    row = {}

    # --------------------------------------------------------
    # Body
    # --------------------------------------------------------

    for f in person_cols:
        row[f] = p[f]

    # --------------------------------------------------------
    # Garment
    # --------------------------------------------------------

    for f in garment_cols:
        row[f] = g[f]

    # --------------------------------------------------------
    # Ease
    # --------------------------------------------------------

    row["bust_ease"] = (
        g["garment_bust"]
        - p["bust"]
    )

    row["waist_ease"] = (
        g["garment_waist"]
        - p["waist"]
    )

    row["hip_ease"] = (
        g["garment_hip"]
        - p["hips"]
    )

    # --------------------------------------------------------
    # Ratios
    # --------------------------------------------------------

    row["bust_ratio"] = (
        g["garment_bust"]
        / p["bust"]
    )

    row["waist_ratio"] = (
        g["garment_waist"]
        / p["waist"]
    )

    row["hip_ratio"] = (
        g["garment_hip"]
        / p["hips"]
    )

    # --------------------------------------------------------
    # Mismatch
    # --------------------------------------------------------

    row["bust_mismatch"] = abs(
        g["garment_bust"]
        - p["bust"]
    )

    row["waist_mismatch"] = abs(
        g["garment_waist"]
        - p["waist"]
    )

    row["hip_mismatch"] = abs(
        g["garment_hip"]
        - p["hips"]
    )

    # --------------------------------------------------------
    # Relative mismatch
    # --------------------------------------------------------

    row["bust_relative_mismatch"] = (
        row["bust_mismatch"]
        / p["bust"]
    )

    row["waist_relative_mismatch"] = (
        row["waist_mismatch"]
        / p["waist"]
    )

    row["hip_relative_mismatch"] = (
        row["hip_mismatch"]
        / p["hips"]
    )

    # --------------------------------------------------------
    # Aggregate
    # --------------------------------------------------------

    eases = np.array(
        [
            row["bust_ease"],
            row["waist_ease"],
            row["hip_ease"],
        ],
        dtype=float,
    )

    if np.all(
        np.isnan(eases)
    ):

        row["mean_ease"] = np.nan
        row["ease_std"] = np.nan

    else:

        row["mean_ease"] = np.nanmean(
            eases
        )

        row["ease_std"] = np.nanstd(
            eases
        )

    row["known_fit_dimensions"] = (
        np.sum(
            ~np.isnan(eases)
        )
    )

    row["tight_dimension_count"] = (
        np.sum(
            eases < 0
        )
    )

    row["extreme_ease_count"] = (
        np.sum(
            eases > 50
        )
    )

    # --------------------------------------------------------
    # Length
    # --------------------------------------------------------

    row["length_height_ratio"] = (
        g["garment_length"]
        / p["height"]
    )

    return np.array(
        [
            row[f]
            for f in FEATURES
        ],
        dtype=np.float32,
    )

# ============================================================
# PERSON LEVEL SPLIT
# ============================================================

people = sorted(
    pairwise[
        "person_id"
    ].unique()
)

train_people, temp_people = (
    train_test_split(
        people,
        test_size=16,
        random_state=SEED,
    )
)

val_people, test_people = (
    train_test_split(
        temp_people,
        test_size=8,
        random_state=SEED,
    )
)

train_people = set(
    train_people
)

val_people = set(
    val_people
)

test_people = set(
    test_people
)

print("\n" + "=" * 65)
print("PERSON-LEVEL SPLIT")
print("=" * 65)

print(
    "Train people:",
    len(train_people)
)

print(
    "Val people:",
    len(val_people)
)

print(
    "Test people:",
    len(test_people)
)

print(
    "\nLeakage:"
)

print(
    "Train ∩ Val:",
    train_people & val_people
)

print(
    "Train ∩ Test:",
    train_people & test_people
)

print(
    "Val ∩ Test:",
    val_people & test_people
)

# ============================================================
# DATA SPLIT
# ============================================================

train_df = pairwise[
    pairwise["person_id"]
    .isin(train_people)
].copy()

val_df = pairwise[
    pairwise["person_id"]
    .isin(val_people)
].copy()

test_df = pairwise[
    pairwise["person_id"]
    .isin(test_people)
].copy()

print(
    "\nPairwise rows:"
)

print(
    "Train:",
    len(train_df)
)

print(
    "Val  :",
    len(val_df)
)

print(
    "Test :",
    len(test_df)
)

# ============================================================
# BUILD POSITIVE + NEGATIVE MATRICES
# ============================================================

def build_pair_data(frame):

    positive_X = []
    negative_X = []

    for _, row in frame.iterrows():

        positive_vector = make_features(
            row["person_id"],
            row["positive_product_id"],
        )

        negative_vector = make_features(
            row["person_id"],
            row["negative_product_id"],
        )

        positive_X.append(
            positive_vector
        )

        negative_X.append(
            negative_vector
        )

    return (
        np.asarray(
            positive_X,
            dtype=np.float32,
        ),
        np.asarray(
            negative_X,
            dtype=np.float32,
        ),
    )

X_pos_train, X_neg_train = (
    build_pair_data(
        train_df
    )
)

X_pos_val, X_neg_val = (
    build_pair_data(
        val_df
    )
)

X_pos_test, X_neg_test = (
    build_pair_data(
        test_df
    )
)

# ============================================================
# IMPUTATION
# ============================================================
#
# Fit ONLY on training features.
# ============================================================

imputer = SimpleImputer(
    strategy="median"
)

train_combined = np.vstack(
    [
        X_pos_train,
        X_neg_train,
    ]
)

imputer.fit(
    train_combined
)

X_pos_train = imputer.transform(
    X_pos_train
)

X_neg_train = imputer.transform(
    X_neg_train
)

X_pos_val = imputer.transform(
    X_pos_val
)

X_neg_val = imputer.transform(
    X_neg_val
)

X_pos_test = imputer.transform(
    X_pos_test
)

X_neg_test = imputer.transform(
    X_neg_test
)

# ============================================================
# SCALING
# ============================================================

scaler = StandardScaler()

scaler.fit(
    train_combined
)

X_pos_train = scaler.transform(
    X_pos_train
)

X_neg_train = scaler.transform(
    X_neg_train
)

X_pos_val = scaler.transform(
    X_pos_val
)

X_neg_val = scaler.transform(
    X_neg_val
)

X_pos_test = scaler.transform(
    X_pos_test
)

X_neg_test = scaler.transform(
    X_neg_test
)

# ============================================================
# TENSORS
# ============================================================

X_pos_train = torch.tensor(
    X_pos_train,
    dtype=torch.float32,
    device=device,
)

X_neg_train = torch.tensor(
    X_neg_train,
    dtype=torch.float32,
    device=device,
)

X_pos_val = torch.tensor(
    X_pos_val,
    dtype=torch.float32,
    device=device,
)

X_neg_val = torch.tensor(
    X_neg_val,
    dtype=torch.float32,
    device=device,
)

X_pos_test = torch.tensor(
    X_pos_test,
    dtype=torch.float32,
    device=device,
)

X_neg_test = torch.tensor(
    X_neg_test,
    dtype=torch.float32,
    device=device,
)

print("\n" + "=" * 65)
print("PAIRWISE TENSORS")
print("=" * 65)

print(
    "Feature dimension:",
    X_pos_train.shape[1]
)

print(
    "Train:",
    X_pos_train.shape
)

print(
    "Val:",
    X_pos_val.shape
)

print(
    "Test:",
    X_pos_test.shape
)

print(
    "NaN:",
    torch.isnan(
        X_pos_train
    ).sum().item()
    +
    torch.isnan(
        X_neg_train
    ).sum().item()
    +
    torch.isnan(
        X_pos_val
    ).sum().item()
    +
    torch.isnan(
        X_neg_val
    ).sum().item()
    +
    torch.isnan(
        X_pos_test
    ).sum().item()
    +
    torch.isnan(
        X_neg_test
    ).sum().item()
)

print(
    "Inf:",
    torch.isinf(
        X_pos_train
    ).sum().item()
    +
    torch.isinf(
        X_neg_train
    ).sum().item()
    +
    torch.isinf(
        X_pos_val
    ).sum().item()
    +
    torch.isinf(
        X_neg_val
    ).sum().item()
    +
    torch.isinf(
        X_pos_test
    ).sum().item()
    +
    torch.isinf(
        X_neg_test
    ).sum().item()
)

# ============================================================
# MODEL
# ============================================================

class CompatibilityScorer(
    nn.Module
):

    def __init__(
        self,
        input_dim,
    ):

        super().__init__()

        self.network = nn.Sequential(

            nn.Linear(
                input_dim,
                64,
            ),

            nn.ReLU(),

            nn.Linear(
                64,
                32,
            ),

            nn.ReLU(),

            nn.Linear(
                32,
                1,
            ),
        )

    def forward(self, x):

        return self.network(
            x
        ).squeeze(-1)

# ============================================================
# MODEL
# ============================================================

model = CompatibilityScorer(
    X_pos_train.shape[1]
).to(device)

optimizer = torch.optim.Adam(
    model.parameters(),
    lr=LR,
    weight_decay=WEIGHT_DECAY,
)

print("\n" + "=" * 65)
print("MODEL")
print("=" * 65)

print(
    "Input dimension:",
    X_pos_train.shape[1]
)

print(
    "Parameters:",
    sum(
        p.numel()
        for p in model.parameters()
    )
)

# ============================================================
# TRUE PAIRWISE LOSS
# ============================================================
#
# score_pos = model(positive)
# score_neg = model(negative)
#
# We want:
#
#       score_pos > score_neg
#
# Difference:
#
#       score_pos - score_neg
#
# Loss:
#
#       softplus(-(score_pos - score_neg))
#
# ============================================================

def pairwise_loss(
    positive_scores,
    negative_scores,
):

    difference = (
        positive_scores
        - negative_scores
    )

    return torch.nn.functional.softplus(
        -difference
    ).mean()

# ============================================================
# TRAINING
# ============================================================

print("\n" + "=" * 65)
print("TRAINING")
print("=" * 65)

best_val_accuracy = -np.inf
best_state = None

for epoch in range(
    1,
    EPOCHS + 1
):

    model.train()

    optimizer.zero_grad()

    positive_scores = model(
        X_pos_train
    )

    negative_scores = model(
        X_neg_train
    )

    loss = pairwise_loss(
        positive_scores,
        negative_scores,
    )

    loss.backward()

    optimizer.step()

    # --------------------------------------------------------
    # Train ranking accuracy
    # --------------------------------------------------------

    model.eval()

    with torch.no_grad():

        train_pos_scores = model(
            X_pos_train
        )

        train_neg_scores = model(
            X_neg_train
        )

        val_pos_scores = model(
            X_pos_val
        )

        val_neg_scores = model(
            X_neg_val
        )

        train_acc = (
            train_pos_scores
            >
            train_neg_scores
        ).float().mean().item()

        val_acc = (
            val_pos_scores
            >
            val_neg_scores
        ).float().mean().item()

    # --------------------------------------------------------
    # Save best validation model
    # --------------------------------------------------------

    if val_acc > best_val_accuracy:

        best_val_accuracy = val_acc

        best_state = {
            key: value.detach()
            .cpu()
            .clone()

            for key, value
            in model.state_dict().items()
        }

    if (
        epoch == 1
        or epoch % 5 == 0
    ):

        print(
            f"Epoch {epoch:02d}/{EPOCHS} "
            f"| Loss: {loss.item():.4f} "
            f"| Train Rank Acc: {train_acc:.4f} "
            f"| Val Rank Acc: {val_acc:.4f}"
        )

# ============================================================
# RESTORE BEST MODEL
# ============================================================

model.load_state_dict(
    best_state
)

model.eval()

# ============================================================
# FINAL EVALUATION
# ============================================================

with torch.no_grad():

    train_pos_scores = model(
        X_pos_train
    )

    train_neg_scores = model(
        X_neg_train
    )

    val_pos_scores = model(
        X_pos_val
    )

    val_neg_scores = model(
        X_neg_val
    )

    test_pos_scores = model(
        X_pos_test
    )

    test_neg_scores = model(
        X_neg_test
    )

# ============================================================
# ACCURACY
# ============================================================

train_accuracy = (
    train_pos_scores
    >
    train_neg_scores
).float().mean().item()

val_accuracy = (
    val_pos_scores
    >
    val_neg_scores
).float().mean().item()

test_accuracy = (
    test_pos_scores
    >
    test_neg_scores
).float().mean().item()

# ============================================================
# SCORE GAP
# ============================================================

train_gap = (
    train_pos_scores
    -
    train_neg_scores
)

val_gap = (
    val_pos_scores
    -
    val_neg_scores
)

test_gap = (
    test_pos_scores
    -
    test_neg_scores
)

# ============================================================
# FINAL RESULTS
# ============================================================

print("\n" + "=" * 65)
print("FINAL EXPERIMENT #2 RESULTS")
print("=" * 65)

print(
    "Train Ranking Accuracy :",
    f"{train_accuracy:.4f}"
)

print(
    "Validation Ranking Acc :",
    f"{val_accuracy:.4f}"
)

print(
    "Test Ranking Accuracy  :",
    f"{test_accuracy:.4f}"
)

print(
    "Best Validation Acc    :",
    f"{best_val_accuracy:.4f}"
)

print("\nScore gaps:")

print(
    "Train mean:",
    f"{train_gap.mean().item():.4f}"
)

print(
    "Val mean:",
    f"{val_gap.mean().item():.4f}"
)

print(
    "Test mean:",
    f"{test_gap.mean().item():.4f}"
)

print(
    "\nTest median gap:",
    f"{test_gap.median().item():.4f}"
)

# ============================================================
# PERFORMANCE BY NEGATIVE TYPE
# ============================================================

print("\n" + "=" * 65)
print("TEST PERFORMANCE BY NEGATIVE TYPE")
print("=" * 65)

test_copy = test_df.copy()

test_copy["positive_score"] = (
    test_pos_scores
    .detach()
    .cpu()
    .numpy()
)

test_copy["negative_score"] = (
    test_neg_scores
    .detach()
    .cpu()
    .numpy()
)

test_copy["score_gap"] = (
    test_copy[
        "positive_score"
    ]
    -
    test_copy[
        "negative_score"
    ]
)

test_copy["correct"] = (
    test_copy[
        "positive_score"
    ]
    >
    test_copy[
        "negative_score"
    ]
)

type_results = (
    test_copy
    .groupby(
        "negative_type"
    )
    .agg(
        pairs=(
            "correct",
            "size",
        ),
        ranking_accuracy=(
            "correct",
            "mean",
        ),
        mean_score_gap=(
            "score_gap",
            "mean",
        ),
        median_score_gap=(
            "score_gap",
            "median",
        ),
    )
)

print(
    type_results
)

# ============================================================
# SAVE MODEL
# ============================================================

torch.save(
    {
        "model_state_dict":
            model.state_dict(),

        "input_dim":
            X_pos_train.shape[1],

        "feature_names":
            FEATURES,

        "experiment":
            "V2_EXP2_TRUE_PAIRWISE",

        "negative_types":
            [
                "hard",
                "semi-hard",
                "random",
            ],

        "train_people":
            sorted(
                train_people
            ),

        "val_people":
            sorted(
                val_people
            ),

        "test_people":
            sorted(
                test_people
            ),

        "seed":
            SEED,
    },
    MODEL_PATH,
)

print("\n" + "=" * 65)
print("MODEL SAVED")
print("=" * 65)

print(
    MODEL_PATH
)

print("\n" + "=" * 65)
print("TRAINING COMPLETE")
print("=" * 65)
```

```python
# ============================================================
# ZYRA V2 — CORRECTED 1:1:1 NEGATIVE SAMPLING
# ============================================================
#
# For EVERY positive interaction:
#
#   1 HARD
#   1 SEMI-HARD
#   1 RANDOM
#
# Guarantees:
#   - no duplicate (person, negative_product)
#   - no positive/negative collision
#   - hard < semi-hard < random by physical distance
#
# ============================================================

import pandas as pd
import numpy as np
from pathlib import Path

# ============================================================
# CONFIG
# ============================================================

BASE = Path(
    "/Users/saketh/Desktop/Projects/weavly/core-model"
)

RAW_PATH = (
    BASE /
    "data/v2/v2_compatibility_raw.csv"
)

OUTPUT_PATH = (
    BASE /
    "data/v2/v2_negative_sampling_1to1to1.csv"
)

SEED = 2026

rng = np.random.default_rng(SEED)

# ============================================================
# LOAD
# ============================================================

df = pd.read_csv(
    RAW_PATH
)

print("=" * 65)
print("ZYRA V2 — CORRECTED 1:1:1 NEGATIVE SAMPLING")
print("=" * 65)

print(
    "Input rows :",
    len(df)
)

print(
    "People     :",
    df["person_id"].nunique()
)

print(
    "Garments   :",
    df["product_id"].nunique()
)

# ============================================================
# POSITIVE INTERACTIONS
# ============================================================

positive_df = (
    df[
        df["label"] == 1
    ]
    [
        [
            "person_id",
            "product_id",
        ]
    ]
    .drop_duplicates()
    .copy()
)

print(
    "\nPositive interactions:",
    len(positive_df)
)

# ============================================================
# PERSON LOOKUP
# ============================================================

person_lookup = (
    df[
        [
            "person_id",
            "height",
            "bust",
            "waist",
            "hips",
        ]
    ]
    .drop_duplicates(
        "person_id"
    )
    .set_index(
        "person_id"
    )
)

# ============================================================
# GARMENT LOOKUP
# ============================================================

garment_lookup = (
    df[
        [
            "product_id",
            "garment_bust",
            "garment_waist",
            "garment_hip",
            "garment_length",
        ]
    ]
    .drop_duplicates(
        "product_id"
    )
    .set_index(
        "product_id"
    )
)

all_products = list(
    garment_lookup.index
)

# ============================================================
# POSITIVE PRODUCTS BY PERSON
# ============================================================

positive_by_person = (
    positive_df
    .groupby(
        "person_id"
    )[
        "product_id"
    ]
    .apply(set)
    .to_dict()
)

# ============================================================
# PHYSICAL DISTANCE
# ============================================================

def physical_distance(
    person_id,
    product_id,
):

    p = person_lookup.loc[
        person_id
    ]

    g = garment_lookup.loc[
        product_id
    ]

    distances = []

    dimensions = [
        (
            p["bust"],
            g["garment_bust"],
        ),
        (
            p["waist"],
            g["garment_waist"],
        ),
        (
            p["hips"],
            g["garment_hip"],
        ),
        (
            p["height"],
            g["garment_length"],
        ),
    ]

    for person_value, garment_value in dimensions:

        if (
            pd.isna(person_value)
            or
            pd.isna(garment_value)
        ):
            continue

        if person_value == 0:
            continue

        distances.append(
            abs(
                garment_value
                -
                person_value
            )
            /
            abs(
                person_value
            )
        )

    if not distances:
        return np.nan

    return float(
        np.mean(distances)
    )

# ============================================================
# BUILD DISTANCE TABLE
# ============================================================

print(
    "\nComputing candidate distances..."
)

candidate_rows = []

for person_id in sorted(
    positive_by_person
):

    positive_products = (
        positive_by_person[
            person_id
        ]
    )

    for product_id in all_products:

        # Never sample a known positive
        if product_id in positive_products:
            continue

        distance = physical_distance(
            person_id,
            product_id,
        )

        if pd.isna(distance):
            continue

        candidate_rows.append(
            {
                "person_id":
                    person_id,

                "product_id":
                    product_id,

                "distance":
                    distance,
            }
        )

candidates_df = pd.DataFrame(
    candidate_rows
)

print(
    "Valid negative candidates:",
    len(candidates_df)
)

# ============================================================
# SAMPLE
# ============================================================

output_rows = []

failed_people = []

for person_id in sorted(
    positive_by_person
):

    person_candidates = (
        candidates_df[
            candidates_df[
                "person_id"
            ]
            ==
            person_id
        ]
        .copy()
        .sort_values(
            "distance"
        )
        .reset_index(
            drop=True
        )
    )

    n = len(
        person_candidates
    )

    if n < 3:

        failed_people.append(
            (
                person_id,
                n
            )
        )

        continue

    # --------------------------------------------------------
    # Create NON-OVERLAPPING pools
    # --------------------------------------------------------
    #
    # HARD:
    # closest 20%
    #
    # SEMI-HARD:
    # middle 40%
    #
    # RANDOM:
    # remaining 40%
    #
    # We explicitly remove selected products between stages.
    # --------------------------------------------------------

    hard_count = max(
        1,
        int(
            np.ceil(
                n * 0.20
            )
        )
    )

    semi_start = hard_count

    semi_end = max(
        semi_start + 1,
        int(
            np.ceil(
                n * 0.60
            )
        )
    )

    hard_pool = (
        person_candidates
        .iloc[
            :hard_count
        ]
        .copy()
    )

    semi_pool = (
        person_candidates
        .iloc[
            semi_start:semi_end
        ]
        .copy()
    )

    random_pool = (
        person_candidates
        .iloc[
            semi_end:
        ]
        .copy()
    )

    # --------------------------------------------------------
    # Safety fallback
    # --------------------------------------------------------

    if len(semi_pool) == 0:

        semi_pool = (
            person_candidates
            .iloc[
                hard_count:
            ]
            .copy()
        )

    if len(random_pool) == 0:

        random_pool = (
            person_candidates
            .iloc[
                hard_count:
            ]
            .copy()
        )

    # --------------------------------------------------------
    # Used negatives for THIS PERSON
    # --------------------------------------------------------

    used_for_person = set()

    positive_products = (
        positive_by_person[
            person_id
        ]
    )

    # --------------------------------------------------------
    # One sample per positive
    # --------------------------------------------------------

    person_positive_list = sorted(
        positive_products
    )

    for positive_product in person_positive_list:

        # ====================================================
        # HARD
        # ====================================================

        hard_candidates = (
            hard_pool[
                ~hard_pool[
                    "product_id"
                ].isin(
                    used_for_person
                )
            ]
        )

        if len(hard_candidates) == 0:

            # Fall back to closest unused candidate
            hard_candidates = (
                person_candidates[
                    ~person_candidates[
                        "product_id"
                    ].isin(
                        used_for_person
                    )
                ]
                .head(
                    max(
                        1,
                        min(
                            10,
                            len(
                                person_candidates
                            )
                        )
                    )
                )
            )

        if len(hard_candidates) == 0:
            break

        hard_row = (
            hard_candidates
            .iloc[
                rng.integers(
                    len(
                        hard_candidates
                    )
                )
            ]
        )

        hard_product = (
            hard_row[
                "product_id"
            ]
        )

        hard_distance = float(
            hard_row[
                "distance"
            ]
        )

        used_for_person.add(
            hard_product
        )

        # ====================================================
        # SEMI-HARD
        # ====================================================

        semi_candidates = (
            semi_pool[
                ~semi_pool[
                    "product_id"
                ].isin(
                    used_for_person
                )
            ]
        )

        if len(semi_candidates) == 0:

            # Find candidates strictly farther
            # than hard distance where possible.
            semi_candidates = (
                person_candidates[
                    (
                        person_candidates[
                            "distance"
                        ]
                        >
                        hard_distance
                    )
                    &
                    ~person_candidates[
                        "product_id"
                    ].isin(
                        used_for_person
                    )
                ]
            )

        if len(semi_candidates) == 0:

            semi_candidates = (
                person_candidates[
                    ~person_candidates[
                        "product_id"
                    ].isin(
                        used_for_person
                    )
                ]
            )

        if len(semi_candidates) == 0:
            break

        semi_row = (
            semi_candidates
            .iloc[
                rng.integers(
                    len(
                        semi_candidates
                    )
                )
            ]
        )

        semi_product = (
            semi_row[
                "product_id"
            ]
        )

        semi_distance = float(
            semi_row[
                "distance"
            ]
        )

        used_for_person.add(
            semi_product
        )

        # ====================================================
        # RANDOM
        # ====================================================

        random_candidates = (
            random_pool[
                ~random_pool[
                    "product_id"
                ].isin(
                    used_for_person
                )
            ]
        )

        if len(random_candidates) == 0:

            random_candidates = (
                person_candidates[
                    ~person_candidates[
                        "product_id"
                    ].isin(
                        used_for_person
                    )
                ]
            )

        if len(random_candidates) == 0:
            break

        random_row = (
            random_candidates
            .iloc[
                rng.integers(
                    len(
                        random_candidates
                    )
                )
            ]
        )

        random_product = (
            random_row[
                "product_id"
            ]
        )

        random_distance = float(
            random_row[
                "distance"
            ]
        )

        used_for_person.add(
            random_product
        )

        # ====================================================
        # STORE
        # ====================================================

        output_rows.extend(
            [
                {
                    "person_id":
                        person_id,

                    "positive_product_id":
                        positive_product,

                    "negative_product_id":
                        hard_product,

                    "negative_type":
                        "hard",

                    "negative_distance":
                        hard_distance,
                },

                {
                    "person_id":
                        person_id,

                    "positive_product_id":
                        positive_product,

                    "negative_product_id":
                        semi_product,

                    "negative_type":
                        "semi-hard",

                    "negative_distance":
                        semi_distance,
                },

                {
                    "person_id":
                        person_id,

                    "positive_product_id":
                        positive_product,

                    "negative_product_id":
                        random_product,

                    "negative_type":
                        "random",

                    "negative_distance":
                        random_distance,
                },
            ]
        )

# ============================================================
# RESULT DATAFRAME
# ============================================================

result = pd.DataFrame(
    output_rows
)

# ============================================================
# RESULTS
# ============================================================

print("\n" + "=" * 65)
print("SAMPLING RESULTS")
print("=" * 65)

print(
    "Positive interactions:",
    len(positive_df)
)

print(
    "Generated ranking rows:",
    len(result)
)

print(
    "\nNegative type distribution:"
)

print(
    result[
        "negative_type"
    ].value_counts()
)

# ============================================================
# EXPECTED COUNTS
# ============================================================

expected_count = len(
    positive_df
)

hard_count_actual = (
    result[
        "negative_type"
    ]
    .eq("hard")
    .sum()
)

semi_count_actual = (
    result[
        "negative_type"
    ]
    .eq("semi-hard")
    .sum()
)

random_count_actual = (
    result[
        "negative_type"
    ]
    .eq("random")
    .sum()
)

print("\nExpected:")
print(
    "Hard      :",
    expected_count
)

print(
    "Semi-hard :",
    expected_count
)

print(
    "Random    :",
    expected_count
)

print("\nActual:")
print(
    "Hard      :",
    hard_count_actual
)

print(
    "Semi-hard :",
    semi_count_actual
)

print(
    "Random    :",
    random_count_actual
)

# ============================================================
# DUPLICATE CHECK
# ============================================================

duplicate_rows = (
    result[
        [
            "person_id",
            "positive_product_id",
            "negative_product_id",
            "negative_type",
        ]
    ]
    .duplicated()
    .sum()
)

duplicate_person_negative = (
    result[
        [
            "person_id",
            "negative_product_id",
        ]
    ]
    .duplicated()
    .sum()
)

print("\n" + "=" * 65)
print("INTEGRITY")
print("=" * 65)

print(
    "Duplicate full rows:",
    duplicate_rows
)

print(
    "Duplicate person-negative pairs:",
    duplicate_person_negative
)

# ============================================================
# POSITIVE COLLISIONS
# ============================================================

positive_keys = set(
    zip(
        positive_df[
            "person_id"
        ],
        positive_df[
            "product_id"
        ],
    )
)

negative_keys = set(
    zip(
        result[
            "person_id"
        ],
        result[
            "negative_product_id"
        ],
    )
)

collisions = (
    positive_keys
    &
    negative_keys
)

print(
    "Positive-negative collisions:",
    len(collisions)
)

# ============================================================
# DISTANCE DISTRIBUTION
# ============================================================

print("\n" + "=" * 65)
print("DISTANCE DISTRIBUTION")
print("=" * 65)

distance_stats = (
    result
    .groupby(
        "negative_type"
    )[
        "negative_distance"
    ]
    .describe()
)

print(
    distance_stats
)

# ============================================================
# MEAN DISTANCE
# ============================================================

means = (
    result
    .groupby(
        "negative_type"
    )[
        "negative_distance"
    ]
    .mean()
)

hard_mean = means.get(
    "hard",
    np.nan
)

semi_mean = means.get(
    "semi-hard",
    np.nan
)

random_mean = means.get(
    "random",
    np.nan
)

print("\nMean distances:")

print(
    "Hard      :",
    f"{hard_mean:.6f}"
)

print(
    "Semi-hard :",
    f"{semi_mean:.6f}"
)

print(
    "Random    :",
    f"{random_mean:.6f}"
)

# ============================================================
# DIFFICULTY ORDER CHECK
# ============================================================

difficulty_ok = (
    hard_mean
    <
    semi_mean
    <
    random_mean
)

print(
    "\nHard < Semi-hard < Random:",
    difficulty_ok
)

# ============================================================
# PER PERSON COUNTS
# ============================================================

person_type_counts = (
    result
    .groupby(
        [
            "person_id",
            "negative_type",
        ]
    )
    .size()
    .unstack(
        fill_value=0
    )
)

print("\n" + "=" * 65)
print("PER-PERSON NEGATIVE COUNTS")
print("=" * 65)

print(
    person_type_counts
)

# ============================================================
# FAILED PEOPLE
# ============================================================

print("\n" + "=" * 65)
print("PEOPLE WITH INSUFFICIENT CANDIDATES")
print("=" * 65)

if failed_people:

    for person_id, count in failed_people:

        print(
            person_id,
            "candidates:",
            count
        )

else:

    print(
        "None"
    )

# ============================================================
# SAMPLE
# ============================================================

print("\n" + "=" * 65)
print("EXAMPLE SAMPLES")
print("=" * 65)

display(
    result.head(20)
)

# ============================================================
# FINAL PASS/FAIL
# ============================================================

counts_ok = (
    hard_count_actual
    ==
    expected_count
    and
    semi_count_actual
    ==
    expected_count
    and
    random_count_actual
    ==
    expected_count
)

integrity_ok = (
    duplicate_rows == 0
    and
    duplicate_person_negative == 0
    and
    len(collisions) == 0
)

final_pass = (
    counts_ok
    and
    integrity_ok
    and
    difficulty_ok
)

print("\n" + "=" * 65)
print("FINAL VALIDATION")
print("=" * 65)

print(
    "1:1:1 counts:",
    "PASS" if counts_ok else "FAIL"
)

print(
    "No duplicate rows:",
    "PASS"
    if duplicate_rows == 0
    else "FAIL"
)

print(
    "No duplicate person-negative pairs:",
    "PASS"
    if duplicate_person_negative == 0
    else "FAIL"
)

print(
    "No positive collisions:",
    "PASS"
    if len(collisions) == 0
    else "FAIL"
)

print(
    "Hard < Semi-hard < Random:",
    "PASS"
    if difficulty_ok
    else "FAIL"
)

print(
    "\nOVERALL:",
    "PASS" if final_pass else "FAIL"
)

# ============================================================
# SAVE ONLY IF VALID
# ============================================================

if final_pass:

    result.to_csv(
        OUTPUT_PATH,
        index=False,
    )

    print(
        "\nSaved:",
        OUTPUT_PATH
    )

else:

    print(
        "\nDataset NOT saved as valid."
    )

print(
    "\nCorrected negative sampling complete."
)
```

```python
# ============================================================
# ZYRA V2 — FINAL 1:1:1 NEGATIVE SAMPLER
# ============================================================

import pandas as pd
import numpy as np
from pathlib import Path

# ============================================================
# CONFIG
# ============================================================

BASE = Path(
    "/Users/saketh/Desktop/Projects/weavly/core-model"
)

RAW_PATH = (
    BASE /
    "data/v2/v2_compatibility_raw.csv"
)

OUTPUT_PATH = (
    BASE /
    "data/v2/v2_negative_sampling_1to1to1.csv"
)

SEED = 2026

rng = np.random.default_rng(SEED)

# ============================================================
# LOAD
# ============================================================

df = pd.read_csv(
    RAW_PATH
)

positive_df = (
    df[
        df["label"] == 1
    ][
        [
            "person_id",
            "product_id",
        ]
    ]
    .drop_duplicates()
    .copy()
)

person_lookup = (
    df[
        [
            "person_id",
            "height",
            "bust",
            "waist",
            "hips",
        ]
    ]
    .drop_duplicates(
        "person_id"
    )
    .set_index(
        "person_id"
    )
)

garment_lookup = (
    df[
        [
            "product_id",
            "garment_bust",
            "garment_waist",
            "garment_hip",
            "garment_length",
        ]
    ]
    .drop_duplicates(
        "product_id"
    )
    .set_index(
        "product_id"
    )
)

all_products = list(
    garment_lookup.index
)

positive_by_person = (
    positive_df
    .groupby(
        "person_id"
    )[
        "product_id"
    ]
    .apply(set)
    .to_dict()
)

# ============================================================
# DISTANCE
# ============================================================

def physical_distance(
    person_id,
    product_id,
):

    p = person_lookup.loc[
        person_id
    ]

    g = garment_lookup.loc[
        product_id
    ]

    values = []

    dimensions = [
        (
            p["bust"],
            g["garment_bust"],
        ),
        (
            p["waist"],
            g["garment_waist"],
        ),
        (
            p["hips"],
            g["garment_hip"],
        ),
        (
            p["height"],
            g["garment_length"],
        ),
    ]

    for pv, gv in dimensions:

        if (
            pd.isna(pv)
            or
            pd.isna(gv)
        ):
            continue

        if pv == 0:
            continue

        values.append(
            abs(
                gv - pv
            )
            /
            abs(pv)
        )

    if not values:
        return np.nan

    return float(
        np.mean(values)
    )

# ============================================================
# BUILD CANDIDATES
# ============================================================

candidate_rows = []

for person_id in sorted(
    positive_by_person
):

    positives = (
        positive_by_person[
            person_id
        ]
    )

    for product_id in all_products:

        if product_id in positives:
            continue

        distance = physical_distance(
            person_id,
            product_id,
        )

        if pd.isna(distance):
            continue

        candidate_rows.append(
            {
                "person_id":
                    person_id,

                "product_id":
                    product_id,

                "distance":
                    distance,
            }
        )

candidates = pd.DataFrame(
    candidate_rows
)

# ============================================================
# SAMPLING
# ============================================================

rows = []

for person_id in sorted(
    positive_by_person
):

    positives = sorted(
        positive_by_person[
            person_id
        ]
    )

    person_candidates = (
        candidates[
            candidates[
                "person_id"
            ]
            ==
            person_id
        ]
        .sort_values(
            "distance"
        )
        .reset_index(
            drop=True
        )
    )

    n = len(
        person_candidates
    )

    if n < 3:

        raise ValueError(
            f"Person {person_id} has "
            f"only {n} valid negative candidates."
        )

    # --------------------------------------------------------
    # GLOBAL POOLS
    # --------------------------------------------------------

    hard_end = max(
        1,
        int(
            np.ceil(
                n * 0.20
            )
        )
    )

    semi_start = hard_end

    semi_end = max(
        semi_start + 1,
        int(
            np.ceil(
                n * 0.60
            )
        )
    )

    hard_pool = (
        person_candidates
        .iloc[
            :hard_end
        ]
        .copy()
    )

    semi_pool = (
        person_candidates
        .iloc[
            semi_start:semi_end
        ]
        .copy()
    )

    random_pool = (
        person_candidates
        .iloc[
            semi_end:
        ]
        .copy()
    )

    # Safety
    if len(semi_pool) == 0:
        semi_pool = person_candidates

    if len(random_pool) == 0:
        random_pool = person_candidates

    # --------------------------------------------------------
    # USED FULL KEYS
    #
    # IMPORTANT:
    # Same negative garment MAY be used for another positive
    # interaction.
    #
    # But the exact:
    #
    # person + positive + negative + type
    #
    # combination can never repeat.
    # --------------------------------------------------------

    used_keys = set()

    # --------------------------------------------------------
    # EVERY POSITIVE GETS THREE NEGATIVES
    # --------------------------------------------------------

    for positive_product in positives:

        # ====================================================
        # HARD
        # ====================================================

        hard_options = []

        for _, candidate in hard_pool.iterrows():

            key = (
                person_id,
                positive_product,
                candidate["product_id"],
                "hard",
            )

            if key not in used_keys:

                hard_options.append(
                    candidate
                )

        if not hard_options:

            raise RuntimeError(
                f"No hard negative available "
                f"for person={person_id}, "
                f"positive={positive_product}"
            )

        hard_row = hard_options[
            rng.integers(
                len(hard_options)
            )
        ]

        hard_product = (
            hard_row["product_id"]
        )

        hard_distance = float(
            hard_row["distance"]
        )

        hard_key = (
            person_id,
            positive_product,
            hard_product,
            "hard",
        )

        used_keys.add(
            hard_key
        )

        # ====================================================
        # SEMI-HARD
        # ====================================================

        semi_options = []

        for _, candidate in semi_pool.iterrows():

            key = (
                person_id,
                positive_product,
                candidate["product_id"],
                "semi-hard",
            )

            if key not in used_keys:

                # Prefer a garment farther away
                # than the hard negative.
                if (
                    candidate["distance"]
                    >
                    hard_distance
                ):

                    semi_options.append(
                        candidate
                    )

        if not semi_options:

            for _, candidate in semi_pool.iterrows():

                key = (
                    person_id,
                    positive_product,
                    candidate["product_id"],
                    "semi-hard",
                )

                if key not in used_keys:

                    semi_options.append(
                        candidate
                    )

        if not semi_options:

            raise RuntimeError(
                f"No semi-hard negative available "
                f"for person={person_id}, "
                f"positive={positive_product}"
            )

        semi_row = semi_options[
            rng.integers(
                len(semi_options)
            )
        ]

        semi_product = (
            semi_row["product_id"]
        )

        semi_distance = float(
            semi_row["distance"]
        )

        semi_key = (
            person_id,
            positive_product,
            semi_product,
            "semi-hard",
        )

        used_keys.add(
            semi_key
        )

        # ====================================================
        # RANDOM
        # ====================================================

        random_options = []

        for _, candidate in random_pool.iterrows():

            key = (
                person_id,
                positive_product,
                candidate["product_id"],
                "random",
            )

            if key not in used_keys:

                # Prefer distance greater than semi-hard.
                if (
                    candidate["distance"]
                    >
                    semi_distance
                ):

                    random_options.append(
                        candidate
                    )

        if not random_options:

            for _, candidate in random_pool.iterrows():

                key = (
                    person_id,
                    positive_product,
                    candidate["product_id"],
                    "random",
                )

                if key not in used_keys:

                    random_options.append(
                        candidate
                    )

        if not random_options:

            raise RuntimeError(
                f"No random negative available "
                f"for person={person_id}, "
                f"positive={positive_product}"
            )

        random_row = random_options[
            rng.integers(
                len(random_options)
            )
        ]

        random_product = (
            random_row["product_id"]
        )

        random_distance = float(
            random_row["distance"]
        )

        random_key = (
            person_id,
            positive_product,
            random_product,
            "random",
        )

        used_keys.add(
            random_key
        )

        # ====================================================
        # SAVE THREE
        # ====================================================

        rows.extend(
            [
                {
                    "person_id":
                        person_id,

                    "positive_product_id":
                        positive_product,

                    "negative_product_id":
                        hard_product,

                    "negative_type":
                        "hard",

                    "negative_distance":
                        hard_distance,
                },

                {
                    "person_id":
                        person_id,

                    "positive_product_id":
                        positive_product,

                    "negative_product_id":
                        semi_product,

                    "negative_type":
                        "semi-hard",

                    "negative_distance":
                        semi_distance,
                },

                {
                    "person_id":
                        person_id,

                    "positive_product_id":
                        positive_product,

                    "negative_product_id":
                        random_product,

                    "negative_type":
                        "random",

                    "negative_distance":
                        random_distance,
                },
            ]
        )

# ============================================================
# DATAFRAME
# ============================================================

result = pd.DataFrame(
    rows
)

# ============================================================
# COUNTS
# ============================================================

print("=" * 65)
print("FINAL 1:1:1 SAMPLING")
print("=" * 65)

print(
    "Positive interactions:",
    len(positive_df)
)

print(
    "Expected total:",
    len(positive_df) * 3
)

print(
    "Actual total:",
    len(result)
)

print(
    "\nNegative distribution:"
)

print(
    result[
        "negative_type"
    ].value_counts()
)

# ============================================================
# DUPLICATES
# ============================================================

duplicate_full = (
    result[
        [
            "person_id",
            "positive_product_id",
            "negative_product_id",
            "negative_type",
        ]
    ]
    .duplicated()
    .sum()
)

print(
    "\nDuplicate full rows:",
    duplicate_full
)

# ============================================================
# POSITIVE COLLISIONS
# ============================================================

positive_keys = set(
    zip(
        positive_df[
            "person_id"
        ],
        positive_df[
            "product_id"
        ],
    )
)

negative_keys = set(
    zip(
        result[
            "person_id"
        ],
        result[
            "negative_product_id"
        ],
    )
)

collisions = (
    positive_keys
    &
    negative_keys
)

print(
    "Positive-negative collisions:",
    len(collisions)
)

# ============================================================
# DISTANCE
# ============================================================

distance_stats = (
    result
    .groupby(
        "negative_type"
    )[
        "negative_distance"
    ]
    .agg(
        [
            "count",
            "mean",
            "median",
            "min",
            "max",
        ]
    )
)

print(
    "\nDistance statistics:"
)

print(
    distance_stats
)

# ============================================================
# DIFFICULTY
# ============================================================

means = (
    result
    .groupby(
        "negative_type"
    )[
        "negative_distance"
    ]
    .mean()
)

hard_mean = means[
    "hard"
]

semi_mean = means[
    "semi-hard"
]

random_mean = means[
    "random"
]

difficulty_ok = (
    hard_mean
    <
    semi_mean
    <
    random_mean
)

print(
    "\nHard mean      :",
    hard_mean
)

print(
    "Semi-hard mean :",
    semi_mean
)

print(
    "Random mean    :",
    random_mean
)

print(
    "\nHard < Semi-hard < Random:",
    difficulty_ok
)

# ============================================================
# 1:1:1 VALIDATION
# ============================================================

hard_count = (
    result[
        "negative_type"
    ]
    .eq("hard")
    .sum()
)

semi_count = (
    result[
        "negative_type"
    ]
    .eq("semi-hard")
    .sum()
)

random_count = (
    result[
        "negative_type"
    ]
    .eq("random")
    .sum()
)

expected = len(
    positive_df
)

counts_ok = (
    hard_count == expected
    and
    semi_count == expected
    and
    random_count == expected
    and
    len(result)
    == expected * 3
)

print("\n" + "=" * 65)
print("FINAL VALIDATION")
print("=" * 65)

print(
    "Hard count:",
    hard_count,
    "/",
    expected
)

print(
    "Semi-hard count:",
    semi_count,
    "/",
    expected
)

print(
    "Random count:",
    random_count,
    "/",
    expected
)

print(
    "1:1:1 counts:",
    "PASS" if counts_ok else "FAIL"
)

print(
    "No duplicate rows:",
    "PASS"
    if duplicate_full == 0
    else "FAIL"
)

print(
    "No positive collisions:",
    "PASS"
    if len(collisions) == 0
    else "FAIL"
)

print(
    "Difficulty ordering:",
    "PASS"
    if difficulty_ok
    else "FAIL"
)

final_pass = (
    counts_ok
    and duplicate_full == 0
    and len(collisions) == 0
    and difficulty_ok
)

print(
    "\nOVERALL:",
    "PASS"
    if final_pass
    else "FAIL"
)

# ============================================================
# SAVE
# ============================================================

if final_pass:

    result.to_csv(
        OUTPUT_PATH,
        index=False,
    )

    print(
        "\nSaved:",
        OUTPUT_PATH
    )

else:

    print(
        "\nNOT SAVED — validation failed."
    )

print(
    "\nSampling complete."
)
```

```python
# ============================================================
# ZYRA V2 — EXPERIMENT #2
# TRUE PAIRWISE RANKER — FINAL 1:1:1 DATASET
# ============================================================

import pandas as pd
import numpy as np
import torch
import torch.nn as nn
import torch.nn.functional as F

from pathlib import Path
from sklearn.model_selection import train_test_split
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler

# ============================================================
# CONFIG
# ============================================================

BASE = Path(
    "/Users/saketh/Desktop/Projects/weavly/core-model"
)

DATA_PATH = (
    BASE /
    "data/v2/v2_negative_sampling_1to1to1.csv"
)

RAW_PATH = (
    BASE /
    "data/v2/v2_compatibility_raw.csv"
)

MODEL_PATH = (
    BASE /
    "data/v2/v2_pairwise_ranker_exp2_final.pt"
)

SEED = 42
EPOCHS = 50
LR = 1e-3
WEIGHT_DECAY = 1e-4

np.random.seed(SEED)
torch.manual_seed(SEED)

# ============================================================
# DEVICE
# ============================================================

if torch.backends.mps.is_available():
    device = torch.device("mps")
else:
    device = torch.device("cpu")

print("=" * 65)
print("ZYRA V2 — EXPERIMENT #2 FINAL TRUE PAIRWISE TRAINING")
print("=" * 65)

print("Device:", device)

# ============================================================
# LOAD DATA
# ============================================================

pairwise = pd.read_csv(DATA_PATH)
raw = pd.read_csv(RAW_PATH)

print("\nPairwise rows:", len(pairwise))
print("People:", pairwise["person_id"].nunique())

print("\nNegative types:")
print(pairwise["negative_type"].value_counts())

# ============================================================
# VERIFY 1:1:1 DATASET BEFORE TRAINING
# ============================================================

positive_count = pairwise[
    ["person_id", "positive_product_id"]
].drop_duplicates().shape[0]

hard_count = (
    pairwise["negative_type"]
    .eq("hard")
    .sum()
)

semi_count = (
    pairwise["negative_type"]
    .eq("semi-hard")
    .sum()
)

random_count = (
    pairwise["negative_type"]
    .eq("random")
    .sum()
)

print("\n" + "=" * 65)
print("INPUT DATASET VALIDATION")
print("=" * 65)

print("Positive interactions:", positive_count)
print("Hard negatives:", hard_count)
print("Semi-hard negatives:", semi_count)
print("Random negatives:", random_count)
print("Total rows:", len(pairwise))

assert hard_count == positive_count
assert semi_count == positive_count
assert random_count == positive_count
assert len(pairwise) == positive_count * 3

print("\n1:1:1 DATASET: PASS")

# ============================================================
# FEATURE ENGINEERING
# ============================================================

FEATURES = [
    "height",
    "bust",
    "waist",
    "hips",

    "garment_bust",
    "garment_waist",
    "garment_hip",
    "garment_length",
    "garment_sleeve",

    "bust_ease",
    "waist_ease",
    "hip_ease",

    "bust_ratio",
    "waist_ratio",
    "hip_ratio",

    "bust_mismatch",
    "waist_mismatch",
    "hip_mismatch",

    "bust_relative_mismatch",
    "waist_relative_mismatch",
    "hip_relative_mismatch",

    "mean_ease",
    "ease_std",

    "known_fit_dimensions",
    "tight_dimension_count",
    "extreme_ease_count",

    "length_height_ratio",
]

# ============================================================
# LOOKUPS
# ============================================================

person_lookup = (
    raw[
        [
            "person_id",
            "height",
            "bust",
            "waist",
            "hips",
        ]
    ]
    .drop_duplicates("person_id")
    .set_index("person_id")
)

garment_lookup = (
    raw[
        [
            "product_id",
            "garment_bust",
            "garment_waist",
            "garment_hip",
            "garment_length",
            "garment_sleeve",
        ]
    ]
    .drop_duplicates("product_id")
    .set_index("product_id")
)

# ============================================================
# FEATURE BUILDER
# ============================================================

def make_features(person_id, product_id):

    p = person_lookup.loc[person_id]
    g = garment_lookup.loc[product_id]

    row = {}

    # --------------------------------------------------------
    # Body
    # --------------------------------------------------------

    row["height"] = p["height"]
    row["bust"] = p["bust"]
    row["waist"] = p["waist"]
    row["hips"] = p["hips"]

    # --------------------------------------------------------
    # Garment
    # --------------------------------------------------------

    row["garment_bust"] = g["garment_bust"]
    row["garment_waist"] = g["garment_waist"]
    row["garment_hip"] = g["garment_hip"]
    row["garment_length"] = g["garment_length"]
    row["garment_sleeve"] = g["garment_sleeve"]

    # --------------------------------------------------------
    # Ease
    # --------------------------------------------------------

    row["bust_ease"] = (
        g["garment_bust"] - p["bust"]
    )

    row["waist_ease"] = (
        g["garment_waist"] - p["waist"]
    )

    row["hip_ease"] = (
        g["garment_hip"] - p["hips"]
    )

    # --------------------------------------------------------
    # Ratios
    # --------------------------------------------------------

    row["bust_ratio"] = (
        g["garment_bust"] / p["bust"]
    )

    row["waist_ratio"] = (
        g["garment_waist"] / p["waist"]
    )

    row["hip_ratio"] = (
        g["garment_hip"] / p["hips"]
    )

    # --------------------------------------------------------
    # Absolute mismatch
    # --------------------------------------------------------

    row["bust_mismatch"] = abs(
        g["garment_bust"] - p["bust"]
    )

    row["waist_mismatch"] = abs(
        g["garment_waist"] - p["waist"]
    )

    row["hip_mismatch"] = abs(
        g["garment_hip"] - p["hips"]
    )

    # --------------------------------------------------------
    # Relative mismatch
    # --------------------------------------------------------

    row["bust_relative_mismatch"] = (
        row["bust_mismatch"] / p["bust"]
    )

    row["waist_relative_mismatch"] = (
        row["waist_mismatch"] / p["waist"]
    )

    row["hip_relative_mismatch"] = (
        row["hip_mismatch"] / p["hips"]
    )

    # --------------------------------------------------------
    # Aggregate ease
    # --------------------------------------------------------

    eases = np.array(
        [
            row["bust_ease"],
            row["waist_ease"],
            row["hip_ease"],
        ],
        dtype=float,
    )

    valid_eases = eases[
        ~np.isnan(eases)
    ]

    if len(valid_eases) == 0:

        row["mean_ease"] = np.nan
        row["ease_std"] = np.nan

    else:

        row["mean_ease"] = np.mean(
            valid_eases
        )

        row["ease_std"] = np.std(
            valid_eases
        )

    row["known_fit_dimensions"] = (
        np.sum(
            ~np.isnan(eases)
        )
    )

    row["tight_dimension_count"] = (
        np.sum(
            eases < 0
        )
    )

    row["extreme_ease_count"] = (
        np.sum(
            eases > 50
        )
    )

    # --------------------------------------------------------
    # Length / height
    # --------------------------------------------------------

    if (
        pd.isna(g["garment_length"])
        or
        pd.isna(p["height"])
    ):

        row["length_height_ratio"] = np.nan

    else:

        row["length_height_ratio"] = (
            g["garment_length"]
            /
            p["height"]
        )

    return np.array(
        [
            row[f]
            for f in FEATURES
        ],
        dtype=np.float32,
    )

print("\nFeature dimension:", len(FEATURES))

# ============================================================
# PERSON-LEVEL SPLIT
# ============================================================

people = sorted(
    pairwise["person_id"].unique()
)

train_people, temp_people = train_test_split(
    people,
    test_size=16,
    random_state=SEED,
)

val_people, test_people = train_test_split(
    temp_people,
    test_size=8,
    random_state=SEED,
)

train_people = set(train_people)
val_people = set(val_people)
test_people = set(test_people)

print("\n" + "=" * 65)
print("PERSON-LEVEL SPLIT")
print("=" * 65)

print("Train people:", len(train_people))
print("Validation people:", len(val_people))
print("Test people:", len(test_people))

print("\nLeakage:")
print(
    "Train ∩ Val:",
    train_people & val_people
)
print(
    "Train ∩ Test:",
    train_people & test_people
)
print(
    "Val ∩ Test:",
    val_people & test_people
)

# ============================================================
# SPLIT PAIRWISE DATA
# ============================================================

train_df = pairwise[
    pairwise["person_id"].isin(train_people)
].copy()

val_df = pairwise[
    pairwise["person_id"].isin(val_people)
].copy()

test_df = pairwise[
    pairwise["person_id"].isin(test_people)
].copy()

print("\nPairwise rows:")
print("Train:", len(train_df))
print("Val  :", len(val_df))
print("Test :", len(test_df))

# ============================================================
# BUILD POSITIVE / NEGATIVE FEATURES
# ============================================================

def build_pair_features(frame):

    pos = []
    neg = []

    for _, row in frame.iterrows():

        pos.append(
            make_features(
                row["person_id"],
                row["positive_product_id"],
            )
        )

        neg.append(
            make_features(
                row["person_id"],
                row["negative_product_id"],
            )
        )

    return (
        np.asarray(pos, dtype=np.float32),
        np.asarray(neg, dtype=np.float32),
    )

X_pos_train, X_neg_train = build_pair_features(
    train_df
)

X_pos_val, X_neg_val = build_pair_features(
    val_df
)

X_pos_test, X_neg_test = build_pair_features(
    test_df
)

# ============================================================
# IMPUTATION
# ============================================================

imputer = SimpleImputer(
    strategy="median"
)

train_reference = np.vstack(
    [
        X_pos_train,
        X_neg_train,
    ]
)

imputer.fit(train_reference)

X_pos_train = imputer.transform(
    X_pos_train
)

X_neg_train = imputer.transform(
    X_neg_train
)

X_pos_val = imputer.transform(
    X_pos_val
)

X_neg_val = imputer.transform(
    X_neg_val
)

X_pos_test = imputer.transform(
    X_pos_test
)

X_neg_test = imputer.transform(
    X_neg_test
)

# ============================================================
# SCALING
# ============================================================

scaler = StandardScaler()

scaler.fit(
    train_reference
)

X_pos_train = scaler.transform(
    X_pos_train
)

X_neg_train = scaler.transform(
    X_neg_train
)

X_pos_val = scaler.transform(
    X_pos_val
)

X_neg_val = scaler.transform(
    X_neg_val
)

X_pos_test = scaler.transform(
    X_pos_test
)

X_neg_test = scaler.transform(
    X_neg_test
)

# ============================================================
# TENSORS
# ============================================================

def tensor(x):

    return torch.tensor(
        x,
        dtype=torch.float32,
        device=device,
    )

X_pos_train = tensor(X_pos_train)
X_neg_train = tensor(X_neg_train)

X_pos_val = tensor(X_pos_val)
X_neg_val = tensor(X_neg_val)

X_pos_test = tensor(X_pos_test)
X_neg_test = tensor(X_neg_test)

print("\n" + "=" * 65)
print("PAIRWISE TENSORS")
print("=" * 65)

print(
    "Train:",
    X_pos_train.shape
)

print(
    "Val:",
    X_pos_val.shape
)

print(
    "Test:",
    X_pos_test.shape
)

print(
    "NaN:",
    (
        torch.isnan(X_pos_train).sum()
        + torch.isnan(X_neg_train).sum()
        + torch.isnan(X_pos_val).sum()
        + torch.isnan(X_neg_val).sum()
        + torch.isnan(X_pos_test).sum()
        + torch.isnan(X_neg_test).sum()
    ).item()
)

print(
    "Inf:",
    (
        torch.isinf(X_pos_train).sum()
        + torch.isinf(X_neg_train).sum()
        + torch.isinf(X_pos_val).sum()
        + torch.isinf(X_neg_val).sum()
        + torch.isinf(X_pos_test).sum()
        + torch.isinf(X_neg_test).sum()
    ).item()
)

# ============================================================
# MODEL
# ============================================================

class CompatibilityScorer(nn.Module):

    def __init__(self, input_dim):

        super().__init__()

        self.network = nn.Sequential(

            nn.Linear(
                input_dim,
                64
            ),

            nn.ReLU(),

            nn.Linear(
                64,
                32
            ),

            nn.ReLU(),

            nn.Linear(
                32,
                1
            ),
        )

    def forward(self, x):

        return self.network(
            x
        ).squeeze(-1)


model = CompatibilityScorer(
    len(FEATURES)
).to(device)

optimizer = torch.optim.Adam(
    model.parameters(),
    lr=LR,
    weight_decay=WEIGHT_DECAY,
)

print("\n" + "=" * 65)
print("MODEL")
print("=" * 65)

print(
    "Input dimension:",
    len(FEATURES)
)

print(
    "Parameters:",
    sum(
        p.numel()
        for p in model.parameters()
    )
)

# ============================================================
# TRUE PAIRWISE LOSS
# ============================================================

def pairwise_loss(
    positive_scores,
    negative_scores,
):

    margin = (
        positive_scores
        -
        negative_scores
    )

    return F.softplus(
        -margin
    ).mean()

# ============================================================
# TRAIN
# ============================================================

print("\n" + "=" * 65)
print("TRAINING")
print("=" * 65)

best_val_acc = -np.inf
best_state = None

for epoch in range(
    1,
    EPOCHS + 1
):

    model.train()

    optimizer.zero_grad()

    positive_scores = model(
        X_pos_train
    )

    negative_scores = model(
        X_neg_train
    )

    loss = pairwise_loss(
        positive_scores,
        negative_scores,
    )

    loss.backward()

    optimizer.step()

    # --------------------------------------------------------
    # Evaluation
    # --------------------------------------------------------

    model.eval()

    with torch.no_grad():

        train_pos = model(
            X_pos_train
        )

        train_neg = model(
            X_neg_train
        )

        val_pos = model(
            X_pos_val
        )

        val_neg = model(
            X_neg_val
        )

        train_acc = (
            train_pos > train_neg
        ).float().mean().item()

        val_acc = (
            val_pos > val_neg
        ).float().mean().item()

    # --------------------------------------------------------
    # Best checkpoint
    # --------------------------------------------------------

    if val_acc > best_val_acc:

        best_val_acc = val_acc

        best_state = {
            k: v.detach()
            .cpu()
            .clone()

            for k, v
            in model.state_dict().items()
        }

    if (
        epoch == 1
        or epoch % 5 == 0
    ):

        print(
            f"Epoch {epoch:02d}/{EPOCHS} "
            f"| Loss: {loss.item():.4f} "
            f"| Train Rank Acc: {train_acc:.4f} "
            f"| Val Rank Acc: {val_acc:.4f}"
        )

# ============================================================
# RESTORE BEST CHECKPOINT
# ============================================================

model.load_state_dict(
    best_state
)

model.eval()

# ============================================================
# FINAL EVALUATION
# ============================================================

with torch.no_grad():

    train_pos = model(
        X_pos_train
    )

    train_neg = model(
        X_neg_train
    )

    val_pos = model(
        X_pos_val
    )

    val_neg = model(
        X_neg_val
    )

    test_pos = model(
        X_pos_test
    )

    test_neg = model(
        X_neg_test
    )

# ============================================================
# METRICS
# ============================================================

train_acc = (
    train_pos > train_neg
).float().mean().item()

val_acc = (
    val_pos > val_neg
).float().mean().item()

test_acc = (
    test_pos > test_neg
).float().mean().item()

train_gap = (
    train_pos - train_neg
)

val_gap = (
    val_pos - val_neg
)

test_gap = (
    test_pos - test_neg
)

# ============================================================
# FINAL RESULTS
# ============================================================

print("\n" + "=" * 65)
print("FINAL EXPERIMENT #2 RESULTS")
print("=" * 65)

print(
    "Train Ranking Accuracy :",
    f"{train_acc:.4f}"
)

print(
    "Validation Ranking Acc :",
    f"{val_acc:.4f}"
)

print(
    "Test Ranking Accuracy  :",
    f"{test_acc:.4f}"
)

print(
    "Best Validation Acc    :",
    f"{best_val_acc:.4f}"
)

print("\nScore gaps:")

print(
    "Train mean:",
    f"{train_gap.mean().item():.4f}"
)

print(
    "Validation mean:",
    f"{val_gap.mean().item():.4f}"
)

print(
    "Test mean:",
    f"{test_gap.mean().item():.4f}"
)

print(
    "Test median:",
    f"{test_gap.median().item():.4f}"
)

# ============================================================
# TEST PERFORMANCE BY NEGATIVE TYPE
# ============================================================

test_results = test_df.copy()

test_results["positive_score"] = (
    test_pos
    .detach()
    .cpu()
    .numpy()
)

test_results["negative_score"] = (
    test_neg
    .detach()
    .cpu()
    .numpy()
)

test_results["score_gap"] = (
    test_results["positive_score"]
    -
    test_results["negative_score"]
)

test_results["correct"] = (
    test_results["positive_score"]
    >
    test_results["negative_score"]
)

type_results = (
    test_results
    .groupby("negative_type")
    .agg(
        pairs=(
            "correct",
            "size"
        ),

        ranking_accuracy=(
            "correct",
            "mean"
        ),

        mean_score_gap=(
            "score_gap",
            "mean"
        ),

        median_score_gap=(
            "score_gap",
            "median"
        ),
    )
)

print("\n" + "=" * 65)
print("TEST PERFORMANCE BY NEGATIVE TYPE")
print("=" * 65)

print(
    type_results
)

# ============================================================
# SAVE MODEL
# ============================================================

torch.save(
    {
        "model_state_dict":
            model.state_dict(),

        "input_dim":
            len(FEATURES),

        "feature_names":
            FEATURES,

        "experiment":
            "V2_EXP2_FINAL_TRUE_PAIRWISE",

        "negative_sampling":
            "1:1:1",

        "negative_types":
            [
                "hard",
                "semi-hard",
                "random",
            ],

        "train_people":
            sorted(train_people),

        "val_people":
            sorted(val_people),

        "test_people":
            sorted(test_people),

        "seed":
            SEED,
    },
    MODEL_PATH,
)

print("\n" + "=" * 65)
print("MODEL SAVED")
print("=" * 65)

print(
    MODEL_PATH
)

print("\n" + "=" * 65)
print("EXPERIMENT #2 TRAINING COMPLETE")
print("=" * 65)
```

```python
# ============================================================
# ZYRA V2 — EXPERIMENT #2 RANKING DIAGNOSTIC
# ============================================================
#
# Purpose:
#   Understand why:
#       HARD       -> 86.14%
#       SEMI-HARD  -> 74.63%
#       RANDOM     -> 29.50%
#
# This cell:
#   1. Loads the saved Experiment #2 model
#   2. Recreates the SAME test split
#   3. Scores positive / hard / semi-hard / random garments
#   4. Finds cases where random negatives beat positives
#   5. Compares physical compatibility features
#   6. Reports per-person behaviour
#
# NO TRAINING.
# ============================================================

import pandas as pd
import numpy as np
import torch
import torch.nn as nn

from pathlib import Path
from sklearn.model_selection import train_test_split
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler

# ============================================================
# PATHS
# ============================================================

BASE = Path(
    "/Users/saketh/Desktop/Projects/weavly/core-model"
)

PAIRWISE_PATH = (
    BASE /
    "data/v2/v2_negative_sampling_1to1to1.csv"
)

RAW_PATH = (
    BASE /
    "data/v2/v2_compatibility_raw.csv"
)

MODEL_PATH = (
    BASE /
    "data/v2/v2_pairwise_ranker_exp2_final.pt"
)

# ============================================================
# CONFIG
# ============================================================

SEED = 42

np.random.seed(SEED)
torch.manual_seed(SEED)

if torch.backends.mps.is_available():
    device = torch.device("mps")
else:
    device = torch.device("cpu")

print("=" * 70)
print("ZYRA V2 — EXPERIMENT #2 RANKING DIAGNOSTIC")
print("=" * 70)

print("Device:", device)

# ============================================================
# LOAD
# ============================================================

pairwise = pd.read_csv(
    PAIRWISE_PATH
)

raw = pd.read_csv(
    RAW_PATH
)

print(
    "\nPairwise rows:",
    len(pairwise)
)

print(
    "People:",
    pairwise["person_id"].nunique()
)

# ============================================================
# FEATURES
# ============================================================

FEATURES = [
    "height",
    "bust",
    "waist",
    "hips",

    "garment_bust",
    "garment_waist",
    "garment_hip",
    "garment_length",
    "garment_sleeve",

    "bust_ease",
    "waist_ease",
    "hip_ease",

    "bust_ratio",
    "waist_ratio",
    "hip_ratio",

    "bust_mismatch",
    "waist_mismatch",
    "hip_mismatch",

    "bust_relative_mismatch",
    "waist_relative_mismatch",
    "hip_relative_mismatch",

    "mean_ease",
    "ease_std",

    "known_fit_dimensions",
    "tight_dimension_count",
    "extreme_ease_count",

    "length_height_ratio",
]

# ============================================================
# LOOKUPS
# ============================================================

person_lookup = (
    raw[
        [
            "person_id",
            "height",
            "bust",
            "waist",
            "hips",
        ]
    ]
    .drop_duplicates(
        "person_id"
    )
    .set_index(
        "person_id"
    )
)

garment_lookup = (
    raw[
        [
            "product_id",
            "garment_bust",
            "garment_waist",
            "garment_hip",
            "garment_length",
            "garment_sleeve",
        ]
    ]
    .drop_duplicates(
        "product_id"
    )
    .set_index(
        "product_id"
    )
)

# ============================================================
# FEATURE FUNCTION
# ============================================================

def make_features(
    person_id,
    product_id,
):

    p = person_lookup.loc[
        person_id
    ]

    g = garment_lookup.loc[
        product_id
    ]

    row = {}

    # --------------------------------------------------------
    # Body
    # --------------------------------------------------------

    row["height"] = p["height"]
    row["bust"] = p["bust"]
    row["waist"] = p["waist"]
    row["hips"] = p["hips"]

    # --------------------------------------------------------
    # Garment
    # --------------------------------------------------------

    row["garment_bust"] = g["garment_bust"]
    row["garment_waist"] = g["garment_waist"]
    row["garment_hip"] = g["garment_hip"]
    row["garment_length"] = g["garment_length"]
    row["garment_sleeve"] = g["garment_sleeve"]

    # --------------------------------------------------------
    # Ease
    # --------------------------------------------------------

    row["bust_ease"] = (
        g["garment_bust"]
        -
        p["bust"]
    )

    row["waist_ease"] = (
        g["garment_waist"]
        -
        p["waist"]
    )

    row["hip_ease"] = (
        g["garment_hip"]
        -
        p["hips"]
    )

    # --------------------------------------------------------
    # Ratios
    # --------------------------------------------------------

    row["bust_ratio"] = (
        g["garment_bust"]
        /
        p["bust"]
    )

    row["waist_ratio"] = (
        g["garment_waist"]
        /
        p["waist"]
    )

    row["hip_ratio"] = (
        g["garment_hip"]
        /
        p["hips"]
    )

    # --------------------------------------------------------
    # Mismatch
    # --------------------------------------------------------

    row["bust_mismatch"] = abs(
        g["garment_bust"]
        -
        p["bust"]
    )

    row["waist_mismatch"] = abs(
        g["garment_waist"]
        -
        p["waist"]
    )

    row["hip_mismatch"] = abs(
        g["garment_hip"]
        -
        p["hips"]
    )

    # --------------------------------------------------------
    # Relative mismatch
    # --------------------------------------------------------

    row["bust_relative_mismatch"] = (
        row["bust_mismatch"]
        /
        p["bust"]
    )

    row["waist_relative_mismatch"] = (
        row["waist_mismatch"]
        /
        p["waist"]
    )

    row["hip_relative_mismatch"] = (
        row["hip_mismatch"]
        /
        p["hips"]
    )

    # --------------------------------------------------------
    # Aggregate ease
    # --------------------------------------------------------

    eases = np.array(
        [
            row["bust_ease"],
            row["waist_ease"],
            row["hip_ease"],
        ],
        dtype=float,
    )

    valid = eases[
        ~np.isnan(eases)
    ]

    if len(valid) == 0:

        row["mean_ease"] = np.nan
        row["ease_std"] = np.nan

    else:

        row["mean_ease"] = np.mean(
            valid
        )

        row["ease_std"] = np.std(
            valid
        )

    row["known_fit_dimensions"] = (
        np.sum(
            ~np.isnan(eases)
        )
    )

    row["tight_dimension_count"] = (
        np.sum(
            eases < 0
        )
    )

    row["extreme_ease_count"] = (
        np.sum(
            eases > 50
        )
    )

    # --------------------------------------------------------
    # Length
    # --------------------------------------------------------

    if (
        pd.isna(
            g["garment_length"]
        )
        or
        pd.isna(
            p["height"]
        )
    ):

        row[
            "length_height_ratio"
        ] = np.nan

    else:

        row[
            "length_height_ratio"
        ] = (
            g["garment_length"]
            /
            p["height"]
        )

    return np.array(
        [
            row[f]
            for f in FEATURES
        ],
        dtype=np.float32,
    )

# ============================================================
# RECREATE EXACT PERSON SPLIT
# ============================================================

people = sorted(
    pairwise[
        "person_id"
    ].unique()
)

train_people, temp_people = (
    train_test_split(
        people,
        test_size=16,
        random_state=SEED,
    )
)

val_people, test_people = (
    train_test_split(
        temp_people,
        test_size=8,
        random_state=SEED,
    )
)

test_people = set(
    test_people
)

print("\n" + "=" * 70)
print("TEST PEOPLE")
print("=" * 70)

print(
    sorted(test_people)
)

# ============================================================
# TEST DATA
# ============================================================

test_df = pairwise[
    pairwise[
        "person_id"
    ].isin(
        test_people
    )
].copy()

print(
    "\nTest rows:",
    len(test_df)
)

# ============================================================
# BUILD TEST FEATURES
# ============================================================

positive_features = []
negative_features = []

for _, row in test_df.iterrows():

    positive_features.append(
        make_features(
            row["person_id"],
            row["positive_product_id"],
        )
    )

    negative_features.append(
        make_features(
            row["person_id"],
            row["negative_product_id"],
        )
    )

positive_features = np.asarray(
    positive_features,
    dtype=np.float32,
)

negative_features = np.asarray(
    negative_features,
    dtype=np.float32,
)

# ============================================================
# IMPUTE
# ============================================================

imputer = SimpleImputer(
    strategy="median"
)

# ------------------------------------------------------------
# IMPORTANT:
# Refit only using training people.
# ------------------------------------------------------------

train_df = pairwise[
    pairwise[
        "person_id"
    ].isin(
        set(train_people)
    )
].copy()

train_pos = []
train_neg = []

for _, row in train_df.iterrows():

    train_pos.append(
        make_features(
            row["person_id"],
            row["positive_product_id"],
        )
    )

    train_neg.append(
        make_features(
            row["person_id"],
            row["negative_product_id"],
        )
    )

train_reference = np.vstack(
    [
        np.asarray(
            train_pos,
            dtype=np.float32,
        ),
        np.asarray(
            train_neg,
            dtype=np.float32,
        ),
    ]
)

imputer.fit(
    train_reference
)

positive_features = imputer.transform(
    positive_features
)

negative_features = imputer.transform(
    negative_features
)

# ============================================================
# SCALE
# ============================================================

scaler = StandardScaler()

scaler.fit(
    train_reference
)

positive_features = scaler.transform(
    positive_features
)

negative_features = scaler.transform(
    negative_features
)

# ============================================================
# MODEL
# ============================================================

class CompatibilityScorer(
    nn.Module
):

    def __init__(
        self,
        input_dim,
    ):

        super().__init__()

        self.network = nn.Sequential(

            nn.Linear(
                input_dim,
                64,
            ),

            nn.ReLU(),

            nn.Linear(
                64,
                32,
            ),

            nn.ReLU(),

            nn.Linear(
                32,
                1,
            ),
        )

    def forward(
        self,
        x,
    ):

        return self.network(
            x
        ).squeeze(-1)

# ============================================================
# LOAD MODEL
# ============================================================

checkpoint = torch.load(
    MODEL_PATH,
    map_location=device,
    weights_only=False,
)

model = CompatibilityScorer(
    len(FEATURES)
).to(device)

model.load_state_dict(
    checkpoint[
        "model_state_dict"
    ]
)

model.eval()

print(
    "\nModel loaded."
)

# ============================================================
# SCORE
# ============================================================

pos_tensor = torch.tensor(
    positive_features,
    dtype=torch.float32,
    device=device,
)

neg_tensor = torch.tensor(
    negative_features,
    dtype=torch.float32,
    device=device,
)

with torch.no_grad():

    pos_scores = (
        model(
            pos_tensor
        )
        .detach()
        .cpu()
        .numpy()
    )

    neg_scores = (
        model(
            neg_tensor
        )
        .detach()
        .cpu()
        .numpy()
    )

test_df[
    "positive_score"
] = pos_scores

test_df[
    "negative_score"
] = neg_scores

test_df[
    "score_gap"
] = (
    test_df[
        "positive_score"
    ]
    -
    test_df[
        "negative_score"
    ]
)

test_df[
    "correct"
] = (
    test_df[
        "positive_score"
    ]
    >
    test_df[
        "negative_score"
    ]
)

# ============================================================
# OVERALL RESULTS
# ============================================================

print("\n" + "=" * 70)
print("OVERALL TEST DIAGNOSTIC")
print("=" * 70)

print(
    "Positive mean:",
    f"{pos_scores.mean():.6f}"
)

print(
    "Negative mean:",
    f"{neg_scores.mean():.6f}"
)

print(
    "Mean score gap:",
    f"{(pos_scores - neg_scores).mean():.6f}"
)

print(
    "Ranking accuracy:",
    f"{(pos_scores > neg_scores).mean():.4f}"
)

# ============================================================
# BY TYPE
# ============================================================

print("\n" + "=" * 70)
print("PER NEGATIVE TYPE")
print("=" * 70)

type_stats = (
    test_df
    .groupby(
        "negative_type"
    )
    .agg(
        pairs=(
            "correct",
            "size",
        ),

        ranking_accuracy=(
            "correct",
            "mean",
        ),

        positive_score=(
            "positive_score",
            "mean",
        ),

        negative_score=(
            "negative_score",
            "mean",
        ),

        mean_gap=(
            "score_gap",
            "mean",
        ),

        median_gap=(
            "score_gap",
            "median",
        ),
    )
)

print(
    type_stats
)

# ============================================================
# RANDOM FAILURES
# ============================================================

random_failures = test_df[
    (
        test_df[
            "negative_type"
        ]
        ==
        "random"
    )
    &
    (
        test_df[
            "negative_score"
        ]
        >=
        test_df[
            "positive_score"
        ]
    )
].copy()

print("\n" + "=" * 70)
print("RANDOM NEGATIVES BEATING POSITIVES")
print("=" * 70)

print(
    "Failures:",
    len(random_failures)
)

print(
    "Total random:",
    (
        test_df[
            "negative_type"
        ]
        ==
        "random"
    ).sum()
)

print(
    "Failure rate:",
    f"{len(random_failures) / max(1, (test_df['negative_type'] == 'random').sum()):.4f}"
)

# ============================================================
# WORST RANDOM FAILURES
# ============================================================

print("\n" + "=" * 70)
print("WORST RANDOM FAILURES")
print("=" * 70)

worst_random = (
    random_failures
    .sort_values(
        "score_gap"
    )
)

display(
    worst_random[
        [
            "person_id",
            "positive_product_id",
            "negative_product_id",
            "negative_distance",
            "positive_score",
            "negative_score",
            "score_gap",
        ]
    ]
    .head(30)
)

# ============================================================
# BEST RANDOM NEGATIVES
# ============================================================

print("\n" + "=" * 70)
print("RANDOM NEGATIVES CORRECTLY REJECTED")
print("=" * 70)

random_correct = test_df[
    (
        test_df[
            "negative_type"
        ]
        ==
        "random"
    )
    &
    (
        test_df[
            "positive_score"
        ]
        >
        test_df[
            "negative_score"
        ]
    )
].copy()

display(
    random_correct[
        [
            "person_id",
            "positive_product_id",
            "negative_product_id",
            "negative_distance",
            "positive_score",
            "negative_score",
            "score_gap",
        ]
    ]
    .sort_values(
        "score_gap"
    )
    .head(20)
)

# ============================================================
# PHYSICAL FEATURE COMPARISON
# ============================================================

diagnostic_features = [
    "bust_ease",
    "waist_ease",
    "hip_ease",
    "mean_ease",

    "bust_ratio",
    "waist_ratio",
    "hip_ratio",

    "bust_mismatch",
    "waist_mismatch",
    "hip_mismatch",

    "bust_relative_mismatch",
    "waist_relative_mismatch",
    "hip_relative_mismatch",

    "length_height_ratio",
    "garment_sleeve",

    "known_fit_dimensions",
    "tight_dimension_count",
    "extreme_ease_count",
]

# ============================================================
# CREATE FEATURE DATAFRAME
# ============================================================

feature_rows = []

for _, row in test_df.iterrows():

    positive_raw = make_features(
        row["person_id"],
        row["positive_product_id"],
    )

    negative_raw = make_features(
        row["person_id"],
        row["negative_product_id"],
    )

    item = {
        "person_id":
            row["person_id"],

        "negative_type":
            row["negative_type"],

        "score_gap":
            row["score_gap"],

        "correct":
            row["correct"],
    }

    for feature_name in diagnostic_features:

        idx = FEATURES.index(
            feature_name
        )

        item[
            "pos_" + feature_name
        ] = positive_raw[idx]

        item[
            "neg_" + feature_name
        ] = negative_raw[idx]

        item[
            "diff_" + feature_name
        ] = (
            positive_raw[idx]
            -
            negative_raw[idx]
        )

    feature_rows.append(
        item
    )

feature_df = pd.DataFrame(
    feature_rows
)

# ============================================================
# RANDOM FAILURE FEATURE ANALYSIS
# ============================================================

random_feature_df = feature_df[
    feature_df[
        "negative_type"
    ]
    ==
    "random"
].copy()

random_failure_features = (
    random_feature_df[
        random_feature_df[
            "correct"
        ]
        ==
        False
    ]
)

random_success_features = (
    random_feature_df[
        random_feature_df[
            "correct"
        ]
        ==
        True
    ]
)

print("\n" + "=" * 70)
print("RANDOM FAILURE FEATURE COMPARISON")
print("=" * 70)

comparison_rows = []

for feature in diagnostic_features:

    failure_mean = (
        random_failure_features[
            "diff_" + feature
        ]
        .mean()
    )

    success_mean = (
        random_success_features[
            "diff_" + feature
        ]
        .mean()
    )

    comparison_rows.append(
        {
            "feature":
                feature,

            "failure_mean_diff":
                failure_mean,

            "success_mean_diff":
                success_mean,

            "difference":
                failure_mean
                -
                success_mean,
        }
    )

comparison_df = (
    pd.DataFrame(
        comparison_rows
    )
    .sort_values(
        "difference",
        ascending=False,
    )
)

display(
    comparison_df
)

# ============================================================
# PER PERSON
# ============================================================

print("\n" + "=" * 70)
print("PER-PERSON RANDOM PERFORMANCE")
print("=" * 70)

random_person = (
    test_df[
        test_df[
            "negative_type"
        ]
        ==
        "random"
    ]
    .groupby(
        "person_id"
    )
    .agg(
        pairs=(
            "correct",
            "size",
        ),

        accuracy=(
            "correct",
            "mean",
        ),

        mean_gap=(
            "score_gap",
            "mean",
        ),

        median_gap=(
            "score_gap",
            "median",
        ),
    )
    .sort_values(
        "accuracy"
    )
)

display(
    random_person
)

# ============================================================
# SCORE DISTRIBUTIONS
# ============================================================

print("\n" + "=" * 70)
print("SCORE DISTRIBUTION")
print("=" * 70)

for negative_type in [
    "hard",
    "semi-hard",
    "random",
]:

    subset = test_df[
        test_df[
            "negative_type"
        ]
        ==
        negative_type
    ]

    print(
        f"\n{negative_type.upper()}"
    )

    print(
        "Positive score:"
    )

    print(
        subset[
            "positive_score"
        ]
        .describe()
    )

    print(
        "Negative score:"
    )

    print(
        subset[
            "negative_score"
        ]
        .describe()
    )

# ============================================================
# FINAL INTERPRETATION FLAGS
# ============================================================

random_accuracy = (
    test_df[
        test_df[
            "negative_type"
        ]
        ==
        "random"
    ][
        "correct"
    ]
    .mean()
)

hard_accuracy = (
    test_df[
        test_df[
            "negative_type"
        ]
        ==
        "hard"
    ][
        "correct"
    ]
    .mean()
)

semi_accuracy = (
    test_df[
        test_df[
            "negative_type"
        ]
        ==
        "semi-hard"
    ][
        "correct"
    ]
    .mean()
)

print("\n" + "=" * 70)
print("DIAGNOSTIC SUMMARY")
print("=" * 70)

print(
    f"Hard accuracy      : {hard_accuracy:.4f}"
)

print(
    f"Semi-hard accuracy : {semi_accuracy:.4f}"
)

print(
    f"Random accuracy    : {random_accuracy:.4f}"
)

print(
    f"Random failure rate: {1 - random_accuracy:.4f}"
)

print("\nInterpretation:")

if random_accuracy < 0.50:

    print(
        "CRITICAL: model systematically "
        "prefers many unseen random negatives."
    )

elif random_accuracy < 0.65:

    print(
        "WEAK: random-negative "
        "generalization is insufficient."
    )

else:

    print(
        "ACCEPTABLE: random-negative "
        "generalization is reasonable."
    )

if (
    hard_accuracy > 0.80
    and
    random_accuracy < 0.50
):

    print(
        "\nIMPORTANT:"
    )

    print(
        "The model is strongly sensitive "
        "to local physical similarity but "
        "does not learn a robust global "
        "compatibility function."
    )

print(
    "\nDiagnostic complete."
)
```

```python
# ============================================================
# ZYRA V2 — NEGATIVE LABEL VALIDITY AUDIT
# ============================================================
#
# Goal:
# Determine whether "negative" garments are actually
# physically less compatible than positive garments.
#
# NO MODEL TRAINING.
# ============================================================

import pandas as pd
import numpy as np
from pathlib import Path

from scipy.stats import mannwhitneyu
from sklearn.metrics import roc_auc_score

# ============================================================
# PATHS
# ============================================================

BASE = Path(
    "/Users/saketh/Desktop/Projects/weavly/core-model"
)

PAIRWISE_PATH = (
    BASE /
    "data/v2/v2_negative_sampling_1to1to1.csv"
)

RAW_PATH = (
    BASE /
    "data/v2/v2_compatibility_raw.csv"
)

# ============================================================
# LOAD
# ============================================================

pairwise = pd.read_csv(
    PAIRWISE_PATH
)

raw = pd.read_csv(
    RAW_PATH
)

print("=" * 70)
print("ZYRA V2 — NEGATIVE LABEL VALIDITY AUDIT")
print("=" * 70)

print(
    "Pairwise rows:",
    len(pairwise)
)

print(
    "People:",
    pairwise["person_id"].nunique()
)

# ============================================================
# LOOKUPS
# ============================================================

person_lookup = (
    raw[
        [
            "person_id",
            "height",
            "bust",
            "waist",
            "hips",
        ]
    ]
    .drop_duplicates(
        "person_id"
    )
    .set_index(
        "person_id"
    )
)

garment_lookup = (
    raw[
        [
            "product_id",
            "garment_bust",
            "garment_waist",
            "garment_hip",
            "garment_length",
            "garment_sleeve",
        ]
    ]
    .drop_duplicates(
        "product_id"
    )
    .set_index(
        "product_id"
    )
)

# ============================================================
# PHYSICAL COMPATIBILITY FUNCTION
# ============================================================

def compatibility_features(
    person_id,
    product_id,
):

    p = person_lookup.loc[
        person_id
    ]

    g = garment_lookup.loc[
        product_id
    ]

    # --------------------------------------------------------
    # Raw ease
    # --------------------------------------------------------

    bust_ease = (
        g["garment_bust"]
        - p["bust"]
    )

    waist_ease = (
        g["garment_waist"]
        - p["waist"]
    )

    hip_ease = (
        g["garment_hip"]
        - p["hips"]
    )

    # --------------------------------------------------------
    # Ratios
    # --------------------------------------------------------

    bust_ratio = (
        g["garment_bust"]
        / p["bust"]
    )

    waist_ratio = (
        g["garment_waist"]
        / p["waist"]
    )

    hip_ratio = (
        g["garment_hip"]
        / p["hips"]
    )

    # --------------------------------------------------------
    # Absolute mismatch
    # --------------------------------------------------------

    bust_mismatch = abs(
        g["garment_bust"]
        - p["bust"]
    )

    waist_mismatch = abs(
        g["garment_waist"]
        - p["waist"]
    )

    hip_mismatch = abs(
        g["garment_hip"]
        - p["hips"]
    )

    # --------------------------------------------------------
    # Relative mismatch
    # --------------------------------------------------------

    bust_relative = (
        bust_mismatch
        /
        p["bust"]
    )

    waist_relative = (
        waist_mismatch
        /
        p["waist"]
    )

    hip_relative = (
        hip_mismatch
        /
        p["hips"]
    )

    # --------------------------------------------------------
    # Aggregate mismatch
    # --------------------------------------------------------

    mismatches = np.array(
        [
            bust_relative,
            waist_relative,
            hip_relative,
        ],
        dtype=float,
    )

    valid_mismatches = (
        mismatches[
            ~np.isnan(
                mismatches
            )
        ]
    )

    if len(valid_mismatches) > 0:

        mean_relative_mismatch = (
            np.mean(
                valid_mismatches
            )
        )

        mismatch_std = (
            np.std(
                valid_mismatches
            )
        )

    else:

        mean_relative_mismatch = np.nan
        mismatch_std = np.nan

    # --------------------------------------------------------
    # Ease aggregate
    # --------------------------------------------------------

    eases = np.array(
        [
            bust_ease,
            waist_ease,
            hip_ease,
        ],
        dtype=float,
    )

    valid_eases = (
        eases[
            ~np.isnan(
                eases
            )
        ]
    )

    if len(valid_eases) > 0:

        mean_ease = np.mean(
            valid_eases
        )

        ease_std = np.std(
            valid_eases
        )

    else:

        mean_ease = np.nan
        ease_std = np.nan

    # --------------------------------------------------------
    # Tight dimensions
    # --------------------------------------------------------

    tight_count = np.sum(
        eases < 0
    )

    # --------------------------------------------------------
    # Extreme ease
    # --------------------------------------------------------

    extreme_count = np.sum(
        eases > 50
    )

    # --------------------------------------------------------
    # Length ratio
    # --------------------------------------------------------

    if (
        pd.isna(
            g["garment_length"]
        )
        or
        pd.isna(
            p["height"]
        )
    ):

        length_ratio = np.nan

    else:

        length_ratio = (
            g["garment_length"]
            /
            p["height"]
        )

    return {
        "bust_ease":
            bust_ease,

        "waist_ease":
            waist_ease,

        "hip_ease":
            hip_ease,

        "bust_ratio":
            bust_ratio,

        "waist_ratio":
            waist_ratio,

        "hip_ratio":
            hip_ratio,

        "bust_mismatch":
            bust_mismatch,

        "waist_mismatch":
            waist_mismatch,

        "hip_mismatch":
            hip_mismatch,

        "bust_relative_mismatch":
            bust_relative,

        "waist_relative_mismatch":
            waist_relative,

        "hip_relative_mismatch":
            hip_relative,

        "mean_relative_mismatch":
            mean_relative_mismatch,

        "mismatch_std":
            mismatch_std,

        "mean_ease":
            mean_ease,

        "ease_std":
            ease_std,

        "tight_dimension_count":
            tight_count,

        "extreme_ease_count":
            extreme_count,

        "length_height_ratio":
            length_ratio,
    }


# ============================================================
# BUILD AUDIT DATASET
# ============================================================

rows = []

for _, row in pairwise.iterrows():

    person_id = row[
        "person_id"
    ]

    positive_product = row[
        "positive_product_id"
    ]

    negative_product = row[
        "negative_product_id"
    ]

    positive = compatibility_features(
        person_id,
        positive_product,
    )

    negative = compatibility_features(
        person_id,
        negative_product,
    )

    record = {
        "person_id":
            person_id,

        "positive_product_id":
            positive_product,

        "negative_product_id":
            negative_product,

        "negative_type":
            row["negative_type"],

        "negative_distance":
            row["negative_distance"],
    }

    for feature in positive:

        record[
            "positive_" + feature
        ] = positive[
            feature
        ]

        record[
            "negative_" + feature
        ] = negative[
            feature
        ]

        record[
            "difference_" + feature
        ] = (
            positive[
                feature
            ]
            -
            negative[
                feature
            ]
        )

    rows.append(
        record
    )

audit = pd.DataFrame(
    rows
)

print(
    "\nAudit rows:",
    len(audit)
)

# ============================================================
# DEFINE "MORE PHYSICALLY COMPATIBLE"
# ============================================================
#
# Smaller relative mismatch = better physical fit.
#
# This is an analytical proxy only.
# It is NOT assumed to be ground truth.
# ============================================================

audit[
    "positive_better_mismatch"
] = (
    audit[
        "positive_mean_relative_mismatch"
    ]
    <
    audit[
        "negative_mean_relative_mismatch"
    ]
)

# ============================================================
# OVERALL NEGATIVE VALIDITY
# ============================================================

print("\n" + "=" * 70)
print("OVERALL PHYSICAL COMPATIBILITY")
print("=" * 70)

overall = (
    audit[
        "positive_better_mismatch"
    ]
    .mean()
)

print(
    "Positive has lower physical mismatch:",
    f"{overall:.4f}"
)

print(
    "Negative has lower physical mismatch:",
    f"{1 - overall:.4f}"
)

# ============================================================
# BY NEGATIVE TYPE
# ============================================================

print("\n" + "=" * 70)
print("PHYSICAL COMPATIBILITY BY NEGATIVE TYPE")
print("=" * 70)

type_validity = (
    audit
    .groupby(
        "negative_type"
    )
    .agg(
        pairs=(
            "positive_better_mismatch",
            "size",
        ),

        positive_more_compatible=(
            "positive_better_mismatch",
            "mean",
        ),

        positive_mismatch=(
            "positive_mean_relative_mismatch",
            "mean",
        ),

        negative_mismatch=(
            "negative_mean_relative_mismatch",
            "mean",
        ),
    )
)

print(
    type_validity
)

# ============================================================
# POSITIVE VS NEGATIVE MISMATCH
# ============================================================

print("\n" + "=" * 70)
print("MISMATCH DISTRIBUTION")
print("=" * 70)

for negative_type in [
    "hard",
    "semi-hard",
    "random",
]:

    subset = audit[
        audit[
            "negative_type"
        ]
        ==
        negative_type
    ]

    print(
        f"\n{negative_type.upper()}"
    )

    print(
        "Positive mismatch:"
    )

    print(
        subset[
            "positive_mean_relative_mismatch"
        ]
        .describe()
    )

    print(
        "Negative mismatch:"
    )

    print(
        subset[
            "negative_mean_relative_mismatch"
        ]
        .describe()
    )

# ============================================================
# RANDOM NEGATIVES — CRITICAL ANALYSIS
# ============================================================

random = audit[
    audit[
        "negative_type"
    ]
    ==
    "random"
].copy()

print("\n" + "=" * 70)
print("RANDOM NEGATIVE VALIDITY")
print("=" * 70)

random_positive_better = (
    random[
        "positive_better_mismatch"
    ]
    .mean()
)

print(
    "Positive physically better:",
    f"{random_positive_better:.4f}"
)

print(
    "Random negative physically better:",
    f"{1 - random_positive_better:.4f}"
)

# ============================================================
# CASES WHERE RANDOM NEGATIVE IS PHYSICALLY BETTER
# ============================================================

random_suspicious = random[
    random[
        "negative_mean_relative_mismatch"
    ]
    <
    random[
        "positive_mean_relative_mismatch"
    ]
].copy()

random_suspicious[
    "mismatch_advantage"
] = (
    random_suspicious[
        "positive_mean_relative_mismatch"
    ]
    -
    random_suspicious[
        "negative_mean_relative_mismatch"
    ]
)

print("\n" + "=" * 70)
print("RANDOM NEGATIVES MORE PHYSICALLY COMPATIBLE")
print("=" * 70)

print(
    "Count:",
    len(random_suspicious)
)

print(
    "Percentage:",
    f"{len(random_suspicious) / max(1, len(random)):.4f}"
)

display(
    random_suspicious[
        [
            "person_id",
            "positive_product_id",
            "negative_product_id",
            "positive_mean_relative_mismatch",
            "negative_mean_relative_mismatch",
            "mismatch_advantage",
            "positive_mean_ease",
            "negative_mean_ease",
        ]
    ]
    .sort_values(
        "mismatch_advantage",
        ascending=False,
    )
    .head(50)
)

# ============================================================
# RANDOM NEGATIVES THAT LOOK EXTREMELY COMPATIBLE
# ============================================================

print("\n" + "=" * 70)
print("SUSPICIOUS RANDOM NEGATIVES")
print("=" * 70)

threshold = (
    random[
        "positive_mean_relative_mismatch"
    ]
    .median()
)

very_suspicious = random[
    random[
        "negative_mean_relative_mismatch"
    ]
    <
    threshold
].copy()

print(
    "Random negatives with mismatch better "
    "than positive median:",
    len(very_suspicious)
)

# ============================================================
# MANN-WHITNEY TEST
# ============================================================

print("\n" + "=" * 70)
print("STATISTICAL COMPARISON")
print("=" * 70)

for negative_type in [
    "hard",
    "semi-hard",
    "random",
]:

    subset = audit[
        audit[
            "negative_type"
        ]
        ==
        negative_type
    ]

    positive_values = (
        subset[
            "positive_mean_relative_mismatch"
        ]
        .dropna()
    )

    negative_values = (
        subset[
            "negative_mean_relative_mismatch"
        ]
        .dropna()
    )

    if (
        len(positive_values) > 0
        and
        len(negative_values) > 0
    ):

        stat, p = mannwhitneyu(
            positive_values,
            negative_values,
            alternative="less",
        )

        print(
            f"\n{negative_type}"
        )

        print(
            "Positive median:",
            positive_values.median()
        )

        print(
            "Negative median:",
            negative_values.median()
        )

        print(
            "p-value:",
            f"{p:.8f}"
        )

# ============================================================
# CAN SIMPLE PHYSICAL MISMATCH DISTINGUISH LABELS?
# ============================================================

print("\n" + "=" * 70)
print("MISMATCH AUC")
print("=" * 70)

# Build balanced positive/negative examples
classification_rows = []

for _, row in audit.iterrows():

    classification_rows.append(
        {
            "mismatch":
                row[
                    "positive_mean_relative_mismatch"
                ],

            "label":
                1,
        }
    )

    classification_rows.append(
        {
            "mismatch":
                row[
                    "negative_mean_relative_mismatch"
                ],

            "label":
                0,
        }
    )

classification = pd.DataFrame(
    classification_rows
).dropna()

# Smaller mismatch should indicate positive.
# Therefore invert it.
auc = roc_auc_score(
    classification["label"],
    -classification["mismatch"],
)

print(
    "Physical mismatch ROC-AUC:",
    f"{auc:.4f}"
)

# ============================================================
# PER-PERSON VALIDITY
# ============================================================

print("\n" + "=" * 70)
print("PER-PERSON RANDOM NEGATIVE VALIDITY")
print("=" * 70)

person_validity = (
    random
    .groupby(
        "person_id"
    )
    .agg(
        pairs=(
            "positive_better_mismatch",
            "size",
        ),

        positive_better_pct=(
            "positive_better_mismatch",
            "mean",
        ),

        positive_mismatch=(
            "positive_mean_relative_mismatch",
            "mean",
        ),

        negative_mismatch=(
            "negative_mean_relative_mismatch",
            "mean",
        ),
    )
    .sort_values(
        "positive_better_pct"
    )
)

display(
    person_validity
)

# ============================================================
# POSITIVE/NEGATIVE EASE COMPARISON
# ============================================================

print("\n" + "=" * 70)
print("EASE COMPARISON")
print("=" * 70)

ease_features = [
    "bust_ease",
    "waist_ease",
    "hip_ease",
    "mean_ease",
]

ease_rows = []

for feature in ease_features:

    pos = (
        random[
            "positive_" + feature
        ]
        .dropna()
    )

    neg = (
        random[
            "negative_" + feature
        ]
        .dropna()
    )

    ease_rows.append(
        {
            "feature":
                feature,

            "positive_mean":
                pos.mean(),

            "negative_mean":
                neg.mean(),

            "difference":
                pos.mean()
                -
                neg.mean(),
        }
    )

ease_comparison = pd.DataFrame(
    ease_rows
)

display(
    ease_comparison
)

# ============================================================
# FINAL DIAGNOSIS
# ============================================================

print("\n" + "=" * 70)
print("FINAL NEGATIVE-LABEL DIAGNOSIS")
print("=" * 70)

if random_positive_better >= 0.70:

    print(
        "RANDOM NEGATIVES LOOK GENERALLY VALID."
    )

    print(
        "Most random negatives are physically "
        "less compatible than positives."
    )

elif random_positive_better >= 0.55:

    print(
        "RANDOM NEGATIVES SHOW MODERATE VALIDITY."
    )

    print(
        "There is meaningful label noise."
    )

else:

    print(
        "RANDOM NEGATIVES ARE HIGHLY QUESTIONABLE."
    )

    print(
        "A large fraction are physically "
        "more compatible than the recorded positives."
    )

print(
    "\nRandom positive-better rate:",
    f"{random_positive_better:.4f}"
)

print(
    "Physical mismatch AUC:",
    f"{auc:.4f}"
)

print(
    "\nAudit complete."
)
```

```python
# ============================================================
# ZYRA V2 — STEP 1
# COMPLETE PERSON × GARMENT CANDIDATE MATRIX
# ============================================================

import pandas as pd
import numpy as np
from pathlib import Path

# ============================================================
# PATHS
# ============================================================

BASE = Path(
    "/Users/saketh/Desktop/Projects/weavly/core-model"
)

RAW_PATH = (
    BASE /
    "data/v2/v2_compatibility_raw.csv"
)

OUTPUT_PATH = (
    BASE /
    "data/v2/v2_person_garment_candidates.csv"
)

# ============================================================
# LOAD
# ============================================================

df = pd.read_csv(RAW_PATH)

print("=" * 70)
print("ZYRA V2 — COMPLETE CANDIDATE MATRIX")
print("=" * 70)

print("Raw rows:", len(df))

# ============================================================
# PEOPLE
# ============================================================

people = (
    df[
        [
            "person_id",
            "height",
            "bust",
            "waist",
            "hips",
        ]
    ]
    .drop_duplicates(
        "person_id"
    )
    .reset_index(drop=True)
)

# ============================================================
# GARMENTS
# ============================================================

garments = (
    df[
        [
            "product_id",
            "garment_bust",
            "garment_waist",
            "garment_hip",
            "garment_length",
            "garment_sleeve",
        ]
    ]
    .drop_duplicates(
        "product_id"
    )
    .reset_index(drop=True)
)

print(
    "\nPeople:",
    len(people)
)

print(
    "Garments:",
    len(garments)
)

# ============================================================
# KNOWN POSITIVE INTERACTIONS
# ============================================================

positive_pairs = (
    df[
        df["label"] == 1
    ][
        [
            "person_id",
            "product_id",
        ]
    ]
    .drop_duplicates()
    .copy()
)

positive_keys = set(
    zip(
        positive_pairs[
            "person_id"
        ],
        positive_pairs[
            "product_id"
        ],
    )
)

print(
    "\nKnown positive interactions:",
    len(positive_keys)
)

# ============================================================
# COMPLETE CARTESIAN PRODUCT
# ============================================================

people["_tmp"] = 1
garments["_tmp"] = 1

candidates = (
    people
    .merge(
        garments,
        on="_tmp",
    )
    .drop(
        columns="_tmp"
    )
)

# ============================================================
# LABEL STATUS
# ============================================================

candidates[
    "label_status"
] = [
    (
        "positive"
        if (
            person_id,
            product_id,
        )
        in positive_keys
        else "unknown"
    )

    for person_id, product_id
    in zip(
        candidates[
            "person_id"
        ],
        candidates[
            "product_id"
        ],
    )
]

# ============================================================
# BASIC COUNTS
# ============================================================

print("\n" + "=" * 70)
print("CANDIDATE MATRIX")
print("=" * 70)

print(
    "Total candidates:",
    len(candidates)
)

print(
    "Expected:",
    len(people) * len(garments)
)

print(
    "\nLabel status:"
)

print(
    candidates[
        "label_status"
    ].value_counts()
)

# ============================================================
# VERIFY POSITIVES
# ============================================================

matrix_positive_count = (
    candidates[
        "label_status"
    ]
    .eq("positive")
    .sum()
)

print(
    "\nMatrix positives:",
    matrix_positive_count
)

assert (
    matrix_positive_count
    ==
    len(positive_keys)
)

# ============================================================
# VERIFY ONE ROW PER PERSON-GARMENT
# ============================================================

duplicate_pairs = (
    candidates[
        [
            "person_id",
            "product_id",
        ]
    ]
    .duplicated()
    .sum()
)

print(
    "Duplicate person-garment pairs:",
    duplicate_pairs
)

assert duplicate_pairs == 0

# ============================================================
# PHYSICAL FEATURES
# ============================================================

print("\n" + "=" * 70)
print("PHYSICAL COMPATIBILITY FEATURES")
print("=" * 70)


def safe_ratio(
    numerator,
    denominator,
):

    if (
        pd.isna(numerator)
        or
        pd.isna(denominator)
        or
        denominator == 0
    ):
        return np.nan

    return (
        numerator
        /
        denominator
    )


# ------------------------------------------------------------
# Ease
# ------------------------------------------------------------

candidates[
    "bust_ease"
] = (
    candidates[
        "garment_bust"
    ]
    -
    candidates[
        "bust"
    ]
)

candidates[
    "waist_ease"
] = (
    candidates[
        "garment_waist"
    ]
    -
    candidates[
        "waist"
    ]
)

candidates[
    "hip_ease"
] = (
    candidates[
        "garment_hip"
    ]
    -
    candidates[
        "hips"
    ]
)

# ------------------------------------------------------------
# Ratios
# ------------------------------------------------------------

candidates[
    "bust_ratio"
] = (
    candidates[
        "garment_bust"
    ]
    /
    candidates[
        "bust"
    ]
)

candidates[
    "waist_ratio"
] = (
    candidates[
        "garment_waist"
    ]
    /
    candidates[
        "waist"
    ]
)

candidates[
    "hip_ratio"
] = (
    candidates[
        "garment_hip"
    ]
    /
    candidates[
        "hips"
    ]
)

# ------------------------------------------------------------
# Absolute mismatch
# ------------------------------------------------------------

candidates[
    "bust_mismatch"
] = abs(
    candidates[
        "garment_bust"
    ]
    -
    candidates[
        "bust"
    ]
)

candidates[
    "waist_mismatch"
] = abs(
    candidates[
        "garment_waist"
    ]
    -
    candidates[
        "waist"
    ]
)

candidates[
    "hip_mismatch"
] = abs(
    candidates[
        "garment_hip"
    ]
    -
    candidates[
        "hips"
    ]
)

# ------------------------------------------------------------
# Relative mismatch
# ------------------------------------------------------------

candidates[
    "bust_relative_mismatch"
] = (
    candidates[
        "bust_mismatch"
    ]
    /
    candidates[
        "bust"
    ]
)

candidates[
    "waist_relative_mismatch"
] = (
    candidates[
        "waist_mismatch"
    ]
    /
    candidates[
        "waist"
    ]
)

candidates[
    "hip_relative_mismatch"
] = (
    candidates[
        "hip_mismatch"
    ]
    /
    candidates[
        "hips"
    ]
)

# ------------------------------------------------------------
# Mean ease
# ------------------------------------------------------------

ease_columns = [
    "bust_ease",
    "waist_ease",
    "hip_ease",
]

candidates[
    "mean_ease"
] = candidates[
    ease_columns
].mean(
    axis=1
)

candidates[
    "ease_std"
] = candidates[
    ease_columns
].std(
    axis=1
)

# ------------------------------------------------------------
# Known dimensions
# ------------------------------------------------------------

candidates[
    "known_fit_dimensions"
] = (
    candidates[
        ease_columns
    ]
    .notna()
    .sum(axis=1)
)

# ------------------------------------------------------------
# Tight dimensions
# ------------------------------------------------------------

candidates[
    "tight_dimension_count"
] = (
    candidates[
        ease_columns
    ]
    .lt(0)
    .sum(axis=1)
)

# ------------------------------------------------------------
# Extreme ease
# ------------------------------------------------------------

candidates[
    "extreme_ease_count"
] = (
    candidates[
        ease_columns
    ]
    .gt(50)
    .sum(axis=1)
)

# ------------------------------------------------------------
# Length / height
# ------------------------------------------------------------

candidates[
    "length_height_ratio"
] = (
    candidates[
        "garment_length"
    ]
    /
    candidates[
        "height"
    ]
)

# ============================================================
# PHYSICAL COMPATIBILITY PROXY
# ============================================================
#
# IMPORTANT:
# This is NOT a training label.
#
# It is only used later to identify confidently incompatible
# garments.
# ============================================================

relative_mismatch_columns = [
    "bust_relative_mismatch",
    "waist_relative_mismatch",
    "hip_relative_mismatch",
]

candidates[
    "physical_mismatch"
] = candidates[
    relative_mismatch_columns
].mean(
    axis=1
)

# ============================================================
# MISSINGNESS
# ============================================================

measurement_columns = [
    "garment_bust",
    "garment_waist",
    "garment_hip",
    "garment_length",
    "garment_sleeve",
]

candidates[
    "missing_measurement_count"
] = (
    candidates[
        measurement_columns
    ]
    .isna()
    .sum(axis=1)
)

candidates[
    "measurement_complete"
] = (
    candidates[
        "missing_measurement_count"
    ]
    == 0
)

# ============================================================
# CHECK POSITIVE / UNKNOWN
# ============================================================

print("\n" + "=" * 70)
print("LABEL SUMMARY")
print("=" * 70)

print(
    candidates[
        "label_status"
    ].value_counts()
)

print(
    "\nPositive percentage:",
    f"{matrix_positive_count / len(candidates) * 100:.3f}%"
)

print(
    "Unknown percentage:",
    f"{(len(candidates) - matrix_positive_count) / len(candidates) * 100:.3f}%"
)

# ============================================================
# PER PERSON
# ============================================================

print("\n" + "=" * 70)
print("PER-PERSON SUMMARY")
print("=" * 70)

person_summary = (
    candidates
    .groupby(
        "person_id"
    )
    .agg(
        total_garments=(
            "product_id",
            "count",
        ),

        positives=(
            "label_status",
            lambda x:
                (x == "positive").sum(),
        ),

        unknown=(
            "label_status",
            lambda x:
                (x == "unknown").sum(),
        ),
    )
)

person_summary[
    "positive_rate"
] = (
    person_summary[
        "positives"
    ]
    /
    person_summary[
        "total_garments"
    ]
)

print(
    person_summary
)

# ============================================================
# GARMENT SUMMARY
# ============================================================

print("\n" + "=" * 70)
print("PER-GARMENT SUMMARY")
print("=" * 70)

garment_summary = (
    candidates
    .groupby(
        "product_id"
    )
    .agg(
        people=(
            "person_id",
            "count",
        ),

        positive_count=(
            "label_status",
            lambda x:
                (x == "positive").sum(),
        ),
    )
)

garment_summary[
    "positive_rate"
] = (
    garment_summary[
        "positive_count"
    ]
    /
    garment_summary[
        "people"
    ]
)

print(
    garment_summary[
        "positive_count"
    ]
    .describe()
)

# ============================================================
# DATA QUALITY
# ============================================================

print("\n" + "=" * 70)
print("DATA QUALITY")
print("=" * 70)

numeric_columns = [
    "height",
    "bust",
    "waist",
    "hips",
    "garment_bust",
    "garment_waist",
    "garment_hip",
    "garment_length",
    "garment_sleeve",
    "bust_ease",
    "waist_ease",
    "hip_ease",
    "bust_ratio",
    "waist_ratio",
    "hip_ratio",
    "bust_mismatch",
    "waist_mismatch",
    "hip_mismatch",
    "bust_relative_mismatch",
    "waist_relative_mismatch",
    "hip_relative_mismatch",
    "mean_ease",
    "ease_std",
    "length_height_ratio",
    "physical_mismatch",
]

nan_count = (
    candidates[
        numeric_columns
    ]
    .isna()
    .sum()
    .sum()
)

inf_count = (
    np.isinf(
        candidates[
            numeric_columns
        ]
        .select_dtypes(
            include=np.number
        )
        .to_numpy()
    )
    .sum()
)

print(
    "NaN values:",
    nan_count
)

print(
    "Inf values:",
    inf_count
)

# ============================================================
# SAVE
# ============================================================

candidates.to_csv(
    OUTPUT_PATH,
    index=False,
)

print("\n" + "=" * 70)
print("CANDIDATE MATRIX SAVED")
print("=" * 70)

print(
    OUTPUT_PATH
)

print(
    "\nZYRA V2 STEP 1 COMPLETE"
)
```

```python
# ============================================================
# ZYRA V2 — STEP 2
# CONFIDENT NEGATIVE MINING
# ============================================================

import pandas as pd
import numpy as np
from pathlib import Path

# ============================================================
# PATHS
# ============================================================

BASE = Path(
    "/Users/saketh/Desktop/Projects/weavly/core-model"
)

INPUT_PATH = (
    BASE /
    "data/v2/v2_person_garment_candidates.csv"
)

OUTPUT_PATH = (
    BASE /
    "data/v2/v2_confident_training_dataset.csv"
)

# ============================================================
# LOAD
# ============================================================

df = pd.read_csv(
    INPUT_PATH
)

print("=" * 70)
print("ZYRA V2 — CONFIDENT NEGATIVE MINING")
print("=" * 70)

print(
    "Candidate rows:",
    len(df)
)

# ============================================================
# SEPARATE KNOWN POSITIVES / UNKNOWNS
# ============================================================

positive = df[
    df["label_status"] == "positive"
].copy()

unknown = df[
    df["label_status"] == "unknown"
].copy()

print(
    "\nKnown positives:",
    len(positive)
)

print(
    "Unknown pairs:",
    len(unknown)
)

# ============================================================
# IMPORTANT
# ============================================================
#
# We do NOT assume:
#
# unknown == negative
#
# Instead we look for garments that are clearly
# physically incompatible.
#
# ------------------------------------------------------------
# Primary signal:
#
# physical_mismatch
#
# This is the mean relative mismatch across:
#
# bust
# waist
# hips
#
# Higher = worse physical compatibility.
# ============================================================

print("\n" + "=" * 70)
print("PHYSICAL MISMATCH DISTRIBUTION")
print("=" * 70)

print(
    unknown[
        "physical_mismatch"
    ].describe(
        percentiles=[
            0.50,
            0.75,
            0.80,
            0.85,
            0.90,
            0.95,
            0.975,
            0.99,
        ]
    )
)

# ============================================================
# THRESHOLD
# ============================================================
#
# We intentionally use a conservative threshold.
#
# A pair is considered a confident negative when:
#
#   relative mismatch >= 0.30
#
# Meaning the average relative dimensional mismatch is
# at least 30%.
#
# We also require at least TWO known body/garment
# dimensions so that one missing measurement cannot
# create a false negative.
# ============================================================

MISMATCH_THRESHOLD = 0.30

MIN_KNOWN_DIMENSIONS = 2

confident_negative_mask = (
    (
        unknown[
            "physical_mismatch"
        ]
        >=
        MISMATCH_THRESHOLD
    )
    &
    (
        unknown[
            "known_fit_dimensions"
        ]
        >=
        MIN_KNOWN_DIMENSIONS
    )
)

confident_negative = (
    unknown[
        confident_negative_mask
    ]
    .copy()
)

remaining_unknown = (
    unknown[
        ~confident_negative_mask
    ]
    .copy()
)

print("\n" + "=" * 70)
print("CONFIDENT NEGATIVE MINING")
print("=" * 70)

print(
    "Threshold:",
    MISMATCH_THRESHOLD
)

print(
    "Minimum known dimensions:",
    MIN_KNOWN_DIMENSIONS
)

print(
    "\nConfident negatives:",
    len(confident_negative)
)

print(
    "Remaining unknown:",
    len(remaining_unknown)
)

# ============================================================
# ADD TRAINING LABEL
# ============================================================

positive[
    "training_label"
] = 1

confident_negative[
    "training_label"
] = 0

# ============================================================
# NEGATIVE QUALITY
# ============================================================

print("\n" + "=" * 70)
print("CONFIDENT NEGATIVE QUALITY")
print("=" * 70)

print(
    confident_negative[
        "physical_mismatch"
    ]
    .describe()
)

# ============================================================
# NEGATIVE MISMATCH BUCKETS
# ============================================================

def mismatch_bucket(x):

    if x < 0.30:
        return "ambiguous"

    if x < 0.40:
        return "moderate"

    if x < 0.50:
        return "strong"

    return "very_strong"


confident_negative[
    "negative_strength"
] = (
    confident_negative[
        "physical_mismatch"
    ]
    .apply(
        mismatch_bucket
    )
)

print(
    "\nNegative strength:"
)

print(
    confident_negative[
        "negative_strength"
    ]
    .value_counts()
)

# ============================================================
# DIMENSION AVAILABILITY
# ============================================================

print("\n" + "=" * 70)
print("KNOWN DIMENSIONS")
print("=" * 70)

print(
    confident_negative[
        "known_fit_dimensions"
    ]
    .value_counts()
    .sort_index()
)

# ============================================================
# POSITIVE VS NEGATIVE PHYSICAL MISMATCH
# ============================================================

print("\n" + "=" * 70)
print("POSITIVE VS CONFIDENT NEGATIVE")
print("=" * 70)

print(
    "Positive physical mismatch:"
)

print(
    positive[
        "physical_mismatch"
    ]
    .describe()
)

print(
    "\nConfident negative physical mismatch:"
)

print(
    confident_negative[
        "physical_mismatch"
    ]
    .describe()
)

# ============================================================
# PER PERSON
# ============================================================

print("\n" + "=" * 70)
print("PER-PERSON NEGATIVE COUNTS")
print("=" * 70)

person_positive_counts = (
    positive
    .groupby(
        "person_id"
    )
    .size()
    .rename(
        "positive_count"
    )
)

person_negative_counts = (
    confident_negative
    .groupby(
        "person_id"
    )
    .size()
    .rename(
        "negative_count"
    )
)

person_summary = pd.concat(
    [
        person_positive_counts,
        person_negative_counts,
    ],
    axis=1
).fillna(0)

person_summary[
    "negative_positive_ratio"
] = (
    person_summary[
        "negative_count"
    ]
    /
    person_summary[
        "positive_count"
    ]
)

print(
    person_summary
)

# ============================================================
# NEGATIVE-TO-POSITIVE RATIO
# ============================================================

total_positive = len(
    positive
)

total_negative = len(
    confident_negative
)

print("\n" + "=" * 70)
print("DATASET RATIO")
print("=" * 70)

print(
    "Positive:",
    total_positive
)

print(
    "Confident negative:",
    total_negative
)

print(
    "Negative / Positive:",
    f"{total_negative / max(1, total_positive):.3f}"
)

# ============================================================
# CAP NEGATIVES PER PERSON
# ============================================================
#
# We don't want people with many candidate garments to
# dominate training.
#
# Maximum negatives per person:
#
#   5 × number of positives
#
# ============================================================

MAX_NEGATIVE_MULTIPLIER = 5

selected_negative_parts = []

for person_id, group in (
    confident_negative
    .groupby(
        "person_id"
    )
):

    positive_count = (
        person_positive_counts
        .get(
            person_id,
            0
        )
    )

    max_negatives = max(
        1,
        int(
            positive_count
            *
            MAX_NEGATIVE_MULTIPLIER
        )
    )

    if len(group) <= max_negatives:

        selected = group

    else:

        # Prefer stronger physical incompatibility.
        selected = (
            group
            .sort_values(
                "physical_mismatch",
                ascending=False,
            )
            .head(
                max_negatives
            )
        )

    selected_negative_parts.append(
        selected
    )

if selected_negative_parts:

    selected_negative = pd.concat(
        selected_negative_parts,
        ignore_index=True,
    )

else:

    selected_negative = (
        confident_negative
        .iloc[0:0]
        .copy()
    )

# ============================================================
# FINAL TRAINING DATASET
# ============================================================

training = pd.concat(
    [
        positive,
        selected_negative,
    ],
    ignore_index=True,
)

# ============================================================
# SHUFFLE
# ============================================================

training = (
    training
    .sample(
        frac=1,
        random_state=42,
    )
    .reset_index(
        drop=True
    )
)

# ============================================================
# DUPLICATE CHECK
# ============================================================

duplicate_pairs = (
    training[
        [
            "person_id",
            "product_id",
        ]
    ]
    .duplicated()
    .sum()
)

label_conflicts = (
    training
    .groupby(
        [
            "person_id",
            "product_id",
        ]
    )[
        "training_label"
    ]
    .nunique()
)

label_conflicts = (
    label_conflicts
    .gt(1)
    .sum()
)

# ============================================================
# FINAL SUMMARY
# ============================================================

print("\n" + "=" * 70)
print("FINAL CONFIDENT TRAINING DATASET")
print("=" * 70)

print(
    "Positive:",
    (
        training[
            "training_label"
        ]
        == 1
    ).sum()
)

print(
    "Negative:",
    (
        training[
            "training_label"
        ] == 0
    ).sum()
)

print(
    "Total:",
    len(training)
)

print(
    "\nLabels:"
)

print(
    training[
        "training_label"
    ]
    .value_counts()
)

print(
    "\nDuplicate person-garment pairs:",
    duplicate_pairs
)

print(
    "Label conflicts:",
    label_conflicts
)

# ============================================================
# PER PERSON BALANCE
# ============================================================

balance = (
    training
    .groupby(
        [
            "person_id",
            "training_label",
        ]
    )
    .size()
    .unstack(
        fill_value=0
    )
)

if 0 not in balance.columns:
    balance[0] = 0

if 1 not in balance.columns:
    balance[1] = 0

balance[
    "difference"
] = (
    balance[1]
    -
    balance[0]
)

print("\n" + "=" * 70)
print("PERSON BALANCE")
print("=" * 70)

print(
    balance
)

# ============================================================
# SAVE
# ============================================================

assert duplicate_pairs == 0
assert label_conflicts == 0

training.to_csv(
    OUTPUT_PATH,
    index=False,
)

print("\n" + "=" * 70)
print("CONFIDENT TRAINING DATASET SAVED")
print("=" * 70)

print(
    OUTPUT_PATH
)

print(
    "\nZYRA V2 STEP 2 COMPLETE"
)
```

```python
# ============================================================
# ZYRA V2 — STEP 3
# TRAIN COMPATIBILITY RANKER
# ============================================================

import pandas as pd
import numpy as np
import torch
import torch.nn as nn
import torch.nn.functional as F

from pathlib import Path
from sklearn.model_selection import train_test_split
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler

# ============================================================
# PATHS
# ============================================================

BASE = Path(
    "/Users/saketh/Desktop/Projects/weavly/core-model"
)

DATA_PATH = (
    BASE /
    "data/v2/v2_confident_training_dataset.csv"
)

RAW_PATH = (
    BASE /
    "data/v2/v2_compatibility_raw.csv"
)

MODEL_PATH = (
    BASE /
    "data/v2/v2_compatibility_ranker.pt"
)

SEED = 42
EPOCHS = 50
LR = 1e-3

np.random.seed(SEED)
torch.manual_seed(SEED)

# ============================================================
# DEVICE
# ============================================================

device = (
    torch.device("mps")
    if torch.backends.mps.is_available()
    else torch.device("cpu")
)

print("=" * 70)
print("ZYRA V2 — STEP 3 MODEL TRAINING")
print("=" * 70)

print("Device:", device)

# ============================================================
# LOAD
# ============================================================

df = pd.read_csv(DATA_PATH)
raw = pd.read_csv(RAW_PATH)

print("\nTraining rows:", len(df))
print("People:", df["person_id"].nunique())

print("\nLabels:")
print(df["training_label"].value_counts())

# ============================================================
# FEATURES
# ============================================================

FEATURES = [
    "height",
    "bust",
    "waist",
    "hips",

    "garment_bust",
    "garment_waist",
    "garment_hip",
    "garment_length",
    "garment_sleeve",

    "bust_ease",
    "waist_ease",
    "hip_ease",

    "bust_ratio",
    "waist_ratio",
    "hip_ratio",

    "bust_mismatch",
    "waist_mismatch",
    "hip_mismatch",

    "bust_relative_mismatch",
    "waist_relative_mismatch",
    "hip_relative_mismatch",

    "mean_ease",
    "ease_std",

    "known_fit_dimensions",
    "tight_dimension_count",
    "extreme_ease_count",

    "length_height_ratio",
]

# ============================================================
# LOOKUPS
# ============================================================

person_lookup = (
    raw[
        [
            "person_id",
            "height",
            "bust",
            "waist",
            "hips",
        ]
    ]
    .drop_duplicates("person_id")
    .set_index("person_id")
)

garment_lookup = (
    raw[
        [
            "product_id",
            "garment_bust",
            "garment_waist",
            "garment_hip",
            "garment_length",
            "garment_sleeve",
        ]
    ]
    .drop_duplicates("product_id")
    .set_index("product_id")
)

# ============================================================
# FEATURE BUILDER
# ============================================================

def make_features(person_id, product_id):

    p = person_lookup.loc[person_id]
    g = garment_lookup.loc[product_id]

    row = {}

    row["height"] = p["height"]
    row["bust"] = p["bust"]
    row["waist"] = p["waist"]
    row["hips"] = p["hips"]

    row["garment_bust"] = g["garment_bust"]
    row["garment_waist"] = g["garment_waist"]
    row["garment_hip"] = g["garment_hip"]
    row["garment_length"] = g["garment_length"]
    row["garment_sleeve"] = g["garment_sleeve"]

    row["bust_ease"] = (
        g["garment_bust"] - p["bust"]
    )

    row["waist_ease"] = (
        g["garment_waist"] - p["waist"]
    )

    row["hip_ease"] = (
        g["garment_hip"] - p["hips"]
    )

    row["bust_ratio"] = (
        g["garment_bust"] / p["bust"]
    )

    row["waist_ratio"] = (
        g["garment_waist"] / p["waist"]
    )

    row["hip_ratio"] = (
        g["garment_hip"] / p["hips"]
    )

    row["bust_mismatch"] = abs(
        g["garment_bust"] - p["bust"]
    )

    row["waist_mismatch"] = abs(
        g["garment_waist"] - p["waist"]
    )

    row["hip_mismatch"] = abs(
        g["garment_hip"] - p["hips"]
    )

    row["bust_relative_mismatch"] = (
        row["bust_mismatch"] / p["bust"]
    )

    row["waist_relative_mismatch"] = (
        row["waist_mismatch"] / p["waist"]
    )

    row["hip_relative_mismatch"] = (
        row["hip_mismatch"] / p["hips"]
    )

    eases = np.array(
        [
            row["bust_ease"],
            row["waist_ease"],
            row["hip_ease"],
        ],
        dtype=float,
    )

    valid = eases[~np.isnan(eases)]

    if len(valid):

        row["mean_ease"] = np.mean(valid)
        row["ease_std"] = np.std(valid)

    else:

        row["mean_ease"] = np.nan
        row["ease_std"] = np.nan

    row["known_fit_dimensions"] = (
        np.sum(~np.isnan(eases))
    )

    row["tight_dimension_count"] = (
        np.sum(eases < 0)
    )

    row["extreme_ease_count"] = (
        np.sum(eases > 50)
    )

    if (
        pd.isna(g["garment_length"])
        or
        pd.isna(p["height"])
    ):

        row["length_height_ratio"] = np.nan

    else:

        row["length_height_ratio"] = (
            g["garment_length"]
            /
            p["height"]
        )

    return np.array(
        [row[f] for f in FEATURES],
        dtype=np.float32,
    )

# ============================================================
# PERSON-LEVEL SPLIT
# ============================================================

people = sorted(
    df["person_id"].unique()
)

train_people, temp_people = train_test_split(
    people,
    test_size=16,
    random_state=SEED,
)

val_people, test_people = train_test_split(
    temp_people,
    test_size=8,
    random_state=SEED,
)

train_people = set(train_people)
val_people = set(val_people)
test_people = set(test_people)

print("\n" + "=" * 70)
print("PERSON-LEVEL SPLIT")
print("=" * 70)

print("Train:", len(train_people))
print("Validation:", len(val_people))
print("Test:", len(test_people))

print(
    "Train ∩ Val:",
    train_people & val_people
)

print(
    "Train ∩ Test:",
    train_people & test_people
)

print(
    "Val ∩ Test:",
    val_people & test_people
)

# ============================================================
# SPLIT
# ============================================================

train_df = df[
    df["person_id"].isin(train_people)
].copy()

val_df = df[
    df["person_id"].isin(val_people)
].copy()

test_df = df[
    df["person_id"].isin(test_people)
].copy()

print("\nRows:")
print("Train:", len(train_df))
print("Val:", len(val_df))
print("Test:", len(test_df))

# ============================================================
# BUILD FEATURES
# ============================================================

def build_xy(frame):

    X = []
    y = []

    for _, row in frame.iterrows():

        X.append(
            make_features(
                row["person_id"],
                row["product_id"],
            )
        )

        y.append(
            row["training_label"]
        )

    return (
        np.asarray(X, dtype=np.float32),
        np.asarray(y, dtype=np.float32),
    )

X_train, y_train = build_xy(train_df)
X_val, y_val = build_xy(val_df)
X_test, y_test = build_xy(test_df)

# ============================================================
# IMPUTATION
# ============================================================

imputer = SimpleImputer(
    strategy="median"
)

imputer.fit(X_train)

X_train = imputer.transform(X_train)
X_val = imputer.transform(X_val)
X_test = imputer.transform(X_test)

# ============================================================
# SCALING
# ============================================================

scaler = StandardScaler()

scaler.fit(X_train)

X_train = scaler.transform(X_train)
X_val = scaler.transform(X_val)
X_test = scaler.transform(X_test)

# ============================================================
# TENSORS
# ============================================================

X_train = torch.tensor(
    X_train,
    dtype=torch.float32,
    device=device,
)

y_train = torch.tensor(
    y_train,
    dtype=torch.float32,
    device=device,
)

X_val = torch.tensor(
    X_val,
    dtype=torch.float32,
    device=device,
)

y_val = torch.tensor(
    y_val,
    dtype=torch.float32,
    device=device,
)

X_test = torch.tensor(
    X_test,
    dtype=torch.float32,
    device=device,
)

y_test = torch.tensor(
    y_test,
    dtype=torch.float32,
    device=device,
)

print("\n" + "=" * 70)
print("TENSORS")
print("=" * 70)

print("Feature dimension:", X_train.shape[1])
print("Train:", X_train.shape)
print("Val:", X_val.shape)
print("Test:", X_test.shape)

print(
    "NaN:",
    torch.isnan(X_train).sum().item()
    +
    torch.isnan(X_val).sum().item()
    +
    torch.isnan(X_test).sum().item()
)

# ============================================================
# MODEL
# ============================================================

class CompatibilityRanker(nn.Module):

    def __init__(self, input_dim):

        super().__init__()

        self.network = nn.Sequential(

            nn.Linear(
                input_dim,
                64,
            ),

            nn.ReLU(),

            nn.Linear(
                64,
                32,
            ),

            nn.ReLU(),

            nn.Linear(
                32,
                1,
            ),
        )

    def forward(self, x):

        return self.network(
            x
        ).squeeze(-1)


model = CompatibilityRanker(
    X_train.shape[1]
).to(device)

optimizer = torch.optim.Adam(
    model.parameters(),
    lr=LR,
    weight_decay=1e-4,
)

criterion = nn.BCEWithLogitsLoss()

print("\n" + "=" * 70)
print("MODEL")
print("=" * 70)

print(
    "Input dimension:",
    X_train.shape[1]
)

print(
    "Parameters:",
    sum(
        p.numel()
        for p in model.parameters()
    )
)

# ============================================================
# TRAINING
# ============================================================

print("\n" + "=" * 70)
print("TRAINING")
print("=" * 70)

best_val_loss = float("inf")
best_state = None

for epoch in range(
    1,
    EPOCHS + 1
):

    model.train()

    optimizer.zero_grad()

    logits = model(
        X_train
    )

    loss = criterion(
        logits,
        y_train,
    )

    loss.backward()

    optimizer.step()

    # --------------------------------------------------------
    # Validation
    # --------------------------------------------------------

    model.eval()

    with torch.no_grad():

        val_logits = model(
            X_val
        )

        val_loss = criterion(
            val_logits,
            y_val,
        )

        train_pred = (
            torch.sigmoid(logits)
            >= 0.5
        )

        val_pred = (
            torch.sigmoid(val_logits)
            >= 0.5
        )

        train_acc = (
            train_pred
            ==
            y_train.bool()
        ).float().mean().item()

        val_acc = (
            val_pred
            ==
            y_val.bool()
        ).float().mean().item()

    # --------------------------------------------------------
    # Checkpoint
    # --------------------------------------------------------

    if val_loss.item() < best_val_loss:

        best_val_loss = (
            val_loss.item()
        )

        best_state = {
            k: v.detach()
            .cpu()
            .clone()

            for k, v
            in model.state_dict().items()
        }

    if (
        epoch == 1
        or epoch % 5 == 0
    ):

        print(
            f"Epoch {epoch:02d}/{EPOCHS} "
            f"| Loss: {loss.item():.4f} "
            f"| Train Acc: {train_acc:.4f} "
            f"| Val Loss: {val_loss.item():.4f} "
            f"| Val Acc: {val_acc:.4f}"
        )

# ============================================================
# RESTORE BEST MODEL
# ============================================================

model.load_state_dict(
    best_state
)

model.eval()

# ============================================================
# FINAL SCORES
# ============================================================

with torch.no_grad():

    train_prob = torch.sigmoid(
        model(X_train)
    )

    val_prob = torch.sigmoid(
        model(X_val)
    )

    test_prob = torch.sigmoid(
        model(X_test)
    )

# ============================================================
# ACCURACY
# ============================================================

train_acc = (
    (train_prob >= 0.5)
    ==
    y_train.bool()
).float().mean().item()

val_acc = (
    (val_prob >= 0.5)
    ==
    y_val.bool()
).float().mean().item()

test_acc = (
    (test_prob >= 0.5)
    ==
    y_test.bool()
).float().mean().item()

# ============================================================
# SCORE SEPARATION
# ============================================================

test_prob_np = (
    test_prob
    .detach()
    .cpu()
    .numpy()
)

y_test_np = (
    y_test
    .detach()
    .cpu()
    .numpy()
)

positive_scores = (
    test_prob_np[
        y_test_np == 1
    ]
)

negative_scores = (
    test_prob_np[
        y_test_np == 0
    ]
)

# ============================================================
# RESULTS
# ============================================================

print("\n" + "=" * 70)
print("FINAL V2 RESULTS")
print("=" * 70)

print(
    "Train Accuracy:",
    f"{train_acc:.4f}"
)

print(
    "Validation Accuracy:",
    f"{val_acc:.4f}"
)

print(
    "Test Accuracy:",
    f"{test_acc:.4f}"
)

print(
    "\nPositive score mean:",
    f"{positive_scores.mean():.4f}"
)

print(
    "Negative score mean:",
    f"{negative_scores.mean():.4f}"
)

print(
    "Score gap:",
    f"{positive_scores.mean() - negative_scores.mean():.4f}"
)

# ============================================================
# SAVE MODEL
# ============================================================

torch.save(
    {
        "model_state_dict":
            model.state_dict(),

        "feature_names":
            FEATURES,

        "input_dim":
            len(FEATURES),

        "train_people":
            sorted(train_people),

        "val_people":
            sorted(val_people),

        "test_people":
            sorted(test_people),

        "imputer_statistics":
            imputer.statistics_,

        "scaler_mean":
            scaler.mean_,

        "scaler_scale":
            scaler.scale_,

        "seed":
            SEED,

        "dataset":
            "v2_confident_training_dataset",
    },
    MODEL_PATH,
)

print("\n" + "=" * 70)
print("MODEL SAVED")
print("=" * 70)

print(
    MODEL_PATH
)

print(
    "\nZYRA V2 STEP 3 COMPLETE"
)
```

```python
# ============================================================
# ZYRA V2 — STEP 4
# FULL PERSON × GARMENT TOP-K RANKING EVALUATION
# ============================================================

import pandas as pd
import numpy as np
import torch
import torch.nn as nn

from pathlib import Path

# ============================================================
# PATHS
# ============================================================

BASE = Path(
    "/Users/saketh/Desktop/Projects/weavly/core-model"
)

CANDIDATE_PATH = (
    BASE /
    "data/v2/v2_person_garment_candidates.csv"
)

RAW_PATH = (
    BASE /
    "data/v2/v2_compatibility_raw.csv"
)

MODEL_PATH = (
    BASE /
    "data/v2/v2_compatibility_ranker.pt"
)

OUTPUT_PATH = (
    BASE /
    "data/v2/v2_topk_ranking_results.csv"
)

# ============================================================
# DEVICE
# ============================================================

device = (
    torch.device("mps")
    if torch.backends.mps.is_available()
    else torch.device("cpu")
)

print("=" * 70)
print("ZYRA V2 — FULL TOP-K RANKING EVALUATION")
print("=" * 70)

print("Device:", device)

# ============================================================
# LOAD DATA
# ============================================================

candidates = pd.read_csv(
    CANDIDATE_PATH
)

raw = pd.read_csv(
    RAW_PATH
)

checkpoint = torch.load(
    MODEL_PATH,
    map_location=device,
    weights_only=False,
)

FEATURES = checkpoint[
    "feature_names"
]

TEST_PEOPLE = set(
    checkpoint[
        "test_people"
    ]
)

print(
    "\nCandidate rows:",
    len(candidates)
)

print(
    "People:",
    candidates[
        "person_id"
    ].nunique()
)

print(
    "Garments:",
    candidates[
        "product_id"
    ].nunique()
)

print(
    "Held-out test people:",
    len(TEST_PEOPLE)
)

print(
    "Test people:",
    sorted(TEST_PEOPLE)
)

# ============================================================
# MODEL
# ============================================================

class CompatibilityRanker(nn.Module):

    def __init__(self, input_dim):

        super().__init__()

        self.network = nn.Sequential(

            nn.Linear(
                input_dim,
                64,
            ),

            nn.ReLU(),

            nn.Linear(
                64,
                32,
            ),

            nn.ReLU(),

            nn.Linear(
                32,
                1,
            ),
        )

    def forward(self, x):

        return self.network(
            x
        ).squeeze(-1)


model = CompatibilityRanker(
    len(FEATURES)
)

model.load_state_dict(
    checkpoint[
        "model_state_dict"
    ]
)

model = model.to(device)
model.eval()

# ============================================================
# IMPUTER / SCALER
# ============================================================

imputer_statistics = np.asarray(
    checkpoint[
        "imputer_statistics"
    ],
    dtype=np.float32,
)

scaler_mean = np.asarray(
    checkpoint[
        "scaler_mean"
    ],
    dtype=np.float32,
)

scaler_scale = np.asarray(
    checkpoint[
        "scaler_scale"
    ],
    dtype=np.float32,
)

# ============================================================
# FEATURE MATRIX
# ============================================================

X = candidates[
    FEATURES
].copy()

# ------------------------------------------------------------
# Impute using TRAINING statistics
# ------------------------------------------------------------

for i, feature in enumerate(
    FEATURES
):

    X[feature] = (
        X[feature]
        .fillna(
            imputer_statistics[i]
        )
    )

# ------------------------------------------------------------
# Scale using TRAINING statistics
# ------------------------------------------------------------

X = (
    X
    .to_numpy(
        dtype=np.float32
    )
)

X = (
    X
    -
    scaler_mean
)

X = (
    X
    /
    np.where(
        scaler_scale == 0,
        1.0,
        scaler_scale,
    )
)

# ============================================================
# PREDICT ALL CANDIDATES
# ============================================================

X_tensor = torch.tensor(
    X,
    dtype=torch.float32,
    device=device,
)

print("\n" + "=" * 70)
print("SCORING ALL CANDIDATES")
print("=" * 70)

with torch.no_grad():

    scores = (
        model(
            X_tensor
        )
        .detach()
        .cpu()
        .numpy()
    )

candidates[
    "model_score"
] = scores

print(
    "Scored rows:",
    len(scores)
)

# ============================================================
# TEST SET ONLY
# ============================================================

test_candidates = candidates[
    candidates[
        "person_id"
    ].isin(TEST_PEOPLE)
].copy()

print(
    "Test candidate rows:",
    len(test_candidates)
)

# ============================================================
# VERIFY 897 GARMENTS PER PERSON
# ============================================================

garment_counts = (
    test_candidates
    .groupby(
        "person_id"
    )[
        "product_id"
    ]
    .nunique()
)

print("\nGarments per test person:")
print(garment_counts)

assert (
    garment_counts == 897
).all()

# ============================================================
# RANK
# ============================================================

test_candidates[
    "rank"
] = (
    test_candidates
    .groupby(
        "person_id"
    )[
        "model_score"
    ]
    .rank(
        method="first",
        ascending=False,
    )
)

# ============================================================
# KNOWN POSITIVES
# ============================================================

test_candidates[
    "is_positive"
] = (
    test_candidates[
        "label_status"
    ]
    ==
    "positive"
)

# ============================================================
# METRIC FUNCTIONS
# ============================================================

def recall_at_k(
    ranks,
    total_positives,
    k,
):

    if total_positives == 0:
        return np.nan

    hits = np.sum(
        ranks <= k
    )

    return (
        hits
        /
        total_positives
    )


def average_precision_at_k(
    ranks,
    total_positives,
    k,
):

    if total_positives == 0:
        return np.nan

    sorted_ranks = np.sort(
        ranks
    )

    hits = 0
    precision_sum = 0.0

    for rank in sorted_ranks:

        if rank > k:
            break

        hits += 1

        precision_sum += (
            hits
            /
            rank
        )

    return (
        precision_sum
        /
        min(
            total_positives,
            k,
        )
    )


def ndcg_at_k(
    ranks,
    k,
):

    if len(ranks) == 0:
        return 0.0

    dcg = 0.0

    for rank in ranks:

        if rank <= k:

            dcg += (
                1.0
                /
                np.log2(
                    rank + 1
                )
            )

    ideal_count = min(
        len(ranks),
        k,
    )

    if ideal_count == 0:
        return 0.0

    idcg = sum(
        1.0
        /
        np.log2(
            i + 2
        )

        for i in range(
            ideal_count
        )
    )

    return dcg / idcg


def mrr(
    ranks
):

    if len(ranks) == 0:
        return 0.0

    return (
        1.0
        /
        np.min(ranks)
    )

# ============================================================
# PER PERSON EVALUATION
# ============================================================

results = []

print("\n" + "=" * 70)
print("PER-PERSON TOP-K RESULTS")
print("=" * 70)

for person_id, group in (
    test_candidates
    .groupby(
        "person_id"
    )
):

    positive_ranks = (
        group[
            group[
                "is_positive"
            ]
        ][
            "rank"
        ]
        .to_numpy()
    )

    total_positives = len(
        positive_ranks
    )

    result = {
        "person_id":
            person_id,

        "positive_count":
            total_positives,

        "recall@5":
            recall_at_k(
                positive_ranks,
                total_positives,
                5,
            ),

        "recall@10":
            recall_at_k(
                positive_ranks,
                total_positives,
                10,
            ),

        "recall@20":
            recall_at_k(
                positive_ranks,
                total_positives,
                20,
            ),

        "recall@50":
            recall_at_k(
                positive_ranks,
                total_positives,
                50,
            ),

        "recall@100":
            recall_at_k(
                positive_ranks,
                total_positives,
                100,
            ),

        "ndcg@10":
            ndcg_at_k(
                positive_ranks,
                10,
            ),

        "ndcg@20":
            ndcg_at_k(
                positive_ranks,
                20,
            ),

        "mrr":
            mrr(
                positive_ranks
            ),

        "average_precision@20":
            average_precision_at_k(
                positive_ranks,
                total_positives,
                20,
            ),

        "best_positive_rank":
            np.min(
                positive_ranks
            )
            if total_positives
            else np.nan,

        "median_positive_rank":
            np.median(
                positive_ranks
            )
            if total_positives
            else np.nan,
    }

    results.append(
        result
    )

results_df = pd.DataFrame(
    results
)

# ============================================================
# GLOBAL METRICS
# ============================================================

print("\n" + "=" * 70)
print("GLOBAL TOP-K RESULTS")
print("=" * 70)

metric_columns = [
    "recall@5",
    "recall@10",
    "recall@20",
    "recall@50",
    "recall@100",
    "ndcg@10",
    "ndcg@20",
    "mrr",
    "average_precision@20",
]

global_metrics = (
    results_df[
        metric_columns
    ]
    .mean()
)

for metric, value in (
    global_metrics.items()
):

    print(
        f"{metric:25s}: "
        f"{value:.4f}"
    )

# ============================================================
# POSITIVE RANK DISTRIBUTION
# ============================================================

all_positive_ranks = (
    test_candidates[
        test_candidates[
            "is_positive"
        ]
    ][
        "rank"
    ]
    .to_numpy()
)

print("\n" + "=" * 70)
print("POSITIVE RANK DISTRIBUTION")
print("=" * 70)

print(
    pd.Series(
        all_positive_ranks
    ).describe(
        percentiles=[
            0.25,
            0.50,
            0.75,
            0.90,
            0.95,
        ]
    )
)

# ============================================================
# POSITIVES FOUND IN TOP K
# ============================================================

print("\n" + "=" * 70)
print("KNOWN POSITIVES FOUND IN TOP-K")
print("=" * 70)

total_positive_count = len(
    all_positive_ranks
)

for k in [
    1,
    5,
    10,
    20,
    50,
    100,
]:

    count = np.sum(
        all_positive_ranks
        <= k
    )

    percentage = (
        count
        /
        total_positive_count
        *
        100
    )

    print(
        f"Top-{k:3d}: "
        f"{count:4d} / "
        f"{total_positive_count} "
        f"({percentage:.2f}%)"
    )

# ============================================================
# TOP 10 RECOMMENDATIONS PER TEST PERSON
# ============================================================

print("\n" + "=" * 70)
print("TOP 10 RECOMMENDATIONS")
print("=" * 70)

top10 = (
    test_candidates[
        test_candidates[
            "rank"
        ] <= 10
    ]
    .sort_values(
        [
            "person_id",
            "rank",
        ]
    )
)

for person_id, group in (
    top10
    .groupby(
        "person_id"
    )
):

    print(
        f"\nPerson: {person_id}"
    )

    display(
        group[
            [
                "rank",
                "product_id",
                "model_score",
                "is_positive",
                "physical_mismatch",
                "garment_bust",
                "garment_waist",
                "garment_hip",
                "garment_length",
                "garment_sleeve",
            ]
        ]
    )

# ============================================================
# POSITIVE RANKS PER PERSON
# ============================================================

print("\n" + "=" * 70)
print("POSITIVE RANKS PER PERSON")
print("=" * 70)

for _, row in (
    results_df
    .sort_values(
        "person_id"
    )
    .iterrows()
):

    print(
        f"{row['person_id']}: "
        f"positives={int(row['positive_count'])} | "
        f"best={int(row['best_positive_rank'])} | "
        f"median={row['median_positive_rank']:.1f} | "
        f"R@10={row['recall@10']:.3f} | "
        f"NDCG@10={row['ndcg@10']:.3f}"
    )

# ============================================================
# SAVE RESULTS
# ============================================================

results_df.to_csv(
    OUTPUT_PATH,
    index=False,
)

print("\n" + "=" * 70)
print("TOP-K EVALUATION SAVED")
print("=" * 70)

print(
    OUTPUT_PATH
)

print(
    "\nZYRA V2 STEP 4 COMPLETE"
)
```

```python
# ============================================================
# ZYRA V2.1 — RELATIONSHIP-ONLY PAIRWISE RANKER
# ============================================================
#
# Goal:
# Prevent the model from learning global garment-size shortcuts.
#
# Model sees:
#   person ↔ garment relationships
#
# Model does NOT directly see:
#   raw person measurements
#   raw garment measurements
#
# Evaluation:
#   8 completely unseen people
#   × 897 garments
#   → full ranking
#
# ============================================================

import pandas as pd
import numpy as np
import torch
import torch.nn as nn

from pathlib import Path
from sklearn.model_selection import train_test_split
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler

# ============================================================
# PATHS
# ============================================================

BASE = Path(
    "/Users/saketh/Desktop/Projects/weavly/core-model"
)

DATA_PATH = (
    BASE /
    "data/v2/v2_confident_training_dataset.csv"
)

CANDIDATE_PATH = (
    BASE /
    "data/v2/v2_person_garment_candidates.csv"
)

MODEL_PATH = (
    BASE /
    "data/v2/v2_compatibility_ranker_v21.pt"
)

RANKING_OUTPUT = (
    BASE /
    "data/v2/v2_v21_topk_results.csv"
)

SEED = 42
EPOCHS = 75
LR = 5e-4

np.random.seed(SEED)
torch.manual_seed(SEED)

# ============================================================
# DEVICE
# ============================================================

device = (
    torch.device("mps")
    if torch.backends.mps.is_available()
    else torch.device("cpu")
)

print("=" * 70)
print("ZYRA V2.1 — RELATIONSHIP-ONLY RANKER")
print("=" * 70)

print("Device:", device)

# ============================================================
# LOAD
# ============================================================

train_df = pd.read_csv(
    DATA_PATH
)

candidates = pd.read_csv(
    CANDIDATE_PATH
)

print(
    "\nTraining rows:",
    len(train_df)
)

print(
    "Candidate rows:",
    len(candidates)
)

print(
    "People:",
    candidates["person_id"].nunique()
)

print(
    "Garments:",
    candidates["product_id"].nunique()
)

# ============================================================
# RELATIONSHIP FEATURES
# ============================================================
#
# IMPORTANT:
#
# No raw:
#   height
#   bust
#   waist
#   hips
#   garment_bust
#   garment_waist
#   garment_hip
#   garment_length
#   garment_sleeve
#
# are given directly to the model.
#
# Only person-garment relationships are used.
# ============================================================

FEATURES = [

    # --------------------------------------------------------
    # Absolute ease
    # --------------------------------------------------------

    "bust_ease",
    "waist_ease",
    "hip_ease",

    # --------------------------------------------------------
    # Relative fit
    # --------------------------------------------------------

    "bust_ratio",
    "waist_ratio",
    "hip_ratio",

    # --------------------------------------------------------
    # Absolute mismatch
    # --------------------------------------------------------

    "bust_mismatch",
    "waist_mismatch",
    "hip_mismatch",

    # --------------------------------------------------------
    # Relative mismatch
    # --------------------------------------------------------

    "bust_relative_mismatch",
    "waist_relative_mismatch",
    "hip_relative_mismatch",

    # --------------------------------------------------------
    # Aggregate relationship features
    # --------------------------------------------------------

    "mean_ease",
    "ease_std",

    "known_fit_dimensions",

    "tight_dimension_count",

    "extreme_ease_count",

    "length_height_ratio",

    # --------------------------------------------------------
    # NEW NONLINEAR FIT FEATURES
    # --------------------------------------------------------

    "abs_mean_ease",
    "positive_ease_count",
    "negative_ease_count",

    "max_relative_mismatch",
    "mean_relative_mismatch",

    # --------------------------------------------------------
    # FIT BALANCE
    # --------------------------------------------------------

    "bust_hip_balance",
    "bust_waist_balance",
    "waist_hip_balance",

    # --------------------------------------------------------
    # SOFT FIT PENALTY
    # --------------------------------------------------------

    "soft_fit_penalty",

]

print(
    "\nFeature count:",
    len(FEATURES)
)

# ============================================================
# FEATURE BUILDER
# ============================================================

def build_relationship_features(
    person_id,
    product_id,
    people_lookup,
    garment_lookup,
):

    p = people_lookup.loc[
        person_id
    ]

    g = garment_lookup.loc[
        product_id
    ]

    # --------------------------------------------------------
    # Raw relationships
    # --------------------------------------------------------

    bust_ease = (
        g["garment_bust"]
        -
        p["bust"]
    )

    waist_ease = (
        g["garment_waist"]
        -
        p["waist"]
    )

    hip_ease = (
        g["garment_hip"]
        -
        p["hips"]
    )

    # --------------------------------------------------------
    # Ratios
    # --------------------------------------------------------

    bust_ratio = (
        g["garment_bust"]
        /
        p["bust"]
    )

    waist_ratio = (
        g["garment_waist"]
        /
        p["waist"]
    )

    hip_ratio = (
        g["garment_hip"]
        /
        p["hips"]
    )

    # --------------------------------------------------------
    # Absolute mismatch
    # --------------------------------------------------------

    bust_mismatch = abs(
        bust_ease
    )

    waist_mismatch = abs(
        waist_ease
    )

    hip_mismatch = abs(
        hip_ease
    )

    # --------------------------------------------------------
    # Relative mismatch
    # --------------------------------------------------------

    bust_relative = (
        bust_mismatch
        /
        p["bust"]
    )

    waist_relative = (
        waist_mismatch
        /
        p["waist"]
    )

    hip_relative = (
        hip_mismatch
        /
        p["hips"]
    )

    # --------------------------------------------------------
    # Ease vector
    # --------------------------------------------------------

    eases = np.array(
        [
            bust_ease,
            waist_ease,
            hip_ease,
        ],
        dtype=float,
    )

    valid_eases = (
        eases[
            ~np.isnan(eases)
        ]
    )

    if len(valid_eases) > 0:

        mean_ease = (
            np.mean(
                valid_eases
            )
        )

        ease_std = (
            np.std(
                valid_eases
            )
        )

        abs_mean_ease = (
            np.mean(
                np.abs(
                    valid_eases
                )
            )
        )

        positive_ease_count = (
            np.sum(
                valid_eases > 0
            )
        )

        negative_ease_count = (
            np.sum(
                valid_eases < 0
            )
        )

    else:

        mean_ease = np.nan
        ease_std = np.nan
        abs_mean_ease = np.nan
        positive_ease_count = 0
        negative_ease_count = 0

    # --------------------------------------------------------
    # Known dimensions
    # --------------------------------------------------------

    known_fit_dimensions = (
        np.sum(
            ~np.isnan(eases)
        )
    )

    # --------------------------------------------------------
    # Tight dimensions
    # --------------------------------------------------------

    tight_dimension_count = (
        np.sum(
            eases < 0
        )
    )

    # --------------------------------------------------------
    # Extreme ease
    # --------------------------------------------------------

    extreme_ease_count = (
        np.sum(
            eases > 50
        )
    )

    # --------------------------------------------------------
    # Relative mismatch vector
    # --------------------------------------------------------

    relative_mismatches = np.array(
        [
            bust_relative,
            waist_relative,
            hip_relative,
        ],
        dtype=float,
    )

    valid_relative = (
        relative_mismatches[
            ~np.isnan(
                relative_mismatches
            )
        ]
    )

    if len(valid_relative) > 0:

        max_relative_mismatch = (
            np.max(
                valid_relative
            )
        )

        mean_relative_mismatch = (
            np.mean(
                valid_relative
            )
        )

    else:

        max_relative_mismatch = np.nan
        mean_relative_mismatch = np.nan

    # --------------------------------------------------------
    # Length / height relationship
    # --------------------------------------------------------

    if (
        pd.isna(
            g["garment_length"]
        )
        or
        pd.isna(
            p["height"]
        )
        or
        p["height"] == 0
    ):

        length_height_ratio = np.nan

    else:

        length_height_ratio = (
            g["garment_length"]
            /
            p["height"]
        )

    # --------------------------------------------------------
    # Balance features
    # --------------------------------------------------------

    bust_hip_balance = (
        bust_relative
        -
        hip_relative
    )

    bust_waist_balance = (
        bust_relative
        -
        waist_relative
    )

    waist_hip_balance = (
        waist_relative
        -
        hip_relative
    )

    # --------------------------------------------------------
    # Soft fit penalty
    #
    # Penalizes large mismatch smoothly instead of using
    # binary tight/extreme flags alone.
    # --------------------------------------------------------

    valid_for_penalty = (
        relative_mismatches[
            ~np.isnan(
                relative_mismatches
            )
        ]
    )

    if len(valid_for_penalty) > 0:

        soft_fit_penalty = np.mean(
            np.square(
                valid_for_penalty
            )
        )

    else:

        soft_fit_penalty = np.nan

    # --------------------------------------------------------
    # RETURN
    # --------------------------------------------------------

    return {

        "bust_ease":
            bust_ease,

        "waist_ease":
            waist_ease,

        "hip_ease":
            hip_ease,

        "bust_ratio":
            bust_ratio,

        "waist_ratio":
            waist_ratio,

        "hip_ratio":
            hip_ratio,

        "bust_mismatch":
            bust_mismatch,

        "waist_mismatch":
            waist_mismatch,

        "hip_mismatch":
            hip_mismatch,

        "bust_relative_mismatch":
            bust_relative,

        "waist_relative_mismatch":
            waist_relative,

        "hip_relative_mismatch":
            hip_relative,

        "mean_ease":
            mean_ease,

        "ease_std":
            ease_std,

        "known_fit_dimensions":
            known_fit_dimensions,

        "tight_dimension_count":
            tight_dimension_count,

        "extreme_ease_count":
            extreme_ease_count,

        "length_height_ratio":
            length_height_ratio,

        "abs_mean_ease":
            abs_mean_ease,

        "positive_ease_count":
            positive_ease_count,

        "negative_ease_count":
            negative_ease_count,

        "max_relative_mismatch":
            max_relative_mismatch,

        "mean_relative_mismatch":
            mean_relative_mismatch,

        "bust_hip_balance":
            bust_hip_balance,

        "bust_waist_balance":
            bust_waist_balance,

        "waist_hip_balance":
            waist_hip_balance,

        "soft_fit_penalty":
            soft_fit_penalty,
    }


# ============================================================
# LOOKUPS
# ============================================================

people_lookup = (
    candidates[
        [
            "person_id",
            "height",
            "bust",
            "waist",
            "hips",
        ]
    ]
    .drop_duplicates(
        "person_id"
    )
    .set_index(
        "person_id"
    )
)

garment_lookup = (
    candidates[
        [
            "product_id",
            "garment_bust",
            "garment_waist",
            "garment_hip",
            "garment_length",
            "garment_sleeve",
        ]
    ]
    .drop_duplicates(
        "product_id"
    )
    .set_index(
        "product_id"
    )
)

# ============================================================
# BUILD TRAINING FEATURES
# ============================================================

print("\n" + "=" * 70)
print("BUILDING RELATIONSHIP FEATURES")
print("=" * 70)

def build_matrix(frame):

    X = []
    y = []

    for _, row in frame.iterrows():

        features = (
            build_relationship_features(
                row["person_id"],
                row["product_id"],
                people_lookup,
                garment_lookup,
            )
        )

        X.append(
            [
                features[f]
                for f in FEATURES
            ]
        )

        y.append(
            row["training_label"]
        )

    return (
        np.asarray(
            X,
            dtype=np.float32,
        ),
        np.asarray(
            y,
            dtype=np.float32,
        ),
    )

# ============================================================
# PERSON SPLIT
# ============================================================

people = sorted(
    train_df[
        "person_id"
    ]
    .unique()
)

train_people, temp_people = (
    train_test_split(
        people,
        test_size=16,
        random_state=SEED,
    )
)

val_people, test_people = (
    train_test_split(
        temp_people,
        test_size=8,
        random_state=SEED,
    )
)

train_people = set(
    train_people
)

val_people = set(
    val_people
)

test_people = set(
    test_people
)

print(
    "Train people:",
    len(train_people)
)

print(
    "Validation people:",
    len(val_people)
)

print(
    "Test people:",
    len(test_people)
)

# ============================================================
# SPLIT DATA
# ============================================================

train_split = train_df[
    train_df[
        "person_id"
    ].isin(
        train_people
    )
].copy()

val_split = train_df[
    train_df[
        "person_id"
    ].isin(
        val_people
    )
].copy()

test_split = train_df[
    train_df[
        "person_id"
    ].isin(
        test_people
    )
].copy()

print(
    "\nTraining rows:",
    len(train_split)
)

print(
    "Validation rows:",
    len(val_split)
)

print(
    "Test rows:",
    len(test_split)
)

# ============================================================
# BUILD
# ============================================================

X_train, y_train = build_matrix(
    train_split
)

X_val, y_val = build_matrix(
    val_split
)

X_test, y_test = build_matrix(
    test_split
)

# ============================================================
# IMPUTATION
# ============================================================

imputer = SimpleImputer(
    strategy="median"
)

imputer.fit(
    X_train
)

X_train = imputer.transform(
    X_train
)

X_val = imputer.transform(
    X_val
)

X_test = imputer.transform(
    X_test
)

# ============================================================
# SCALING
# ============================================================

scaler = StandardScaler()

scaler.fit(
    X_train
)

X_train = scaler.transform(
    X_train
)

X_val = scaler.transform(
    X_val
)

X_test = scaler.transform(
    X_test
)

print("\n" + "=" * 70)
print("FEATURE MATRICES")
print("=" * 70)

print(
    "Feature dimension:",
    X_train.shape[1]
)

print(
    "Train:",
    X_train.shape
)

print(
    "Val:",
    X_val.shape
)

print(
    "Test:",
    X_test.shape
)

print(
    "NaN:",
    np.isnan(
        X_train
    ).sum()
    +
    np.isnan(
        X_val
    ).sum()
    +
    np.isnan(
        X_test
    ).sum()
)

# ============================================================
# TENSORS
# ============================================================

X_train_t = torch.tensor(
    X_train,
    dtype=torch.float32,
    device=device,
)

y_train_t = torch.tensor(
    y_train,
    dtype=torch.float32,
    device=device,
)

X_val_t = torch.tensor(
    X_val,
    dtype=torch.float32,
    device=device,
)

y_val_t = torch.tensor(
    y_val,
    dtype=torch.float32,
    device=device,
)

X_test_t = torch.tensor(
    X_test,
    dtype=torch.float32,
    device=device,
)

y_test_t = torch.tensor(
    y_test,
    dtype=torch.float32,
    device=device,
)

# ============================================================
# MODEL
# ============================================================

class V21Ranker(nn.Module):

    def __init__(
        self,
        input_dim,
    ):

        super().__init__()

        self.network = nn.Sequential(

            nn.Linear(
                input_dim,
                64,
            ),

            nn.ReLU(),

            nn.Dropout(
                0.10
            ),

            nn.Linear(
                64,
                32,
            ),

            nn.ReLU(),

            nn.Dropout(
                0.10
            ),

            nn.Linear(
                32,
                1,
            ),
        )

    def forward(self, x):

        return self.network(
            x
        ).squeeze(-1)


model = V21Ranker(
    len(FEATURES)
).to(device)

optimizer = torch.optim.AdamW(
    model.parameters(),
    lr=LR,
    weight_decay=1e-3,
)

criterion = nn.BCEWithLogitsLoss()

print("\n" + "=" * 70)
print("MODEL")
print("=" * 70)

print(
    "Input dimension:",
    len(FEATURES)
)

print(
    "Parameters:",
    sum(
        p.numel()
        for p in model.parameters()
    )
)

# ============================================================
# TRAIN
# ============================================================

print("\n" + "=" * 70)
print("TRAINING")
print("=" * 70)

best_val_loss = float(
    "inf"
)

best_state = None

patience = 12
patience_counter = 0

for epoch in range(
    1,
    EPOCHS + 1,
):

    model.train()

    optimizer.zero_grad()

    train_logits = model(
        X_train_t
    )

    loss = criterion(
        train_logits,
        y_train_t,
    )

    loss.backward()

    optimizer.step()

    # --------------------------------------------------------
    # Validation
    # --------------------------------------------------------

    model.eval()

    with torch.no_grad():

        val_logits = model(
            X_val_t
        )

        val_loss = criterion(
            val_logits,
            y_val_t,
        )

        train_pred = (
            torch.sigmoid(
                train_logits
            )
            >= 0.5
        )

        val_pred = (
            torch.sigmoid(
                val_logits
            )
            >= 0.5
        )

        train_acc = (
            train_pred
            ==
            y_train_t.bool()
        ).float().mean().item()

        val_acc = (
            val_pred
            ==
            y_val_t.bool()
        ).float().mean().item()

    # --------------------------------------------------------
    # Best checkpoint
    # --------------------------------------------------------

    if (
        val_loss.item()
        <
        best_val_loss
    ):

        best_val_loss = (
            val_loss.item()
        )

        best_state = {
            k:
                v.detach()
                .cpu()
                .clone()

            for k, v
            in model.state_dict().items()
        }

        patience_counter = 0

    else:

        patience_counter += 1

    if (
        epoch == 1
        or epoch % 5 == 0
    ):

        print(
            f"Epoch {epoch:02d}/{EPOCHS} "
            f"| Loss {loss.item():.4f} "
            f"| Train {train_acc:.4f} "
            f"| Val Loss {val_loss.item():.4f} "
            f"| Val {val_acc:.4f}"
        )

    if patience_counter >= patience:

        print(
            "\nEarly stopping at epoch",
            epoch,
        )

        break

# ============================================================
# RESTORE BEST
# ============================================================

model.load_state_dict(
    best_state
)

model.eval()

# ============================================================
# CLASSIFICATION RESULTS
# ============================================================

with torch.no_grad():

    train_scores = torch.sigmoid(
        model(
            X_train_t
        )
    ).cpu().numpy()

    val_scores = torch.sigmoid(
        model(
            X_val_t
        )
    ).cpu().numpy()

    test_scores = torch.sigmoid(
        model(
            X_test_t
        )
    ).cpu().numpy()

train_acc = (
    (
        train_scores >= 0.5
    )
    ==
    y_train.astype(bool)
).mean()

val_acc = (
    (
        val_scores >= 0.5
    )
    ==
    y_val.astype(bool)
).mean()

test_acc = (
    (
        test_scores >= 0.5
    )
    ==
    y_test.astype(bool)
).mean()

positive_scores = (
    test_scores[
        y_test == 1
    ]
)

negative_scores = (
    test_scores[
        y_test == 0
    ]
)

print("\n" + "=" * 70)
print("V2.1 CLASSIFICATION RESULTS")
print("=" * 70)

print(
    "Train accuracy:",
    f"{train_acc:.4f}"
)

print(
    "Validation accuracy:",
    f"{val_acc:.4f}"
)

print(
    "Test accuracy:",
    f"{test_acc:.4f}"
)

print(
    "Positive mean score:",
    f"{positive_scores.mean():.4f}"
)

print(
    "Negative mean score:",
    f"{negative_scores.mean():.4f}"
)

print(
    "Score gap:",
    f"{positive_scores.mean() - negative_scores.mean():.4f}"
)

# ============================================================
# FULL CANDIDATE MATRIX FEATURES
# ============================================================

print("\n" + "=" * 70)
print("SCORING ALL 50,232 CANDIDATES")
print("=" * 70)

candidate_features = []

for _, row in candidates.iterrows():

    features = (
        build_relationship_features(
            row["person_id"],
            row["product_id"],
            people_lookup,
            garment_lookup,
        )
    )

    candidate_features.append(
        [
            features[f]
            for f in FEATURES
        ]
    )

candidate_X = np.asarray(
    candidate_features,
    dtype=np.float32,
)

candidate_X = imputer.transform(
    candidate_X
)

candidate_X = scaler.transform(
    candidate_X
)

candidate_tensor = torch.tensor(
    candidate_X,
    dtype=torch.float32,
    device=device,
)

with torch.no_grad():

    candidate_scores = (
        torch.sigmoid(
            model(
                candidate_tensor
            )
        )
        .cpu()
        .numpy()
    )

candidates[
    "v21_score"
] = candidate_scores

# ============================================================
# RANK ALL GARMENTS FOR TEST PEOPLE
# ============================================================

test_candidates = candidates[
    candidates[
        "person_id"
    ].isin(
        test_people
    )
].copy()

test_candidates[
    "rank"
] = (
    test_candidates
    .groupby(
        "person_id"
    )[
        "v21_score"
    ]
    .rank(
        method="first",
        ascending=False,
    )
)

test_candidates[
    "is_positive"
] = (
    test_candidates[
        "label_status"
    ]
    ==
    "positive"
)

# ============================================================
# METRICS
# ============================================================

def recall_at_k(
    ranks,
    k,
):

    if len(ranks) == 0:
        return 0.0

    return np.mean(
        ranks <= k
    )


def ndcg_at_k(
    ranks,
    k,
):

    if len(ranks) == 0:
        return 0.0

    dcg = sum(
        1.0
        /
        np.log2(
            r + 1
        )

        for r in ranks
        if r <= k
    )

    ideal_count = min(
        len(ranks),
        k,
    )

    idcg = sum(
        1.0
        /
        np.log2(
            i + 2
        )

        for i in range(
            ideal_count
        )
    )

    if idcg == 0:
        return 0.0

    return dcg / idcg


def reciprocal_rank(
    ranks
):

    if len(ranks) == 0:
        return 0.0

    return (
        1.0
        /
        np.min(ranks)
    )

# ============================================================
# PER PERSON
# ============================================================

ranking_results = []

for person_id, group in (
    test_candidates
    .groupby(
        "person_id"
    )
):

    positive_ranks = (
        group[
            group[
                "is_positive"
            ]
        ][
            "rank"
        ]
        .to_numpy()
    )

    ranking_results.append(
        {
            "person_id":
                person_id,

            "positive_count":
                len(
                    positive_ranks
                ),

            "recall@5":
                recall_at_k(
                    positive_ranks,
                    5,
                ),

            "recall@10":
                recall_at_k(
                    positive_ranks,
                    10,
                ),

            "recall@20":
                recall_at_k(
                    positive_ranks,
                    20,
                ),

            "recall@50":
                recall_at_k(
                    positive_ranks,
                    50,
                ),

            "recall@100":
                recall_at_k(
                    positive_ranks,
                    100,
                ),

            "ndcg@10":
                ndcg_at_k(
                    positive_ranks,
                    10,
                ),

            "ndcg@20":
                ndcg_at_k(
                    positive_ranks,
                    20,
                ),

            "mrr":
                reciprocal_rank(
                    positive_ranks
                ),

            "best_rank":
                np.min(
                    positive_ranks
                )
                if len(
                    positive_ranks
                )
                else np.nan,

            "median_rank":
                np.median(
                    positive_ranks
                )
                if len(
                    positive_ranks
                )
                else np.nan,
        }
    )

ranking_results = pd.DataFrame(
    ranking_results
)

# ============================================================
# GLOBAL
# ============================================================

print("\n" + "=" * 70)
print("V2.1 GLOBAL RANKING RESULTS")
print("=" * 70)

for metric in [
    "recall@5",
    "recall@10",
    "recall@20",
    "recall@50",
    "recall@100",
    "ndcg@10",
    "ndcg@20",
    "mrr",
]:

    print(
        f"{metric:15s}: "
        f"{ranking_results[metric].mean():.4f}"
    )

# ============================================================
# POSITIVE RANK DISTRIBUTION
# ============================================================

positive_ranks = (
    test_candidates[
        test_candidates[
            "is_positive"
        ]
    ][
        "rank"
    ]
    .to_numpy()
)

print("\n" + "=" * 70)
print("POSITIVE RANK DISTRIBUTION")
print("=" * 70)

print(
    pd.Series(
        positive_ranks
    ).describe(
        percentiles=[
            0.25,
            0.50,
            0.75,
            0.90,
            0.95,
        ]
    )
)

# ============================================================
# TOP-K COUNTS
# ============================================================

print("\n" + "=" * 70)
print("KNOWN POSITIVES IN TOP-K")
print("=" * 70)

for k in [
    1,
    5,
    10,
    20,
    50,
    100,
]:

    count = np.sum(
        positive_ranks <= k
    )

    percentage = (
        count
        /
        len(
            positive_ranks
        )
        *
        100
    )

    print(
        f"Top-{k:3d}: "
        f"{count:4d} / "
        f"{len(positive_ranks)} "
        f"({percentage:.2f}%)"
    )

# ============================================================
# TOP 10 FOR EACH TEST PERSON
# ============================================================

print("\n" + "=" * 70)
print("TOP 10 V2.1 RECOMMENDATIONS")
print("=" * 70)

for person_id, group in (
    test_candidates
    .groupby(
        "person_id"
    )
):

    top10 = (
        group
        .sort_values(
            "rank"
        )
        .head(10)
    )

    print(
        f"\nPerson: {person_id}"
    )

    display(
        top10[
            [
                "rank",
                "product_id",
                "v21_score",
                "is_positive",
                "physical_mismatch",
                "garment_bust",
                "garment_waist",
                "garment_hip",
                "garment_length",
                "garment_sleeve",
            ]
        ]
    )

# ============================================================
# SAVE MODEL
# ============================================================

torch.save(
    {
        "model_state_dict":
            model.state_dict(),

        "feature_names":
            FEATURES,

        "input_dim":
            len(FEATURES),

        "train_people":
            sorted(
                train_people
            ),

        "val_people":
            sorted(
                val_people
            ),

        "test_people":
            sorted(
                test_people
            ),

        "imputer_statistics":
            imputer.statistics_,

        "scaler_mean":
            scaler.mean_,

        "scaler_scale":
            scaler.scale_,

        "seed":
            SEED,

        "version":
            "V2.1_relationship_only",
    },
    MODEL_PATH,
)

ranking_results.to_csv(
    RANKING_OUTPUT,
    index=False,
)

print("\n" + "=" * 70)
print("V2.1 COMPLETE")
print("=" * 70)

print(
    "Model:",
    MODEL_PATH
)

print(
    "Ranking results:",
    RANKING_OUTPUT
)
```

```python
# ============================================================
# ZYRA V2 — STEP 5
# GARMENT DATA INVENTORY
# ============================================================

import pandas as pd
import numpy as np
from pathlib import Path

# ============================================================
# PATH
# ============================================================

BASE = Path(
    "/Users/saketh/Desktop/Projects/weavly/core-model"
)

RAW_PATH = (
    BASE /
    "data/v2/v2_compatibility_raw.csv"
)

CANDIDATE_PATH = (
    BASE /
    "data/v2/v2_person_garment_candidates.csv"
)

print("=" * 70)
print("ZYRA V2 — GARMENT DATA INVENTORY")
print("=" * 70)

# ============================================================
# LOAD
# ============================================================

raw = pd.read_csv(
    RAW_PATH
)

candidates = pd.read_csv(
    CANDIDATE_PATH
)

print(
    "\nRaw rows:",
    len(raw)
)

print(
    "Raw columns:",
    len(raw.columns)
)

print(
    "Candidate rows:",
    len(candidates)
)

# ============================================================
# ALL COLUMNS
# ============================================================

print("\n" + "=" * 70)
print("ALL RAW COLUMNS")
print("=" * 70)

for i, col in enumerate(
    raw.columns,
    start=1
):

    dtype = raw[col].dtype

    unique = raw[col].nunique(
        dropna=True
    )

    missing = raw[col].isna().sum()

    missing_pct = (
        missing
        /
        len(raw)
        *
        100
    )

    print(
        f"{i:02d}. "
        f"{col:35s} "
        f"dtype={str(dtype):12s} "
        f"unique={unique:6d} "
        f"missing={missing:6d} "
        f"({missing_pct:6.2f}%)"
    )

# ============================================================
# GARMENT-LEVEL DATASET
# ============================================================

print("\n" + "=" * 70)
print("GARMENT-LEVEL INVENTORY")
print("=" * 70)

if "product_id" not in raw.columns:

    raise ValueError(
        "product_id column not found"
    )

garments = (
    raw
    .drop_duplicates(
        "product_id"
    )
    .copy()
)

print(
    "Unique garments:",
    len(garments)
)

# ============================================================
# GARMENT COLUMNS
# ============================================================

print("\n" + "=" * 70)
print("GARMENT COLUMNS")
print("=" * 70)

garment_columns = []

for col in garments.columns:

    # Exclude obvious person fields
    person_words = [
        "person",
        "user",
        "profile",
        "height",
        "bust",
        "waist",
        "hips",
    ]

    if any(
        word in col.lower()
        for word in person_words
    ):

        continue

    garment_columns.append(
        col
    )

for col in garment_columns:

    unique = garments[
        col
    ].nunique(
        dropna=True
    )

    missing = garments[
        col
    ].isna().sum()

    print(
        f"{col:35s} "
        f"unique={unique:6d} "
        f"missing={missing:6d}"
    )

# ============================================================
# TEXT / CATEGORICAL FEATURES
# ============================================================

print("\n" + "=" * 70)
print("CATEGORICAL / TEXT FEATURES")
print("=" * 70)

categorical_candidates = []

for col in garment_columns:

    series = garments[col]

    if (
        series.dtype == "object"
        or
        str(
            series.dtype
        ).startswith(
            "category"
        )
    ):

        categorical_candidates.append(
            col
        )

        print(
            f"\n{col}"
        )

        print(
            "Unique:",
            series.nunique(
                dropna=True
            )
        )

        print(
            "Missing:",
            series.isna().sum()
        )

        values = (
            series
            .dropna()
            .astype(str)
            .value_counts()
            .head(20)
        )

        if len(values):

            print(
                values.to_string()
            )

# ============================================================
# NUMERIC GARMENT FEATURES
# ============================================================

print("\n" + "=" * 70)
print("NUMERIC GARMENT FEATURES")
print("=" * 70)

numeric_candidates = []

for col in garment_columns:

    if pd.api.types.is_numeric_dtype(
        garments[col]
    ):

        numeric_candidates.append(
            col
        )

        print(
            f"\n{col}"
        )

        print(
            garments[col]
            .describe()
            .to_string()
        )

# ============================================================
# POTENTIAL PRODUCT METADATA
# ============================================================

print("\n" + "=" * 70)
print("POTENTIAL PRODUCT METADATA")
print("=" * 70)

metadata_keywords = [
    "category",
    "type",
    "style",
    "color",
    "colour",
    "fabric",
    "material",
    "brand",
    "designer",
    "silhouette",
    "pattern",
    "occasion",
    "season",
    "sleeve",
    "neck",
    "collar",
    "length",
    "fit",
    "texture",
    "description",
    "name",
    "title",
    "product",
]

metadata_columns = []

for col in garments.columns:

    col_lower = col.lower()

    if any(
        keyword in col_lower
        for keyword in metadata_keywords
    ):

        metadata_columns.append(
            col
        )

        print(
            "FOUND:",
            col
        )

# ============================================================
# PRODUCT ID EXAMPLES
# ============================================================

print("\n" + "=" * 70)
print("PRODUCT ID EXAMPLES")
print("=" * 70)

print(
    garments[
        "product_id"
    ]
    .head(30)
    .to_string(
        index=False
    )
)

# ============================================================
# COMPLETE GARMENT EXAMPLES
# ============================================================

print("\n" + "=" * 70)
print("SAMPLE GARMENTS")
print("=" * 70)

sample_columns = [
    col
    for col in [
        "product_id",
        "product_name",
        "name",
        "title",
        "category",
        "type",
        "style",
        "color",
        "colour",
        "fabric",
        "material",
        "brand",
        "silhouette",
        "pattern",
        "occasion",
        "description",
        "garment_bust",
        "garment_waist",
        "garment_hip",
        "garment_length",
        "garment_sleeve",
    ]
    if col in garments.columns
]

if len(sample_columns):

    print(
        garments[
            sample_columns
        ]
        .head(10)
        .to_string(
            index=False
        )
    )

else:

    print(
        "No expected metadata columns found."
    )

# ============================================================
# VARIANT ANALYSIS
# ============================================================

print("\n" + "=" * 70)
print("PRODUCT VARIANT ANALYSIS")
print("=" * 70)

if "product_id" in raw.columns:

    variant_counts = (
        raw
        .groupby(
            "product_id"
        )
        .size()
    )

    print(
        "Rows per product:"
    )

    print(
        variant_counts
        .describe()
        .to_string()
    )

# ============================================================
# DUPLICATE PHYSICAL PRODUCTS
# ============================================================

physical_columns = [
    col
    for col in [
        "garment_bust",
        "garment_waist",
        "garment_hip",
        "garment_length",
        "garment_sleeve",
    ]
    if col in garments.columns
]

if len(
    physical_columns
) >= 2:

    duplicate_physical = (
        garments
        .groupby(
            physical_columns,
            dropna=False
        )
        .size()
        .sort_values(
            ascending=False
        )
    )

    print("\n" + "=" * 70)
    print("PHYSICAL-SPEC DUPLICATES")
    print("=" * 70)

    print(
        duplicate_physical
        .head(20)
        .to_string()
    )

# ============================================================
# INFORMATION DENSITY
# ============================================================

print("\n" + "=" * 70)
print("INFORMATION DENSITY")
print("=" * 70)

for col in garments.columns:

    non_null = (
        garments[col]
        .notna()
        .sum()
    )

    percentage = (
        non_null
        /
        len(garments)
        *
        100
    )

    if (
        percentage >= 50
    ):

        quality = "GOOD"

    elif (
        percentage >= 20
    ):

        quality = "PARTIAL"

    else:

        quality = "SPARSE"

    print(
        f"{col:35s} "
        f"{percentage:6.2f}% "
        f"{quality}"
    )

# ============================================================
# FINAL SUMMARY
# ============================================================

print("\n" + "=" * 70)
print("FINAL GARMENT DATA SUMMARY")
print("=" * 70)

print(
    "Unique garments:",
    len(garments)
)

print(
    "Categorical/text columns:",
    len(categorical_candidates)
)

print(
    "Numeric columns:",
    len(numeric_candidates)
)

print(
    "Potential metadata columns:",
    len(metadata_columns)
)

print(
    "\nCategorical:",
    categorical_candidates
)

print(
    "\nMetadata:",
    metadata_columns
)

print(
    "\nNumeric:",
    numeric_candidates
)

print("\n" + "=" * 70)
print("ZYRA V2 STEP 5 COMPLETE")
print("=" * 70)
```

```python
# ============================================================
# ZYRA V2 — STEP 6
# PRODUCT_ID SEMANTIC FEATURE EXTRACTION
# ============================================================

import pandas as pd
import numpy as np
import re
from pathlib import Path

# ============================================================
# PATHS
# ============================================================

BASE = Path(
    "/Users/saketh/Desktop/Projects/weavly/core-model"
)

INPUT_PATH = (
    BASE /
    "data/v2/v2_person_garment_candidates.csv"
)

OUTPUT_PATH = (
    BASE /
    "data/v2/v2_garment_semantic_features.csv"
)

# ============================================================
# LOAD UNIQUE GARMENTS
# ============================================================

df = pd.read_csv(INPUT_PATH)

garments = (
    df[
        ["product_id"]
    ]
    .drop_duplicates()
    .reset_index(drop=True)
)

print("=" * 70)
print("ZYRA V2 — PRODUCT SEMANTIC EXTRACTION")
print("=" * 70)

print(
    "Unique garments:",
    len(garments)
)

# ============================================================
# DICTIONARIES
# ============================================================

CATEGORY_WORDS = {
    "top": [
        "top",
        "tee",
        "tank",
        "shirt",
        "blouse",
        "tunic",
        "pullover",
        "sweater",
        "hoodie",
        "jersey",
    ],

    "cardigan": [
        "cardigan",
        "cowl",
    ],

    "kimono": [
        "kimono",
        "poncho",
    ],

    "dress": [
        "dress",
    ],

    "skirt": [
        "skirt",
    ],

    "pants": [
        "pant",
        "trouser",
        "legging",
        "jean",
    ],

    "shorts": [
        "short",
    ],

    "jacket": [
        "jacket",
        "blazer",
        "coat",
    ],

    "outerwear": [
        "cape",
        "wrap",
    ],
}

STYLE_WORDS = {
    "relaxed": [
        "relaxed",
        "cosy",
        "oversized",
        "swing",
        "batwing",
        "loose",
    ],

    "fitted": [
        "fitted",
        "stretch",
        "slim",
        "fitted",
        "tailored",
    ],

    "casual": [
        "tee",
        "hoodie",
        "jersey",
        "tank",
        "pullover",
    ],

    "elegant": [
        "blouse",
        "cowl",
        "drape",
        "luxe",
        "silk",
    ],

    "romantic": [
        "love",
        "belle",
        "ruffle",
        "floral",
        "bell",
    ],

    "sporty": [
        "jersey",
        "crew",
        "sport",
    ],

    "layering": [
        "cardigan",
        "kimono",
        "poncho",
        "wrap",
    ],
}

PATTERN_WORDS = {
    "stripe": [
        "stripe",
        "striped",
    ],

    "floral": [
        "floral",
        "flower",
    ],

    "animal": [
        "leopard",
        "animal",
        "tiger",
        "zebra",
    ],

    "check": [
        "check",
        "checked",
        "plaid",
    ],

    "spot": [
        "spot",
        "dot",
        "polka",
    ],

    "sequin": [
        "sequin",
        "sequins",
    ],

    "plain": [
        "plain",
        "solid",
        "essential",
    ],
}

MATERIAL_WORDS = {
    "knit": [
        "knit",
        "knitted",
    ],

    "jersey": [
        "jersey",
    ],

    "stretch": [
        "stretch",
    ],

    "cosy": [
        "cosy",
        "cozy",
    ],

    "textured": [
        "textured",
        "texture",
    ],

    "sequin": [
        "sequin",
        "sequins",
    ],
}

SILHOUETTE_WORDS = {
    "swing": [
        "swing",
    ],

    "oversized": [
        "oversized",
        "oversize",
        "loose",
    ],

    "relaxed": [
        "relaxed",
        "cosy",
        "batwing",
    ],

    "fitted": [
        "fitted",
        "stretch",
        "slim",
    ],

    "longline": [
        "longline",
        "long",
    ],

    "cropped": [
        "crop",
        "cropped",
    ],
}

SLEEVE_WORDS = {
    "long": [
        "long-sleeve",
        "longsleeve",
    ],

    "short": [
        "short-sleeve",
        "shortsleeve",
        "half-sleeve",
    ],

    "flute": [
        "flute-sleeve",
    ],

    "bell": [
        "bell-sleeve",
    ],

    "batwing": [
        "batwing",
    ],
}

NECKLINE_WORDS = {
    "cowl": [
        "cowl",
    ],

    "v_neck": [
        "vneck",
        "v-neck",
        "v-neck",
    ],

    "scoop": [
        "scoop",
    ],

    "off_shoulder": [
        "off-the-shoulder",
        "off-shoulder",
    ],

    "crew": [
        "crew",
    ],

    "collared": [
        "collar",
        "collared",
    ],
}

LENGTH_WORDS = {
    "long": [
        "long",
        "longline",
    ],

    "short": [
        "short",
        "cropped",
    ],

    "mid": [
        "midi",
        "mid",
    ],
}

COLOR_WORDS = [
    "black",
    "white",
    "pink",
    "blue",
    "navy",
    "green",
    "jade",
    "khaki",
    "grey",
    "gray",
    "nude",
    "blush",
    "teal",
    "wine",
    "red",
    "orange",
    "yellow",
    "lime",
    "purple",
    "violet",
    "brown",
    "cream",
    "natural",
    "opal",
    "gold",
    "silver",
    "olive",
    "stone",
    "stripe",
]

# ============================================================
# HELPERS
# ============================================================

def normalize_text(text):

    text = str(text).lower()

    text = (
        text
        .replace("%20", "-")
        .replace("_", "-")
    )

    text = re.sub(
        r"[^a-z0-9#\-]",
        "-",
        text,
    )

    text = re.sub(
        r"-+",
        "-",
        text,
    )

    return text.strip("-")


def extract_name_and_color(product_id):

    product_id = str(
        product_id
    )

    if "#" in product_id:

        name, color = (
            product_id
            .split(
                "#",
                1,
            )
        )

    else:

        name = product_id
        color = ""

    return (
        normalize_text(name),
        normalize_text(color),
    )


def find_matches(
    text,
    dictionary,
):

    matches = []

    for category, words in (
        dictionary.items()
    ):

        for word in words:

            word = normalize_text(
                word
            )

            if (
                word in text
            ):

                matches.append(
                    category
                )

                break

    return matches


def find_color(
    color_text,
    name_text,
):

    text = (
        color_text
        +
        "-"
        +
        name_text
    )

    matches = []

    for color in COLOR_WORDS:

        if (
            normalize_text(color)
            in text
        ):

            matches.append(
                color
            )

    # Preserve order
    matches = list(
        dict.fromkeys(
            matches
        )
    )

    return matches


# ============================================================
# EXTRACT
# ============================================================

rows = []

for product_id in garments[
    "product_id"
]:

    name, color = (
        extract_name_and_color(
            product_id
        )
    )

    combined = (
        name
        +
        "-"
        +
        color
    )

    categories = find_matches(
        combined,
        CATEGORY_WORDS,
    )

    styles = find_matches(
        combined,
        STYLE_WORDS,
    )

    patterns = find_matches(
        combined,
        PATTERN_WORDS,
    )

    materials = find_matches(
        combined,
        MATERIAL_WORDS,
    )

    silhouettes = find_matches(
        combined,
        SILHOUETTE_WORDS,
    )

    sleeves = find_matches(
        combined,
        SLEEVE_WORDS,
    )

    necklines = find_matches(
        combined,
        NECKLINE_WORDS,
    )

    lengths = find_matches(
        combined,
        LENGTH_WORDS,
    )

    colors = find_color(
        color,
        name,
    )

    # --------------------------------------------------------
    # Primary values
    # --------------------------------------------------------

    primary_category = (
        categories[0]
        if categories
        else "unknown"
    )

    primary_style = (
        styles[0]
        if styles
        else "unknown"
    )

    primary_pattern = (
        patterns[0]
        if patterns
        else "plain"
    )

    primary_material = (
        materials[0]
        if materials
        else "unknown"
    )

    primary_silhouette = (
        silhouettes[0]
        if silhouettes
        else "unknown"
    )

    primary_sleeve = (
        sleeves[0]
        if sleeves
        else "unknown"
    )

    primary_neckline = (
        necklines[0]
        if necklines
        else "unknown"
    )

    primary_length = (
        lengths[0]
        if lengths
        else "unknown"
    )

    primary_color = (
        colors[0]
        if colors
        else "unknown"
    )

    # --------------------------------------------------------
    # Binary semantic features
    # --------------------------------------------------------

    rows.append(
        {
            "product_id":
                product_id,

            "product_name":
                name,

            "product_color_raw":
                color,

            "category":
                primary_category,

            "style":
                primary_style,

            "pattern":
                primary_pattern,

            "material":
                primary_material,

            "silhouette":
                primary_silhouette,

            "sleeve_type":
                primary_sleeve,

            "neckline":
                primary_neckline,

            "length_style":
                primary_length,

            "color":
                primary_color,

            "category_count":
                len(categories),

            "style_count":
                len(styles),

            "pattern_count":
                len(patterns),

            "material_count":
                len(materials),

            "semantic_token_count":
                len(
                    set(
                        categories
                        +
                        styles
                        +
                        patterns
                        +
                        materials
                        +
                        silhouettes
                        +
                        sleeves
                        +
                        necklines
                        +
                        lengths
                        +
                        colors
                    )
                ),

            # Binary flags

            "is_striped":
                int(
                    "stripe"
                    in patterns
                ),

            "is_floral":
                int(
                    "floral"
                    in patterns
                ),

            "is_animal_print":
                int(
                    "animal"
                    in patterns
                ),

            "is_checked":
                int(
                    "check"
                    in patterns
                ),

            "is_sequin":
                int(
                    "sequin"
                    in patterns
                ),

            "is_knit":
                int(
                    "knit"
                    in materials
                ),

            "is_stretch":
                int(
                    "stretch"
                    in materials
                ),

            "is_relaxed":
                int(
                    (
                        "relaxed"
                        in styles
                    )
                    or
                    (
                        "relaxed"
                        in silhouettes
                    )
                ),

            "is_fitted":
                int(
                    (
                        "fitted"
                        in styles
                    )
                    or
                    (
                        "fitted"
                        in silhouettes
                    )
                ),

            "is_swing":
                int(
                    "swing"
                    in silhouettes
                ),

            "is_longline":
                int(
                    "longline"
                    in silhouettes
                ),

            "is_off_shoulder":
                int(
                    "off_shoulder"
                    in necklines
                ),

            "is_batwing":
                int(
                    "batwing"
                    in silhouettes
                ),

            "is_long_sleeve":
                int(
                    "long"
                    in sleeves
                ),

            "is_short_sleeve":
                int(
                    "short"
                    in sleeves
                ),
        }
    )

# ============================================================
# DATAFRAME
# ============================================================

semantic = pd.DataFrame(
    rows
)

# ============================================================
# CHECK
# ============================================================

print("\n" + "=" * 70)
print("SEMANTIC EXTRACTION SUMMARY")
print("=" * 70)

print(
    "Rows:",
    len(semantic)
)

print(
    "Unique products:",
    semantic[
        "product_id"
    ].nunique()
)

assert (
    len(semantic)
    ==
    garments[
        "product_id"
    ].nunique()
)

assert (
    semantic[
        "product_id"
    ].duplicated()
    .sum()
    ==
    0
)

# ============================================================
# DISTRIBUTIONS
# ============================================================

for column in [
    "category",
    "style",
    "pattern",
    "material",
    "silhouette",
    "sleeve_type",
    "neckline",
    "length_style",
    "color",
]:

    print(
        "\n" + "-" * 60
    )

    print(
        column.upper()
    )

    print(
        semantic[
            column
        ]
        .value_counts()
        .head(20)
        .to_string()
    )

# ============================================================
# COVERAGE
# ============================================================

print("\n" + "=" * 70)
print("SEMANTIC COVERAGE")
print("=" * 70)

semantic_columns = [
    "category",
    "style",
    "pattern",
    "material",
    "silhouette",
    "sleeve_type",
    "neckline",
    "length_style",
    "color",
]

for column in semantic_columns:

    known = (
        semantic[
            column
        ]
        != "unknown"
    ).sum()

    coverage = (
        known
        /
        len(semantic)
        *
        100
    )

    print(
        f"{column:20s}: "
        f"{known:4d}/{len(semantic)} "
        f"({coverage:6.2f}%)"
    )

# ============================================================
# EXAMPLES
# ============================================================

print("\n" + "=" * 70)
print("PARSED PRODUCT EXAMPLES")
print("=" * 70)

display(
    semantic[
        [
            "product_id",
            "product_name",
            "product_color_raw",
            "category",
            "style",
            "pattern",
            "material",
            "silhouette",
            "sleeve_type",
            "neckline",
            "length_style",
            "color",
        ]
    ]
    .head(30)
)

# ============================================================
# SAVE
# ============================================================

semantic.to_csv(
    OUTPUT_PATH,
    index=False,
)

print("\n" + "=" * 70)
print("SEMANTIC FEATURES SAVED")
print("=" * 70)

print(
    OUTPUT_PATH
)

print(
    "\nZYRA V2 STEP 6 COMPLETE"
)
```

```python
# ============================================================
# ZYRA V2 — STEP 7 CORRECTED
# HYBRID PERSON × GARMENT DATASET
# ============================================================

import pandas as pd
import numpy as np
from pathlib import Path

# ============================================================
# PATHS
# ============================================================

BASE = Path(
    "/Users/saketh/Desktop/Projects/weavly/core-model"
)

CANDIDATE_PATH = (
    BASE /
    "data/v2/v2_person_garment_candidates.csv"
)

TRAINING_PATH = (
    BASE /
    "data/v2/v2_confident_training_dataset.csv"
)

SEMANTIC_PATH = (
    BASE /
    "data/v2/v2_garment_semantic_features.csv"
)

OUTPUT_CANDIDATE_PATH = (
    BASE /
    "data/v2/v2_hybrid_candidates.csv"
)

OUTPUT_TRAINING_PATH = (
    BASE /
    "data/v2/v2_hybrid_training_dataset.csv"
)

# ============================================================
# LOAD
# ============================================================

candidates = pd.read_csv(
    CANDIDATE_PATH
)

training = pd.read_csv(
    TRAINING_PATH
)

semantic = pd.read_csv(
    SEMANTIC_PATH
)

print("=" * 70)
print("ZYRA V2 — STEP 7 CORRECTED HYBRID DATASET")
print("=" * 70)

print(
    "Candidates:",
    len(candidates)
)

print(
    "Training:",
    len(training)
)

print(
    "Semantic garments:",
    len(semantic)
)

# ============================================================
# VALIDATE SEMANTIC TABLE
# ============================================================

assert "product_id" in semantic.columns

assert (
    semantic["product_id"]
    .duplicated()
    .sum()
    == 0
)

# ============================================================
# SEMANTIC COLUMNS
# ============================================================

SEMANTIC_CATEGORICAL = [
    "category",
    "style",
    "pattern",
    "material",
    "silhouette",
    "sleeve_type",
    "neckline",
    "length_style",
    "color",
]

SEMANTIC_NUMERIC = [
    "category_count",
    "style_count",
    "pattern_count",
    "material_count",
    "semantic_token_count",

    "is_striped",
    "is_floral",
    "is_animal_print",
    "is_checked",
    "is_sequin",

    "is_knit",
    "is_stretch",

    "is_relaxed",
    "is_fitted",
    "is_swing",
    "is_longline",

    "is_off_shoulder",
    "is_batwing",

    "is_long_sleeve",
    "is_short_sleeve",
]

# ============================================================
# KEEP ONLY EXISTING SEMANTIC COLUMNS
# ============================================================

available_cat = [
    c
    for c in SEMANTIC_CATEGORICAL
    if c in semantic.columns
]

available_num = [
    c
    for c in SEMANTIC_NUMERIC
    if c in semantic.columns
]

print("\n" + "=" * 70)
print("AVAILABLE SEMANTIC FEATURES")
print("=" * 70)

print(
    "Categorical:",
    available_cat
)

print(
    "Numeric:",
    available_num
)

# ============================================================
# MERGE
# ============================================================

semantic_subset = semantic[
    [
        "product_id"
    ]
    +
    available_cat
    +
    available_num
].copy()

hybrid_candidates = candidates.merge(
    semantic_subset,
    on="product_id",
    how="left",
    validate="many_to_one",
)

hybrid_training = training.merge(
    semantic_subset,
    on="product_id",
    how="left",
    validate="many_to_one",
)

assert (
    len(hybrid_candidates)
    ==
    len(candidates)
)

assert (
    len(hybrid_training)
    ==
    len(training)
)

# ============================================================
# NORMALIZE CATEGORICALS
# ============================================================

for col in available_cat:

    hybrid_candidates[col] = (
        hybrid_candidates[col]
        .fillna("unknown")
        .astype(str)
        .str.strip()
        .str.lower()
    )

    hybrid_training[col] = (
        hybrid_training[col]
        .fillna("unknown")
        .astype(str)
        .str.strip()
        .str.lower()
    )

# ============================================================
# ONE-HOT ENCODING
# ============================================================

candidate_cat = pd.get_dummies(
    hybrid_candidates[
        available_cat
    ],
    prefix=available_cat,
    dtype=np.float32,
)

training_cat = pd.get_dummies(
    hybrid_training[
        available_cat
    ],
    prefix=available_cat,
    dtype=np.float32,
)

# ============================================================
# ALIGN COLUMNS
# ============================================================
#
# This is critical.
#
# Candidate and training matrices must have exactly the
# same semantic feature columns.
# ============================================================

candidate_cat, training_cat = (
    candidate_cat.align(
        training_cat,
        join="outer",
        axis=1,
        fill_value=0.0,
    )
)

# ============================================================
# BASE COLUMNS
# ============================================================

IDENTIFIERS = [
    "person_id",
    "product_id",
]

LABEL_COLUMNS = [
    "label_status",
    "training_label",
]

# ------------------------------------------------------------
# Physical features
# ------------------------------------------------------------

PHYSICAL_FEATURES = [
    "height",
    "bust",
    "waist",
    "hips",

    "garment_bust",
    "garment_waist",
    "garment_hip",
    "garment_length",
    "garment_sleeve",

    "bust_ease",
    "waist_ease",
    "hip_ease",

    "bust_ratio",
    "waist_ratio",
    "hip_ratio",

    "bust_mismatch",
    "waist_mismatch",
    "hip_mismatch",

    "bust_relative_mismatch",
    "waist_relative_mismatch",
    "hip_relative_mismatch",

    "mean_ease",
    "ease_std",

    "known_fit_dimensions",
    "tight_dimension_count",
    "extreme_ease_count",

    "length_height_ratio",
]

# ============================================================
# SELECT EXISTING PHYSICAL FEATURES
# ============================================================

available_physical = [
    c
    for c in PHYSICAL_FEATURES
    if c in hybrid_training.columns
]

print("\n" + "=" * 70)
print("PHYSICAL FEATURES")
print("=" * 70)

print(
    "Available:",
    len(available_physical)
)

# ============================================================
# BUILD BASE MATRICES
# ============================================================

candidate_base = hybrid_candidates[
    available_physical
].copy()

training_base = hybrid_training[
    available_physical
].copy()

# ============================================================
# FORCE NUMERIC
# ============================================================

for col in available_physical:

    candidate_base[col] = pd.to_numeric(
        candidate_base[col],
        errors="coerce",
    )

    training_base[col] = pd.to_numeric(
        training_base[col],
        errors="coerce",
    )

# ============================================================
# SEMANTIC NUMERIC FEATURES
# ============================================================

candidate_num = hybrid_candidates[
    available_num
].copy()

training_num = hybrid_training[
    available_num
].copy()

for col in available_num:

    candidate_num[col] = pd.to_numeric(
        candidate_num[col],
        errors="coerce",
    )

    training_num[col] = pd.to_numeric(
        training_num[col],
        errors="coerce",
    )

# ============================================================
# RESET INDEX
# ============================================================

candidate_base = (
    candidate_base
    .reset_index(drop=True)
)

training_base = (
    training_base
    .reset_index(drop=True)
)

candidate_num = (
    candidate_num
    .reset_index(drop=True)
)

training_num = (
    training_num
    .reset_index(drop=True)
)

candidate_cat = (
    candidate_cat
    .reset_index(drop=True)
)

training_cat = (
    training_cat
    .reset_index(drop=True)
)

# ============================================================
# COMBINE
# ============================================================

candidate_features = pd.concat(
    [
        candidate_base,
        candidate_num,
        candidate_cat,
    ],
    axis=1,
)

training_features = pd.concat(
    [
        training_base,
        training_num,
        training_cat,
    ],
    axis=1,
)

# ============================================================
# FORCE ALL FEATURES NUMERIC
# ============================================================

for col in candidate_features.columns:

    candidate_features[col] = pd.to_numeric(
        candidate_features[col],
        errors="coerce",
    )

for col in training_features.columns:

    training_features[col] = pd.to_numeric(
        training_features[col],
        errors="coerce",
    )

# ============================================================
# FINAL COLUMN ALIGNMENT
# ============================================================

candidate_features, training_features = (
    candidate_features.align(
        training_features,
        join="outer",
        axis=1,
        fill_value=0.0,
    )
)

# ============================================================
# ADD IDENTIFIERS
# ============================================================

candidate_output = pd.concat(
    [
        hybrid_candidates[
            IDENTIFIERS
        ].reset_index(drop=True),

        candidate_features.reset_index(
            drop=True
        ),
    ],
    axis=1,
)

training_output = pd.concat(
    [
        hybrid_training[
            IDENTIFIERS
            +
            LABEL_COLUMNS
        ].reset_index(drop=True),

        training_features.reset_index(
            drop=True
        ),
    ],
    axis=1,
)

# ============================================================
# FEATURE LIST
# ============================================================

HYBRID_FEATURES = [
    col
    for col in training_output.columns
    if col not in (
        IDENTIFIERS
        +
        LABEL_COLUMNS
    )
]

# ============================================================
# VALIDATION
# ============================================================

print("\n" + "=" * 70)
print("FINAL HYBRID VALIDATION")
print("=" * 70)

print(
    "Hybrid feature count:",
    len(HYBRID_FEATURES)
)

print(
    "Candidate shape:",
    candidate_output.shape
)

print(
    "Training shape:",
    training_output.shape
)

# ------------------------------------------------------------
# Numeric validation
# ------------------------------------------------------------

non_numeric_train = [
    col
    for col in HYBRID_FEATURES
    if not pd.api.types.is_numeric_dtype(
        training_output[col]
    )
]

non_numeric_candidate = [
    col
    for col in HYBRID_FEATURES
    if not pd.api.types.is_numeric_dtype(
        candidate_output[col]
    )
]

print(
    "\nNon-numeric training:",
    non_numeric_train
)

print(
    "Non-numeric candidate:",
    non_numeric_candidate
)

assert (
    len(non_numeric_train)
    == 0
)

assert (
    len(non_numeric_candidate)
    == 0
)

# ------------------------------------------------------------
# NaN / Inf
# ------------------------------------------------------------

train_matrix = (
    training_output[
        HYBRID_FEATURES
    ]
    .to_numpy(
        dtype=np.float32
    )
)

candidate_matrix = (
    candidate_output[
        HYBRID_FEATURES
    ]
    .to_numpy(
        dtype=np.float32
    )
)

train_nan = np.isnan(
    train_matrix
).sum()

candidate_nan = np.isnan(
    candidate_matrix
).sum()

train_inf = np.isinf(
    train_matrix
).sum()

candidate_inf = np.isinf(
    candidate_matrix
).sum()

print(
    "\nTraining NaN:",
    train_nan
)

print(
    "Candidate NaN:",
    candidate_nan
)

print(
    "Training Inf:",
    train_inf
)

print(
    "Candidate Inf:",
    candidate_inf
)

# NaN is allowed at this stage for physical measurements.
# It will be imputed during training.

assert train_inf == 0
assert candidate_inf == 0

# ============================================================
# DUPLICATE PAIRS
# ============================================================

candidate_duplicates = (
    candidate_output[
        [
            "person_id",
            "product_id",
        ]
    ]
    .duplicated()
    .sum()
)

training_duplicates = (
    training_output[
        [
            "person_id",
            "product_id",
        ]
    ]
    .duplicated()
    .sum()
)

print(
    "\nCandidate duplicate pairs:",
    candidate_duplicates
)

print(
    "Training duplicate pairs:",
    training_duplicates
)

assert candidate_duplicates == 0
assert training_duplicates == 0

# ============================================================
# LABEL COUNTS
# ============================================================

print("\n" + "=" * 70)
print("TRAINING LABELS")
print("=" * 70)

print(
    training_output[
        "training_label"
    ]
    .value_counts()
)

# ============================================================
# SAVE
# ============================================================

candidate_output.to_csv(
    OUTPUT_CANDIDATE_PATH,
    index=False,
)

training_output.to_csv(
    OUTPUT_TRAINING_PATH,
    index=False,
)

# ============================================================
# FINAL
# ============================================================

print("\n" + "=" * 70)
print("HYBRID DATASET SAVED")
print("=" * 70)

print(
    "Candidate:",
    OUTPUT_CANDIDATE_PATH
)

print(
    "Training:",
    OUTPUT_TRAINING_PATH
)

print(
    "\nFeature count:",
    len(HYBRID_FEATURES)
)

print(
    "\nZYRA V2 STEP 7 COMPLETE"
)
```

```python
# ============================================================
# ZYRA V2 — STEP 8
# HYBRID FULL-CATALOG RANKING EVALUATION
# ============================================================

import pandas as pd
import numpy as np
import torch
import torch.nn as nn
from pathlib import Path

# ============================================================
# PATHS
# ============================================================

BASE = Path(
    "/Users/saketh/Desktop/Projects/weavly/core-model"
)

MODEL_PATH = (
    BASE /
    "data/v2/v2_hybrid_ranker.pt"
)

HYBRID_PATH = (
    BASE /
    "data/v2/v2_hybrid_candidates.csv"
)

ORIGINAL_PATH = (
    BASE /
    "data/v2/v2_person_garment_candidates.csv"
)

RESULT_PATH = (
    BASE /
    "data/v2/v2_hybrid_ranking_results.csv"
)

# ============================================================
# LOAD
# ============================================================

print("=" * 70)
print("ZYRA V2 — HYBRID FULL-CATALOG EVALUATION")
print("=" * 70)

hybrid = pd.read_csv(
    HYBRID_PATH
)

original = pd.read_csv(
    ORIGINAL_PATH
)

checkpoint = torch.load(
    MODEL_PATH,
    map_location="cpu",
    weights_only=False,
)

FEATURES = checkpoint[
    "feature_names"
]

TEST_PEOPLE = set(
    checkpoint[
        "test_people"
    ]
)

print(
    "Hybrid rows:",
    len(hybrid)
)

print(
    "Original rows:",
    len(original)
)

print(
    "Feature count:",
    len(FEATURES)
)

print(
    "Test people:",
    sorted(TEST_PEOPLE)
)

# ============================================================
# MODEL
# ============================================================

class HybridRanker(nn.Module):

    def __init__(
        self,
        input_dim,
    ):

        super().__init__()

        self.network = nn.Sequential(

            nn.Linear(
                input_dim,
                128,
            ),

            nn.ReLU(),

            nn.BatchNorm1d(
                128
            ),

            nn.Dropout(
                0.15
            ),

            nn.Linear(
                128,
                64,
            ),

            nn.ReLU(),

            nn.Dropout(
                0.10
            ),

            nn.Linear(
                64,
                32,
            ),

            nn.ReLU(),

            nn.Linear(
                32,
                1,
            ),
        )

    def forward(self, x):

        return self.network(
            x
        ).squeeze(-1)


model = HybridRanker(
    len(FEATURES)
)

model.load_state_dict(
    checkpoint[
        "model_state_dict"
    ]
)

model.eval()

print(
    "\nModel loaded successfully."
)

# ============================================================
# RECONSTRUCT GROUND TRUTH
# ============================================================

ground_truth = original[
    [
        "person_id",
        "product_id",
        "label_status",
    ]
].copy()

# ============================================================
# VERIFY PAIRS
# ============================================================

hybrid_pairs = set(
    zip(
        hybrid[
            "person_id"
        ],
        hybrid[
            "product_id"
        ],
    )
)

truth_pairs = set(
    zip(
        ground_truth[
            "person_id"
        ],
        ground_truth[
            "product_id"
        ],
    )
)

print("\n" + "=" * 70)
print("GROUND-TRUTH ALIGNMENT")
print("=" * 70)

print(
    "Hybrid pairs:",
    len(hybrid_pairs)
)

print(
    "Ground-truth pairs:",
    len(truth_pairs)
)

print(
    "Missing pairs:",
    len(
        truth_pairs
        -
        hybrid_pairs
    )
)

print(
    "Extra pairs:",
    len(
        hybrid_pairs
        -
        truth_pairs
    )
)

assert (
    hybrid_pairs
    ==
    truth_pairs
)

# ============================================================
# MERGE GROUND TRUTH
# ============================================================

ranking_df = hybrid.merge(
    ground_truth,
    on=[
        "person_id",
        "product_id",
    ],
    how="left",
    validate="one_to_one",
)

assert (
    ranking_df[
        "label_status"
    ]
    .notna()
    .all()
)

print(
    "\nGround truth successfully attached."
)

# ============================================================
# FEATURE MATRIX
# ============================================================

X = (
    ranking_df[
        FEATURES
    ]
    .to_numpy(
        dtype=np.float32
    )
)

print("\n" + "=" * 70)
print("FEATURE MATRIX")
print("=" * 70)

print(
    "Shape:",
    X.shape
)

# ============================================================
# IMPUTATION
# ============================================================

imputer_values = np.asarray(
    checkpoint[
        "imputer_statistics"
    ],
    dtype=np.float32,
)

for i in range(
    X.shape[1]
):

    missing = np.isnan(
        X[:, i]
    )

    if missing.any():

        X[
            missing,
            i
        ] = imputer_values[i]

# ============================================================
# SCALING
# ============================================================

scaler_mean = np.asarray(
    checkpoint[
        "scaler_mean"
    ],
    dtype=np.float32,
)

scaler_scale = np.asarray(
    checkpoint[
        "scaler_scale"
    ],
    dtype=np.float32,
)

scaler_scale = np.where(
    scaler_scale == 0,
    1.0,
    scaler_scale,
)

X = (
    X
    -
    scaler_mean
)

X = (
    X
    /
    scaler_scale
)

print(
    "NaN after preprocessing:",
    np.isnan(X).sum()
)

print(
    "Inf after preprocessing:",
    np.isinf(X).sum()
)

assert np.isnan(X).sum() == 0
assert np.isinf(X).sum() == 0

# ============================================================
# SCORE ALL CANDIDATES
# ============================================================

X_tensor = torch.tensor(
    X,
    dtype=torch.float32,
)

with torch.no_grad():

    scores = (
        torch.sigmoid(
            model(
                X_tensor
            )
        )
        .numpy()
    )

# ============================================================
# ADD SCORES
# ============================================================

ranking_df = pd.concat(
    [
        ranking_df.reset_index(
            drop=True
        ),

        pd.DataFrame(
            {
                "hybrid_score":
                    scores
            }
        ),
    ],
    axis=1,
)

# ============================================================
# TEST PEOPLE ONLY
# ============================================================

ranking_df = ranking_df[
    ranking_df[
        "person_id"
    ].isin(
        TEST_PEOPLE
    )
].copy()

# ============================================================
# POSITIVE FLAG
# ============================================================

ranking_df[
    "is_positive"
] = (
    ranking_df[
        "label_status"
    ]
    ==
    "positive"
)

# ============================================================
# FULL CATALOG VALIDATION
# ============================================================

print("\n" + "=" * 70)
print("CATALOG VALIDATION")
print("=" * 70)

catalog_counts = (
    ranking_df
    .groupby(
        "person_id"
    )[
        "product_id"
    ]
    .nunique()
)

print(
    catalog_counts
)

assert (
    catalog_counts == 897
).all()

# ============================================================
# RANK
# ============================================================

ranking_df[
    "rank"
] = (
    ranking_df
    .groupby(
        "person_id"
    )[
        "hybrid_score"
    ]
    .rank(
        method="first",
        ascending=False,
    )
)

# ============================================================
# METRIC FUNCTIONS
# ============================================================

def recall_at_k(
    ranks,
    k,
):

    if len(ranks) == 0:

        return 0.0

    return float(
        np.mean(
            ranks <= k
        )
    )


def ndcg_at_k(
    ranks,
    k,
):

    if len(ranks) == 0:

        return 0.0

    dcg = sum(

        1.0
        /
        np.log2(
            rank + 1
        )

        for rank in ranks

        if rank <= k
    )

    ideal_count = min(
        len(ranks),
        k,
    )

    if ideal_count == 0:

        return 0.0

    idcg = sum(

        1.0
        /
        np.log2(
            i + 2
        )

        for i in range(
            ideal_count
        )
    )

    return float(
        dcg / idcg
    )


def reciprocal_rank(
    ranks
):

    if len(ranks) == 0:

        return 0.0

    return float(
        1.0
        /
        np.min(ranks)
    )

# ============================================================
# PER PERSON
# ============================================================

person_results = []

for person_id, group in (
    ranking_df
    .groupby(
        "person_id"
    )
):

    positive_ranks = (
        group[
            group[
                "is_positive"
            ]
        ][
            "rank"
        ]
        .to_numpy()
    )

    person_results.append(
        {
            "person_id":
                person_id,

            "positive_count":
                len(
                    positive_ranks
                ),

            "recall@5":
                recall_at_k(
                    positive_ranks,
                    5,
                ),

            "recall@10":
                recall_at_k(
                    positive_ranks,
                    10,
                ),

            "recall@20":
                recall_at_k(
                    positive_ranks,
                    20,
                ),

            "recall@50":
                recall_at_k(
                    positive_ranks,
                    50,
                ),

            "recall@100":
                recall_at_k(
                    positive_ranks,
                    100,
                ),

            "ndcg@10":
                ndcg_at_k(
                    positive_ranks,
                    10,
                ),

            "ndcg@20":
                ndcg_at_k(
                    positive_ranks,
                    20,
                ),

            "mrr":
                reciprocal_rank(
                    positive_ranks
                ),

            "best_rank":
                (
                    np.min(
                        positive_ranks
                    )
                    if len(
                        positive_ranks
                    )
                    else np.nan
                ),

            "median_rank":
                (
                    np.median(
                        positive_ranks
                    )
                    if len(
                        positive_ranks
                    )
                    else np.nan
                ),
        }
    )

person_results = pd.DataFrame(
    person_results
)

# ============================================================
# GLOBAL METRICS
# ============================================================

print("\n" + "=" * 70)
print("GLOBAL HYBRID RANKING RESULTS")
print("=" * 70)

metrics = [
    "recall@5",
    "recall@10",
    "recall@20",
    "recall@50",
    "recall@100",
    "ndcg@10",
    "ndcg@20",
    "mrr",
]

for metric in metrics:

    print(
        f"{metric:15s}: "
        f"{person_results[metric].mean():.4f}"
    )

# ============================================================
# ALL POSITIVE RANKS
# ============================================================

positive_ranks = (
    ranking_df[
        ranking_df[
            "is_positive"
        ]
    ][
        "rank"
    ]
    .to_numpy()
)

print("\n" + "=" * 70)
print("POSITIVE RANK DISTRIBUTION")
print("=" * 70)

print(
    pd.Series(
        positive_ranks
    ).describe(
        percentiles=[
            0.25,
            0.50,
            0.75,
            0.90,
            0.95,
        ]
    )
)

# ============================================================
# TOP-K HIT COUNTS
# ============================================================

print("\n" + "=" * 70)
print("KNOWN POSITIVES IN TOP-K")
print("=" * 70)

total_positive = len(
    positive_ranks
)

for k in [
    1,
    5,
    10,
    20,
    50,
    100,
]:

    hits = np.sum(
        positive_ranks <= k
    )

    percentage = (
        hits
        /
        total_positive
        *
        100
    )

    print(
        f"Top-{k:3d}: "
        f"{hits:4d} / "
        f"{total_positive:4d} "
        f"({percentage:6.2f}%)"
    )

# ============================================================
# PER-PERSON RESULTS
# ============================================================

print("\n" + "=" * 70)
print("PER-PERSON RESULTS")
print("=" * 70)

print(
    person_results[
        [
            "person_id",
            "positive_count",
            "recall@10",
            "recall@20",
            "recall@100",
            "ndcg@10",
            "ndcg@20",
            "mrr",
            "best_rank",
            "median_rank",
        ]
    ]
    .sort_values(
        "person_id"
    )
    .to_string(
        index=False
    )
)

# ============================================================
# TOP 10 RECOMMENDATIONS
# ============================================================

print("\n" + "=" * 70)
print("TOP 10 HYBRID RECOMMENDATIONS")
print("=" * 70)

for person_id, group in (
    ranking_df
    .groupby(
        "person_id"
    )
):

    top10 = (
        group
        .sort_values(
            "rank"
        )
        .head(10)
    )

    print(
        f"\nPerson: {person_id}"
    )

    display(
        top10[
            [
                "rank",
                "product_id",
                "hybrid_score",
                "is_positive",
            ]
        ]
    )

# ============================================================
# SAVE DETAILED RANKINGS
# ============================================================

ranking_df.to_csv(
    RESULT_PATH,
    index=False,
)

print("\n" + "=" * 70)
print("EVALUATION SAVED")
print("=" * 70)

print(
    RESULT_PATH
)

print("\n" + "=" * 70)
print("ZYRA V2 STEP 8 COMPLETE")
print("=" * 70)
```

```python
# ============================================================
# ZYRA V2 — STEP 9
# PERSON-SPECIFIC PREFERENCE PROFILES
# ============================================================

import pandas as pd
import numpy as np
from pathlib import Path

# ============================================================
# PATHS
# ============================================================

BASE = Path(
    "/Users/saketh/Desktop/Projects/weavly/core-model"
)

TRAINING_PATH = (
    BASE /
    "data/v2/v2_confident_training_dataset.csv"
)

SEMANTIC_PATH = (
    BASE /
    "data/v2/v2_garment_semantic_features.csv"
)

OUTPUT_PATH = (
    BASE /
    "data/v2/v2_person_preference_profiles.csv"
)

# ============================================================
# LOAD
# ============================================================

training = pd.read_csv(
    TRAINING_PATH
)

semantic = pd.read_csv(
    SEMANTIC_PATH
)

print("=" * 70)
print("ZYRA V2 — STEP 9 PERSON PREFERENCE PROFILES")
print("=" * 70)

# ============================================================
# POSITIVE INTERACTIONS ONLY
# ============================================================

positive = training[
    training[
        "training_label"
    ] == 1
].copy()

print(
    "Positive interactions:",
    len(positive)
)

print(
    "People:",
    positive[
        "person_id"
    ].nunique()
)

# ============================================================
# MERGE PRODUCT SEMANTICS
# ============================================================

semantic_columns = [
    "product_id",
    "category",
    "style",
    "pattern",
    "material",
    "silhouette",
    "sleeve_type",
    "neckline",
    "length_style",
    "color",
]

semantic_subset = semantic[
    [
        c
        for c in semantic_columns
        if c in semantic.columns
    ]
].copy()

positive = positive.merge(
    semantic_subset,
    on="product_id",
    how="left",
    validate="many_to_one",
)

# ============================================================
# UNKNOWN NORMALIZATION
# ============================================================

profile_columns = [
    "category",
    "style",
    "pattern",
    "material",
    "silhouette",
    "sleeve_type",
    "neckline",
    "length_style",
    "color",
]

for col in profile_columns:

    if col in positive.columns:

        positive[col] = (
            positive[col]
            .fillna("unknown")
            .astype(str)
            .str.lower()
            .str.strip()
        )

# ============================================================
# CREATE PROFILE ROWS
# ============================================================

profile_rows = []

for person_id, group in (
    positive.groupby(
        "person_id"
    )
):

    row = {
        "person_id":
            person_id,

        "positive_count":
            len(group),
    }

    # --------------------------------------------------------
    # CATEGORY
    # --------------------------------------------------------

    for column in profile_columns:

        if column not in group.columns:

            continue

        counts = (
            group[
                column
            ]
            .value_counts(
                normalize=True
            )
        )

        # Top preference
        if len(counts):

            row[
                f"{column}_top"
            ] = counts.index[0]

            row[
                f"{column}_top_share"
            ] = counts.iloc[0]

        else:

            row[
                f"{column}_top"
            ] = "unknown"

            row[
                f"{column}_top_share"
            ] = 0.0

        # ----------------------------------------------------
        # TOP 3 DISTRIBUTION
        # ----------------------------------------------------

        top3 = counts.head(3)

        for rank in range(
            3
        ):

            if rank < len(top3):

                value = (
                    top3.index[
                        rank
                    ]
                )

                share = (
                    top3.iloc[
                        rank
                    ]
                )

            else:

                value = "unknown"
                share = 0.0

            row[
                f"{column}_{rank+1}"
            ] = value

            row[
                f"{column}_{rank+1}_share"
            ] = share

    profile_rows.append(
        row
    )

# ============================================================
# DATAFRAME
# ============================================================

profiles = pd.DataFrame(
    profile_rows
)

# ============================================================
# DISPLAY
# ============================================================

print("\n" + "=" * 70)
print("PERSON PROFILES")
print("=" * 70)

print(
    "Profiles:",
    len(profiles)
)

print(
    "Columns:",
    len(profiles.columns)
)

display(
    profiles
)

# ============================================================
# DETAILED PROFILE REPORT
# ============================================================

print("\n" + "=" * 70)
print("DETAILED PREFERENCE PROFILES")
print("=" * 70)

for _, row in profiles.iterrows():

    print(
        f"\n{'=' * 50}"
    )

    print(
        "PERSON:",
        row[
            "person_id"
        ]
    )

    print(
        "Positive interactions:",
        row[
            "positive_count"
        ]
    )

    for column in profile_columns:

        top = row.get(
            f"{column}_top",
            "unknown",
        )

        share = row.get(
            f"{column}_top_share",
            0.0,
        )

        top2 = row.get(
            f"{column}_2",
            "unknown",
        )

        top2_share = row.get(
            f"{column}_2_share",
            0.0,
        )

        top3 = row.get(
            f"{column}_3",
            "unknown",
        )

        top3_share = row.get(
            f"{column}_3_share",
            0.0,
        )

        print(
            f"{column:15s}: "
            f"{top} ({share:.2f}) | "
            f"{top2} ({top2_share:.2f}) | "
            f"{top3} ({top3_share:.2f})"
        )

# ============================================================
# PROFILE DIVERSITY
# ============================================================

print("\n" + "=" * 70)
print("PROFILE DIVERSITY")
print("=" * 70)

for column in profile_columns:

    if column not in positive.columns:

        continue

    diversity = (
        positive
        .groupby(
            "person_id"
        )[column]
        .nunique()
    )

    print(
        f"\n{column}"
    )

    print(
        diversity.describe()
        .to_string()
    )

# ============================================================
# PREFERENCE ENTROPY
# ============================================================

print("\n" + "=" * 70)
print("PREFERENCE ENTROPY")
print("=" * 70)

def entropy(
    values
):

    probabilities = (
        values
        .value_counts(
            normalize=True
        )
        .to_numpy()
    )

    if len(
        probabilities
    ) == 0:

        return 0.0

    return float(
        -np.sum(
            probabilities
            *
            np.log2(
                probabilities
            )
        )
    )


entropy_rows = []

for person_id, group in (
    positive.groupby(
        "person_id"
    )
):

    result = {
        "person_id":
            person_id
    }

    for column in profile_columns:

        if column in group.columns:

            result[
                f"{column}_entropy"
            ] = entropy(
                group[
                    column
                ]
            )

    entropy_rows.append(
        result
    )

entropy_df = pd.DataFrame(
    entropy_rows
)

display(
    entropy_df
)

# ============================================================
# SAVE
# ============================================================

profiles.to_csv(
    OUTPUT_PATH,
    index=False,
)

print("\n" + "=" * 70)
print("PREFERENCE PROFILES SAVED")
print("=" * 70)

print(
    OUTPUT_PATH
)

print("\n" + "=" * 70)
print("ZYRA V2 STEP 9 COMPLETE")
print("=" * 70)
```

```python
# ============================================================
# ZYRA V2 — STEP 10
# PERSONALIZED PREFERENCE + PHYSICAL BASELINE
# ============================================================

import pandas as pd
import numpy as np
from pathlib import Path

# ============================================================
# PATHS
# ============================================================

BASE = Path(
    "/Users/saketh/Desktop/Projects/weavly/core-model"
)

CANDIDATE_PATH = (
    BASE /
    "data/v2/v2_person_garment_candidates.csv"
)

SEMANTIC_PATH = (
    BASE /
    "data/v2/v2_garment_semantic_features.csv"
)

PROFILE_PATH = (
    BASE /
    "data/v2/v2_person_preference_profiles.csv"
)

MODEL_PATH = (
    BASE /
    "data/v2/v2_hybrid_ranker.pt"
)

OUTPUT_PATH = (
    BASE /
    "data/v2/v2_personalized_baseline_results.csv"
)

# ============================================================
# LOAD
# ============================================================

candidates = pd.read_csv(
    CANDIDATE_PATH
)

semantic = pd.read_csv(
    SEMANTIC_PATH
)

profiles = pd.read_csv(
    PROFILE_PATH
)

print("=" * 70)
print("ZYRA V2 — STEP 10 PERSONALIZED BASELINE")
print("=" * 70)

print(
    "Candidate rows:",
    len(candidates)
)

print(
    "Garments:",
    semantic["product_id"].nunique()
)

print(
    "Profiles:",
    profiles["person_id"].nunique()
)

# ============================================================
# TEST PEOPLE
# ============================================================

# Load the exact held-out people used by V2.
# This prevents accidentally evaluating on training people.

import torch

checkpoint = torch.load(
    MODEL_PATH,
    map_location="cpu",
    weights_only=False,
)

TEST_PEOPLE = set(
    checkpoint[
        "test_people"
    ]
)

print(
    "\nTest people:",
    sorted(TEST_PEOPLE)
)

# ============================================================
# MERGE SEMANTICS
# ============================================================

semantic_columns = [
    "product_id",
    "category",
    "style",
    "pattern",
    "material",
    "silhouette",
    "sleeve_type",
    "neckline",
    "length_style",
    "color",
]

semantic_subset = semantic[
    [
        c
        for c in semantic_columns
        if c in semantic.columns
    ]
].copy()

df = candidates.merge(
    semantic_subset,
    on="product_id",
    how="left",
    validate="many_to_one",
)

# ============================================================
# NORMALIZE
# ============================================================

PROFILE_COLUMNS = [
    "category",
    "style",
    "pattern",
    "material",
    "silhouette",
    "sleeve_type",
    "neckline",
    "length_style",
    "color",
]

for col in PROFILE_COLUMNS:

    df[col] = (
        df[col]
        .fillna("unknown")
        .astype(str)
        .str.lower()
        .str.strip()
    )

# ============================================================
# PERSON PROFILE LOOKUP
# ============================================================

profile_lookup = (
    profiles
    .set_index(
        "person_id"
    )
)

# ============================================================
# PREFERENCE SIMILARITY
# ============================================================
#
# For each person × garment:
#
# score = average agreement with that person's
# top preferences.
#
# We deliberately use TOP-3 preference distributions,
# not only the single most common value.
#
# ============================================================

def preference_score(
    person_id,
    garment,
):

    if person_id not in profile_lookup.index:

        return 0.0

    profile = profile_lookup.loc[
        person_id
    ]

    scores = []

    for column in PROFILE_COLUMNS:

        value = garment[column]

        if value == "unknown":

            continue

        # ----------------------------------------------------
        # TOP 1
        # ----------------------------------------------------

        top1 = str(
            profile.get(
                f"{column}_1",
                "unknown",
            )
        )

        share1 = float(
            profile.get(
                f"{column}_1_share",
                0.0,
            )
        )

        if value == top1:

            scores.append(
                share1
            )

            continue

        # ----------------------------------------------------
        # TOP 2
        # ----------------------------------------------------

        top2 = str(
            profile.get(
                f"{column}_2",
                "unknown",
            )
        )

        share2 = float(
            profile.get(
                f"{column}_2_share",
                0.0,
            )
        )

        if value == top2:

            scores.append(
                share2
            )

            continue

        # ----------------------------------------------------
        # TOP 3
        # ----------------------------------------------------

        top3 = str(
            profile.get(
                f"{column}_3",
                "unknown",
            )
        )

        share3 = float(
            profile.get(
                f"{column}_3_share",
                0.0,
            )
        )

        if value == top3:

            scores.append(
                share3
            )

        else:

            scores.append(
                0.0
            )

    if len(scores) == 0:

        return 0.0

    return float(
        np.mean(
            scores
        )
    )

# ============================================================
# CALCULATE PREFERENCE SCORES
# ============================================================

print("\n" + "=" * 70)
print("CALCULATING PERSONALIZED PREFERENCE SCORES")
print("=" * 70)

preference_scores = []

for row in df.itertuples(
    index=False
):

    garment = {
        column:
            getattr(
                row,
                column,
            )
        for column in PROFILE_COLUMNS
    }

    score = preference_score(
        row.person_id,
        garment,
    )

    preference_scores.append(
        score
    )

df["preference_score"] = (
    preference_scores
)

# ============================================================
# PHYSICAL COMPATIBILITY
# ============================================================
#
# We use the existing physical compatibility features.
#
# Rather than relying on one arbitrary raw column, create
# a normalized physical compatibility score.
#
# Lower mismatch/ease violations = better.
#
# ============================================================

physical_score_columns = [
    "bust_mismatch",
    "waist_mismatch",
    "hip_mismatch",
    "bust_relative_mismatch",
    "waist_relative_mismatch",
    "hip_relative_mismatch",
]

available_physical = [
    c
    for c in physical_score_columns
    if c in df.columns
]

print(
    "\nPhysical features:",
    available_physical
)

physical_matrix = (
    df[
        available_physical
    ]
    .apply(
        pd.to_numeric,
        errors="coerce",
    )
    .to_numpy(
        dtype=np.float64
    )
)

# ------------------------------------------------------------
# Robust normalization
# ------------------------------------------------------------

physical_components = []

for i in range(
    physical_matrix.shape[1]
):

    values = physical_matrix[
        :,
        i
    ]

    median = np.nanmedian(
        values
    )

    q25 = np.nanpercentile(
        values,
        25
    )

    q75 = np.nanpercentile(
        values,
        75
    )

    iqr = q75 - q25

    if (
        not np.isfinite(iqr)
        or
        iqr < 1e-8
    ):

        iqr = 1.0

    z = (
        values
        -
        median
    ) / iqr

    z = np.nan_to_num(
        z,
        nan=0.0,
        posinf=5.0,
        neginf=-5.0,
    )

    # Smaller mismatch = better.
    compatibility = np.exp(
        -np.abs(z)
    )

    physical_components.append(
        compatibility
    )

physical_components = np.column_stack(
    physical_components
)

df[
    "physical_score"
] = np.mean(
    physical_components,
    axis=1,
)

# ============================================================
# SCORE WEIGHTS
# ============================================================
#
# Start with equal contribution.
#
# This is a BASELINE, not the final model.
# ============================================================

PHYSICAL_WEIGHT = 0.50
PREFERENCE_WEIGHT = 0.50

df[
    "personalized_score"
] = (
    PHYSICAL_WEIGHT
    *
    df[
        "physical_score"
    ]
    +
    PREFERENCE_WEIGHT
    *
    df[
        "preference_score"
    ]
)

# ============================================================
# TEST PEOPLE
# ============================================================

test_df = df[
    df[
        "person_id"
    ].isin(
        TEST_PEOPLE
    )
].copy()

# ============================================================
# FULL CATALOG VALIDATION
# ============================================================

print("\n" + "=" * 70)
print("CATALOG VALIDATION")
print("=" * 70)

catalog_counts = (
    test_df
    .groupby(
        "person_id"
    )[
        "product_id"
    ]
    .nunique()
)

print(
    catalog_counts
)

assert (
    catalog_counts == 897
).all()

# ============================================================
# POSITIVE FLAG
# ============================================================

test_df[
    "is_positive"
] = (
    test_df[
        "label_status"
    ]
    ==
    "positive"
)

# ============================================================
# RANK
# ============================================================

test_df[
    "rank"
] = (
    test_df
    .groupby(
        "person_id"
    )[
        "personalized_score"
    ]
    .rank(
        method="first",
        ascending=False,
    )
)

# ============================================================
# METRICS
# ============================================================

def recall_at_k(
    ranks,
    k,
):

    if len(ranks) == 0:

        return 0.0

    return float(
        np.mean(
            ranks <= k
        )
    )


def ndcg_at_k(
    ranks,
    k,
):

    if len(ranks) == 0:

        return 0.0

    dcg = sum(

        1.0
        /
        np.log2(
            rank + 1
        )

        for rank in ranks

        if rank <= k
    )

    ideal_count = min(
        len(ranks),
        k,
    )

    if ideal_count == 0:

        return 0.0

    idcg = sum(

        1.0
        /
        np.log2(
            i + 2
        )

        for i in range(
            ideal_count
        )
    )

    return float(
        dcg / idcg
    )


def mrr(
    ranks
):

    if len(ranks) == 0:

        return 0.0

    return float(
        1.0
        /
        np.min(ranks)
    )

# ============================================================
# PER PERSON
# ============================================================

results = []

for person_id, group in (
    test_df
    .groupby(
        "person_id"
    )
):

    ranks = (
        group[
            group[
                "is_positive"
            ]
        ][
            "rank"
        ]
        .to_numpy()
    )

    results.append(
        {
            "person_id":
                person_id,

            "positive_count":
                len(ranks),

            "recall@5":
                recall_at_k(
                    ranks,
                    5,
                ),

            "recall@10":
                recall_at_k(
                    ranks,
                    10,
                ),

            "recall@20":
                recall_at_k(
                    ranks,
                    20,
                ),

            "recall@50":
                recall_at_k(
                    ranks,
                    50,
                ),

            "recall@100":
                recall_at_k(
                    ranks,
                    100,
                ),

            "ndcg@10":
                ndcg_at_k(
                    ranks,
                    10,
                ),

            "ndcg@20":
                ndcg_at_k(
                    ranks,
                    20,
                ),

            "mrr":
                mrr(
                    ranks
                ),

            "best_rank":
                np.min(ranks),

            "median_rank":
                np.median(ranks),
        }
    )

results_df = pd.DataFrame(
    results
)

# ============================================================
# GLOBAL RESULTS
# ============================================================

print("\n" + "=" * 70)
print("PERSONALIZED BASELINE RESULTS")
print("=" * 70)

for metric in [
    "recall@5",
    "recall@10",
    "recall@20",
    "recall@50",
    "recall@100",
    "ndcg@10",
    "ndcg@20",
    "mrr",
]:

    print(
        f"{metric:15s}: "
        f"{results_df[metric].mean():.4f}"
    )

# ============================================================
# POSITIVE RANKS
# ============================================================

positive_ranks = (
    test_df[
        test_df[
            "is_positive"
        ]
    ][
        "rank"
    ]
    .to_numpy()
)

print("\n" + "=" * 70)
print("POSITIVE RANK DISTRIBUTION")
print("=" * 70)

print(
    pd.Series(
        positive_ranks
    ).describe(
        percentiles=[
            0.25,
            0.50,
            0.75,
            0.90,
            0.95,
        ]
    )
)

# ============================================================
# TOP-K
# ============================================================

print("\n" + "=" * 70)
print("KNOWN POSITIVES IN TOP-K")
print("=" * 70)

for k in [
    1,
    5,
    10,
    20,
    50,
    100,
]:

    hits = np.sum(
        positive_ranks <= k
    )

    total = len(
        positive_ranks
    )

    pct = (
        hits
        /
        total
        *
        100
    )

    print(
        f"Top-{k:3d}: "
        f"{hits:4d} / "
        f"{total:4d} "
        f"({pct:6.2f}%)"
    )

# ============================================================
# PER-PERSON
# ============================================================

print("\n" + "=" * 70)
print("PER-PERSON RESULTS")
print("=" * 70)

print(
    results_df[
        [
            "person_id",
            "positive_count",
            "recall@10",
            "recall@20",
            "recall@100",
            "ndcg@10",
            "mrr",
            "best_rank",
            "median_rank",
        ]
    ]
    .sort_values(
        "person_id"
    )
    .to_string(
        index=False
    )
)

# ============================================================
# TOP 10
# ============================================================

print("\n" + "=" * 70)
print("TOP 10 PERSONALIZED RECOMMENDATIONS")
print("=" * 70)

for person_id, group in (
    test_df
    .groupby(
        "person_id"
    )
):

    top10 = (
        group
        .sort_values(
            "rank"
        )
        .head(10)
    )

    print(
        f"\nPerson: {person_id}"
    )

    display(
        top10[
            [
                "rank",
                "product_id",
                "personalized_score",
                "physical_score",
                "preference_score",
                "is_positive",
            ]
        ]
    )

# ============================================================
# SAVE
# ============================================================

test_df.to_csv(
    OUTPUT_PATH,
    index=False,
)

print("\n" + "=" * 70)
print("PERSONALIZED BASELINE SAVED")
print("=" * 70)

print(
    OUTPUT_PATH
)

print("\n" + "=" * 70)
print("ZYRA V2 STEP 10 COMPLETE")
print("=" * 70)
```

```python
# ============================================================
# ZYRA V2 — STEP 11
# ACTUAL RECOMMENDATION INFERENCE TEST
# ============================================================

import pandas as pd
import numpy as np
import torch
import torch.nn as nn
from pathlib import Path

# ============================================================
# PATHS
# ============================================================

BASE = Path(
    "/Users/saketh/Desktop/Projects/weavly/core-model"
)

MODEL_PATH = (
    BASE /
    "data/v2/v2_hybrid_ranker.pt"
)

CANDIDATE_PATH = (
    BASE /
    "data/v2/v2_hybrid_candidates.csv"
)

ORIGINAL_PATH = (
    BASE /
    "data/v2/v2_person_garment_candidates.csv"
)

OUTPUT_PATH = (
    BASE /
    "data/v2/v2_actual_inference_results.csv"
)

# ============================================================
# LOAD MODEL
# ============================================================

print("=" * 70)
print("ZYRA V2 — ACTUAL RECOMMENDATION INFERENCE")
print("=" * 70)

checkpoint = torch.load(
    MODEL_PATH,
    map_location="cpu",
    weights_only=False,
)

FEATURES = checkpoint[
    "feature_names"
]

TEST_PEOPLE = checkpoint[
    "test_people"
]

print(
    "Model features:",
    len(FEATURES)
)

print(
    "Test people:",
    TEST_PEOPLE
)

# ============================================================
# MODEL
# ============================================================

class HybridRanker(nn.Module):

    def __init__(
        self,
        input_dim,
    ):

        super().__init__()

        self.network = nn.Sequential(

            nn.Linear(
                input_dim,
                128,
            ),

            nn.ReLU(),

            nn.BatchNorm1d(
                128,
            ),

            nn.Dropout(
                0.15,
            ),

            nn.Linear(
                128,
                64,
            ),

            nn.ReLU(),

            nn.Dropout(
                0.10,
            ),

            nn.Linear(
                64,
                32,
            ),

            nn.ReLU(),

            nn.Linear(
                32,
                1,
            ),
        )

    def forward(
        self,
        x,
    ):

        return self.network(
            x
        ).squeeze(-1)


model = HybridRanker(
    len(FEATURES)
)

model.load_state_dict(
    checkpoint[
        "model_state_dict"
    ]
)

model.eval()

print(
    "\nModel loaded successfully."
)

# ============================================================
# LOAD CANDIDATES
# ============================================================

candidates = pd.read_csv(
    CANDIDATE_PATH
)

original = pd.read_csv(
    ORIGINAL_PATH
)

print(
    "\nCandidate rows:",
    len(candidates)
)

print(
    "Original rows:",
    len(original)
)

# ============================================================
# GROUND TRUTH
# ============================================================

ground_truth = original[
    [
        "person_id",
        "product_id",
        "label_status",
    ]
].copy()

# ============================================================
# MERGE LABELS
# ============================================================

df = candidates.merge(
    ground_truth,
    on=[
        "person_id",
        "product_id",
    ],
    how="left",
    validate="one_to_one",
)

assert (
    df[
        "label_status"
    ]
    .notna()
    .all()
)

# ============================================================
# FEATURE MATRIX
# ============================================================

X = (
    df[
        FEATURES
    ]
    .to_numpy(
        dtype=np.float32
    )
)

# ============================================================
# IMPUTATION
# ============================================================

imputer_values = np.asarray(
    checkpoint[
        "imputer_statistics"
    ],
    dtype=np.float32,
)

for i in range(
    X.shape[1]
):

    missing = np.isnan(
        X[:, i]
    )

    if missing.any():

        X[
            missing,
            i
        ] = imputer_values[i]

# ============================================================
# SCALING
# ============================================================

scaler_mean = np.asarray(
    checkpoint[
        "scaler_mean"
    ],
    dtype=np.float32,
)

scaler_scale = np.asarray(
    checkpoint[
        "scaler_scale"
    ],
    dtype=np.float32,
)

scaler_scale = np.where(
    scaler_scale == 0,
    1.0,
    scaler_scale,
)

X = (
    X
    -
    scaler_mean
)

X = (
    X
    /
    scaler_scale
)

# ============================================================
# VALIDATE
# ============================================================

assert (
    np.isnan(X).sum()
    == 0
)

assert (
    np.isinf(X).sum()
    == 0
)

# ============================================================
# MODEL INFERENCE
# ============================================================

with torch.no_grad():

    tensor = torch.tensor(
        X,
        dtype=torch.float32,
    )

    scores = (
        torch.sigmoid(
            model(
                tensor
            )
        )
        .numpy()
    )

df["zyra_score"] = scores

# ============================================================
# TEST PEOPLE
# ============================================================

df = df[
    df[
        "person_id"
    ].isin(
        TEST_PEOPLE
    )
].copy()

# ============================================================
# RANK
# ============================================================

df["rank"] = (
    df
    .groupby(
        "person_id"
    )[
        "zyra_score"
    ]
    .rank(
        method="first",
        ascending=False,
    )
)

df["is_positive"] = (
    df[
        "label_status"
    ]
    == "positive"
)

# ============================================================
# VERIFY CATALOG
# ============================================================

catalog_counts = (
    df
    .groupby(
        "person_id"
    )[
        "product_id"
    ]
    .nunique()
)

print("\n" + "=" * 70)
print("CATALOG")
print("=" * 70)

print(
    catalog_counts
)

assert (
    catalog_counts == 897
).all()

# ============================================================
# ACTUAL RECOMMENDATIONS
# ============================================================

print("\n" + "=" * 70)
print("ZYRA TOP-10 RECOMMENDATIONS")
print("=" * 70)

for person_id in TEST_PEOPLE:

    person_df = (
        df[
            df[
                "person_id"
            ]
            ==
            person_id
        ]
        .sort_values(
            "rank"
        )
    )

    top10 = (
        person_df
        .head(10)
        .copy()
    )

    print(
        "\n" + "-" * 70
    )

    print(
        f"PERSON: {person_id}"
    )

    print(
        "-" * 70
    )

    display(
        top10[
            [
                "rank",
                "product_id",
                "zyra_score",
                "is_positive",
            ]
        ]
    )

# ============================================================
# TOP-K POSITIVE COVERAGE
# ============================================================

print("\n" + "=" * 70)
print("ACTUAL RECOMMENDATION COVERAGE")
print("=" * 70)

all_positive_ranks = (
    df[
        df[
            "is_positive"
        ]
    ][
        "rank"
    ]
    .to_numpy()
)

total = len(
    all_positive_ranks
)

for k in [
    1,
    5,
    10,
    20,
    50,
    100,
]:

    hits = np.sum(
        all_positive_ranks <= k
    )

    print(
        f"Top-{k:3d}: "
        f"{hits:4d}/{total:4d} "
        f"("
        f"{100 * hits / total:.2f}%"
        f")"
    )

# ============================================================
# BEST POSITIVE PER PERSON
# ============================================================

print("\n" + "=" * 70)
print("BEST POSITIVE RANK PER PERSON")
print("=" * 70)

for person_id in TEST_PEOPLE:

    person_positive = (
        df[
            (
                df[
                    "person_id"
                ]
                ==
                person_id
            )
            &
            (
                df[
                    "is_positive"
                ]
            )
        ]
    )

    if len(
        person_positive
    ) == 0:

        print(
            person_id,
            ": NO POSITIVES"
        )

        continue

    best = (
        person_positive[
            "rank"
        ]
        .min()
    )

    median = (
        person_positive[
            "rank"
        ]
        .median()
    )

    print(
        f"{person_id:15s} "
        f"best={best:.0f} "
        f"median={median:.0f}"
    )

# ============================================================
# SCORE DISTRIBUTION
# ============================================================

print("\n" + "=" * 70)
print("SCORE DISTRIBUTION")
print("=" * 70)

print(
    df[
        "zyra_score"
    ].describe()
)

# ============================================================
# SAVE
# ============================================================

df.to_csv(
    OUTPUT_PATH,
    index=False,
)

print("\n" + "=" * 70)
print("INFERENCE RESULTS SAVED")
print("=" * 70)

print(
    OUTPUT_PATH
)

print("\n" + "=" * 70)
print("ZYRA V2 STEP 11 COMPLETE")
print("=" * 70)
```
